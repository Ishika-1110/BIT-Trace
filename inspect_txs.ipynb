```python
import pandas as pd
from pathlib import Path

PROJECT_ROOT = Path(r"C:\Users\YASHIKA\Desktop\bit-trace")

TX_FEATURES = (
    PROJECT_ROOT
    / "EllipticPlusPlus"
    / "Transactions Dataset"
    / "txs_features.csv"
)

print("Loading:")
print(TX_FEATURES)

df = pd.read_csv(TX_FEATURES)

print("\nShape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())
```

```python
# ============================================================
# STEP 1 — INVESTIGATE THE 965 INCOMPLETE TRANSACTIONS
# ============================================================

# The 17 transaction-level statistical columns
transaction_stat_columns = [
    "in_txs_degree",
    "out_txs_degree",
    "total_BTC",
    "fees",
    "size",
    "num_input_addresses",
    "num_output_addresses",
    "in_BTC_min",
    "in_BTC_max",
    "in_BTC_mean",
    "in_BTC_median",
    "in_BTC_total",
    "out_BTC_min",
    "out_BTC_max",
    "out_BTC_mean",
    "out_BTC_median",
    "out_BTC_total"
]

# Find transactions where all 17 transaction-stat columns are missing
affected_mask = df[transaction_stat_columns].isna().all(axis=1)

affected_df = df.loc[affected_mask].copy()

print("Total affected transactions:", len(affected_df))

# ------------------------------------------------------------
# Check missing values in the OTHER feature groups
# ------------------------------------------------------------

local_columns = [
    col for col in df.columns
    if col.startswith("Local_feature_")
]

aggregate_columns = [
    col for col in df.columns
    if col.startswith("Aggregate_feature_")
]

print("\nLocal features:", len(local_columns))
print("Aggregate features:", len(aggregate_columns))

print("\nMissing Local-feature values in affected transactions:")
print(
    affected_df[local_columns]
    .isna()
    .sum()
    .sum()
)

print("\nMissing Aggregate-feature values in affected transactions:")
print(
    affected_df[aggregate_columns]
    .isna()
    .sum()
    .sum()
)

# ------------------------------------------------------------
# Check txId and Time step
# ------------------------------------------------------------

print("\nMissing txId values:")
print(affected_df["txId"].isna().sum())

print("\nMissing Time step values:")
print(affected_df["Time step"].isna().sum())
```

```python
# ============================================================
# STEP 2 — CHECK EXTREME VALUES
# ============================================================

numeric_columns = df.select_dtypes(include="number").columns

outlier_summary = pd.DataFrame({
    "min": df[numeric_columns].min(),
    "25%": df[numeric_columns].quantile(0.25),
    "median": df[numeric_columns].median(),
    "75%": df[numeric_columns].quantile(0.75),
    "max": df[numeric_columns].max()
})

# Range between Q1 and Q3
outlier_summary["IQR"] = (
    outlier_summary["75%"] - outlier_summary["25%"]
)

# IQR-based boundaries
outlier_summary["lower_bound"] = (
    outlier_summary["25%"] - 1.5 * outlier_summary["IQR"]
)

outlier_summary["upper_bound"] = (
    outlier_summary["75%"] + 1.5 * outlier_summary["IQR"]
)

# Count values outside IQR boundaries
outlier_summary["outlier_count"] = 0

for col in numeric_columns:
    lower = outlier_summary.loc[col, "lower_bound"]
    upper = outlier_summary.loc[col, "upper_bound"]

    outlier_summary.loc[col, "outlier_count"] = (
        ((df[col] < lower) | (df[col] > upper))
        .sum()
    )

display(
    outlier_summary.sort_values(
        "outlier_count",
        ascending=False
    ).head(30)
)
```

```python
import matplotlib.pyplot as plt

features_to_check = [
    "out_txs_degree",
    "num_input_addresses",
    "total_BTC",
    "fees",
    "size",
    "Aggregate_feature_72",
    "Local_feature_4"
]

for col in features_to_check:
    plt.figure(figsize=(8, 4))
    df[col].dropna().hist(bins=50)
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Number of transactions")
    plt.show()
```

```python
# ============================================================
# STEP 4 — FEATURE CORRELATION ANALYSIS
# ============================================================

# Select only feature columns
feature_cols = [
    col for col in df.columns
    if col.startswith("Local_feature_")
    or col.startswith("Aggregate_feature_")
    or col in [
        "in_txs_degree",
        "out_txs_degree",
        "total_BTC",
        "fees",
        "size",
        "num_input_addresses",
        "num_output_addresses",
        "in_BTC_min",
        "in_BTC_max",
        "in_BTC_mean",
        "in_BTC_median",
        "in_BTC_total",
        "out_BTC_min",
        "out_BTC_max",
        "out_BTC_mean",
        "out_BTC_median",
        "out_BTC_total"
    ]
]

print("Number of feature columns:", len(feature_cols))

# Calculate Pearson correlation
corr_matrix = df[feature_cols].corr()

print("\nCorrelation matrix shape:")
print(corr_matrix.shape)

# Find highly correlated feature pairs
high_corr_pairs = []

for i in range(len(corr_matrix.columns)):
    for j in range(i + 1, len(corr_matrix.columns)):

        corr_value = corr_matrix.iloc[i, j]

        if abs(corr_value) >= 0.90:
            high_corr_pairs.append(
                (
                    corr_matrix.columns[i],
                    corr_matrix.columns[j],
                    corr_value
                )
            )

# Convert to DataFrame
high_corr_df = pd.DataFrame(
    high_corr_pairs,
    columns=["Feature_1", "Feature_2", "Correlation"]
)

# Sort by absolute correlation
if len(high_corr_df) > 0:
    high_corr_df["Absolute_Correlation"] = (
        high_corr_df["Correlation"].abs()
    )

    high_corr_df = high_corr_df.sort_values(
        "Absolute_Correlation",
        ascending=False
    )

print("\nHighly correlated feature pairs (|correlation| >= 0.90):")
print(high_corr_df.head(50).to_string(index=False))
```

```python
# ============================================================
# STEP 5 — FEATURE GROUP STRUCTURE
# ============================================================

local_features = [
    col for col in df.columns
    if col.startswith("Local_feature_")
]

aggregate_features = [
    col for col in df.columns
    if col.startswith("Aggregate_feature_")
]

transaction_features = [
    col for col in df.columns
    if col not in local_features
    and col not in aggregate_features
    and col not in ["txId", "Time step"]
]

print("LOCAL FEATURES")
print("Count:", len(local_features))
print(local_features)

print("\n" + "="*70)

print("AGGREGATE FEATURES")
print("Count:", len(aggregate_features))
print(aggregate_features)

print("\n" + "="*70)

print("OTHER TRANSACTION FEATURES")
print("Count:", len(transaction_features))
print(transaction_features)
```

```python
# ============================================================
# STEP 11 — EDGE ID INTEGRITY CHECK
# ============================================================

feature_txids = set(features_df["txId"])

edge_txids = (
    set(edge_df["txId1"]) |
    set(edge_df["txId2"])
)

missing_from_features = edge_txids - feature_txids

print("Unique transaction IDs in features:",
      len(feature_txids))

print("Unique transaction IDs appearing in edges:",
      len(edge_txids))

print("\nEdge transaction IDs NOT found in features:",
      len(missing_from_features))

if len(missing_from_features) == 0:
    print("\n✅ All edge transaction IDs exist in txs_features.csv")
else:
    print("\nFirst 20 missing IDs:")
    print(list(missing_from_features)[:20])
```

```python
# ============================================================
# STEP 12 — EDGE COVERAGE AGAINST CLASSES
# ============================================================

class_txids = set(classes_df["txId"])

missing_from_classes = edge_txids - class_txids

print("Unique transaction IDs in classes:",
      len(class_txids))

print("Unique transaction IDs in edges:",
      len(edge_txids))

print("\nEdge transaction IDs NOT found in classes:",
      len(missing_from_classes))

if len(missing_from_classes) == 0:
    print("\n✅ All edge transaction IDs have a class entry.")
else:
    print("\nFirst 20 missing IDs:")
    print(list(missing_from_classes)[:20])
```

```python
# ============================================================
# STEP 13 — FEATURE GROUP SUMMARY
# ============================================================

local_features = [
    c for c in features_df.columns
    if c.startswith("Local_feature_")
]

aggregate_features = [
    c for c in features_df.columns
    if c.startswith("Aggregate_feature_")
]

transaction_features = [
    c for c in features_df.columns
    if c not in local_features
    and c not in aggregate_features
    and c not in ["txId", "Time step"]
]

print("Local features:", len(local_features))
print("Aggregate features:", len(aggregate_features))
print("Transaction-level features:", len(transaction_features))

print("\nLocal feature range:")
print(local_features[:5], "...", local_features[-5:])

print("\nAggregate feature range:")
print(aggregate_features[:5], "...", aggregate_features[-5:])

print("\nTransaction-level features:")
print(transaction_features)
```

```python
# ============================================================
# STEP 14 — FEATURE STATISTICS BY GROUP
# ============================================================

local_stats = features_df[local_features].describe().T
aggregate_stats = features_df[aggregate_features].describe().T
transaction_stats = features_df[transaction_features].describe().T

print("LOCAL FEATURES")
display(local_stats.head(10))

print("\nAGGREGATE FEATURES")
display(aggregate_stats.head(10))

print("\nTRANSACTION-LEVEL FEATURES")
display(transaction_stats)
```

```python
# ============================================================
# STEP 15 — LOW VARIANCE / NEAR-CONSTANT FEATURES
# ============================================================

model_features = local_features + aggregate_features + transaction_features

variance_df = (
    features_df[model_features]
    .var()
    .sort_values()
    .to_frame(name="variance")
)

print("Total model features:", len(model_features))

print("\n10 lowest-variance features:")
display(variance_df.head(10))

print("\n10 highest-variance features:")
display(variance_df.tail(10))
```

```python
# ============================================================
# STEP 16 — SYSTEMATIC REDUNDANCY CHECK
# ============================================================

corr = features_df[model_features].corr().abs()

# Keep only upper triangle so we don't count A-B and B-A twice
upper = corr.where(
    np.triu(np.ones(corr.shape), k=1).astype(bool)
)

high_corr_pairs = (
    upper.stack()
    .reset_index()
)

high_corr_pairs.columns = [
    "Feature_1",
    "Feature_2",
    "Absolute_Correlation"
]

high_corr_pairs = (
    high_corr_pairs[
        high_corr_pairs["Absolute_Correlation"] >= 0.90
    ]
    .sort_values(
        "Absolute_Correlation",
        ascending=False
    )
)

print("Total highly correlated pairs (|r| >= 0.90):",
      len(high_corr_pairs))

display(high_corr_pairs.head(30))
```

```python
# ============================================================
# STEP 17 — UNIQUE FEATURES INVOLVED IN HIGH CORRELATION
# ============================================================

high_corr_features = set(
    high_corr_pairs["Feature_1"]
).union(
    set(high_corr_pairs["Feature_2"])
)

print("Highly correlated pairs:", len(high_corr_pairs))

print("Unique features involved:",
      len(high_corr_features))

print("\nHow many are Local features?")
print(
    sum(f.startswith("Local_feature_")
        for f in high_corr_features)
)

print("\nHow many are Aggregate features?")
print(
    sum(f.startswith("Aggregate_feature_")
        for f in high_corr_features)
)

print("\nHow many are transaction-level features?")
print(
    sum(
        f in transaction_features
        for f in high_corr_features
    )
)
```

```python
# ============================================================
# STEP 18 — WHERE IS THE REDUNDANCY COMING FROM?
# ============================================================

def feature_group(feature):
    if feature.startswith("Local_feature_"):
        return "Local"
    elif feature.startswith("Aggregate_feature_"):
        return "Aggregate"
    elif feature in transaction_features:
        return "Transaction"
    else:
        return "Other"


high_corr_pairs["Group_1"] = (
    high_corr_pairs["Feature_1"].apply(feature_group)
)

high_corr_pairs["Group_2"] = (
    high_corr_pairs["Feature_2"].apply(feature_group)
)

group_counts = (
    high_corr_pairs
    .groupby(["Group_1", "Group_2"])
    .size()
    .sort_values(ascending=False)
)

print("High-correlation pairs by feature group:\n")
print(group_counts)
```

```python

# ============================================================
# STEP 19 — FIND REDUNDANT FEATURE GROUPS
# NetworkX-free version
# ============================================================

# Each feature initially belongs to its own group
parent = {feature: feature for feature in model_features}


def find(x):
    while parent[x] != x:
        parent[x] = parent[parent[x]]
        x = parent[x]
    return x


def union(a, b):
    root_a = find(a)
    root_b = find(b)

    if root_a != root_b:
        parent[root_b] = root_a


# Connect highly correlated features
for _, row in high_corr_pairs.iterrows():
    union(
        row["Feature_1"],
        row["Feature_2"]
    )


# Build groups
groups = {}

for feature in model_features:
    root = find(feature)

    if root not in groups:
        groups[root] = []

    groups[root].append(feature)


# Keep only groups containing 2+ features
redundant_groups = [
    sorted(group)
    for group in groups.values()
    if len(group) > 1
]

# Largest groups first
redundant_groups = sorted(
    redundant_groups,
    key=len,
    reverse=True
)

print("Number of redundant groups:",
      len(redundant_groups))

print("\nLargest redundant groups:\n")

for i, group in enumerate(redundant_groups[:20], 1):
    print(f"Group {i} ({len(group)} features):")
    print(group)
    print()
```

```python
# ============================================================
# STEP 45 — RANDOM FOREST FEATURE IMPORTANCE
# ============================================================

importance_df = pd.DataFrame({
    "Feature": X_train.columns,
    "Importance": weighted_RF.feature_importances_
})

importance_df = importance_df.sort_values(
    by="Importance",
    ascending=False
).reset_index(drop=True)

print("Top 20 most important features:")
print(importance_df.head(20).to_string(index=False))

print("\nBottom 20 least important features:")
print(importance_df.tail(20).to_string(index=False))
```

```python
# ============================================================
# STEP 46 — IMPORTANT FEATURES vs CORRELATED FEATURES
# ============================================================

# Top 20 important features
top20_features = importance_df.head(20)["Feature"].tolist()

# Check which top features participate in high-correlation pairs
top20_correlated = correlated_pairs[
    correlated_pairs["Feature_1"].isin(top20_features) |
    correlated_pairs["Feature_2"].isin(top20_features)
].copy()

print("Top 20 important features:")
print(top20_features)

print("\nTop important features involved in |r| >= 0.90 correlations:")
print(top20_correlated.to_string(index=False))

print("\nNumber of top-20 features involved in high correlations:")

involved_features = set(
    top20_correlated["Feature_1"]
).union(
    set(top20_correlated["Feature_2"])
)

print(
    len(set(top20_features).intersection(involved_features))
)
```

```python
# ============================================================
# STEP 46A — RECREATE HIGH-CORRELATION PAIRS
# ============================================================

import numpy as np
import pandas as pd

# Calculate absolute correlation matrix
corr_matrix = X_train.corr().abs()

# Keep only upper triangle to avoid duplicate pairs
upper = corr_matrix.where(
    np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
)

# Extract feature pairs with |r| >= 0.90
correlated_pairs = (
    upper.stack()
    .reset_index()
)

correlated_pairs.columns = [
    "Feature_1",
    "Feature_2",
    "Absolute_Correlation"
]

correlated_pairs = correlated_pairs[
    correlated_pairs["Absolute_Correlation"] >= 0.90
].sort_values(
    "Absolute_Correlation",
    ascending=False
).reset_index(drop=True)

print("Total highly correlated pairs:", len(correlated_pairs))

print("\nTop 20 highly correlated pairs:")
print(
    correlated_pairs.head(20).to_string(index=False)
)
```

```python
# ============================================================
# STEP 46B — IMPORTANT FEATURES vs CORRELATED FEATURES
# ============================================================

# Top 20 features according to Random Forest
top20_features = importance_df.head(20)["Feature"].tolist()

# Find correlation pairs involving any top-20 feature
top20_correlated = correlated_pairs[
    correlated_pairs["Feature_1"].isin(top20_features) |
    correlated_pairs["Feature_2"].isin(top20_features)
].copy()

print("Top 20 important features:")
print(top20_features)

print("\nTop important features involved in |r| >= 0.90 correlations:")

if len(top20_correlated) > 0:
    print(top20_correlated.to_string(index=False))
else:
    print("None")

involved_features = set(
    top20_correlated["Feature_1"]
).union(
    set(top20_correlated["Feature_2"])
)

top20_involved = set(top20_features).intersection(involved_features)

print("\nNumber of top-20 features involved in high correlations:")
print(len(top20_involved))

print("\nThose top-20 features:")
print(sorted(top20_involved))
```

```python
# ============================================================
# STEP 47A — IDENTIFY REDUNDANT FEATURES
# ============================================================

# Work with a copy of the correlation matrix
corr = X_train.corr().abs()

# Upper triangle only
upper = corr.where(
    np.triu(np.ones(corr.shape), k=1).astype(bool)
)

# Features having correlation >= 0.90 with an earlier feature
to_drop = [
    column
    for column in upper.columns
    if any(upper[column] >= 0.90)
]

print("Original number of features:", X_train.shape[1])
print("Features marked as redundant:", len(to_drop))
print("Features remaining:", X_train.shape[1] - len(to_drop))

print("\nFirst 30 features marked for removal:")
print(to_drop[:30])
```

```python
# ============================================================
# STEP 47B — CORRELATED FEATURES + RF IMPORTANCE
# ============================================================

# Create importance lookup
importance_lookup = dict(
    zip(
        importance_df["Feature"],
        importance_df["Importance"]
    )
)

# Show the first 30 automatically marked redundant features
# together with their RF importance
redundant_importance = pd.DataFrame({
    "Feature": to_drop,
    "RF_Importance": [
        importance_lookup.get(feature, 0)
        for feature in to_drop
    ]
})

redundant_importance = redundant_importance.sort_values(
    by="RF_Importance",
    ascending=False
).reset_index(drop=True)

print("Redundant features ranked by RF importance:")
print(
    redundant_importance.head(30).to_string(index=False)
)
```

```python
# ============================================================
# STEP 48 — PERMUTATION FEATURE IMPORTANCE
# ============================================================

from sklearn.inspection import permutation_importance

print("Calculating permutation importance...")
print("This may take some time...")

perm_result = permutation_importance(
    weighted_RF,
    X_test,
    y_test,
    n_repeats=3,
    random_state=42,
    scoring="f1_macro",
    n_jobs=-1
)

permutation_df = pd.DataFrame({
    "Feature": X_test.columns,
    "Permutation_Importance": perm_result.importances_mean,
    "Std": perm_result.importances_std
})

permutation_df = permutation_df.sort_values(
    by="Permutation_Importance",
    ascending=False
).reset_index(drop=True)

print("\nTop 20 features by permutation importance:")
print(
    permutation_df.head(20).to_string(index=False)
)

print("\nBottom 20 features by permutation importance:")
print(
    permutation_df.tail(20).to_string(index=False)
)
```

```python
# ============================================================
# STEP 49 — CREATE CANDIDATE REDUCED FEATURE SET
# ============================================================

# Features that were important according to RF impurity importance
rf_top20 = set(
    importance_df.head(20)["Feature"]
)

# Features with positive permutation importance
positive_perm_features = set(
    permutation_df[
        permutation_df["Permutation_Importance"] > 0
    ]["Feature"]
)

# Combine both sources of evidence
candidate_features = sorted(
    positive_perm_features.union(rf_top20)
)

print("Original number of features:", X_train.shape[1])
print("Candidate retained features:", len(candidate_features))
print("Candidate features for removal:", 
      X_train.shape[1] - len(candidate_features))

print("\nFirst 30 candidate retained features:")
print(candidate_features[:30])

print("\nTop RF features that are retained:")
print(
    sorted(
        rf_top20.intersection(candidate_features)
    )
)
```

```python
# ============================================================
# STEP 50 — CREATE REDUCED TRAIN/TEST DATASETS
# ============================================================

X_train_reduced = X_train[candidate_features].copy()
X_test_reduced = X_test[candidate_features].copy()

print("Original X_train shape :", X_train.shape)
print("Original X_test shape  :", X_test.shape)

print("\nReduced X_train shape  :", X_train_reduced.shape)
print("Reduced X_test shape   :", X_test_reduced.shape)

print("\nNumber of retained features:", len(candidate_features))
```

```python
# ============================================================
# STEP 51 — REDUCED-FEATURE CLASS-WEIGHTED RANDOM FOREST
# ============================================================

reduced_RF = RandomForestClassifier(
    n_estimators=200,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

print("Training reduced-feature Random Forest...")

reduced_RF.fit(
    X_train_reduced,
    y_train
)

print("✅ Reduced-feature Random Forest trained successfully.")
print("Number of trees:", len(reduced_RF.estimators_))
```

```python
# ============================================================
# STEP 52 — REDUCED RF PREDICTIONS
# ============================================================

y_pred_reduced = reduced_RF.predict(X_test_reduced)

print("Reduced RF predictions generated successfully.")
print("Number of predictions:", len(y_pred_reduced))

print("\nFirst 20 predictions:")
print(y_pred_reduced[:20])
```

```python
# ============================================================
# STEP 53 — EVALUATE REDUCED RANDOM FOREST
# ============================================================

from sklearn.metrics import classification_report, confusion_matrix

print("Reduced RF Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_reduced))

print("\nReduced RF Classification Report:")
print(
    classification_report(
        y_test,
        y_pred_reduced,
        target_names=["Illicit (Class 1)", "Licit (Class 2)"],
        digits=4
    )
)
```

```python
# ============================================================
# STEP 54 — INSPECT TIME-STEP DISTRIBUTION
# ============================================================

print("Unique time steps:")
print(sorted(supervised_df["Time step"].unique()))

print("\nNumber of unique time steps:",
      supervised_df["Time step"].nunique())

print("\nTransactions per time step:")
print(
    supervised_df["Time step"]
    .value_counts()
    .sort_index()
)

print("\nClass distribution by time step:")
print(
    pd.crosstab(
        supervised_df["Time step"],
        supervised_df["class"]
    )
)
```

```python
# ============================================================
# STEP 55 — LABELLED TRANSACTIONS BY TIME STEP
# ============================================================

time_class = (
    supervised_df
    .groupby(["Time step", "class"])
    .size()
    .unstack(fill_value=0)
)

time_class.columns = [
    f"class_{int(c)}"
    for c in time_class.columns
]

time_class["total_labelled"] = time_class.sum(axis=1)

print("Labelled transactions by time step:")
print(time_class.to_string())

print("\nTotal labelled transactions:",
      time_class["total_labelled"].sum())
```

```python
# ============================================================
# STEP 56 — CHECK POSSIBLE TIME-BASED SPLITS
# ============================================================

time_summary = time_class.copy()

# Cumulative labelled counts
time_summary["cumulative_labelled"] = (
    time_summary["total_labelled"].cumsum()
)

time_summary["cumulative_class_1"] = (
    time_summary["class_1"].cumsum()
)

time_summary["cumulative_class_2"] = (
    time_summary["class_2"].cumsum()
)

total_labelled = time_summary["total_labelled"].sum()
total_class_1 = time_summary["class_1"].sum()

# Candidate cutoffs
candidate_cutoffs = [30, 32, 34, 36, 38, 40]

rows = []

for cutoff in candidate_cutoffs:

    train_part = time_summary.loc[
        time_summary.index <= cutoff
    ]

    test_part = time_summary.loc[
        time_summary.index > cutoff
    ]

    rows.append({
        "Train_time_steps": f"1-{cutoff}",
        "Test_time_steps": f"{cutoff+1}-49",

        "Train_labelled": train_part["total_labelled"].sum(),
        "Test_labelled": test_part["total_labelled"].sum(),

        "Train_class_1": train_part["class_1"].sum(),
        "Test_class_1": test_part["class_1"].sum(),

        "Train_class_2": train_part["class_2"].sum(),
        "Test_class_2": test_part["class_2"].sum()
    })

split_check = pd.DataFrame(rows)

print("Possible chronological splits:\n")
print(split_check.to_string(index=False))
```

```python
# ============================================================
# STEP 57 — TIME-BASED TRAIN / TEST SPLIT
# ============================================================

TIME_CUTOFF = 30

# Use the supervised dataset's time step
train_mask = supervised_df["Time step"] <= TIME_CUTOFF
test_mask = supervised_df["Time step"] > TIME_CUTOFF

# Features
X_temporal = supervised_df[candidate_features]

# Target
y_temporal = supervised_df["class"]

# Temporal split
X_train_time = X_temporal[train_mask].copy()
X_test_time = X_temporal[test_mask].copy()

y_train_time = y_temporal[train_mask].copy()
y_test_time = y_temporal[test_mask].copy()

print("Temporal split:")
print(f"Training time steps: 1-{TIME_CUTOFF}")
print(f"Testing time steps : {TIME_CUTOFF + 1}-49")

print("\nShapes:")
print("X_train_time:", X_train_time.shape)
print("X_test_time :", X_test_time.shape)

print("\nTraining class distribution:")
print(y_train_time.value_counts().sort_index())

print("\nTesting class distribution:")
print(y_test_time.value_counts().sort_index())
```

```python
# ============================================================
# STEP 58 — TEMPORAL CLASS-WEIGHTED RANDOM FOREST
# ============================================================

temporal_RF = RandomForestClassifier(
    n_estimators=200,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

print("Training temporal Random Forest...")

temporal_RF.fit(
    X_train_time,
    y_train_time
)

print("✅ Temporal Random Forest trained successfully.")
print("Number of trees:", len(temporal_RF.estimators_))
```

```python
# ============================================================
# STEP 59 — TEMPORAL RF PREDICTIONS
# ============================================================

y_pred_temporal = temporal_RF.predict(X_test_time)

print("Temporal RF predictions generated successfully.")
print("Number of predictions:", len(y_pred_temporal))

print("\nFirst 20 predictions:")
print(y_pred_temporal[:20])
```

```python
# ============================================================
# STEP 60 — EVALUATE TEMPORAL RANDOM FOREST
# ============================================================

from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    accuracy_score
)

cm_temporal = confusion_matrix(
    y_test_time,
    y_pred_temporal
)

print("Temporal RF Confusion Matrix:")
print(cm_temporal)

print("\nTemporal RF Classification Report:")

print(
    classification_report(
        y_test_time,
        y_pred_temporal,
        target_names=[
            "Illicit (Class 1)",
            "Licit (Class 2)"
        ],
        digits=4
    )
)

print(
    "\nTemporal RF Accuracy:",
    round(
        accuracy_score(
            y_test_time,
            y_pred_temporal
        ),
        4
    )
)
```

```python
# ============================================================
# STEP 61 — TEMPORAL PERFORMANCE BY TIME STEP
# ============================================================

temporal_results = supervised_df.loc[
    test_mask,
    ["txId", "Time step", "class"]
].copy()

temporal_results["prediction"] = y_pred_temporal

# Calculate performance separately for each future time step
rows = []

for timestep in sorted(temporal_results["Time step"].unique()):

    subset = temporal_results[
        temporal_results["Time step"] == timestep
    ]

    actual_illicit = (subset["class"] == 1).sum()
    predicted_illicit = (subset["prediction"] == 1).sum()

    true_positive = (
        (subset["class"] == 1) &
        (subset["prediction"] == 1)
    ).sum()

    false_negative = (
        (subset["class"] == 1) &
        (subset["prediction"] == 2)
    ).sum()

    recall = (
        true_positive / actual_illicit
        if actual_illicit > 0
        else np.nan
    )

    rows.append({
        "Time_step": timestep,
        "Actual_illicit": actual_illicit,
        "Predicted_illicit": predicted_illicit,
        "True_positive": true_positive,
        "False_negative": false_negative,
        "Illicit_recall": recall
    })

temporal_by_time = pd.DataFrame(rows)

print("Illicit detection by future time step:")
print(
    temporal_by_time.to_string(index=False)
)

print("\nAverage illicit recall across future time steps:")
print(
    round(
        temporal_by_time["Illicit_recall"].mean(),
        4
    )
)
```

```python
# ============================================================
# STEP 62 — CLASS RATE BY TIME STEP
# ============================================================

class_rate_by_time = time_class.copy()

class_rate_by_time["illicit_rate"] = (
    class_rate_by_time["class_1"]
    / class_rate_by_time["total_labelled"]
    * 100
)

print("Illicit percentage by time step:\n")

print(
    class_rate_by_time[
        ["class_1", "class_2", "total_labelled", "illicit_rate"]
    ].to_string()
)

print("\nOverall illicit percentage:")
print(
    round(
        total_class_1 / total_labelled * 100,
        2
    ),
    "%"
)

print("\nAverage illicit percentage:")
print(
    round(
        class_rate_by_time["illicit_rate"].mean(),
        2
    ),
    "%"
)
```

```python
# ============================================================
# STEP 63 — TRAIN vs FUTURE CLASS DISTRIBUTION
# ============================================================

train_stats = pd.DataFrame({
    "class": y_train_time.value_counts().sort_index()
})

test_stats = pd.DataFrame({
    "class": y_test_time.value_counts().sort_index()
})

train_stats["percentage"] = (
    train_stats["class"] /
    train_stats["class"].sum() * 100
)

test_stats["percentage"] = (
    test_stats["class"] /
    test_stats["class"].sum() * 100
)

print("TRAINING PERIOD — Time steps 1-30")
print(train_stats)

print("\nFUTURE TEST PERIOD — Time steps 31-49")
print(test_stats)

print("\nComparison:")
print(
    pd.DataFrame({
        "Training_%": train_stats["percentage"],
        "Future_%": test_stats["percentage"]
    })
)
```

```python
# ============================================================
# STEP 64 — FEATURE DISTRIBUTION SHIFT
# ============================================================

from scipy.stats import wasserstein_distance
import pandas as pd
import numpy as np

shift_results = []

for feature in candidate_features:

    train_values = X_train_time[feature].dropna().values
    test_values = X_test_time[feature].dropna().values

    # Skip completely empty features
    if len(train_values) == 0 or len(test_values) == 0:
        continue

    distance = wasserstein_distance(
        train_values,
        test_values
    )

    # Normalize by combined standard deviation
    combined_std = np.std(
        np.concatenate([train_values, test_values])
    )

    if combined_std > 0:
        normalized_distance = distance / combined_std
    else:
        normalized_distance = 0.0

    shift_results.append({
        "Feature": feature,
        "Wasserstein_Distance": distance,
        "Normalized_Shift": normalized_distance
    })

shift_df = pd.DataFrame(shift_results)

shift_df = shift_df.sort_values(
    "Normalized_Shift",
    ascending=False
).reset_index(drop=True)

print("Total features analyzed:", len(shift_df))

print("\nTop 20 features with largest train → future distribution shift:")
print(
    shift_df.head(20).to_string(index=False)
)

print("\nShift summary:")
print(shift_df["Normalized_Shift"].describe())
```

```python
# ============================================================
# STEP 65 — CLASS-CONDITIONAL DISTRIBUTION SHIFT
# ============================================================

from scipy.stats import wasserstein_distance
import numpy as np
import pandas as pd

class_shift_results = []

for feature in candidate_features:

    # Training: illicit and licit separately
    train_illicit = X_train_time.loc[
        y_train_time == 1, feature
    ].dropna().values

    train_licit = X_train_time.loc[
        y_train_time == 2, feature
    ].dropna().values

    # Future: illicit and licit separately
    test_illicit = X_test_time.loc[
        y_test_time == 1, feature
    ].dropna().values

    test_licit = X_test_time.loc[
        y_test_time == 2, feature
    ].dropna().values

    # Need data on both sides
    if (
        len(train_illicit) == 0
        or len(test_illicit) == 0
        or len(train_licit) == 0
        or len(test_licit) == 0
    ):
        continue

    # Illicit distribution shift
    illicit_distance = wasserstein_distance(
        train_illicit,
        test_illicit
    )

    # Licit distribution shift
    licit_distance = wasserstein_distance(
        train_licit,
        test_licit
    )

    # Normalize using combined standard deviation
    illicit_std = np.std(
        np.concatenate([
            train_illicit,
            test_illicit
        ])
    )

    licit_std = np.std(
        np.concatenate([
            train_licit,
            test_licit
        ])
    )

    illicit_shift = (
        illicit_distance / illicit_std
        if illicit_std > 0
        else 0
    )

    licit_shift = (
        licit_distance / licit_std
        if licit_std > 0
        else 0
    )

    class_shift_results.append({
        "Feature": feature,
        "Illicit_Shift": illicit_shift,
        "Licit_Shift": licit_shift,
        "Average_Class_Shift": (
            illicit_shift + licit_shift
        ) / 2
    })

class_shift_df = pd.DataFrame(
    class_shift_results
)

class_shift_df = class_shift_df.sort_values(
    "Average_Class_Shift",
    ascending=False
).reset_index(drop=True)

print("Top 20 features with largest class-conditional shift:")
print(
    class_shift_df
    .head(20)
    .to_string(index=False)
)
```

```python
# ============================================================
# STEP 66 — COMBINE IMPORTANCE + TEMPORAL SHIFT
# ============================================================

# RF importance
rf_importance = importance_df[
    ["Feature", "Importance"]
].copy()

rf_importance = rf_importance.rename(
    columns={"Importance": "RF_Importance"}
)

# Permutation importance
perm_importance = permutation_df[
    ["Feature", "Permutation_Importance"]
].copy()

# Temporal shift
shift_info = shift_df[
    ["Feature", "Normalized_Shift"]
].copy()

# Merge all three analyses
feature_diagnostics = (
    rf_importance
    .merge(
        perm_importance,
        on="Feature",
        how="left"
    )
    .merge(
        shift_info,
        on="Feature",
        how="left"
    )
)

# Sort by RF importance
feature_diagnostics = feature_diagnostics.sort_values(
    "RF_Importance",
    ascending=False
).reset_index(drop=True)

print("Feature diagnostic table:")
print(
    feature_diagnostics.head(30).to_string(index=False)
)
```

```python
# ============================================================
# STEP 67 — FEATURE STABILITY GROUPS
# ============================================================

diagnostic = feature_diagnostics.dropna(
    subset=[
        "RF_Importance",
        "Permutation_Importance",
        "Normalized_Shift"
    ]
).copy()

# Use medians rather than arbitrary thresholds
rf_threshold = diagnostic["RF_Importance"].median()
shift_threshold = diagnostic["Normalized_Shift"].median()

diagnostic["Importance_Group"] = np.where(
    diagnostic["RF_Importance"] >= rf_threshold,
    "High importance",
    "Low importance"
)

diagnostic["Shift_Group"] = np.where(
    diagnostic["Normalized_Shift"] >= shift_threshold,
    "High shift",
    "Low shift"
)

diagnostic["Group"] = (
    diagnostic["Importance_Group"]
    + " + "
    + diagnostic["Shift_Group"]
)

print("RF importance median:",
      round(rf_threshold, 6))

print("Temporal shift median:",
      round(shift_threshold, 6))

print("\nFeature groups:")
print(
    diagnostic[
        [
            "Feature",
            "RF_Importance",
            "Permutation_Importance",
            "Normalized_Shift",
            "Group"
        ]
    ]
    .sort_values(
        ["Group", "RF_Importance"],
        ascending=[True, False]
    )
    .to_string(index=False)
)
```

```python
# ============================================================
# STEP 68 — STABLE FEATURE TEMPORAL EXPERIMENT
# ============================================================

stable_features = diagnostic[
    (diagnostic["Importance_Group"] == "High importance") &
    (diagnostic["Shift_Group"] == "Low shift")
]["Feature"].tolist()

print("Number of stable important features:", len(stable_features))

print("\nStable important features:")
print(stable_features)

# Create reduced temporal datasets
X_train_stable = X_train_time[stable_features].copy()
X_test_stable = X_test_time[stable_features].copy()

print("\nX_train_stable shape:", X_train_stable.shape)
print("X_test_stable shape :", X_test_stable.shape)
```

```python
# ============================================================
# STEP 69 — STABLE-FEATURE TEMPORAL RANDOM FOREST
# ============================================================

stable_RF = RandomForestClassifier(
    n_estimators=200,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

print("Training stable-feature temporal Random Forest...")

stable_RF.fit(
    X_train_stable,
    y_train_time
)

print("✅ Stable-feature Random Forest trained successfully.")
print("Number of trees:", len(stable_RF.estimators_))
```

```python
# ============================================================
# STEP 70 — STABLE-FEATURE TEMPORAL PREDICTIONS
# ============================================================

y_pred_stable = stable_RF.predict(X_test_stable)

print("Stable-feature predictions generated successfully.")
print("Number of predictions:", len(y_pred_stable))

print("\nFirst 20 predictions:")
print(y_pred_stable[:20])
```

```python
# ============================================================
# STEP 71 — EVALUATE STABLE-FEATURE TEMPORAL RF
# ============================================================

cm_stable = confusion_matrix(
    y_test_time,
    y_pred_stable
)

print("Stable-feature Temporal RF Confusion Matrix:")
print(cm_stable)

print("\nStable-feature Temporal RF Classification Report:")

print(
    classification_report(
        y_test_time,
        y_pred_stable,
        target_names=[
            "Illicit (Class 1)",
            "Licit (Class 2)"
        ],
        digits=4
    )
)

stable_accuracy = accuracy_score(
    y_test_time,
    y_pred_stable
)

print(
    "\nStable-feature Temporal RF Accuracy:",
    round(stable_accuracy, 4)
)
```

```python
# ============================================================
# STEP 72 — TEMPORAL RF PROBABILITY SCORES
# ============================================================

y_proba_temporal = temporal_RF.predict_proba(
    X_test_time
)

print("Probability matrix shape:")
print(y_proba_temporal.shape)

print("\nClass order:")
print(temporal_RF.classes_)

print("\nFirst 20 illicit probabilities:")
print(
    y_proba_temporal[:20, 0]
)

print("\nProbability range for Class 1:")
print(
    y_proba_temporal[:, 0].min(),
    "to",
    y_proba_temporal[:, 0].max()
)
```

```python
# ============================================================
# STEP 73 — TEMPORAL DECISION THRESHOLD ANALYSIS
# ============================================================

from sklearn.metrics import precision_score, recall_score, f1_score

threshold_results = []

# Probability of Class 1 (Illicit)
illicit_probability = y_proba_temporal[:, 0]

for threshold in np.arange(0.10, 0.91, 0.05):

    # Class 1 if probability >= threshold
    threshold_predictions = np.where(
        illicit_probability >= threshold,
        1,
        2
    )

    precision = precision_score(
        y_test_time,
        threshold_predictions,
        pos_label=1,
        zero_division=0
    )

    recall = recall_score(
        y_test_time,
        threshold_predictions,
        pos_label=1,
        zero_division=0
    )

    f1 = f1_score(
        y_test_time,
        threshold_predictions,
        pos_label=1,
        zero_division=0
    )

    flagged = np.sum(
        threshold_predictions == 1
    )

    threshold_results.append({
        "Threshold": round(threshold, 2),
        "Illicit_Precision": precision,
        "Illicit_Recall": recall,
        "Illicit_F1": f1,
        "Flagged_Transactions": flagged
    })

threshold_df = pd.DataFrame(
    threshold_results
)

print("Threshold analysis on future time steps 31–49:\n")

print(
    threshold_df.to_string(index=False)
)

print("\nBest threshold by illicit F1:")

best_f1 = threshold_df.loc[
    threshold_df["Illicit_F1"].idxmax()
]

print(best_f1)
```

```python
# ============================================================
# STEP 74 — TEMPORAL VALIDATION SPLIT
# ============================================================

from sklearn.model_selection import train_test_split

# Split the existing training period into:
#   Model-training subset
#   Validation subset
#
# IMPORTANT:
# We keep the original future test period (31-49) untouched.

X_train_model, X_val = train_test_split(
    X_train_time,
    test_size=0.20,
    random_state=42,
    stratify=y_train_time
)

y_train_model, y_val = train_test_split(
    y_train_time,
    test_size=0.20,
    random_state=42,
    stratify=y_train_time
)

print("Training subset shape :", X_train_model.shape)
print("Validation subset shape:", X_val.shape)

print("\nTraining subset distribution:")
print(y_train_model.value_counts())

print("\nValidation subset distribution:")
print(y_val.value_counts())
```

```python
# ============================================================
# STEP 75 — VALIDATION RANDOM FOREST
# ============================================================

validation_RF = RandomForestClassifier(
    n_estimators=200,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

print("Training validation Random Forest...")

validation_RF.fit(
    X_train_model,
    y_train_model
)

print("✅ Validation Random Forest trained successfully.")
print("Number of trees:", len(validation_RF.estimators_))
```

```python
# ============================================================
# STEP 76 — VALIDATION PROBABILITY SCORES
# ============================================================

y_proba_val = validation_RF.predict_proba(X_val)

print("Validation probability matrix shape:")
print(y_proba_val.shape)

print("\nClass order:")
print(validation_RF.classes_)

print("\nFirst 20 validation illicit probabilities:")
print(y_proba_val[:20, 0])

print("\nClass 1 probability range:")
print(
    y_proba_val[:, 0].min(),
    "to",
    y_proba_val[:, 0].max()
)
```

```python
# ============================================================
# STEP 77 — VALIDATION THRESHOLD ANALYSIS
# ============================================================

validation_threshold_results = []

illicit_probability_val = y_proba_val[:, 0]

for threshold in np.arange(0.10, 0.91, 0.05):

    threshold_predictions = np.where(
        illicit_probability_val >= threshold,
        1,
        2
    )

    precision = precision_score(
        y_val,
        threshold_predictions,
        pos_label=1,
        zero_division=0
    )

    recall = recall_score(
        y_val,
        threshold_predictions,
        pos_label=1,
        zero_division=0
    )

    f1 = f1_score(
        y_val,
        threshold_predictions,
        pos_label=1,
        zero_division=0
    )

    flagged = np.sum(
        threshold_predictions == 1
    )

    validation_threshold_results.append({
        "Threshold": round(threshold, 2),
        "Illicit_Precision": precision,
        "Illicit_Recall": recall,
        "Illicit_F1": f1,
        "Flagged_Transactions": flagged
    })

validation_threshold_df = pd.DataFrame(
    validation_threshold_results
)

print("Validation threshold analysis:\n")

print(
    validation_threshold_df.to_string(index=False)
)

print("\nBest threshold by validation illicit F1:")

best_validation_threshold = validation_threshold_df.loc[
    validation_threshold_df["Illicit_F1"].idxmax()
]

print(best_validation_threshold)
```

```python
# ============================================================
# STEP 78 — APPLY VALIDATION-SELECTED THRESHOLD TO FUTURE DATA
# ============================================================

selected_threshold = best_validation_threshold["Threshold"]

print("Selected threshold from validation:", selected_threshold)

# Use the ORIGINAL temporal RF probabilities
# on the untouched future period (time steps 31-49)
future_illicit_probability = y_proba_temporal[:, 0]

y_pred_future_threshold = np.where(
    future_illicit_probability >= selected_threshold,
    1,
    2
)

print("\nFuture predictions generated successfully.")
print("Number of future predictions:", len(y_pred_future_threshold))

print("\nFirst 20 future predictions:")
print(y_pred_future_threshold[:20])
```

```python
# ============================================================
# STEP 79 — FINAL FUTURE EVALUATION AT VALIDATION THRESHOLD
# ============================================================

cm_future_threshold = confusion_matrix(
    y_test_time,
    y_pred_future_threshold
)

print("Future Confusion Matrix — Threshold 0.45:")
print(cm_future_threshold)

print("\nFuture Classification Report — Threshold 0.45:")

print(
    classification_report(
        y_test_time,
        y_pred_future_threshold,
        target_names=[
            "Illicit (Class 1)",
            "Licit (Class 2)"
        ],
        digits=4
    )
)

future_accuracy_threshold = accuracy_score(
    y_test_time,
    y_pred_future_threshold
)

print(
    "\nFuture Accuracy:",
    round(future_accuracy_threshold, 4)
)
```

```python
# ============================================================
# STEP 80 — FUTURE ERROR ANALYSIS BY TIME STEP
# ============================================================

future_error_analysis = []

# Use the original temporal RF predictions at threshold 0.50
predicted_labels = y_pred_time

for time_step in sorted(
    df_supervised.loc[
        df_supervised["Time step"].isin(range(31, 50)),
        "Time step"
    ].unique()
):

    mask = (
        df_supervised.loc[
            df_supervised["Time step"].isin(range(31, 50)),
            "Time step"
        ] == time_step
    ).values

    actual = y_test_time[mask]
    predicted = predicted_labels[mask]

    actual_illicit = np.sum(actual == 1)
    predicted_illicit = np.sum(predicted == 1)

    true_positive = np.sum(
        (actual == 1) & (predicted == 1)
    )

    false_negative = np.sum(
        (actual == 1) & (predicted == 2)
    )

    false_positive = np.sum(
        (actual == 2) & (predicted == 1)
    )

    illicit_recall = (
        true_positive / actual_illicit
        if actual_illicit > 0
        else np.nan
    )

    future_error_analysis.append({
        "Time_step": time_step,
        "Actual_illicit": actual_illicit,
        "Predicted_illicit": predicted_illicit,
        "True_positive": true_positive,
        "False_negative": false_negative,
        "False_positive": false_positive,
        "Illicit_recall": illicit_recall
    })

future_error_df = pd.DataFrame(
    future_error_analysis
)

print("Future-period error analysis:\n")

print(
    future_error_df.to_string(index=False)
)

print("\nAverage illicit recall:")
print(
    round(
        future_error_df["Illicit_recall"].mean(),
        4
    )
)
```

```python
# ============================================================
# STEP — CLASS 3 DISTINCTIVE FEATURES
# ============================================================

from sklearn.preprocessing import StandardScaler
import pandas as pd
import numpy as np

# Class-wise means already calculated
class_means = class_feature_df.groupby("class")[feature_cols].mean()

# Difference of Class 3 from Classes 1 and 2
class3_vs_1 = (class_means.loc[3] - class_means.loc[1]).abs()
class3_vs_2 = (class_means.loc[3] - class_means.loc[2]).abs()

# Average difference from both classes
class3_distinctiveness = (
    class3_vs_1 + class3_vs_2
) / 2

class3_diagnostic = pd.DataFrame({
    "Class1_Mean": class_means.loc[1],
    "Class2_Mean": class_means.loc[2],
    "Class3_Mean": class_means.loc[3],
    "Abs_Diff_Class3_vs_Class1": class3_vs_1,
    "Abs_Diff_Class3_vs_Class2": class3_vs_2,
    "Class3_Distinctiveness": class3_distinctiveness
})

class3_diagnostic = class3_diagnostic.sort_values(
    "Class3_Distinctiveness",
    ascending=False
)

print("=" * 80)
print("TOP FEATURES DISTINCTIVE FOR CLASS 3")
print("=" * 80)

display(class3_diagnostic.head(20))
```

```python
# ============================================================
# STEP — CLASS 3 FEATURE SEPARATION
# ============================================================

from sklearn.feature_selection import f_classif

X = class_feature_df[feature_cols]
y = class_feature_df["class"]

# ANOVA F-test
f_scores, p_values = f_classif(X, y)

feature_stats = pd.DataFrame({
    "Feature": feature_cols,
    "F_score": f_scores,
    "p_value": p_values
})

feature_stats = feature_stats.sort_values(
    "F_score",
    ascending=False
)

print("=" * 80)
print("TOP FEATURES SEPARATING CLASS 1 / CLASS 2 / CLASS 3")
print("=" * 80)

display(feature_stats.head(30))
```

```python
# ============================================================
# CHECK NaN VALUES IN CLASS-WISE FEATURE DATA
# ============================================================

import numpy as np
import pandas as pd

feature_cols = [
    c for c in class_feature_df.columns
    if c != "class"
    and pd.api.types.is_numeric_dtype(class_feature_df[c])
]

nan_counts = class_feature_df[feature_cols].isna().sum()

nan_features = nan_counts[nan_counts > 0].sort_values(ascending=False)

print("=" * 70)
print("NaN ANALYSIS")
print("=" * 70)

print("Total numeric features:", len(feature_cols))
print("Features containing NaN:", len(nan_features))
print("Total NaN values:", int(nan_counts.sum()))

print("\nTop features with NaN:")
print(nan_features.head(30))
```

```python
# ============================================================
# ANALYZE CLASSES OF ROWS CONTAINING NaN
# ============================================================

nan_mask = class_feature_df[feature_cols].isna().any(axis=1)

nan_rows = class_feature_df.loc[nan_mask]

print("=" * 70)
print("ROWS CONTAINING NaN")
print("=" * 70)

print("Rows containing at least one NaN:", len(nan_rows))

print("\nClass distribution among NaN rows:")
print(
    nan_rows["class"]
    .value_counts()
    .sort_index()
)

print("\nClass percentages among NaN rows:")
print(
    (nan_rows["class"].value_counts(normalize=True).sort_index() * 100)
    .round(2)
)

print("\nOverall class distribution:")
print(
    class_feature_df["class"]
    .value_counts()
    .sort_index()
)
```

```python
# ============================================================
# CHECK NaN PATTERN
# ============================================================

nan_feature_counts = class_feature_df[feature_cols].isna().sum()

nan_feature_counts = nan_feature_counts[
    nan_feature_counts > 0
].sort_values(ascending=False)

print("=" * 70)
print("NaN PATTERN BY FEATURE")
print("=" * 70)

print(nan_feature_counts)

print("\nUnique NaN counts across affected features:")
print(nan_feature_counts.unique())

print("\nAre all affected features NaN on the same rows?")

affected_features = nan_feature_counts.index.tolist()

nan_pattern = class_feature_df[affected_features].isna()

same_pattern = nan_pattern.eq(
    nan_pattern.iloc[:, 0],
    axis=0
).all().all()

print("Same NaN row pattern:", same_pattern)
```

```python
# ============================================================
# CHECK OTHER FEATURES FOR THE 965 INCOMPLETE TRANSACTIONS
# ============================================================

affected_features = [
    "in_txs_degree",
    "out_txs_degree",
    "total_BTC",
    "fees",
    "size",
    "num_input_addresses",
    "num_output_addresses",
    "in_BTC_min",
    "in_BTC_max",
    "in_BTC_mean",
    "in_BTC_median",
    "in_BTC_total",
    "out_BTC_min",
    "out_BTC_max",
    "out_BTC_mean",
    "out_BTC_median",
    "out_BTC_total"
]

nan_mask = class_feature_df[affected_features].isna().any(axis=1)

other_features = [
    c for c in feature_cols
    if c not in affected_features
]

other_nan_count = (
    class_feature_df.loc[nan_mask, other_features]
    .isna()
    .sum()
)

print("=" * 70)
print("OTHER FEATURES IN THE 965 INCOMPLETE ROWS")
print("=" * 70)

print("Total other numeric features:", len(other_features))
print(
    "Other features containing NaN:",
    int((other_nan_count > 0).sum())
)
print(
    "Total additional NaN values:",
    int(other_nan_count.sum())
)

print("\nFeatures with additional NaN:")
print(
    other_nan_count[other_nan_count > 0]
    .sort_values(ascending=False)
)
```

```python
# ============================================================
# 3-CLASS ANOVA F-TEST — COMPLETE FEATURES ONLY
# ============================================================

from sklearn.feature_selection import f_classif

# Keep only features with NO missing values
complete_feature_cols = [
    c for c in feature_cols
    if class_feature_df[c].isna().sum() == 0
]

X_anova = class_feature_df[complete_feature_cols]
y_anova = class_feature_df["class"]

print("=" * 70)
print("3-CLASS ANOVA F-TEST")
print("=" * 70)

print("Total numeric features:", len(feature_cols))
print("Complete features used:", len(complete_feature_cols))
print("Features excluded because of NaN:",
      len(feature_cols) - len(complete_feature_cols))

# ANOVA
f_scores, p_values = f_classif(X_anova, y_anova)

anova_results = pd.DataFrame({
    "Feature": complete_feature_cols,
    "F_score": f_scores,
    "p_value": p_values
})

# Sort by F-score
anova_results = anova_results.sort_values(
    "F_score",
    ascending=False
).reset_index(drop=True)

print("\nTop 30 features by ANOVA F-score:")
print(anova_results.head(30).to_string(index=False))
```

```python
# ============================================================
# TOP ANOVA FEATURES — CLASS-WISE COMPARISON
# ============================================================

top_features = feature_stats.sort_values(
    "F_score", ascending=False
).head(20)["Feature"].tolist()

comparison = []

for feature in top_features:
    means = class_feature_df.groupby("class")[feature].mean()

    c1 = means.get(1, np.nan)
    c2 = means.get(2, np.nan)
    c3 = means.get(3, np.nan)

    comparison.append({
        "Feature": feature,
        "Class1_Mean": c1,
        "Class2_Mean": c2,
        "Class3_Mean": c3,
        "Abs_Diff_3_vs_1": abs(c3 - c1),
        "Abs_Diff_3_vs_2": abs(c3 - c2)
    })

top_anova_comparison = pd.DataFrame(comparison)

print("=" * 100)
print("TOP ANOVA FEATURES — CLASS-WISE COMPARISON")
print("=" * 100)

display(top_anova_comparison)
```

```python
# ============================================================
# RECREATE feature_stats FROM ANOVA RESULTS
# ============================================================

feature_stats = pd.DataFrame({
    "Feature": complete_features,
    "F_score": f_scores,
    "p_value": p_values
})

feature_stats = feature_stats.sort_values(
    "F_score",
    ascending=False
).reset_index(drop=True)

print("=" * 70)
print("ANOVA FEATURE STATS RECREATED")
print("=" * 70)

print("Total features:", len(feature_stats))
print("\nTop 20 features:")
display(feature_stats.head(20))
```

```python
# ============================================================
# RECREATE 3-CLASS ANOVA RESULTS
# ============================================================

import numpy as np
import pandas as pd
from sklearn.feature_selection import f_classif

# Numeric feature columns — class ko exclude karo
feature_cols = [
    c for c in class_feature_df.columns
    if c != "class"
    and pd.api.types.is_numeric_dtype(class_feature_df[c])
]

# Sirf features jisme koi NaN nahi hai
complete_features = [
    c for c in feature_cols
    if class_feature_df[c].notna().all()
]

print("=" * 70)
print("ANOVA INPUT")
print("=" * 70)

print("Total numeric features:", len(feature_cols))
print("Complete features:", len(complete_features))
print("Excluded due to NaN:", len(feature_cols) - len(complete_features))

# X and y
X = class_feature_df[complete_features]
y = class_feature_df["class"]

# ANOVA
f_scores, p_values = f_classif(X, y)

# Results DataFrame
feature_stats = pd.DataFrame({
    "Feature": complete_features,
    "F_score": f_scores,
    "p_value": p_values
})

feature_stats = feature_stats.sort_values(
    "F_score",
    ascending=False
).reset_index(drop=True)

print("\n" + "=" * 70)
print("TOP 20 ANOVA FEATURES")
print("=" * 70)

display(feature_stats.head(20))
```

```python
# ============================================================
# TOP ANOVA FEATURES — CLASS-WISE COMPARISON
# ============================================================

top_features = feature_stats.head(20)["Feature"].tolist()

class_comparison = (
    class_feature_df
    .groupby("class")[top_features]
    .mean()
    .T
    .reset_index()
    .rename(columns={"index": "Feature"})
)

# Rename class columns
class_comparison = class_comparison.rename(columns={
    1: "Class1_Mean",
    2: "Class2_Mean",
    3: "Class3_Mean"
})

# Difference of Class 3 from other classes
class_comparison["Abs_Diff_Class3_vs_Class1"] = (
    class_comparison["Class3_Mean"] -
    class_comparison["Class1_Mean"]
).abs()

class_comparison["Abs_Diff_Class3_vs_Class2"] = (
    class_comparison["Class3_Mean"] -
    class_comparison["Class2_Mean"]
).abs()

# Overall Class 3 distinctiveness
class_comparison["Class3_Distinctiveness"] = (
    class_comparison["Abs_Diff_Class3_vs_Class1"] +
    class_comparison["Abs_Diff_Class3_vs_Class2"]
) / 2

class_comparison = class_comparison.sort_values(
    "Class3_Distinctiveness",
    ascending=False
)

print("=" * 90)
print("TOP ANOVA FEATURES — CLASS-WISE COMPARISON")
print("=" * 90)

display(class_comparison)
```

```python
# ============================================================
# TOP ANOVA FEATURES — CORRELATION / REDUNDANCY ANALYSIS
# ============================================================

top_features = feature_stats.head(30)["Feature"].tolist()

corr_matrix = class_feature_df[top_features].corr()

# Find highly correlated feature pairs
high_corr_pairs = []

for i in range(len(top_features)):
    for j in range(i + 1, len(top_features)):
        
        f1 = top_features[i]
        f2 = top_features[j]
        corr = corr_matrix.loc[f1, f2]
        
        if abs(corr) >= 0.90:
            high_corr_pairs.append({
                "Feature_1": f1,
                "Feature_2": f2,
                "Correlation": corr,
                "Abs_Correlation": abs(corr)
            })

high_corr_df = pd.DataFrame(high_corr_pairs)

if len(high_corr_df) > 0:
    high_corr_df = high_corr_df.sort_values(
        "Abs_Correlation",
        ascending=False
    ).reset_index(drop=True)

print("=" * 90)
print("HIGH-CORRELATION FEATURE PAIRS")
print("=" * 90)

print("Top ANOVA features analysed:", len(top_features))
print("Highly correlated pairs (|r| >= 0.90):", len(high_corr_df))

display(high_corr_df)
```

```python
# ============================================================
# CLASS 3 vs OTHER CLASSES — FEATURE IMPORTANCE
# ============================================================

from sklearn.feature_selection import f_classif

# Binary target:
# Class 3 = 1
# Class 1 & 2 = 0

y_class3 = (class_feature_df["class"] == 3).astype(int)

# Use the 167 complete features
X_class3 = class_feature_df[complete_features]

# ANOVA: Class 3 vs non-Class-3
f_class3, p_class3 = f_classif(X_class3, y_class3)

class3_feature_stats = pd.DataFrame({
    "Feature": complete_features,
    "F_score": f_class3,
    "p_value": p_class3
})

class3_feature_stats = class3_feature_stats.sort_values(
    "F_score",
    ascending=False
).reset_index(drop=True)

print("=" * 90)
print("CLASS 3 vs NON-CLASS-3 FEATURE ANALYSIS")
print("=" * 90)

print("Class 3 samples:", int(y_class3.sum()))
print("Non-Class-3 samples:", int((y_class3 == 0).sum()))
print("Features analysed:", len(complete_features))

print("\nTop 30 features for Class 3 separation:")
display(class3_feature_stats.head(30))
```

```python
# ============================================================
# CLASS 3 — REDUNDANCY-AWARE FEATURE SELECTION
# ============================================================

# Top 30 Class-3 separation features
top_class3_features = class3_feature_stats.sort_values(
    "F_score",
    ascending=False
).head(30)["Feature"].tolist()

print("=" * 80)
print("TOP CLASS-3 FEATURES")
print("=" * 80)

for i, feature in enumerate(top_class3_features, 1):
    print(f"{i:2}. {feature}")

# Correlation matrix
corr_matrix = class_feature_df[top_class3_features].corr()

# Find highly correlated pairs
high_corr_pairs = []

for i in range(len(top_class3_features)):
    for j in range(i + 1, len(top_class3_features)):
        f1 = top_class3_features[i]
        f2 = top_class3_features[j]
        corr = corr_matrix.loc[f1, f2]

        if abs(corr) >= 0.90:
            high_corr_pairs.append({
                "Feature_1": f1,
                "Feature_2": f2,
                "Correlation": corr,
                "Abs_Correlation": abs(corr)
            })

high_corr_df = pd.DataFrame(high_corr_pairs)

print("\n" + "=" * 80)
print("HIGHLY CORRELATED TOP CLASS-3 FEATURES")
print("=" * 80)

print("Number of correlated pairs:", len(high_corr_df))

if len(high_corr_df) > 0:
    display(
        high_corr_df.sort_values(
            "Abs_Correlation",
            ascending=False
        )
    )
else:
    print("No correlation >= 0.90 found.")
```

```python
# ============================================================
# CLASS 3 — REDUNDANCY-AWARE REPRESENTATIVE FEATURES
# ============================================================

# Top Class-3 features
top_class3_features = class3_feature_stats.sort_values(
    "F_score",
    ascending=False
).head(30)["Feature"].tolist()

# Correlation threshold
CORR_THRESHOLD = 0.90

selected_features = []
removed_features = []
feature_groups = []

for feature in top_class3_features:

    highly_correlated_with_selected = []

    for selected in selected_features:
        corr = corr_matrix.loc[feature, selected]

        if abs(corr) >= CORR_THRESHOLD:
            highly_correlated_with_selected.append(
                (selected, corr)
            )

    if not highly_correlated_with_selected:
        selected_features.append(feature)

    else:
        removed_features.append(feature)

        feature_groups.append({
            "Removed_feature": feature,
            "Correlated_with": highly_correlated_with_selected
        })

print("=" * 80)
print("REDUNDANCY-AWARE FEATURE SELECTION")
print("=" * 80)

print("\nOriginal top features:", len(top_class3_features))
print("Selected representative features:", len(selected_features))
print("Removed redundant features:", len(removed_features))

print("\n" + "=" * 80)
print("SELECTED FEATURES")
print("=" * 80)

for i, feature in enumerate(selected_features, 1):
    print(f"{i:2}. {feature}")

print("\n" + "=" * 80)
print("REMOVED REDUNDANT FEATURES")
print("=" * 80)

for feature in removed_features:
    print("-", feature)
```

```python
# ============================================================
# CLASS 3 — SELECTED FEATURE SET SUMMARY
# ============================================================

print("=" * 80)
print("FINAL CANDIDATE FEATURE SET — CLASS 3")
print("=" * 80)

print("Number of candidate features:", len(selected_features))

print("\nSelected features:")
for i, feature in enumerate(selected_features, 1):
    print(f"{i:2}. {feature}")

# Create reduced dataset
X_class3_selected = class_feature_df[selected_features].copy()
y_class3 = (class_feature_df["class"] == 3).astype(int)

print("\n" + "=" * 80)
print("REDUCED DATASET")
print("=" * 80)

print("X shape:", X_class3_selected.shape)
print("y shape:", y_class3.shape)

print("\nTarget distribution:")
print(y_class3.value_counts())

print("\nTarget percentages:")
print(
    y_class3.value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
# ============================================================
# CLASS 3 vs NON-CLASS-3
# TEMPORAL MODEL INPUT PREPARATION
# ============================================================

# Selected features
X = class_feature_df[selected_features].copy()

# Class 3 = 1, everything else = 0
y = (class_feature_df["class"] == 3).astype(int)

# Get corresponding time-step
time_series = class_feature_df["Time step"].copy()

print("=" * 80)
print("CLASS 3 TEMPORAL MODEL INPUT")
print("=" * 80)

print("X shape:", X.shape)
print("y shape:", y.shape)
print("Time-step shape:", time_series.shape)

print("\nTime-step range:")
print("Minimum:", time_series.min())
print("Maximum:", time_series.max())

print("\nClass distribution:")
print(y.value_counts())

print("\nClass percentage:")
print(
    y.value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("\nSelected features:")
for i, feature in enumerate(selected_features, 1):
    print(f"{i:2}. {feature}")
```

```python
# ============================================================
# TEMPORAL SPLIT — CLASS 3 DETECTION
# ============================================================

temporal_check = pd.DataFrame({
    "Time_step": time_series.to_numpy(),
    "Class3": y.to_numpy()
})

time_distribution = (
    temporal_check
    .groupby("Time_step")
    .agg(
        Total=("Class3", "size"),
        Class3=("Class3", "sum")
    )
)

time_distribution["Non_Class3"] = (
    time_distribution["Total"] - time_distribution["Class3"]
)

time_distribution["Class3_%"] = (
    time_distribution["Class3"] /
    time_distribution["Total"] * 100
).round(2)

print("=" * 90)
print("CLASS 3 TEMPORAL DISTRIBUTION")
print("=" * 90)

display(time_distribution)

print("\nTime steps:", time_distribution.index.min(), "to", time_distribution.index.max())
print("Total transactions:", time_distribution["Total"].sum())
print("Total Class 3:", time_distribution["Class3"].sum())
```

```python
# ============================================================
# BIT-TRACE / ELLIPTIC++ TRANSACTIONS
# CLASS-3 / UNKNOWN TRANSACTION ANOMALY PIPELINE
# ============================================================
#
# PURPOSE:
#   1. Load selected transaction features
#   2. Validate data quality
#   3. Prepare temporal data
#   4. Train UNSUPERVISED Isolation Forest
#   5. Generate anomaly scores
#   6. Analyse thresholds
#   7. Validate anomalies against Class 1/2/3 labels
#   8. Analyse anomaly behaviour by time step
#   9. Perform feature-level anomaly analysis
#  10. Save results for BIT-TRACE
#
# IMPORTANT:
#   Class labels are NOT used for model training.
#   They are used ONLY AFTER prediction for evaluation/analysis.
# ============================================================

import os
import gc
import warnings
import numpy as np
import pandas as pd

from sklearn.ensemble import IsolationForest
from sklearn.impute import SimpleImputer

warnings.filterwarnings("ignore")

print("=" * 80)
print("BIT-TRACE — ELLIPTIC++ UNSUPERVISED ANOMALY PIPELINE")
print("=" * 80)


# ============================================================
# 1. PATHS
# ============================================================

BASE_DIR = r"C:\Users\YASHIKA\Desktop\bit-trace"

TX_DIR = os.path.join(
    BASE_DIR,
    "EllipticPlusPlus",
    "Transactions Dataset"
)

FEATURE_FILE = os.path.join(TX_DIR, "txs_features.csv")
CLASS_FILE = os.path.join(TX_DIR, "txs_classes.csv")

print("\n[1] PATH CHECK")
print("-" * 80)

print("Feature file exists:", os.path.exists(FEATURE_FILE))
print("Class file exists:", os.path.exists(CLASS_FILE))

if not os.path.exists(FEATURE_FILE):
    raise FileNotFoundError(FEATURE_FILE)

if not os.path.exists(CLASS_FILE):
    raise FileNotFoundError(CLASS_FILE)


# ============================================================
# 2. SELECTED FEATURES
# ============================================================
#
# These are the 13 features obtained from your previous
# Class-3 redundancy-aware feature selection.
# ============================================================

SELECTED_FEATURES = [
    "Local_feature_53",
    "Local_feature_54",
    "Local_feature_52",
    "Local_feature_59",
    "Aggregate_feature_21",
    "Local_feature_25",
    "Local_feature_77",
    "Local_feature_89",
    "Local_feature_90",
    "Aggregate_feature_32",
    "Aggregate_feature_14",
    "Local_feature_22",
    "Local_feature_47"
]

print("\n[2] SELECTED FEATURES")
print("-" * 80)

for i, feature in enumerate(SELECTED_FEATURES, 1):
    print(f"{i:2d}. {feature}")

print("\nNumber of selected features:", len(SELECTED_FEATURES))


# ============================================================
# 3. LOAD CLASS DATA
# ============================================================

print("\n[3] LOADING CLASS DATA")
print("-" * 80)

classes = pd.read_csv(CLASS_FILE)

print("Class data shape:", classes.shape)
print("Columns:", classes.columns.tolist())

required_class_cols = {"txId", "class"}

if not required_class_cols.issubset(classes.columns):
    raise ValueError(
        f"Expected columns {required_class_cols}, "
        f"found {set(classes.columns)}"
    )

classes["txId"] = pd.to_numeric(classes["txId"], errors="coerce")
classes["class"] = pd.to_numeric(classes["class"], errors="coerce")

print("\nClass distribution:")
print(classes["class"].value_counts().sort_index())


# ============================================================
# 4. LOAD ONLY NECESSARY COLUMNS FROM HUGE FEATURES FILE
# ============================================================
#
# IMPORTANT:
# We do NOT load all 185 columns.
# This saves memory and time.
#
# We first inspect the header to find the time-step column
# and transaction ID column.
# ============================================================

print("\n[4] READING FEATURE HEADER")
print("-" * 80)

header = pd.read_csv(
    FEATURE_FILE,
    nrows=0
)

all_columns = header.columns.tolist()

print("Total columns in feature file:", len(all_columns))

print("\nFirst columns:")
print(all_columns[:20])


# ------------------------------------------------------------
# Identify ID and time columns robustly
# ------------------------------------------------------------

def find_column(columns, candidates):
    lower_map = {c.lower(): c for c in columns}

    for candidate in candidates:
        if candidate.lower() in lower_map:
            return lower_map[candidate.lower()]

    return None


TX_ID_COL = find_column(
    all_columns,
    ["txId", "txID", "tx_id", "transaction_id"]
)

TIME_COL = find_column(
    all_columns,
    ["time_step", "Time_step", "time step", "time"]
)

print("\nDetected transaction ID column:", TX_ID_COL)
print("Detected time column:", TIME_COL)

if TX_ID_COL is None:
    raise ValueError(
        "Could not automatically identify transaction ID column."
    )

if TIME_COL is None:
    raise ValueError(
        "Could not automatically identify time-step column."
    )


# ------------------------------------------------------------
# Check selected features exist
# ------------------------------------------------------------

missing_features = [
    f for f in SELECTED_FEATURES
    if f not in all_columns
]

if missing_features:
    raise ValueError(
        "These selected features were not found:\n"
        + "\n".join(missing_features)
    )


# ============================================================
# 5. LOAD ONLY REQUIRED DATA
# ============================================================

needed_columns = [
    TX_ID_COL,
    TIME_COL
] + SELECTED_FEATURES

# remove duplicates while preserving order
needed_columns = list(dict.fromkeys(needed_columns))

print("\n[5] LOADING REQUIRED FEATURE COLUMNS")
print("-" * 80)

print("Columns being loaded:", len(needed_columns))

feature_data = pd.read_csv(
    FEATURE_FILE,
    usecols=needed_columns,
    low_memory=False
)

print("\nLoaded shape:", feature_data.shape)


# ============================================================
# 6. BASIC CLEANING
# ============================================================

print("\n[6] BASIC DATA CLEANING")
print("-" * 80)

feature_data[TX_ID_COL] = pd.to_numeric(
    feature_data[TX_ID_COL],
    errors="coerce"
)

feature_data[TIME_COL] = pd.to_numeric(
    feature_data[TIME_COL],
    errors="coerce"
)

for feature in SELECTED_FEATURES:
    feature_data[feature] = pd.to_numeric(
        feature_data[feature],
        errors="coerce"
    )

print("Missing txId:", feature_data[TX_ID_COL].isna().sum())
print("Missing time step:", feature_data[TIME_COL].isna().sum())


# ============================================================
# 7. MERGE WITH CLASS LABELS
# ============================================================

print("\n[7] MERGING FEATURES + LABELS")
print("-" * 80)

data = feature_data.merge(
    classes[["txId", "class"]],
    left_on=TX_ID_COL,
    right_on="txId",
    how="left"
)

# remove duplicate txId column if needed
if "txId" in data.columns and TX_ID_COL != "txId":
    data.drop(columns=["txId"], inplace=True)

print("Merged shape:", data.shape)

print("\nMissing labels:")
print(data["class"].isna().sum())

print("\nClass distribution after merge:")
print(data["class"].value_counts(dropna=False).sort_index())


# ============================================================
# 8. VALIDATE TIME STEP
# ============================================================

print("\n[8] TEMPORAL VALIDATION")
print("-" * 80)

print("Minimum time step:", data[TIME_COL].min())
print("Maximum time step:", data[TIME_COL].max())
print("Number of unique time steps:", data[TIME_COL].nunique())

print("\nTransactions per time step:")
print(
    data[TIME_COL]
    .value_counts()
    .sort_index()
    .to_string()
)


# ============================================================
# 9. DATA QUALITY CHECK
# ============================================================

print("\n[9] DATA QUALITY CHECK")
print("-" * 80)

X_raw = data[SELECTED_FEATURES].copy()

nan_counts = X_raw.isna().sum()
inf_counts = np.isinf(X_raw.to_numpy(dtype=float)).sum(axis=0)

quality_table = pd.DataFrame({
    "NaN_count": nan_counts,
    "Inf_count": inf_counts
})

print(quality_table[quality_table.sum(axis=1) > 0])

print(
    "\nTotal NaN values:",
    int(X_raw.isna().sum().sum())
)

print(
    "Total Inf values:",
    int(np.isinf(X_raw.to_numpy(dtype=float)).sum())
)


# ============================================================
# 10. REPLACE INF WITH NaN
# ============================================================

X_raw = X_raw.replace(
    [np.inf, -np.inf],
    np.nan
)


# ============================================================
# 11. IMPUTATION
# ============================================================
#
# We use median imputation.
#
# IMPORTANT:
# This is preprocessing only.
# Class labels are NOT used.
# ============================================================

print("\n[10] MEDIAN IMPUTATION")
print("-" * 80)

imputer = SimpleImputer(strategy="median")

X = imputer.fit_transform(X_raw)

X = pd.DataFrame(
    X,
    columns=SELECTED_FEATURES,
    index=data.index
)

print("Final X shape:", X.shape)
print("Remaining NaN:", int(X.isna().sum().sum()))


# ============================================================
# 12. FEATURE SUMMARY
# ============================================================

print("\n[11] FEATURE SUMMARY")
print("-" * 80)

summary = pd.DataFrame({
    "mean": X.mean(),
    "std": X.std(),
    "min": X.min(),
    "median": X.median(),
    "max": X.max()
})

print(summary.to_string())


# ============================================================
# 13. TEMPORAL TRAIN / VALIDATION / TEST SPLIT
# ============================================================
#
# We use chronological splitting.
#
# 70% earliest time steps -> TRAIN
# next 15%                 -> VALIDATION
# final 15%                -> TEST
#
# The exact boundaries are derived automatically from the
# 49 available time steps.
# ============================================================

print("\n[12] TEMPORAL SPLIT")
print("-" * 80)

unique_times = np.sort(
    data[TIME_COL].dropna().unique()
)

n_times = len(unique_times)

train_end_idx = int(n_times * 0.70)
val_end_idx = int(n_times * 0.85)

# ensure non-empty ranges
train_end_idx = max(1, train_end_idx)
val_end_idx = max(train_end_idx + 1, val_end_idx)
val_end_idx = min(val_end_idx, n_times - 1)

train_times = unique_times[:train_end_idx]
val_times = unique_times[train_end_idx:val_end_idx]
test_times = unique_times[val_end_idx:]

print("Train time steps:",
      train_times[0], "to", train_times[-1])

print("Validation time steps:",
      val_times[0], "to", val_times[-1])

print("Test time steps:",
      test_times[0], "to", test_times[-1])


train_mask = data[TIME_COL].isin(train_times)
val_mask = data[TIME_COL].isin(val_times)
test_mask = data[TIME_COL].isin(test_times)

X_train = X.loc[train_mask]
X_val = X.loc[val_mask]
X_test = X.loc[test_mask]

print("\nTrain samples:", len(X_train))
print("Validation samples:", len(X_val))
print("Test samples:", len(X_test))


# ============================================================
# 14. TRAINING DATA LABEL INDEPENDENCE CHECK
# ============================================================
#
# This is only for reporting.
# Labels are NOT passed into Isolation Forest.
# ============================================================

print("\n[13] TRAINING DATA LABEL DISTRIBUTION")
print("-" * 80)

print(
    data.loc[train_mask, "class"]
    .value_counts()
    .sort_index()
)

print(
    "\nNOTE: These labels are NOT used for model training."
)


# ============================================================
# 15. ISOLATION FOREST
# ============================================================

print("\n[14] TRAINING ISOLATION FOREST")
print("-" * 80)

model = IsolationForest(
    n_estimators=300,
    max_samples="auto",
    contamination="auto",
    random_state=42,
    n_jobs=-1
)

model.fit(X_train)

print("Isolation Forest training complete.")


# ============================================================
# 16. GENERATE SCORES
# ============================================================
#
# sklearn decision_function:
#   higher = more normal
#
# Therefore:
#   anomaly_score = -decision_function
#
# higher anomaly_score = more anomalous
# ============================================================

print("\n[15] GENERATING ANOMALY SCORES")
print("-" * 80)

data["anomaly_score"] = np.nan
data["iforest_prediction"] = np.nan

train_scores = -model.decision_function(X_train)
val_scores = -model.decision_function(X_val)
test_scores = -model.decision_function(X_test)

train_pred = model.predict(X_train)
val_pred = model.predict(X_val)
test_pred = model.predict(X_test)

data.loc[train_mask, "anomaly_score"] = train_scores
data.loc[val_mask, "anomaly_score"] = val_scores
data.loc[test_mask, "anomaly_score"] = test_scores

data.loc[train_mask, "iforest_prediction"] = train_pred
data.loc[val_mask, "iforest_prediction"] = val_pred
data.loc[test_mask, "iforest_prediction"] = test_pred


# ============================================================
# 17. SCORE DISTRIBUTION
# ============================================================

print("\n[16] ANOMALY SCORE DISTRIBUTION")
print("-" * 80)

score_summary = pd.DataFrame({
    "Train": train_scores,
    "Validation": pd.Series(val_scores),
    "Test": pd.Series(test_scores)
})

print(
    score_summary.describe(
        percentiles=[0.01, 0.02, 0.05, 0.10, 0.25, 0.50, 0.75, 0.90, 0.95, 0.99]
    ).to_string()
)


# ============================================================
# 18. THRESHOLD ANALYSIS
# ============================================================
#
# Instead of trusting the default Isolation Forest threshold,
# analyse several anomaly-rate thresholds.
# ============================================================

print("\n[17] THRESHOLD ANALYSIS")
print("-" * 80)

threshold_rates = [
    0.01,
    0.02,
    0.03,
    0.05,
    0.10
]

threshold_rows = []

test_score_series = pd.Series(
    test_scores,
    index=data.index[test_mask]
)

for rate in threshold_rates:

    threshold = test_score_series.quantile(
        1 - rate
    )

    anomaly_mask = test_score_series >= threshold

    threshold_rows.append({
        "Anomaly_Rate": rate,
        "Threshold": threshold,
        "Anomalies": int(anomaly_mask.sum())
    })

threshold_table = pd.DataFrame(threshold_rows)

print(
    threshold_table.to_string(index=False)
)


# ============================================================
# 19. CHOOSE 5% THRESHOLD FOR INITIAL INVESTIGATION
# ============================================================
#
# NOTE:
# 5% is an investigation threshold, NOT a claim that 5% are
# actually fraudulent.
# ============================================================

chosen_rate = 0.05

chosen_threshold = test_score_series.quantile(
    1 - chosen_rate
)

print("\nChosen investigation threshold:", chosen_threshold)

data["is_anomaly"] = False

data.loc[
    test_mask &
    (data["anomaly_score"] >= chosen_threshold),
    "is_anomaly"
] = True


# ============================================================
# 20. OVERALL TEST ANOMALY SUMMARY
# ============================================================

print("\n[18] TEST ANOMALY SUMMARY")
print("-" * 80)

test_results = data.loc[test_mask].copy()

print("Test transactions:", len(test_results))
print(
    "Detected anomalies:",
    int(test_results["is_anomaly"].sum())
)

print(
    "Observed anomaly percentage:",
    round(
        test_results["is_anomaly"].mean() * 100,
        2
    ),
    "%"
)


# ============================================================
# 21. CLASS-WISE ANOMALY ANALYSIS
# ============================================================
#
# IMPORTANT:
# Labels are used ONLY now for evaluation.
# ============================================================

print("\n[19] CLASS-WISE ANOMALY ANALYSIS")
print("-" * 80)

class_analysis = (
    test_results
    .groupby("class")
    .agg(
        Total=("class", "size"),
        Anomalies=("is_anomaly", "sum"),
        Mean_Anomaly_Score=("anomaly_score", "mean"),
        Median_Anomaly_Score=("anomaly_score", "median"),
        Max_Anomaly_Score=("anomaly_score", "max")
    )
)

class_analysis["Anomaly_%"] = (
    class_analysis["Anomalies"]
    / class_analysis["Total"]
    * 100
)

print(
    class_analysis.to_string()
)


# ============================================================
# 22. CLASS-3 SPECIFIC ANALYSIS
# ============================================================

print("\n[20] CLASS-3 / UNKNOWN ANALYSIS")
print("-" * 80)

class3_test = test_results[
    test_results["class"] == 3
]

non_class3_test = test_results[
    test_results["class"] != 3
]

print("Class-3 test samples:", len(class3_test))
print("Non-Class-3 test samples:", len(non_class3_test))

print(
    "\nClass-3 anomalies:",
    int(class3_test["is_anomaly"].sum())
)

print(
    "Class-3 anomaly percentage:",
    round(
        class3_test["is_anomaly"].mean() * 100,
        2
    ),
    "%"
)

print(
    "\nNon-Class-3 anomalies:",
    int(non_class3_test["is_anomaly"].sum())
)

print(
    "Non-Class-3 anomaly percentage:",
    round(
        non_class3_test["is_anomaly"].mean() * 100,
        2
    ),
    "%"
)


# ============================================================
# 23. CLASS-1 / ILLICIT OVERLAP
# ============================================================

print("\n[21] KNOWN ILLICIT OVERLAP")
print("-" * 80)

class1_test = test_results[
    test_results["class"] == 1
]

class2_test = test_results[
    test_results["class"] == 2
]

print("Known Class-1 samples:", len(class1_test))
print(
    "Class-1 detected as anomaly:",
    int(class1_test["is_anomaly"].sum())
)

if len(class1_test) > 0:
    print(
        "Class-1 anomaly rate:",
        round(
            class1_test["is_anomaly"].mean() * 100,
            2
        ),
        "%"
    )

print("\nKnown Class-2 samples:", len(class2_test))
print(
    "Class-2 detected as anomaly:",
    int(class2_test["is_anomaly"].sum())
)

if len(class2_test) > 0:
    print(
        "Class-2 anomaly rate:",
        round(
            class2_test["is_anomaly"].mean() * 100,
            2
        ),
        "%"
    )


# ============================================================
# 24. TEMPORAL ANOMALY ANALYSIS
# ============================================================

print("\n[22] TEMPORAL ANOMALY ANALYSIS")
print("-" * 80)

temporal_analysis = (
    test_results
    .groupby(TIME_COL)
    .agg(
        Transactions=("class", "size"),
        Anomalies=("is_anomaly", "sum"),
        Mean_Anomaly_Score=("anomaly_score", "mean"),
        Class3_Count=("class", lambda x: (x == 3).sum()),
        Class1_Count=("class", lambda x: (x == 1).sum()),
        Class2_Count=("class", lambda x: (x == 2).sum())
    )
)

temporal_analysis["Anomaly_%"] = (
    temporal_analysis["Anomalies"]
    / temporal_analysis["Transactions"]
    * 100
)

temporal_analysis["Class3_%"] = (
    temporal_analysis["Class3_Count"]
    / temporal_analysis["Transactions"]
    * 100
)

print(
    temporal_analysis.to_string()
)


# ============================================================
# 25. TOP ANOMALOUS TRANSACTIONS
# ============================================================

print("\n[23] TOP 30 ANOMALOUS TEST TRANSACTIONS")
print("-" * 80)

top_anomalies = (
    test_results
    .sort_values(
        "anomaly_score",
        ascending=False
    )
    [[
        TX_ID_COL,
        TIME_COL,
        "class",
        "anomaly_score"
    ]]
    .head(30)
)

print(
    top_anomalies.to_string(index=False)
)


# ============================================================
# 26. FEATURE-LEVEL ANOMALY ANALYSIS
# ============================================================
#
# We compare:
#
#   normal test transactions
#           VS
#   detected anomalies
#
# using standardized absolute deviations based on the
# training distribution.
# ============================================================

print("\n[24] FEATURE-LEVEL ANOMALY ANALYSIS")
print("-" * 80)

train_feature_mean = X_train.mean()
train_feature_std = X_train.std().replace(0, np.nan)

test_feature_values = X.loc[test_mask]

z_test = (
    test_feature_values
    - train_feature_mean
) / train_feature_std

z_test = z_test.abs()

z_test["is_anomaly"] = test_results["is_anomaly"].values

normal_z = z_test[
    z_test["is_anomaly"] == False
].drop(columns="is_anomaly")

anomaly_z = z_test[
    z_test["is_anomaly"] == True
].drop(columns="is_anomaly")

feature_effect = pd.DataFrame({
    "Normal_Mean_Abs_Z": normal_z.mean(),
    "Anomaly_Mean_Abs_Z": anomaly_z.mean()
})

feature_effect["Difference"] = (
    feature_effect["Anomaly_Mean_Abs_Z"]
    - feature_effect["Normal_Mean_Abs_Z"]
)

feature_effect = feature_effect.sort_values(
    "Difference",
    ascending=False
)

print(
    feature_effect.to_string()
)


# ============================================================
# 27. TOP FEATURES ASSOCIATED WITH DETECTED ANOMALIES
# ============================================================

print("\n[25] TOP FEATURES ASSOCIATED WITH DETECTED ANOMALIES")
print("-" * 80)

print(
    feature_effect
    .head(13)
    .to_string()
)


# ============================================================
# 28. CLASS-3 FEATURE COMPARISON
# ============================================================

print("\n[26] CLASS-3 VS NON-CLASS-3 — ANOMALY SCORE")
print("-" * 80)

class3_score = test_results.loc[
    test_results["class"] == 3,
    "anomaly_score"
]

non_class3_score = test_results.loc[
    test_results["class"] != 3,
    "anomaly_score"
]

print(
    "Class-3 mean score:",
    class3_score.mean()
)

print(
    "Non-Class-3 mean score:",
    non_class3_score.mean()
)

print(
    "Class-3 median score:",
    class3_score.median()
)

print(
    "Non-Class-3 median score:",
    non_class3_score.median()
)


# ============================================================
# 29. SAVE RESULTS
# ============================================================

print("\n[27] SAVING RESULTS")
print("-" * 80)

OUTPUT_DIR = os.path.join(
    BASE_DIR,
    "outputs",
    "elliptic_transactions"
)

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)

# Main transaction-level results
result_file = os.path.join(
    OUTPUT_DIR,
    "elliptic_transaction_anomaly_results.csv"
)

data.to_csv(
    result_file,
    index=False
)

# Class analysis
class_file = os.path.join(
    OUTPUT_DIR,
    "class_wise_anomaly_analysis.csv"
)

class_analysis.to_csv(
    class_file
)

# Temporal analysis
temporal_file = os.path.join(
    OUTPUT_DIR,
    "temporal_anomaly_analysis.csv"
)

temporal_analysis.to_csv(
    temporal_file
)

# Feature analysis
feature_file = os.path.join(
    OUTPUT_DIR,
    "anomaly_feature_analysis.csv"
)

feature_effect.to_csv(
    feature_file
)

# Threshold table
threshold_file = os.path.join(
    OUTPUT_DIR,
    "anomaly_threshold_analysis.csv"
)

threshold_table.to_csv(
    threshold_file,
    index=False
)

print("Saved:")
print("1.", result_file)
print("2.", class_file)
print("3.", temporal_file)
print("4.", feature_file)
print("5.", threshold_file)


# ============================================================
# 30. FINAL PIPELINE SUMMARY
# ============================================================

print("\n" + "=" * 80)
print("PIPELINE COMPLETE")
print("=" * 80)

print("\nDataset:")
print("  Transactions:", len(data))
print("  Time steps:", data[TIME_COL].nunique())

print("\nFeatures:")
print("  Original selected:", len(SELECTED_FEATURES))
print("  Used:", len(SELECTED_FEATURES))

print("\nModel:")
print("  Isolation Forest")
print("  Estimators:", 300)
print("  Training:", "earlier time steps only")
print("  Labels used during training: NO")

print("\nInvestigation threshold:")
print("  Anomaly rate:", chosen_rate * 100, "%")
print("  Threshold:", chosen_threshold)

print("\nTest results:")
print("  Test transactions:", len(test_results))
print(
    "  Detected anomalies:",
    int(test_results["is_anomaly"].sum())
)

print("\nClass-3:")
print("  Test Class-3:", len(class3_test))
print(
    "  Class-3 anomalies:",
    int(class3_test["is_anomaly"].sum())
)

print("\nOutput directory:")
print(OUTPUT_DIR)

print("\n" + "=" * 80)
print("NEXT STAGE AFTER THIS:")
print("TRANSACTION GRAPH + ANOMALY CORRELATION")
print("=" * 80)


# Free some temporary memory
gc.collect()
```

```python
# ============================================================
# [16] ANOMALY SCORE DISTRIBUTION — FIXED
# ============================================================

print("=" * 80)
print("ANOMALY SCORE DISTRIBUTION")
print("=" * 80)

score_summary = pd.DataFrame({
    "Split": ["Train", "Validation", "Test"],
    "Samples": [
        len(train_scores),
        len(val_scores),
        len(test_scores)
    ],
    "Mean_Score": [
        np.mean(train_scores),
        np.mean(val_scores),
        np.mean(test_scores)
    ],
    "Std_Score": [
        np.std(train_scores),
        np.std(val_scores),
        np.std(test_scores)
    ],
    "Min_Score": [
        np.min(train_scores),
        np.min(val_scores),
        np.min(test_scores)
    ],
    "Median_Score": [
        np.median(train_scores),
        np.median(val_scores),
        np.median(test_scores)
    ],
    "Max_Score": [
        np.max(train_scores),
        np.max(val_scores),
        np.max(test_scores)
    ]
})

print(score_summary.to_string(index=False))
```

```python
# ============================================================
# [17] ANOMALY SCORES + TRUE CLASSES
#     EVALUATION ONLY — LABELS NOT USED FOR TRAINING
# ============================================================

print("=" * 80)
print("UNSUPERVISED ANOMALY DETECTION — CLASS-WISE EVALUATION")
print("=" * 80)

# Create evaluation dataframe
train_eval = merged.iloc[train_idx][["txId", "class", "Time step"]].copy()
val_eval   = merged.iloc[val_idx][["txId", "class", "Time step"]].copy()
test_eval  = merged.iloc[test_idx][["txId", "class", "Time step"]].copy()

train_eval["anomaly_score"] = train_scores
val_eval["anomaly_score"]   = val_scores
test_eval["anomaly_score"]  = test_scores

eval_df = pd.concat(
    [train_eval, val_eval, test_eval],
    ignore_index=True
)

print("\nEvaluation dataset shape:", eval_df.shape)

print("\nClass distribution:")
print(eval_df["class"].value_counts().sort_index())

print("\nAnomaly score statistics by class:")
class_score_stats = eval_df.groupby("class")["anomaly_score"].agg(
    ["count", "mean", "std", "min", "median", "max"]
)

print(class_score_stats)
```

```python
# ============================================================
# BIT-TRACE — UNSUPERVISED ANOMALY DETECTION
# CLASS-WISE EVALUATION
# ============================================================

import pandas as pd
import numpy as np

print("=" * 80)
print("UNSUPERVISED ANOMALY DETECTION — CLASS-WISE EVALUATION")
print("=" * 80)

# ------------------------------------------------------------
# 1. RECREATE EVALUATION DATA FROM THE DATAFRAME
# ------------------------------------------------------------

# Your main merged dataframe was not retained in memory.
# Recreate only the required metadata from the loaded dataset.

eval_df = pd.DataFrame({
    "txId": data["txId"].values,
    "Time step": data["Time step"].values,
    "class": data["class"].values
})

print("\nEvaluation dataframe:", eval_df.shape)

# ------------------------------------------------------------
# 2. ATTACH ANOMALY SCORES
# ------------------------------------------------------------

eval_df["anomaly_score"] = np.nan

eval_df.loc[train_idx, "anomaly_score"] = train_scores
eval_df.loc[val_idx, "anomaly_score"] = val_scores
eval_df.loc[test_idx, "anomaly_score"] = test_scores

# ------------------------------------------------------------
# 3. VERIFY
# ------------------------------------------------------------

print("\nMissing anomaly scores:",
      eval_df["anomaly_score"].isna().sum())

print("Total rows:", len(eval_df))

# ------------------------------------------------------------
# 4. SCORE SUMMARY BY CLASS
# ------------------------------------------------------------

class_summary = (
    eval_df
    .groupby("class")["anomaly_score"]
    .agg(
        Samples="count",
        Mean="mean",
        Std="std",
        Min="min",
        Median="median",
        Max="max"
    )
    .round(6)
)

print("\n" + "=" * 80)
print("ANOMALY SCORE BY ORIGINAL CLASS")
print("=" * 80)

print(class_summary)

# ------------------------------------------------------------
# 5. CLASS-WISE SCORE QUANTILES
# ------------------------------------------------------------

quantiles = (
    eval_df
    .groupby("class")["anomaly_score"]
    .quantile([0.01, 0.05, 0.10, 0.25, 0.50, 0.75, 0.90, 0.95, 0.99])
    .unstack()
)

quantiles.columns = [
    "Q01", "Q05", "Q10", "Q25",
    "Q50", "Q75", "Q90", "Q95", "Q99"
]

print("\n" + "=" * 80)
print("CLASS-WISE ANOMALY SCORE QUANTILES")
print("=" * 80)

print(quantiles.round(6))

# ------------------------------------------------------------
# 6. MOST ANOMALOUS TRANSACTIONS
# ------------------------------------------------------------

# Isolation Forest:
# lower score = more anomalous

most_anomalous = (
    eval_df
    .sort_values("anomaly_score", ascending=True)
    .head(50)
)

print("\n" + "=" * 80)
print("TOP 50 MOST ANOMALOUS TRANSACTIONS")
print("=" * 80)

print(
    most_anomalous[
        ["txId", "Time step", "class", "anomaly_score"]
    ].to_string(index=False)
)

# ------------------------------------------------------------
# 7. CLASS DISTRIBUTION AMONG TOP ANOMALIES
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("CLASS DISTRIBUTION AMONG TOP ANOMALIES")
print("=" * 80)

for n in [50, 100, 500, 1000, 5000]:

    top_n = eval_df.nsmallest(n, "anomaly_score")

    counts = top_n["class"].value_counts().sort_index()

    percentages = (
        top_n["class"]
        .value_counts(normalize=True)
        .sort_index()
        .mul(100)
        .round(2)
    )

    result = pd.DataFrame({
        "Count": counts,
        "Percentage": percentages
    })

    print(f"\nTop {n} anomalies:")
    print(result)

# ------------------------------------------------------------
# 8. CLASS 3 REPRESENTATION IN ANOMALIES
# ------------------------------------------------------------

overall_class3_pct = (
    (eval_df["class"] == 3).mean() * 100
)

print("\n" + "=" * 80)
print("CLASS 3 ANOMALY REPRESENTATION")
print("=" * 80)

print(f"Overall Class 3 percentage: {overall_class3_pct:.2f}%")

for n in [50, 100, 500, 1000, 5000]:

    top_n = eval_df.nsmallest(n, "anomaly_score")

    class3_count = (top_n["class"] == 3).sum()
    class3_pct = class3_count / n * 100

    enrichment = class3_pct / overall_class3_pct

    print(
        f"Top {n:5d}: "
        f"Class 3 = {class3_count:5d} "
        f"({class3_pct:6.2f}%) | "
        f"Enrichment = {enrichment:.3f}x"
    )

# ------------------------------------------------------------
# 9. TEMPORAL ANOMALY ANALYSIS
# ------------------------------------------------------------

temporal = (
    eval_df
    .groupby("Time step")
    .agg(
        Transactions=("txId", "count"),
        Mean_Anomaly_Score=("anomaly_score", "mean"),
        Median_Anomaly_Score=("anomaly_score", "median"),
        Class3_Count=("class", lambda x: (x == 3).sum())
    )
)

temporal["Class3_%"] = (
    temporal["Class3_Count"]
    / temporal["Transactions"]
    * 100
)

print("\n" + "=" * 80)
print("MOST ANOMALOUS TIME STEPS")
print("=" * 80)

print(
    temporal
    .sort_values("Mean_Anomaly_Score")
    .head(15)
    .round(4)
)

print("\n" + "=" * 80)
print("EVALUATION COMPLETE")
print("=" * 80)
```

```python
# =============================================================================
# BIT-TRACE — CLASS-2-ONLY UNSUPERVISED ANOMALY DETECTION
# =============================================================================
# IMPORTANT:
# - Class labels are NOT used by Isolation Forest during training.
# - Class 2 is used ONLY to define the "known-normal/reference" training pool.
# - Class 1 / Class 2 / Class 3 labels are used AFTER prediction for evaluation.
# - Temporal split:
#       Train      : Time 1–34
#       Validation : Time 35–41
#       Test       : Time 42–49
# =============================================================================

import numpy as np
import pandas as pd

from sklearn.ensemble import IsolationForest
from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    precision_score,
    recall_score,
    confusion_matrix
)

# =============================================================================
# 1. RECOVER DATA
# =============================================================================

print("=" * 90)
print("BIT-TRACE — CLASS-2-ONLY UNSUPERVISED ANOMALY DETECTION")
print("=" * 90)

# Try to locate the dataframe created by the previous pipeline
if "merged" in globals():
    data = merged.copy()

elif "eval_df" in globals():
    data = eval_df.copy()

else:
    raise NameError(
        "Required dataframe not found. "
        "Run the previous Elliptic++ loading/merging cell first."
    )

print("\n[1] DATA CHECK")
print("-" * 90)

print("Rows:", len(data))
print("Columns:", len(data.columns))

# -------------------------------------------------------------------------
# Detect important columns
# -------------------------------------------------------------------------

tx_col = "txId"

if tx_col not in data.columns:
    raise ValueError("txId column not found.")

if "class" not in data.columns:
    raise ValueError("class column not found.")

if "Time step" not in data.columns:
    raise ValueError("'Time step' column not found.")

print("txId column       :", tx_col)
print("class column      : class")
print("time column       : Time step")


# =============================================================================
# 2. SELECT FEATURES
# =============================================================================

selected_features = [
    "Local_feature_53",
    "Local_feature_54",
    "Local_feature_52",
    "Local_feature_59",
    "Aggregate_feature_21",
    "Local_feature_25",
    "Local_feature_77",
    "Local_feature_89",
    "Local_feature_90",
    "Aggregate_feature_32",
    "Aggregate_feature_14",
    "Local_feature_22",
    "Local_feature_47"
]

missing_features = [
    f for f in selected_features
    if f not in data.columns
]

if missing_features:
    raise ValueError(
        f"Missing selected features: {missing_features}"
    )

print("\n[2] SELECTED FEATURES")
print("-" * 90)

for i, f in enumerate(selected_features, 1):
    print(f"{i:2d}. {f}")

print("\nNumber of features:", len(selected_features))


# =============================================================================
# 3. CLEAN / SORT DATA
# =============================================================================

data = data.copy()

data["Time step"] = pd.to_numeric(
    data["Time step"],
    errors="coerce"
)

data["class"] = pd.to_numeric(
    data["class"],
    errors="coerce"
)

data = data.dropna(
    subset=["txId", "Time step", "class"]
).copy()

data = data.sort_values(
    ["Time step", "txId"]
).reset_index(drop=True)

print("\n[3] TEMPORAL RANGE")
print("-" * 90)

print("Minimum time step:", data["Time step"].min())
print("Maximum time step:", data["Time step"].max())
print("Unique time steps:", data["Time step"].nunique())


# =============================================================================
# 4. HANDLE FEATURE VALUES
# =============================================================================

X_all = data[selected_features].copy()

X_all = X_all.replace(
    [np.inf, -np.inf],
    np.nan
)

# Median imputation
X_all = X_all.fillna(
    X_all.median()
)

if X_all.isna().sum().sum() != 0:
    raise ValueError("NaN values remain after imputation.")

print("\n[4] FEATURE MATRIX")
print("-" * 90)

print("X shape:", X_all.shape)
print("Remaining NaN:", X_all.isna().sum().sum())
print("Remaining Inf:", np.isinf(X_all.to_numpy()).sum())


# =============================================================================
# 5. TEMPORAL SPLIT
# =============================================================================

train_mask = data["Time step"].between(1, 34)

val_mask = data["Time step"].between(35, 41)

test_mask = data["Time step"].between(42, 49)

print("\n[5] TEMPORAL SPLIT")
print("-" * 90)

print("Train      : Time 1–34")
print("Validation : Time 35–41")
print("Test       : Time 42–49")

print("\nTrain samples      :", train_mask.sum())
print("Validation samples :", val_mask.sum())
print("Test samples       :", test_mask.sum())


# =============================================================================
# 6. CRITICAL STEP:
#    TRAIN ONLY ON CLASS 2 FROM TRAIN PERIOD
# =============================================================================

class2_train_mask = (
    train_mask &
    (data["class"] == 2)
)

X_train_normal = X_all.loc[
    class2_train_mask
].to_numpy()

print("\n[6] CLASS-2-ONLY TRAINING SET")
print("-" * 90)

print("Training samples available:", len(X_train_normal))

print("\nTraining class distribution:")
print(
    data.loc[
        class2_train_mask,
        "class"
    ].value_counts()
)

if len(X_train_normal) == 0:
    raise ValueError(
        "No Class-2 samples found in training period."
    )


# =============================================================================
# 7. TRAIN ISOLATION FOREST
# =============================================================================

print("\n[7] TRAINING ISOLATION FOREST")
print("-" * 90)

model = IsolationForest(
    n_estimators=300,
    max_samples="auto",
    contamination="auto",
    random_state=42,
    n_jobs=-1
)

model.fit(X_train_normal)

print("✓ Isolation Forest trained")
print("n_estimators :", 300)
print("max_samples  :", "auto")
print("contamination:", "auto")
print("random_state :", 42)


# =============================================================================
# 8. GENERATE SCORES
# =============================================================================
# sklearn:
# decision_function:
#     higher = more normal
#     lower  = more anomalous
#
# We therefore convert it to:
#
# anomaly_score = -decision_function
#
# higher anomaly_score = more anomalous
# =============================================================================

print("\n[8] GENERATING ANOMALY SCORES")
print("-" * 90)

X_train_all = X_all.loc[train_mask].to_numpy()
X_val = X_all.loc[val_mask].to_numpy()
X_test = X_all.loc[test_mask].to_numpy()

train_scores = -model.decision_function(X_train_all)
val_scores = -model.decision_function(X_val)
test_scores = -model.decision_function(X_test)

print("✓ Scores generated")

print("Train:", len(train_scores))
print("Validation:", len(val_scores))
print("Test:", len(test_scores))


# =============================================================================
# 9. ATTACH SCORES
# =============================================================================

data["anomaly_score"] = np.nan

data.loc[train_mask, "anomaly_score"] = train_scores
data.loc[val_mask, "anomaly_score"] = val_scores
data.loc[test_mask, "anomaly_score"] = test_scores

if data["anomaly_score"].isna().any():
    raise ValueError("Some anomaly scores are missing.")

print("\n[9] SCORE ATTACHMENT")
print("-" * 90)

print("✓ All rows have anomaly scores")


# =============================================================================
# 10. SCORE DISTRIBUTION
# =============================================================================

print("\n[10] ANOMALY SCORE DISTRIBUTION")
print("-" * 90)

score_summary = pd.DataFrame({
    "Split": [
        "Train",
        "Validation",
        "Test"
    ],
    "Samples": [
        len(train_scores),
        len(val_scores),
        len(test_scores)
    ],
    "Mean": [
        np.mean(train_scores),
        np.mean(val_scores),
        np.mean(test_scores)
    ],
    "Std": [
        np.std(train_scores),
        np.std(val_scores),
        np.std(test_scores)
    ],
    "Min": [
        np.min(train_scores),
        np.min(val_scores),
        np.min(test_scores)
    ],
    "Median": [
        np.median(train_scores),
        np.median(val_scores),
        np.median(test_scores)
    ],
    "Max": [
        np.max(train_scores),
        np.max(val_scores),
        np.max(test_scores)
    ]
})

print(
    score_summary.to_string(
        index=False,
        float_format=lambda x: f"{x:.6f}"
    )
)


# =============================================================================
# 11. CLASS-WISE SCORE DISTRIBUTION
# =============================================================================

print("\n[11] TEST SCORE DISTRIBUTION BY CLASS")
print("-" * 90)

test_data = data.loc[
    test_mask
].copy()

class_score_summary = (
    test_data
    .groupby("class")["anomaly_score"]
    .agg(
        Samples="count",
        Mean="mean",
        Std="std",
        Min="min",
        Median="median",
        Max="max"
    )
)

print(
    class_score_summary.to_string(
        float_format=lambda x: f"{x:.6f}"
    )
)


# =============================================================================
# 12. CLASS-WISE QUANTILES
# =============================================================================

print("\n[12] TEST SCORE QUANTILES")
print("-" * 90)

quantiles = (
    test_data
    .groupby("class")["anomaly_score"]
    .quantile(
        [0.01, 0.05, 0.10, 0.25,
         0.50, 0.75, 0.90,
         0.95, 0.99]
    )
    .unstack()
)

quantiles.columns = [
    "Q01", "Q05", "Q10",
    "Q25", "Q50", "Q75",
    "Q90", "Q95", "Q99"
]

print(
    quantiles.to_string(
        float_format=lambda x: f"{x:.6f}"
    )
)


# =============================================================================
# 13. BINARY EVALUATION
# =============================================================================
#
# For this experiment:
#
# Class 1 = known illicit
# Class 2 = known licit
#
# Class 3 is deliberately excluded from binary evaluation.
#
# This evaluates whether anomalies are enriched for known illicit
# transactions WITHOUT training on Class-1 labels.
# =============================================================================

binary_test = test_data[
    test_data["class"].isin([1, 2])
].copy()

binary_test["binary_label"] = (
    binary_test["class"] == 1
).astype(int)

y_true = binary_test["binary_label"].to_numpy()
y_score = binary_test["anomaly_score"].to_numpy()

print("\n[13] CLASS 1 vs CLASS 2 — UNSUPERVISED EVALUATION")
print("-" * 90)

print("Class-1 test samples:", (y_true == 1).sum())
print("Class-2 test samples:", (y_true == 0).sum())


# =============================================================================
# 14. ROC-AUC
# =============================================================================

roc_auc = roc_auc_score(
    y_true,
    y_score
)

# PR-AUC
pr_auc = average_precision_score(
    y_true,
    y_score
)

print("\nROC-AUC:", round(roc_auc, 6))
print("PR-AUC :", round(pr_auc, 6))


# =============================================================================
# 15. TOP-K ANALYSIS
# =============================================================================

print("\n[14] TOP-K ANOMALY ANALYSIS")
print("-" * 90)

top_k_values = [
    50,
    100,
    500,
    1000,
    5000
]

# Sort highest anomaly score first
binary_sorted = binary_test.sort_values(
    "anomaly_score",
    ascending=False
).reset_index(drop=True)

total_illicit = (
    binary_sorted["binary_label"] == 1
).sum()

total_licit = (
    binary_sorted["binary_label"] == 0
).sum()

total_binary = len(binary_sorted)

overall_illicit_rate = (
    total_illicit / total_binary
)

results = []

for k in top_k_values:

    k_actual = min(k, len(binary_sorted))

    top = binary_sorted.head(k_actual)

    illicit_count = (
        top["binary_label"] == 1
    ).sum()

    licit_count = (
        top["binary_label"] == 0
    ).sum()

    precision_k = (
        illicit_count / k_actual
    )

    recall_k = (
        illicit_count / total_illicit
        if total_illicit > 0 else np.nan
    )

    enrichment = (
        precision_k / overall_illicit_rate
        if overall_illicit_rate > 0
        else np.nan
    )

    results.append({
        "Top_K": k_actual,
        "Illicit": illicit_count,
        "Licit": licit_count,
        "Illicit_%": precision_k * 100,
        "Recall_%": recall_k * 100,
        "Enrichment_x": enrichment
    })

topk_results = pd.DataFrame(results)

print(
    topk_results.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)


# =============================================================================
# 16. TOP-K INCLUDING CLASS 3
# =============================================================================

print("\n[15] TOP-K — ALL CLASSES")
print("-" * 90)

all_sorted = test_data.sort_values(
    "anomaly_score",
    ascending=False
).reset_index(drop=True)

all_results = []

for k in top_k_values:

    k_actual = min(k, len(all_sorted))

    top = all_sorted.head(k_actual)

    counts = top["class"].value_counts()

    c1 = counts.get(1, 0)
    c2 = counts.get(2, 0)
    c3 = counts.get(3, 0)

    all_results.append({
        "Top_K": k_actual,
        "Class_1": c1,
        "Class_2": c2,
        "Class_3": c3,
        "Class_1_%": c1 / k_actual * 100,
        "Class_2_%": c2 / k_actual * 100,
        "Class_3_%": c3 / k_actual * 100
    })

all_topk_results = pd.DataFrame(
    all_results
)

print(
    all_topk_results.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)


# =============================================================================
# 17. CLASS 3 — SEPARATE EXPLORATORY ANALYSIS
# =============================================================================

print("\n[16] CLASS-3 EXPLORATORY ANALYSIS")
print("-" * 90)

class3_test = test_data[
    test_data["class"] == 3
]

class1_test = test_data[
    test_data["class"] == 1
]

class2_test = test_data[
    test_data["class"] == 2
]

print("Class 1:", len(class1_test))
print("Class 2:", len(class2_test))
print("Class 3:", len(class3_test))

print("\nMean anomaly score:")
print("Class 1:", class1_test["anomaly_score"].mean())
print("Class 2:", class2_test["anomaly_score"].mean())
print("Class 3:", class3_test["anomaly_score"].mean())


# =============================================================================
# 18. TOP 30 TEST ANOMALIES
# =============================================================================

print("\n[17] TOP 30 MOST ANOMALOUS TEST TRANSACTIONS")
print("-" * 90)

top30 = test_data.sort_values(
    "anomaly_score",
    ascending=False
).head(30)[
    [
        "txId",
        "Time step",
        "class",
        "anomaly_score"
    ]
]

print(
    top30.to_string(
        index=False,
        float_format=lambda x: f"{x:.6f}"
    )
)


# =============================================================================
# 19. TEMPORAL PERFORMANCE — CLASS 1 vs CLASS 2
# =============================================================================

print("\n[18] TEMPORAL CLASS-1 vs CLASS-2 ANALYSIS")
print("-" * 90)

temporal_results = []

for t in sorted(
    binary_test["Time step"].unique()
):

    temp = binary_test[
        binary_test["Time step"] == t
    ]

    if temp["binary_label"].nunique() < 2:
        continue

    try:
        auc = roc_auc_score(
            temp["binary_label"],
            temp["anomaly_score"]
        )
    except:
        auc = np.nan

    temporal_results.append({
        "Time_step": t,
        "Samples": len(temp),
        "Class1": (temp["binary_label"] == 1).sum(),
        "Class2": (temp["binary_label"] == 0).sum(),
        "ROC_AUC": auc,
        "Mean_Score": temp["anomaly_score"].mean()
    })

temporal_results = pd.DataFrame(
    temporal_results
)

print(
    temporal_results.to_string(
        index=False,
        float_format=lambda x: f"{x:.6f}"
    )
)


# =============================================================================
# 20. CONFUSION MATRIX AT TOP 5% THRESHOLD
# =============================================================================
#
# This threshold is NOT learned using labels.
# It is simply the top 5% most anomalous transactions in the
# binary test population.
# =============================================================================

print("\n[19] TOP 5% ANOMALY THRESHOLD")
print("-" * 90)

threshold = binary_test[
    "anomaly_score"
].quantile(0.95)

binary_test["predicted_anomaly"] = (
    binary_test["anomaly_score"] >= threshold
).astype(int)

cm = confusion_matrix(
    binary_test["binary_label"],
    binary_test["predicted_anomaly"]
)

tn, fp, fn, tp = cm.ravel()

precision_5 = (
    tp / (tp + fp)
    if (tp + fp) > 0 else 0
)

recall_5 = (
    tp / (tp + fn)
    if (tp + fn) > 0 else 0
)

print("Threshold:", threshold)

print("\nConfusion Matrix:")
print(cm)

print("\nTP:", tp)
print("FP:", fp)
print("TN:", tn)
print("FN:", fn)

print("\nPrecision:", round(precision_5, 6))
print("Recall   :", round(recall_5, 6))


# =============================================================================
# 21. SAVE RESULTS IN MEMORY
# =============================================================================

binary_test_results = binary_test.copy()

test_results = test_data.copy()

print("\n" + "=" * 90)
print("PIPELINE COMPLETE")
print("=" * 90)

print("""
✓ Isolation Forest trained ONLY on Class 2 / Time 1–34
✓ Class-1 labels NOT used during training
✓ Temporal validation preserved
✓ Temporal test preserved
✓ ROC-AUC calculated
✓ PR-AUC calculated
✓ Top-K illicit enrichment calculated
✓ Class-3 separately analysed
✓ Temporal performance calculated
✓ Top anomalies extracted

IMPORTANT:
Class 1 / Class 2 metrics are external evaluation only.
Class 3 remains an exploratory/unknown population.
""")

print("=" * 90)
```

```python
# ============================================================
# BIT-TRACE — RECOVER FINAL ELLIPTIC++ MODEL DATA
# ============================================================

import os
import numpy as np
import pandas as pd

print("=" * 80)
print("RECOVERING FINAL ELLIPTIC++ MODEL DATA")
print("=" * 80)

# ------------------------------------------------------------
# 1. SELECTED FEATURES
# ------------------------------------------------------------

SELECTED_FEATURES = [
    "Local_feature_53",
    "Local_feature_54",
    "Local_feature_52",
    "Local_feature_59",
    "Aggregate_feature_21",
    "Local_feature_25",
    "Local_feature_77",
    "Local_feature_89",
    "Local_feature_90",
    "Aggregate_feature_32",
    "Aggregate_feature_14",
    "Local_feature_22",
    "Local_feature_47"
]

print("\n[1] Selected features:", len(SELECTED_FEATURES))

for i, f in enumerate(SELECTED_FEATURES, 1):
    print(f"{i:2}. {f}")


# ------------------------------------------------------------
# 2. FIND FILES
# ------------------------------------------------------------

possible_feature_paths = [
    "txs_features.csv",
    "./txs_features.csv",
    "EllipticPlusPlus/Transactions Dataset/txs_features.csv",
    "data/EllipticPlusPlus/Transactions Dataset/txs_features.csv",
    "data/raw/EllipticPlusPlus/Transactions Dataset/txs_features.csv",
]

possible_class_paths = [
    "txs_classes.csv",
    "./txs_classes.csv",
    "EllipticPlusPlus/Transactions Dataset/txs_classes.csv",
    "data/EllipticPlusPlus/Transactions Dataset/txs_classes.csv",
    "data/raw/EllipticPlusPlus/Transactions Dataset/txs_classes.csv",
]

feature_path = next(
    (p for p in possible_feature_paths if os.path.exists(p)),
    None
)

class_path = next(
    (p for p in possible_class_paths if os.path.exists(p)),
    None
)

print("\n[2] FILE CHECK")

if feature_path is None:
    raise FileNotFoundError(
        "txs_features.csv not found. "
        "Run: !dir /s /b txs_features.csv"
    )

if class_path is None:
    raise FileNotFoundError(
        "txs_classes.csv not found. "
        "Run: !dir /s /b txs_classes.csv"
    )

print("Feature file:", feature_path)
print("Class file  :", class_path)


# ------------------------------------------------------------
# 3. READ CLASS FILE
# ------------------------------------------------------------

print("\n[3] LOADING CLASS DATA")
print("-" * 80)

classes = pd.read_csv(class_path)

print("Shape:", classes.shape)
print("Columns:", classes.columns.tolist())

if "txId" not in classes.columns:
    raise ValueError("txId not found in class file.")

if "class" not in classes.columns:
    raise ValueError("class not found in class file.")

classes["txId"] = pd.to_numeric(
    classes["txId"],
    errors="coerce"
)

print("\nClass distribution:")
print(classes["class"].value_counts().sort_index())


# ------------------------------------------------------------
# 4. READ ONLY REQUIRED FEATURE COLUMNS
# ------------------------------------------------------------

print("\n[4] READING FEATURE DATA")
print("-" * 80)

# First read only header
header = pd.read_csv(
    feature_path,
    nrows=0
)

all_columns = header.columns.tolist()

print("Total columns in raw feature file:", len(all_columns))

required_columns = [
    "txId",
    "Time step"
] + SELECTED_FEATURES

missing = [
    c for c in required_columns
    if c not in all_columns
]

if missing:
    raise ValueError(
        "These columns are NOT present in txs_features.csv:\n"
        + "\n".join(missing)
    )

print("All 13 selected features found ✓")

# Read only required columns
features = pd.read_csv(
    feature_path,
    usecols=required_columns
)

print("Loaded feature shape:", features.shape)


# ------------------------------------------------------------
# 5. BASIC CLEANING
# ------------------------------------------------------------

print("\n[5] BASIC CLEANING")
print("-" * 80)

features["txId"] = pd.to_numeric(
    features["txId"],
    errors="coerce"
)

features["Time step"] = pd.to_numeric(
    features["Time step"],
    errors="coerce"
)

print("Missing txId:", features["txId"].isna().sum())
print("Missing Time step:", features["Time step"].isna().sum())

# Remove rows without identifiers/time
features = features.dropna(
    subset=["txId", "Time step"]
).copy()

# Ensure numeric features are numeric
for f in SELECTED_FEATURES:
    features[f] = pd.to_numeric(
        features[f],
        errors="coerce"
    )


# ------------------------------------------------------------
# 6. MERGE FEATURES + CLASSES
# ------------------------------------------------------------

print("\n[6] MERGING FEATURES + CLASSES")
print("-" * 80)

data = features.merge(
    classes[["txId", "class"]],
    on="txId",
    how="inner",
    validate="one_to_one"
)

print("Merged shape:", data.shape)

print("Missing labels:", data["class"].isna().sum())

print("\nClass distribution:")
print(data["class"].value_counts().sort_index())


# ------------------------------------------------------------
# 7. CLEAN NaN / INF
# ------------------------------------------------------------

print("\n[7] FEATURE QUALITY")
print("-" * 80)

X_raw = data[SELECTED_FEATURES].copy()

# Replace inf with NaN
X_raw = X_raw.replace(
    [np.inf, -np.inf],
    np.nan
)

nan_before = int(X_raw.isna().sum().sum())

print("NaN/Inf values before imputation:", nan_before)

# Median imputation
for f in SELECTED_FEATURES:
    median_value = X_raw[f].median()

    X_raw[f] = X_raw[f].fillna(
        median_value
    )

nan_after = int(X_raw.isna().sum().sum())

print("NaN values after imputation:", nan_after)


# ------------------------------------------------------------
# 8. FINAL MATRIX
# ------------------------------------------------------------

X_all = X_raw.to_numpy(
    dtype=np.float32
)

y_all = data["class"].to_numpy()

time_steps = data["Time step"].to_numpy()

tx_ids = data["txId"].to_numpy()


# ------------------------------------------------------------
# 9. TEMPORAL SPLIT
# ------------------------------------------------------------

print("\n[8] TEMPORAL SPLIT")
print("-" * 80)

train_idx = np.where(
    time_steps <= 34
)[0]

val_idx = np.where(
    (time_steps >= 35) &
    (time_steps <= 41)
)[0]

test_idx = np.where(
    time_steps >= 42
)[0]

print("Train samples:", len(train_idx))
print("Validation samples:", len(val_idx))
print("Test samples:", len(test_idx))


# ------------------------------------------------------------
# 10. FINAL VALIDATION
# ------------------------------------------------------------

print("\n[9] FINAL OBJECT VALIDATION")
print("-" * 80)

print("X_all shape      :", X_all.shape)
print("y_all shape      :", y_all.shape)
print("time_steps shape :", time_steps.shape)
print("tx_ids shape     :", tx_ids.shape)

print("\nTime range:")
print("Min:", time_steps.min())
print("Max:", time_steps.max())

print("\nSelected features:", X_all.shape[1])

print("\nClass distribution:")
print(
    pd.Series(y_all)
    .value_counts()
    .sort_index()
)

print("\nTemporal split:")
print("Train:", len(train_idx))
print("Val  :", len(val_idx))
print("Test :", len(test_idx))

assert X_all.shape[0] == len(y_all)
assert X_all.shape[0] == len(time_steps)
assert X_all.shape[0] == len(tx_ids)

assert len(train_idx) + len(val_idx) + len(test_idx) == len(X_all)

assert not np.isnan(X_all).any()
assert not np.isinf(X_all).any()

print("\n" + "=" * 80)
print("✓ X_all CREATED SUCCESSFULLY")
print("✓ y_all CREATED SUCCESSFULLY")
print("✓ time_steps CREATED SUCCESSFULLY")
print("✓ tx_ids CREATED SUCCESSFULLY")
print("✓ train_idx CREATED SUCCESSFULLY")
print("✓ val_idx CREATED SUCCESSFULLY")
print("✓ test_idx CREATED SUCCESSFULLY")
print("=" * 80)
```

```python
# ============================================================
# BIT-TRACE — ELLIPTIC++ FINAL UNSUPERVISED PIPELINE
# ============================================================
# Purpose:
#   1. Load Elliptic++ transaction data
#   2. Select previously validated 13 features
#   3. Clean + impute
#   4. Preserve temporal ordering
#   5. Temporal train/validation/test split
#   6. Train Isolation Forest ONLY on train data
#   7. Generate anomaly scores
#   8. Evaluate anomaly enrichment using labels ONLY AFTER training
#   9. Analyse test set
#  10. Save final anomaly results
#
# IMPORTANT:
# Labels are NEVER used to train Isolation Forest.
# ============================================================

import os
import warnings
import numpy as np
import pandas as pd

from pathlib import Path
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import IsolationForest

warnings.filterwarnings("ignore")

print("=" * 80)
print("BIT-TRACE — ELLIPTIC++ FINAL UNSUPERVISED ANOMALY PIPELINE")
print("=" * 80)


# ============================================================
# 1. EXACT DATA PATHS
# ============================================================

FEATURE_PATH = Path(
    r"C:\Users\YASHIKA\Desktop\bit-trace\EllipticPlusPlus\Transactions Dataset\txs_features.csv"
)

CLASS_PATH = Path(
    r"C:\Users\YASHIKA\Desktop\bit-trace\EllipticPlusPlus\Transactions Dataset\txs_classes.csv"
)

EDGE_PATH = Path(
    r"C:\Users\YASHIKA\Desktop\bit-trace\EllipticPlusPlus\Transactions Dataset\txs_edgelist.csv"
)

print("\n[1] PATH CHECK")
print("-" * 80)

print("Feature file:", FEATURE_PATH.exists())
print("Class file:  ", CLASS_PATH.exists())
print("Edge file:   ", EDGE_PATH.exists())

if not FEATURE_PATH.exists():
    raise FileNotFoundError(f"Feature file not found:\n{FEATURE_PATH}")

if not CLASS_PATH.exists():
    raise FileNotFoundError(f"Class file not found:\n{CLASS_PATH}")


# ============================================================
# 2. SELECTED FEATURES
# ============================================================

SELECTED_FEATURES = [
    "Local_feature_53",
    "Local_feature_54",
    "Local_feature_52",
    "Local_feature_59",
    "Aggregate_feature_21",
    "Local_feature_25",
    "Local_feature_77",
    "Local_feature_89",
    "Local_feature_90",
    "Aggregate_feature_32",
    "Aggregate_feature_14",
    "Local_feature_22",
    "Local_feature_47",
]

print("\n[2] SELECTED FEATURES")
print("-" * 80)

for i, f in enumerate(SELECTED_FEATURES, 1):
    print(f"{i:2d}. {f}")

print("Number of features:", len(SELECTED_FEATURES))


# ============================================================
# 3. LOAD CLASS DATA
# ============================================================

print("\n[3] LOADING CLASS DATA")
print("-" * 80)

classes = pd.read_csv(CLASS_PATH)

print("Shape:", classes.shape)
print("Columns:", classes.columns.tolist())

# Standardize column names if necessary
classes.columns = [str(c).strip() for c in classes.columns]

if "txId" not in classes.columns:
    raise ValueError(
        f"txId not found in class file. Columns:\n{classes.columns.tolist()}"
    )

if "class" not in classes.columns:
    raise ValueError(
        f"class not found in class file. Columns:\n{classes.columns.tolist()}"
    )

print("\nClass distribution:")
print(classes["class"].value_counts().sort_index())


# ============================================================
# 4. READ FEATURE HEADER FIRST
# ============================================================

print("\n[4] READING FEATURE HEADER")
print("-" * 80)

header = pd.read_csv(
    FEATURE_PATH,
    nrows=0
)

header.columns = [str(c).strip() for c in header.columns]

print("Total columns:", len(header.columns))

# Detect identifiers
if "txId" in header.columns:
    TX_COL = "txId"
else:
    raise ValueError("txId column not found.")

if "Time step" in header.columns:
    TIME_COL = "Time step"
elif "time_step" in header.columns:
    TIME_COL = "time_step"
else:
    raise ValueError("Time step column not found.")

print("Transaction ID:", TX_COL)
print("Time column:", TIME_COL)


# ============================================================
# 5. VERIFY SELECTED FEATURES
# ============================================================

print("\n[5] VERIFYING SELECTED FEATURES")
print("-" * 80)

missing_features = [
    f for f in SELECTED_FEATURES
    if f not in header.columns
]

if missing_features:
    raise ValueError(
        "Selected features missing from txs_features.csv:\n"
        + "\n".join(missing_features)
    )

print("✓ All 13 selected features exist.")


# ============================================================
# 6. LOAD ONLY REQUIRED COLUMNS
# ============================================================

required_columns = [
    TX_COL,
    TIME_COL,
] + SELECTED_FEATURES

print("\n[6] LOADING REQUIRED FEATURE COLUMNS")
print("-" * 80)

print("Columns loaded:", len(required_columns))

features = pd.read_csv(
    FEATURE_PATH,
    usecols=required_columns
)

features.columns = [str(c).strip() for c in features.columns]

print("Loaded shape:", features.shape)


# ============================================================
# 7. BASIC CLEANING
# ============================================================

print("\n[7] BASIC DATA CLEANING")
print("-" * 80)

features[TX_COL] = pd.to_numeric(
    features[TX_COL],
    errors="coerce"
)

features[TIME_COL] = pd.to_numeric(
    features[TIME_COL],
    errors="coerce"
)

print("Missing txId:", features[TX_COL].isna().sum())
print("Missing time:", features[TIME_COL].isna().sum())

if features[TX_COL].isna().any():
    raise ValueError("Missing transaction IDs detected.")

if features[TIME_COL].isna().any():
    raise ValueError("Missing time steps detected.")


# ============================================================
# 8. MERGE FEATURES + LABELS
# ============================================================

print("\n[8] MERGING FEATURES + LABELS")
print("-" * 80)

data = features.merge(
    classes[[TX_COL, "class"]],
    on=TX_COL,
    how="inner",
    validate="one_to_one"
)

print("Merged shape:", data.shape)

if len(data) != len(features):
    print(
        "WARNING: Some feature rows did not match class labels."
    )

print("Missing labels:", data["class"].isna().sum())

if data["class"].isna().any():
    raise ValueError("Missing class labels after merge.")

print("\nClass distribution:")
print(data["class"].value_counts().sort_index())


# ============================================================
# 9. SORT TEMPORALLY
# ============================================================

print("\n[9] TEMPORAL ORDERING")
print("-" * 80)

data = data.sort_values(
    [TIME_COL, TX_COL]
).reset_index(drop=True)

time_steps = data[TIME_COL].to_numpy()

print("Minimum time step:", time_steps.min())
print("Maximum time step:", time_steps.max())
print("Unique time steps:", data[TIME_COL].nunique())


# ============================================================
# 10. CREATE FEATURE MATRIX
# ============================================================

X_raw = data[SELECTED_FEATURES].copy()

# Convert everything to numeric
for col in SELECTED_FEATURES:
    X_raw[col] = pd.to_numeric(
        X_raw[col],
        errors="coerce"
    )

X_raw = X_raw.replace(
    [np.inf, -np.inf],
    np.nan
)

print("\n[10] DATA QUALITY")
print("-" * 80)

nan_total = int(X_raw.isna().sum().sum())
inf_total = int(
    np.isinf(X_raw.to_numpy()).sum()
)

print("Total NaN:", nan_total)
print("Total Inf:", inf_total)


# ============================================================
# 11. MEDIAN IMPUTATION
# ============================================================

print("\n[11] MEDIAN IMPUTATION")
print("-" * 80)

imputer = SimpleImputer(
    strategy="median"
)

X_imputed = imputer.fit_transform(X_raw)

print("Shape:", X_imputed.shape)
print(
    "Remaining NaN:",
    np.isnan(X_imputed).sum()
)


# ============================================================
# 12. TEMPORAL SPLIT
# ============================================================
# Train      : 1–34
# Validation : 35–41
# Test       : 42–49
#
# IMPORTANT:
# The split is based ONLY on time.
# ============================================================

print("\n[12] TEMPORAL SPLIT")
print("-" * 80)

train_idx = np.where(
    time_steps <= 34
)[0]

val_idx = np.where(
    (time_steps >= 35) &
    (time_steps <= 41)
)[0]

test_idx = np.where(
    time_steps >= 42
)[0]

print(
    f"Train samples:      {len(train_idx):,}"
)

print(
    f"Validation samples: {len(val_idx):,}"
)

print(
    f"Test samples:       {len(test_idx):,}"
)

print(
    "Total:",
    len(train_idx) +
    len(val_idx) +
    len(test_idx)
)


# ============================================================
# 13. SCALE FEATURES
# ============================================================
# Fit scaler ONLY on training data.
# ============================================================

print("\n[13] FEATURE SCALING")
print("-" * 80)

scaler = StandardScaler()

X_train = scaler.fit_transform(
    X_imputed[train_idx]
)

X_val = scaler.transform(
    X_imputed[val_idx]
)

X_test = scaler.transform(
    X_imputed[test_idx]
)

print("Train:", X_train.shape)
print("Validation:", X_val.shape)
print("Test:", X_test.shape)


# ============================================================
# 14. ISOLATION FOREST
# ============================================================

print("\n[14] TRAINING ISOLATION FOREST")
print("-" * 80)

print("Training data:", X_train.shape)

# contamination='auto':
# Let Isolation Forest determine the threshold internally.
#
# n_estimators=300:
# More stable than a very small forest.
#
# n_jobs=-1:
# Use all CPU cores.

iso_forest = IsolationForest(
    n_estimators=300,
    contamination="auto",
    max_samples="auto",
    max_features=1.0,
    random_state=42,
    n_jobs=-1
)

iso_forest.fit(X_train)

print("✓ Isolation Forest training complete.")


# ============================================================
# 15. GENERATE SCORES
# ============================================================
# sklearn:
# decision_function()
# larger = more normal
# smaller = more anomalous
#
# For BIT-TRACE:
# anomaly_score = -decision_function()
# larger = more anomalous
# ============================================================

print("\n[15] GENERATING ANOMALY SCORES")
print("-" * 80)

train_scores = -iso_forest.decision_function(X_train)
val_scores = -iso_forest.decision_function(X_val)
test_scores = -iso_forest.decision_function(X_test)

print("✓ Scores generated.")


# ============================================================
# 16. SCORE DISTRIBUTION
# ============================================================

print("\n[16] ANOMALY SCORE DISTRIBUTION")
print("-" * 80)

score_summary = pd.DataFrame({
    "Split": [
        "Train",
        "Validation",
        "Test"
    ],
    "Samples": [
        len(train_scores),
        len(val_scores),
        len(test_scores)
    ],
    "Mean_Score": [
        np.mean(train_scores),
        np.mean(val_scores),
        np.mean(test_scores)
    ],
    "Std_Score": [
        np.std(train_scores),
        np.std(val_scores),
        np.std(test_scores)
    ],
    "Min_Score": [
        np.min(train_scores),
        np.min(val_scores),
        np.min(test_scores)
    ],
    "Median_Score": [
        np.median(train_scores),
        np.median(val_scores),
        np.median(test_scores)
    ],
    "Max_Score": [
        np.max(train_scores),
        np.max(val_scores),
        np.max(test_scores)
    ]
})

print(
    score_summary.to_string(
        index=False,
        float_format=lambda x: f"{x:.6f}"
    )
)


# ============================================================
# 17. BUILD EVALUATION DATAFRAME
# ============================================================

print("\n[17] BUILDING EVALUATION DATA")
print("-" * 80)

eval_df = data[
    [TX_COL, TIME_COL, "class"]
].copy()

eval_df["split"] = "unknown"
eval_df["anomaly_score"] = np.nan

eval_df.loc[
    train_idx,
    "split"
] = "train"

eval_df.loc[
    val_idx,
    "split"
] = "validation"

eval_df.loc[
    test_idx,
    "split"
] = "test"

eval_df.loc[
    train_idx,
    "anomaly_score"
] = train_scores

eval_df.loc[
    val_idx,
    "anomaly_score"
] = val_scores

eval_df.loc[
    test_idx,
    "anomaly_score"
] = test_scores

print(
    "Missing anomaly scores:",
    eval_df["anomaly_score"].isna().sum()
)


# ============================================================
# 18. CLASS-WISE ANALYSIS
# ============================================================
# IMPORTANT:
# This is evaluation only.
# Labels were NOT used during training.
# ============================================================

print("\n" + "=" * 80)
print("CLASS-WISE UNSUPERVISED ANOMALY ANALYSIS")
print("=" * 80)

class_summary = (
    eval_df
    .groupby("class")["anomaly_score"]
    .agg([
        "count",
        "mean",
        "std",
        "min",
        "median",
        "max"
    ])
)

print(
    class_summary.to_string(
        float_format=lambda x: f"{x:.6f}"
    )
)


# ============================================================
# 19. QUANTILES
# ============================================================

print("\n" + "=" * 80)
print("CLASS-WISE SCORE QUANTILES")
print("=" * 80)

quantiles = (
    eval_df
    .groupby("class")["anomaly_score"]
    .quantile(
        [
            0.01,
            0.05,
            0.10,
            0.25,
            0.50,
            0.75,
            0.90,
            0.95,
            0.99
        ]
    )
    .unstack()
)

quantiles.columns = [
    "Q01",
    "Q05",
    "Q10",
    "Q25",
    "Q50",
    "Q75",
    "Q90",
    "Q95",
    "Q99"
]

print(
    quantiles.to_string(
        float_format=lambda x: f"{x:.6f}"
    )
)


# ============================================================
# 20. TOP ANOMALIES
# ============================================================

print("\n" + "=" * 80)
print("TOP 100 MOST ANOMALOUS TRANSACTIONS")
print("=" * 80)

top_100 = (
    eval_df
    .sort_values(
        "anomaly_score",
        ascending=False
    )
    .head(100)
)

print(
    top_100[
        [TX_COL, TIME_COL, "class", "split", "anomaly_score"]
    ].head(30).to_string(
        index=False,
        float_format=lambda x: f"{x:.6f}"
    )
)


# ============================================================
# 21. TOP-K CLASS DISTRIBUTION
# ============================================================

print("\n" + "=" * 80)
print("CLASS DISTRIBUTION AMONG TOP ANOMALIES")
print("=" * 80)

overall_class3_pct = (
    (eval_df["class"] == 3).mean() * 100
)

for k in [50, 100, 500, 1000, 5000]:

    top_k = (
        eval_df
        .nlargest(
            min(k, len(eval_df)),
            "anomaly_score"
        )
    )

    counts = (
        top_k["class"]
        .value_counts()
        .sort_index()
    )

    print(f"\nTop {k:,} anomalies")

    print(
        counts.to_string()
    )

    print("\nPercentages:")

    print(
        (
            counts / len(top_k) * 100
        ).round(2).to_string()
    )

    class3_count = (
        top_k["class"] == 3
    ).sum()

    class3_pct = (
        class3_count / len(top_k) * 100
    )

    enrichment = (
        class3_pct /
        overall_class3_pct
    )

    print(
        f"Class-3: {class3_pct:.2f}%"
    )

    print(
        f"Class-3 enrichment: {enrichment:.3f}x"
    )


# ============================================================
# 22. TEMPORAL ANOMALY ANALYSIS
# ============================================================

print("\n" + "=" * 80)
print("TEMPORAL ANOMALY ANALYSIS")
print("=" * 80)

temporal_summary = (
    eval_df
    .groupby(TIME_COL)
    .agg(
        Transactions=(TX_COL, "count"),
        Mean_Anomaly_Score=(
            "anomaly_score",
            "mean"
        ),
        Median_Anomaly_Score=(
            "anomaly_score",
            "median"
        ),
        Class3_Count=(
            "class",
            lambda x: (x == 3).sum()
        )
    )
)

temporal_summary["Class3_%"] = (
    temporal_summary["Class3_Count"] /
    temporal_summary["Transactions"] *
    100
)

# Higher anomaly score = more anomalous
temporal_summary = temporal_summary.sort_values(
    "Mean_Anomaly_Score",
    ascending=False
)

print(
    temporal_summary.head(15).to_string(
        float_format=lambda x: f"{x:.4f}"
    )
)


# ============================================================
# 23. TEST SET ANALYSIS
# ============================================================

print("\n" + "=" * 80)
print("TEST SET — FINAL UNSUPERVISED ANALYSIS")
print("=" * 80)

test_eval = eval_df[
    eval_df["split"] == "test"
].copy()

print(
    "Test samples:",
    len(test_eval)
)

test_class_summary = (
    test_eval
    .groupby("class")["anomaly_score"]
    .agg([
        "count",
        "mean",
        "std",
        "min",
        "median",
        "max"
    ])
)

print(
    "\nTest-set anomaly scores by class:"
)

print(
    test_class_summary.to_string(
        float_format=lambda x: f"{x:.6f}"
    )
)


# ============================================================
# 24. TEST TOP ANOMALIES
# ============================================================

print("\n" + "=" * 80)
print("TEST SET — TOP 50 ANOMALOUS TRANSACTIONS")
print("=" * 80)

test_top50 = (
    test_eval
    .nlargest(
        50,
        "anomaly_score"
    )
)

print(
    test_top50[
        [TX_COL, TIME_COL, "class", "anomaly_score"]
    ].to_string(
        index=False,
        float_format=lambda x: f"{x:.6f}"
    )
)


# ============================================================
# 25. SAVE RESULTS
# ============================================================

print("\n" + "=" * 80)
print("SAVING RESULTS")
print("=" * 80)

RESULTS_DIR = (
    Path.cwd().parent /
    "results" /
    "elliptic_unsupervised"
)

RESULTS_DIR.mkdir(
    parents=True,
    exist_ok=True
)

# Complete anomaly results
results_path = (
    RESULTS_DIR /
    "elliptic_transaction_anomaly_scores.csv"
)

eval_df.to_csv(
    results_path,
    index=False
)

# Top anomalies
top_path = (
    RESULTS_DIR /
    "top_100_anomalous_transactions.csv"
)

top_100.to_csv(
    top_path,
    index=False
)

# Temporal summary
temporal_path = (
    RESULTS_DIR /
    "temporal_anomaly_summary.csv"
)

temporal_summary.to_csv(
    temporal_path
)

# Class summary
class_path = (
    RESULTS_DIR /
    "class_anomaly_summary.csv"
)

class_summary.to_csv(
    class_path
)

print("✓ Results saved:")
print(results_path)
print(top_path)
print(temporal_path)
print(class_path)


# ============================================================
# 26. FINAL PIPELINE STATUS
# ============================================================

print("\n" + "=" * 80)
print("PIPELINE STATUS")
print("=" * 80)

print("✓ Elliptic++ transaction data loaded")
print("✓ 13 previously selected features verified")
print("✓ Temporal ordering preserved")
print("✓ Median imputation performed")
print("✓ Scaler fitted only on training period")
print("✓ Temporal train/validation/test split")
print("✓ Isolation Forest trained without labels")
print("✓ Anomaly scores generated")
print("✓ Class-wise evaluation performed AFTER training")
print("✓ Top anomaly analysis completed")
print("✓ Temporal anomaly analysis completed")
print("✓ Test-set analysis completed")
print("✓ Results exported")

print("\nIMPORTANT:")
print("Class labels were used ONLY for post-training evaluation.")
print("They were NOT used to fit Isolation Forest.")

print("\nDone.")
```

```python
# ============================================================
# BIT-TRACE — ELLIPTIC++ PIPELINE
# FIXED PATH + FEATURE LOADING
# ============================================================

import os
import numpy as np
import pandas as pd

print("=" * 80)
print("BIT-TRACE — ELLIPTIC++ DATA LOADING")
print("=" * 80)

# ------------------------------------------------------------
# 1. EXACT FILE PATHS
# ------------------------------------------------------------

FEATURE_PATH = r"C:\Users\YASHIKA\Desktop\bit-trace\EllipticPlusPlus\Transactions Dataset\txs_features.csv"

CLASS_PATH = r"C:\Users\YASHIKA\Desktop\bit-trace\EllipticPlusPlus\Transactions Dataset\txs_classes.csv"

EDGELIST_PATH = r"C:\Users\YASHIKA\Desktop\bit-trace\EllipticPlusPlus\Transactions Dataset\txs_edgelist.csv"

# ------------------------------------------------------------
# 2. CHECK FILES
# ------------------------------------------------------------

print("\n[1] FILE CHECK")
print("-" * 80)

for name, path in [
    ("FEATURE", FEATURE_PATH),
    ("CLASS", CLASS_PATH),
    ("EDGELIST", EDGELIST_PATH)
]:
    exists = os.path.exists(path)
    print(f"{name:<10}: {exists}")
    if exists:
        print(f"Path       : {path}")
        print(f"Size       : {os.path.getsize(path) / (1024**2):.2f} MB")

if not os.path.exists(FEATURE_PATH):
    raise FileNotFoundError(FEATURE_PATH)

if not os.path.exists(CLASS_PATH):
    raise FileNotFoundError(CLASS_PATH)

# ------------------------------------------------------------
# 3. SELECTED FEATURES
# ------------------------------------------------------------

SELECTED_FEATURES = [
    "Local_feature_53",
    "Local_feature_54",
    "Local_feature_52",
    "Local_feature_59",
    "Aggregate_feature_21",
    "Local_feature_25",
    "Local_feature_77",
    "Local_feature_89",
    "Local_feature_90",
    "Aggregate_feature_32",
    "Aggregate_feature_14",
    "Local_feature_22",
    "Local_feature_47"
]

print("\n[2] SELECTED FEATURES")
print("-" * 80)

for i, f in enumerate(SELECTED_FEATURES, 1):
    print(f"{i:2d}. {f}")

# ------------------------------------------------------------
# 4. READ CLASS DATA
# ------------------------------------------------------------

print("\n[3] LOADING CLASS DATA")
print("-" * 80)

classes = pd.read_csv(CLASS_PATH)

print("Shape:", classes.shape)
print("Columns:", classes.columns.tolist())

print("\nFirst rows:")
print(classes.head())

# Normalize column names
classes.columns = classes.columns.str.strip()

# Detect ID and class columns
id_candidates = ["txId", "txid", "tx_id", "transactionId"]

id_col = next(
    (c for c in id_candidates if c in classes.columns),
    None
)

if id_col is None:
    raise ValueError(
        f"Transaction ID column not found. Columns: {classes.columns.tolist()}"
    )

class_candidates = ["class", "Class", "label", "Label"]

class_col = next(
    (c for c in class_candidates if c in classes.columns),
    None
)

if class_col is None:
    raise ValueError(
        f"Class column not found. Columns: {classes.columns.tolist()}"
    )

# Rename consistently
classes = classes.rename(
    columns={
        id_col: "txId",
        class_col: "class"
    }
)

print("\nNormalized columns:", classes.columns.tolist())

print("\nClass distribution:")
print(classes["class"].value_counts(dropna=False))

# ------------------------------------------------------------
# 5. READ ONLY REQUIRED FEATURE COLUMNS
# ------------------------------------------------------------

print("\n[4] READING FEATURE HEADER")
print("-" * 80)

header = pd.read_csv(
    FEATURE_PATH,
    nrows=0
)

header.columns = header.columns.str.strip()

print("Total columns:", len(header.columns))

print("\nFirst 20 columns:")
print(header.columns[:20].tolist())

# Detect columns
feature_id_col = next(
    (c for c in ["txId", "txid", "tx_id"] if c in header.columns),
    None
)

time_col = next(
    (c for c in ["Time step", "time_step", "Time_step", "time step"] if c in header.columns),
    None
)

if feature_id_col is None:
    raise ValueError(
        f"txId column not found in feature file.\n"
        f"Available columns: {header.columns.tolist()[:30]}"
    )

if time_col is None:
    raise ValueError(
        f"Time step column not found.\n"
        f"Available columns: {header.columns.tolist()[:30]}"
    )

print("\nDetected ID column :", feature_id_col)
print("Detected time column:", time_col)

# Check selected features
missing = [
    f for f in SELECTED_FEATURES
    if f not in header.columns
]

if missing:
    print("\nMISSING FEATURES:")
    for f in missing:
        print(" -", f)

    raise ValueError(
        "Selected features do not match this feature file."
    )

print("\n✓ All 13 selected features found.")

# ------------------------------------------------------------
# 6. LOAD ONLY 15 REQUIRED COLUMNS
# ------------------------------------------------------------

required_columns = [
    feature_id_col,
    time_col
] + SELECTED_FEATURES

print("\n[5] LOADING REQUIRED COLUMNS")
print("-" * 80)

print("Columns to load:", len(required_columns))

data = pd.read_csv(
    FEATURE_PATH,
    usecols=required_columns
)

data.columns = data.columns.str.strip()

# Normalize ID/time names
data = data.rename(
    columns={
        feature_id_col: "txId",
        time_col: "Time step"
    }
)

print("Loaded shape:", data.shape)

# ------------------------------------------------------------
# 7. BASIC CLEANING
# ------------------------------------------------------------

print("\n[6] BASIC DATA CLEANING")
print("-" * 80)

print("Missing txId:", data["txId"].isna().sum())
print("Missing Time step:", data["Time step"].isna().sum())

# Remove duplicate transaction IDs if any
duplicate_ids = data["txId"].duplicated().sum()

print("Duplicate txId:", duplicate_ids)

if duplicate_ids > 0:
    print("Removing duplicate txId rows...")
    data = data.drop_duplicates(
        subset="txId",
        keep="first"
    ).reset_index(drop=True)

# ------------------------------------------------------------
# 8. NUMERIC CONVERSION
# ------------------------------------------------------------

for col in SELECTED_FEATURES:
    data[col] = pd.to_numeric(
        data[col],
        errors="coerce"
    )

data["Time step"] = pd.to_numeric(
    data["Time step"],
    errors="coerce"
)

# ------------------------------------------------------------
# 9. MERGE LABELS
# ------------------------------------------------------------

print("\n[7] MERGING FEATURES + CLASSES")
print("-" * 80)

data = data.merge(
    classes[["txId", "class"]],
    on="txId",
    how="inner",
    validate="one_to_one"
)

print("Merged shape:", data.shape)

print("\nMissing class labels:")
print(data["class"].isna().sum())

print("\nClass distribution:")
print(data["class"].value_counts())

# ------------------------------------------------------------
# 10. DATA QUALITY
# ------------------------------------------------------------

print("\n[8] DATA QUALITY")
print("-" * 80)

X_raw = data[SELECTED_FEATURES].copy()

nan_total = int(X_raw.isna().sum().sum())

inf_total = int(
    np.isinf(
        X_raw.to_numpy(dtype=np.float64)
    ).sum()
)

print("Total NaN:", nan_total)
print("Total Inf:", inf_total)

# Replace infinity with NaN
X_raw = X_raw.replace(
    [np.inf, -np.inf],
    np.nan
)

# ------------------------------------------------------------
# 11. MEDIAN IMPUTATION
# ------------------------------------------------------------

print("\n[9] MEDIAN IMPUTATION")
print("-" * 80)

medians = X_raw.median()

X = X_raw.fillna(medians)

print("Remaining NaN:", int(X.isna().sum().sum()))

# ------------------------------------------------------------
# 12. FINAL MATRIX
# ------------------------------------------------------------

X_all = X.to_numpy(dtype=np.float32)

y_all = data["class"].to_numpy()

time_steps = data["Time step"].to_numpy()

tx_ids = data["txId"].to_numpy()

print("\n[10] FINAL DATA")
print("-" * 80)

print("X shape       :", X_all.shape)
print("y shape       :", y_all.shape)
print("time shape    :", time_steps.shape)
print("txId shape    :", tx_ids.shape)

print("\nTime range:")
print("Minimum:", np.nanmin(time_steps))
print("Maximum:", np.nanmax(time_steps))

print("\nClass distribution:")
print(pd.Series(y_all).value_counts().sort_index())

# ------------------------------------------------------------
# 13. TEMPORAL SPLIT
# ------------------------------------------------------------

print("\n[11] TEMPORAL SPLIT")
print("-" * 80)

train_idx = np.where(time_steps <= 34)[0]

val_idx = np.where(
    (time_steps >= 35) &
    (time_steps <= 41)
)[0]

test_idx = np.where(time_steps >= 42)[0]

print("Train:", len(train_idx))
print("Validation:", len(val_idx))
print("Test:", len(test_idx))

# ------------------------------------------------------------
# 14. VERIFY SPLIT
# ------------------------------------------------------------

print("\nTrain time range:")
print(
    time_steps[train_idx].min(),
    "to",
    time_steps[train_idx].max()
)

print("Validation time range:")
print(
    time_steps[val_idx].min(),
    "to",
    time_steps[val_idx].max()
)

print("Test time range:")
print(
    time_steps[test_idx].min(),
    "to",
    time_steps[test_idx].max()
)

# ------------------------------------------------------------
# 15. FINAL STATUS
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("ELLIPTIC++ DATA PREPARATION COMPLETE")
print("=" * 80)

print("✓ Files loaded")
print("✓ 13 selected features loaded")
print("✓ Labels merged")
print("✓ NaN/Inf handled")
print("✓ X_all created")
print("✓ y_all created")
print("✓ time_steps created")
print("✓ train_idx created")
print("✓ val_idx created")
print("✓ test_idx created")

print("\nVariables now available:")
print("X_all")
print("y_all")
print("time_steps")
print("tx_ids")
print("data")
print("train_idx")
print("val_idx")
print("test_idx")

print("\nReady for Isolation Forest.")
```

```python
# ============================================================
# BIT-TRACE — ELLIPTIC++ ISOLATION FOREST
# UNSUPERVISED ANOMALY DETECTION
# ============================================================

import numpy as np
import pandas as pd

from sklearn.ensemble import IsolationForest
from sklearn.metrics import roc_auc_score, average_precision_score

print("=" * 80)
print("BIT-TRACE — ELLIPTIC++ ISOLATION FOREST")
print("=" * 80)

# ------------------------------------------------------------
# 1. VERIFY INPUTS
# ------------------------------------------------------------

print("\n[1] INPUT VERIFICATION")
print("-" * 80)

print("X_all       :", X_all.shape)
print("y_all       :", y_all.shape)
print("time_steps  :", time_steps.shape)
print("train_idx   :", len(train_idx))
print("val_idx     :", len(val_idx))
print("test_idx    :", len(test_idx))

assert len(X_all) == len(y_all) == len(time_steps) == len(tx_ids)

# ------------------------------------------------------------
# 2. TEMPORAL DATA
# ------------------------------------------------------------

X_train = X_all[train_idx]
X_val   = X_all[val_idx]
X_test  = X_all[test_idx]

y_train = y_all.iloc[train_idx].to_numpy()
y_val   = y_all.iloc[val_idx].to_numpy()
y_test  = y_all.iloc[test_idx].to_numpy()

tx_train = tx_ids.iloc[train_idx].to_numpy()
tx_val   = tx_ids.iloc[val_idx].to_numpy()
tx_test  = tx_ids.iloc[test_idx].to_numpy()

time_train = time_steps[train_idx]
time_val   = time_steps[val_idx]
time_test  = time_steps[test_idx]

print("\n[2] TEMPORAL SPLIT")
print("-" * 80)

print(f"Train : {len(X_train):,} | time {time_train.min()}–{time_train.max()}")
print(f"Val   : {len(X_val):,} | time {time_val.min()}–{time_val.max()}")
print(f"Test  : {len(X_test):,} | time {time_test.min()}–{time_test.max()}")

# ------------------------------------------------------------
# 3. TRAINING LABELS ARE NOT USED
# ------------------------------------------------------------

print("\n[3] UNSUPERVISED TRAINING")
print("-" * 80)

print("Training labels available only for post-hoc evaluation.")
print("Isolation Forest will NOT use y_train.")

# ------------------------------------------------------------
# 4. TRAIN ISOLATION FOREST
# ------------------------------------------------------------

print("\n[4] TRAINING ISOLATION FOREST")
print("-" * 80)

iso_forest = IsolationForest(
    n_estimators=300,
    contamination="auto",
    max_samples="auto",
    max_features=1.0,
    bootstrap=False,
    random_state=42,
    n_jobs=-1
)

iso_forest.fit(X_train)

print("✓ Isolation Forest trained")
print("Trees:", iso_forest.n_estimators)

# ------------------------------------------------------------
# 5. GENERATE SCORES
# ------------------------------------------------------------
# sklearn:
# decision_function:
#   lower = more anomalous
#
# We convert it into an anomaly score where:
#   higher = more anomalous

print("\n[5] GENERATING ANOMALY SCORES")
print("-" * 80)

train_raw = iso_forest.decision_function(X_train)
val_raw   = iso_forest.decision_function(X_val)
test_raw  = iso_forest.decision_function(X_test)

train_score = -train_raw
val_score   = -val_raw
test_score  = -test_raw

print("✓ Scores generated")

# ------------------------------------------------------------
# 6. SCORE SUMMARY
# ------------------------------------------------------------

print("\n[6] SCORE DISTRIBUTION")
print("-" * 80)

score_summary = pd.DataFrame({
    "Split": ["Train", "Validation", "Test"],
    "Samples": [
        len(train_score),
        len(val_score),
        len(test_score)
    ],
    "Mean": [
        train_score.mean(),
        val_score.mean(),
        test_score.mean()
    ],
    "Std": [
        train_score.std(),
        val_score.std(),
        test_score.std()
    ],
    "Min": [
        train_score.min(),
        val_score.min(),
        test_score.min()
    ],
    "Median": [
        np.median(train_score),
        np.median(val_score),
        np.median(test_score)
    ],
    "Max": [
        train_score.max(),
        val_score.max(),
        test_score.max()
    ]
})

print(score_summary.to_string(index=False))

# ------------------------------------------------------------
# 7. CREATE EVALUATION DATAFRAME
# ------------------------------------------------------------

print("\n[7] CREATING EVALUATION DATA")
print("-" * 80)

train_eval = pd.DataFrame({
    "txId": tx_train,
    "Time step": time_train,
    "class": y_train,
    "anomaly_score": train_score
})

val_eval = pd.DataFrame({
    "txId": tx_val,
    "Time step": time_val,
    "class": y_val,
    "anomaly_score": val_score
})

test_eval = pd.DataFrame({
    "txId": tx_test,
    "Time step": time_test,
    "class": y_test,
    "anomaly_score": test_score
})

eval_df = pd.concat(
    [train_eval, val_eval, test_eval],
    ignore_index=True
)

print("Evaluation dataframe:", eval_df.shape)

# ------------------------------------------------------------
# 8. CLASS-WISE SCORE ANALYSIS
# ------------------------------------------------------------

print("\n[8] CLASS-WISE ANOMALY SCORES")
print("-" * 80)

class_stats = (
    eval_df
    .groupby("class")["anomaly_score"]
    .agg(
        Samples="count",
        Mean="mean",
        Std="std",
        Min="min",
        Median="median",
        Max="max"
    )
)

print(class_stats.to_string())

# ------------------------------------------------------------
# 9. CLASS-WISE QUANTILES
# ------------------------------------------------------------

print("\n[9] CLASS-WISE SCORE QUANTILES")
print("-" * 80)

quantiles = (
    eval_df
    .groupby("class")["anomaly_score"]
    .quantile(
        [0.01, 0.05, 0.10, 0.25, 0.50,
         0.75, 0.90, 0.95, 0.99]
    )
    .unstack()
)

quantiles.columns = [
    "Q01", "Q05", "Q10", "Q25",
    "Q50", "Q75", "Q90", "Q95", "Q99"
]

print(quantiles.to_string())

# ------------------------------------------------------------
# 10. TOP ANOMALOUS TRANSACTIONS
# ------------------------------------------------------------

print("\n[10] TOP 30 MOST ANOMALOUS TRANSACTIONS")
print("-" * 80)

top30 = (
    eval_df
    .sort_values("anomaly_score", ascending=False)
    .head(30)
)

print(top30[
    ["txId", "Time step", "class", "anomaly_score"]
].to_string(index=False))

# ------------------------------------------------------------
# 11. CLASS DISTRIBUTION AMONG TOP ANOMALIES
# ------------------------------------------------------------

print("\n[11] CLASS DISTRIBUTION AMONG TOP ANOMALIES")
print("-" * 80)

overall_class3_pct = (
    (eval_df["class"] == 3).mean() * 100
)

for k in [50, 100, 500, 1000, 5000]:

    top_k = (
        eval_df
        .nlargest(k, "anomaly_score")
    )

    counts = top_k["class"].value_counts().sort_index()

    class3_count = (top_k["class"] == 3).sum()
    class3_pct = class3_count / k * 100

    enrichment = class3_pct / overall_class3_pct

    print(f"\nTop {k:,} anomalies")
    print(counts.to_string())
    print(
        f"Class-3: {class3_pct:.2f}% | "
        f"Enrichment: {enrichment:.3f}x"
    )

# ------------------------------------------------------------
# 12. TEMPORAL ANALYSIS
# ------------------------------------------------------------

print("\n[12] TEMPORAL ANOMALY ANALYSIS")
print("-" * 80)

temporal_stats = (
    eval_df
    .groupby("Time step")
    .agg(
        Transactions=("txId", "count"),
        Mean_Anomaly_Score=("anomaly_score", "mean"),
        Median_Anomaly_Score=("anomaly_score", "median"),
        Class3_Count=("class", lambda x: (x == 3).sum())
    )
)

temporal_stats["Class3_%"] = (
    temporal_stats["Class3_Count"]
    / temporal_stats["Transactions"]
    * 100
)

# Higher score = more anomalous
most_anomalous_time = (
    temporal_stats
    .sort_values(
        "Mean_Anomaly_Score",
        ascending=False
    )
    .head(15)
)

print(most_anomalous_time.to_string())

# ------------------------------------------------------------
# 13. TEST SET ANALYSIS
# ------------------------------------------------------------

print("\n[13] TEST SET — FINAL UNSUPERVISED ANALYSIS")
print("-" * 80)

test_class_stats = (
    test_eval
    .groupby("class")["anomaly_score"]
    .agg(
        Samples="count",
        Mean="mean",
        Std="std",
        Min="min",
        Median="median",
        Max="max"
    )
)

print(test_class_stats.to_string())

# ------------------------------------------------------------
# 14. OPTIONAL POST-HOC BINARY CLASS-3 EVALUATION
# ------------------------------------------------------------
# IMPORTANT:
# These labels are NOT used to train the model.
#
# This only tells us whether anomaly scores align
# with the existing Class-3 labels.

print("\n[14] POST-HOC CLASS-3 EVALUATION")
print("-" * 80)

y_test_binary = (
    test_eval["class"].to_numpy() == 3
).astype(int)

test_scores_np = test_eval["anomaly_score"].to_numpy()

roc_auc = roc_auc_score(
    y_test_binary,
    test_scores_np
)

pr_auc = average_precision_score(
    y_test_binary,
    test_scores_np
)

print(f"Test ROC-AUC : {roc_auc:.4f}")
print(f"Test PR-AUC  : {pr_auc:.4f}")

# ------------------------------------------------------------
# 15. SAVE RESULTS
# ------------------------------------------------------------

print("\n[15] SAVING RESULTS")
print("-" * 80)

results_path = "ellipticpp_isolation_forest_results.csv"

eval_df.to_csv(
    results_path,
    index=False
)

print(f"✓ Results saved: {results_path}")

print("\n" + "=" * 80)
print("PIPELINE COMPLETE")
print("=" * 80)

print("""
✓ Elliptic++ loaded
✓ 13 selected features
✓ Temporal split preserved
✓ Isolation Forest trained unsupervised
✓ Anomaly scores generated
✓ Class-wise analysis completed
✓ Top anomaly analysis completed
✓ Temporal analysis completed
✓ Test-set analysis completed
✓ Post-hoc Class-3 evaluation completed
""")
```

```python
# ============================================================
# BIT-TRACE — ELLIPTIC++ FINAL UNSUPERVISED PIPELINE
# ISOLATION FOREST + TEMPORAL + CLASS-WISE POST-HOC ANALYSIS
# ============================================================

import numpy as np
import pandas as pd

from sklearn.ensemble import IsolationForest
from sklearn.metrics import roc_auc_score, average_precision_score

print("=" * 80)
print("BIT-TRACE — ELLIPTIC++ ISOLATION FOREST")
print("=" * 80)


# ============================================================
# 1. VERIFY EXISTING VARIABLES
# ============================================================

required = [
    "X_all",
    "y_all",
    "time_steps",
    "tx_ids",
    "train_idx",
    "val_idx",
    "test_idx"
]

missing = [v for v in required if v not in globals()]

if missing:
    raise NameError(
        "Missing variables: " + ", ".join(missing)
        + "\nRun your Elliptic++ data preparation cell first."
    )

print("\n[1] INPUTS VERIFIED")
print("-" * 80)

print("X_all      :", np.shape(X_all))
print("y_all      :", np.shape(y_all))
print("time_steps :", np.shape(time_steps))
print("tx_ids     :", np.shape(tx_ids))

print("Train      :", len(train_idx))
print("Validation :", len(val_idx))
print("Test       :", len(test_idx))


# ============================================================
# 2. CONVERT EVERYTHING TO SAFE NUMPY ARRAYS
# ============================================================

X_all_np = np.asarray(X_all, dtype=np.float32)
y_all_np = np.asarray(y_all)
time_np = np.asarray(time_steps)
tx_np = np.asarray(tx_ids)

train_idx = np.asarray(train_idx, dtype=int)
val_idx   = np.asarray(val_idx, dtype=int)
test_idx  = np.asarray(test_idx, dtype=int)

print("\n[2] ARRAY CONVERSION COMPLETE")
print("-" * 80)

print("X:", X_all_np.shape)
print("Features:", X_all_np.shape[1])


# ============================================================
# 3. TEMPORAL SPLIT
# ============================================================

X_train = X_all_np[train_idx]
X_val   = X_all_np[val_idx]
X_test  = X_all_np[test_idx]

y_train = y_all_np[train_idx]
y_val   = y_all_np[val_idx]
y_test  = y_all_np[test_idx]

tx_train = tx_np[train_idx]
tx_val   = tx_np[val_idx]
tx_test  = tx_np[test_idx]

time_train = time_np[train_idx]
time_val   = time_np[val_idx]
time_test  = time_np[test_idx]

print("\n[3] TEMPORAL SPLIT")
print("-" * 80)

print(
    f"Train      : {len(X_train):,} "
    f"(time {time_train.min()}–{time_train.max()})"
)

print(
    f"Validation : {len(X_val):,} "
    f"(time {time_val.min()}–{time_val.max()})"
)

print(
    f"Test       : {len(X_test):,} "
    f"(time {time_test.min()}–{time_test.max()})"
)


# ============================================================
# 4. DATA QUALITY
# ============================================================

print("\n[4] DATA QUALITY")
print("-" * 80)

print("NaN:", np.isnan(X_all_np).sum())
print("Inf:", np.isinf(X_all_np).sum())

if np.isnan(X_train).any() or np.isinf(X_train).any():
    raise ValueError("Train data contains NaN/Inf.")

if np.isnan(X_val).any() or np.isinf(X_val).any():
    raise ValueError("Validation data contains NaN/Inf.")

if np.isnan(X_test).any() or np.isinf(X_test).any():
    raise ValueError("Test data contains NaN/Inf.")

print("✓ No NaN/Inf")


# ============================================================
# 5. TRAIN ISOLATION FOREST
# ============================================================

print("\n[5] TRAINING ISOLATION FOREST")
print("-" * 80)

print("Training labels are NOT used.")

model = IsolationForest(
    n_estimators=300,
    contamination="auto",
    max_samples="auto",
    max_features=1.0,
    bootstrap=False,
    random_state=42,
    n_jobs=-1
)

model.fit(X_train)

print("✓ Model trained")
print("Trees:", model.n_estimators)


# ============================================================
# 6. ANOMALY SCORES
# ============================================================

print("\n[6] GENERATING ANOMALY SCORES")
print("-" * 80)

# sklearn decision_function:
# lower = more anomalous
#
# We reverse sign:
# higher = more anomalous

train_score = -model.decision_function(X_train)
val_score   = -model.decision_function(X_val)
test_score  = -model.decision_function(X_test)

print("✓ Train scores:", len(train_score))
print("✓ Validation scores:", len(val_score))
print("✓ Test scores:", len(test_score))


# ============================================================
# 7. SCORE DISTRIBUTION
# ============================================================

print("\n[7] ANOMALY SCORE DISTRIBUTION")
print("=" * 80)

score_summary = pd.DataFrame({
    "Split": ["Train", "Validation", "Test"],
    "Samples": [
        len(train_score),
        len(val_score),
        len(test_score)
    ],
    "Mean_Score": [
        train_score.mean(),
        val_score.mean(),
        test_score.mean()
    ],
    "Std_Score": [
        train_score.std(),
        val_score.std(),
        test_score.std()
    ],
    "Min_Score": [
        train_score.min(),
        val_score.min(),
        test_score.min()
    ],
    "Median_Score": [
        np.median(train_score),
        np.median(val_score),
        np.median(test_score)
    ],
    "Max_Score": [
        train_score.max(),
        val_score.max(),
        test_score.max()
    ]
})

print(score_summary.to_string(index=False))


# ============================================================
# 8. CREATE EVALUATION DATA
# ============================================================

train_eval = pd.DataFrame({
    "txId": tx_train,
    "Time step": time_train,
    "class": y_train,
    "anomaly_score": train_score
})

val_eval = pd.DataFrame({
    "txId": tx_val,
    "Time step": time_val,
    "class": y_val,
    "anomaly_score": val_score
})

test_eval = pd.DataFrame({
    "txId": tx_test,
    "Time step": time_test,
    "class": y_test,
    "anomaly_score": test_score
})

eval_df = pd.concat(
    [train_eval, val_eval, test_eval],
    ignore_index=True
)

print("\n[8] EVALUATION DATA CREATED")
print("-" * 80)

print("Shape:", eval_df.shape)
print("Missing scores:", eval_df["anomaly_score"].isna().sum())


# ============================================================
# 9. CLASS-WISE ANOMALY ANALYSIS
# ============================================================

print("\n[9] CLASS-WISE ANOMALY SCORE ANALYSIS")
print("=" * 80)

class_stats = (
    eval_df
    .groupby("class")["anomaly_score"]
    .agg(
        Samples="count",
        Mean="mean",
        Std="std",
        Min="min",
        Median="median",
        Max="max"
    )
)

print(class_stats.to_string())


# ============================================================
# 10. CLASS-WISE QUANTILES
# ============================================================

print("\n[10] CLASS-WISE SCORE QUANTILES")
print("=" * 80)

quantiles = (
    eval_df
    .groupby("class")["anomaly_score"]
    .quantile(
        [0.01, 0.05, 0.10, 0.25,
         0.50, 0.75, 0.90, 0.95, 0.99]
    )
    .unstack()
)

quantiles.columns = [
    "Q01", "Q05", "Q10", "Q25",
    "Q50", "Q75", "Q90", "Q95", "Q99"
]

print(quantiles.to_string())


# ============================================================
# 11. TOP ANOMALIES
# ============================================================

print("\n[11] TOP 30 MOST ANOMALOUS TRANSACTIONS")
print("=" * 80)

top30 = (
    eval_df
    .nlargest(30, "anomaly_score")
)

print(
    top30[
        ["txId", "Time step", "class", "anomaly_score"]
    ].to_string(index=False)
)


# ============================================================
# 12. TOP-K CLASS DISTRIBUTION
# ============================================================

print("\n[12] CLASS DISTRIBUTION AMONG TOP ANOMALIES")
print("=" * 80)

overall_class3_pct = (
    (eval_df["class"] == 3).mean() * 100
)

for k in [50, 100, 500, 1000, 5000]:

    top_k = eval_df.nlargest(k, "anomaly_score")

    counts = (
        top_k["class"]
        .value_counts()
        .sort_index()
    )

    class3_count = (
        top_k["class"] == 3
    ).sum()

    class3_pct = (
        class3_count / k * 100
    )

    enrichment = (
        class3_pct / overall_class3_pct
    )

    print(f"\nTop {k:,}")
    print(counts.to_string())

    print(
        f"Class-3 = {class3_pct:.2f}% | "
        f"Overall = {overall_class3_pct:.2f}% | "
        f"Enrichment = {enrichment:.3f}x"
    )


# ============================================================
# 13. TEMPORAL ANOMALY ANALYSIS
# ============================================================

print("\n[13] TEMPORAL ANOMALY ANALYSIS")
print("=" * 80)

temporal_stats = (
    eval_df
    .groupby("Time step")
    .agg(
        Transactions=("txId", "count"),
        Mean_Anomaly_Score=("anomaly_score", "mean"),
        Median_Anomaly_Score=("anomaly_score", "median"),
        Class3_Count=(
            "class",
            lambda x: (x == 3).sum()
        )
    )
)

temporal_stats["Class3_%"] = (
    temporal_stats["Class3_Count"]
    / temporal_stats["Transactions"]
    * 100
)

# Highest anomaly score = most anomalous
most_anomalous_time = (
    temporal_stats
    .sort_values(
        "Mean_Anomaly_Score",
        ascending=False
    )
    .head(15)
)

print(
    most_anomalous_time.to_string()
)


# ============================================================
# 14. TEST SET — FINAL ANALYSIS
# ============================================================

print("\n[14] TEST SET — FINAL UNSUPERVISED ANALYSIS")
print("=" * 80)

test_class_stats = (
    test_eval
    .groupby("class")["anomaly_score"]
    .agg(
        Samples="count",
        Mean="mean",
        Std="std",
        Min="min",
        Median="median",
        Max="max"
    )
)

print(test_class_stats.to_string())


# ============================================================
# 15. POST-HOC CLASS-3 EVALUATION
# ============================================================
#
# IMPORTANT:
# Class labels are NOT used during model training.
#
# They are used ONLY here to check whether the
# unsupervised anomaly ranking aligns with Class 3.
# ============================================================

print("\n[15] POST-HOC CLASS-3 EVALUATION")
print("=" * 80)

y_test_binary = (
    test_eval["class"].to_numpy() == 3
).astype(int)

scores_test = (
    test_eval["anomaly_score"].to_numpy()
)

roc_auc = roc_auc_score(
    y_test_binary,
    scores_test
)

pr_auc = average_precision_score(
    y_test_binary,
    scores_test
)

print(f"Test ROC-AUC : {roc_auc:.4f}")
print(f"Test PR-AUC  : {pr_auc:.4f}")


# ============================================================
# 16. TOP CLASS-3 ENRICHMENT ON TEST ONLY
# ============================================================

print("\n[16] TEST-SET CLASS-3 ENRICHMENT")
print("=" * 80)

test_overall_class3 = (
    (test_eval["class"] == 3).mean() * 100
)

for k in [50, 100, 500, 1000, 5000]:

    k = min(k, len(test_eval))

    top_k = test_eval.nlargest(
        k,
        "anomaly_score"
    )

    class3_count = (
        top_k["class"] == 3
    ).sum()

    class3_pct = (
        class3_count / k * 100
    )

    enrichment = (
        class3_pct / test_overall_class3
    )

    print(
        f"Top {k:5d}: "
        f"Class3 = {class3_pct:6.2f}% | "
        f"Enrichment = {enrichment:.3f}x"
    )


# ============================================================
# 17. SAVE RESULTS
# ============================================================

print("\n[17] SAVING RESULTS")
print("=" * 80)

output_file = "ellipticpp_isolation_forest_results.csv"

eval_df.to_csv(
    output_file,
    index=False
)

print("✓ Saved:", output_file)


# ============================================================
# 18. FINAL STATUS
# ============================================================

print("\n" + "=" * 80)
print("ELLIPTIC++ UNSUPERVISED PIPELINE COMPLETE")
print("=" * 80)

print("""
✓ 13 selected features
✓ Temporal split preserved
✓ Isolation Forest trained
✓ Labels NOT used during training
✓ Anomaly scores generated
✓ Class-wise analysis completed
✓ Top anomaly analysis completed
✓ Temporal analysis completed
✓ Test-set analysis completed
✓ ROC-AUC / PR-AUC calculated post-hoc
✓ Results saved
""")
```

```python
# =============================================================================
# BIT-TRACE — LABEL-INDEPENDENT UNSUPERVISED ANOMALY PIPELINE
# =============================================================================
# Pipeline:
#   1. Reuse existing loaded data OR reload automatically
#   2. Remove ID/time/label from feature candidates
#   3. Median imputation
#   4. Remove low-variance features
#   5. Remove highly correlated redundant features (|r| >= 0.90)
#   6. Temporal split: 1-34 / 35-41 / 42-49
#   7. Isolation Forest
#   8. LOF using a controlled training sample
#   9. Normalize model scores
#  10. Ensemble anomaly score
#  11. Post-hoc class evaluation ONLY
#  12. Temporal analysis
#  13. Save results
# =============================================================================

import os
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import VarianceThreshold
from sklearn.ensemble import IsolationForest
from sklearn.neighbors import LocalOutlierFactor

# =============================================================================
# CONFIGURATION
# =============================================================================

RANDOM_STATE = 42

CORRELATION_THRESHOLD = 0.90
VARIANCE_THRESHOLD = 1e-8

# Isolation Forest
IF_N_ESTIMATORS = 300
IF_CONTAMINATION = "auto"
IF_MAX_SAMPLES = "auto"

# LOF
LOF_TRAIN_SAMPLE = 30000
LOF_NEIGHBORS = 35

# Ensemble weights
IF_WEIGHT = 0.60
LOF_WEIGHT = 0.40

# =============================================================================
# HEADER
# =============================================================================

print("=" * 90)
print("BIT-TRACE — LABEL-INDEPENDENT UNSUPERVISED ANOMALY PIPELINE")
print("=" * 90)

# =============================================================================
# 1. GET DATA
# =============================================================================

print("\n[1] PREPARING INPUT DATA")
print("-" * 90)

# -------------------------------------------------------------------------
# Reuse existing variables if available
# -------------------------------------------------------------------------

required_existing = [
    "data",
    "X_all",
    "time_steps",
    "tx_ids"
]

existing_ok = all(name in globals() for name in required_existing)

if existing_ok:
    print("✓ Existing loaded variables detected")
    print("✓ Reusing current notebook data")

else:
    print("Existing variables not complete.")
    print("Searching for Elliptic++ files...")

    possible_feature_paths = [
        r"C:\Users\YASHIKA\Desktop\bit-trace\EllipticPlusPlus\Transactions Dataset\txs_features.csv",
        r"C:\Users\YASHIKA\Desktop\bit-trace\EllipticPlusPlus\txs_features.csv"
    ]

    possible_class_paths = [
        r"C:\Users\YASHIKA\Desktop\bit-trace\EllipticPlusPlus\Transactions Dataset\txs_classes.csv",
        r"C:\Users\YASHIKA\Desktop\bit-trace\EllipticPlusPlus\txs_classes.csv"
    ]

    feature_path = next(
        (p for p in possible_feature_paths if os.path.exists(p)),
        None
    )

    class_path = next(
        (p for p in possible_class_paths if os.path.exists(p)),
        None
    )

    if feature_path is None:
        raise FileNotFoundError(
            "txs_features.csv not found. "
            "Please set FEATURE_PATH manually."
        )

    if class_path is None:
        raise FileNotFoundError(
            "txs_classes.csv not found. "
            "Please set CLASS_PATH manually."
        )

    print("Feature file:", feature_path)
    print("Class file:", class_path)

    # ---------------------------------------------------------------------
    # Load class data
    # ---------------------------------------------------------------------

    classes = pd.read_csv(class_path)

    classes.columns = [
        str(c).strip() for c in classes.columns
    ]

    # Normalize names
    rename_map = {}

    for c in classes.columns:
        lc = c.lower().replace(" ", "").replace("_", "")

        if lc == "txid":
            rename_map[c] = "txId"

        elif lc == "class":
            rename_map[c] = "class"

    classes = classes.rename(columns=rename_map)

    # ---------------------------------------------------------------------
    # Read feature header first
    # ---------------------------------------------------------------------

    header = pd.read_csv(
        feature_path,
        nrows=0
    )

    feature_columns = list(header.columns)

    print("Total feature-file columns:", len(feature_columns))

    # Detect ID and time columns
    id_candidates = [
        c for c in feature_columns
        if str(c).lower() in ["txid", "tx_id", "transactionid"]
    ]

    time_candidates = [
        c for c in feature_columns
        if str(c).lower().replace("_", " ").strip()
        in ["time step", "timestep", "time"]
    ]

    if not id_candidates:
        raise ValueError("Could not detect txId column.")

    if not time_candidates:
        raise ValueError("Could not detect Time step column.")

    id_col = id_candidates[0]
    time_col = time_candidates[0]

    print("ID column  :", id_col)
    print("Time column:", time_col)

    # ---------------------------------------------------------------------
    # Load all numeric feature columns
    # ---------------------------------------------------------------------

    data = pd.read_csv(
        feature_path,
        usecols=feature_columns
    )

    # Normalize ID name
    if id_col != "txId":
        data = data.rename(columns={id_col: "txId"})
        id_col = "txId"

    if time_col != "Time step":
        data = data.rename(columns={time_col: "Time step"})
        time_col = "Time step"

    # ---------------------------------------------------------------------
    # Merge labels
    # ---------------------------------------------------------------------

    data = data.merge(
        classes[["txId", "class"]],
        on="txId",
        how="inner"
    )

    print("Loaded merged shape:", data.shape)

    # ---------------------------------------------------------------------
    # Create base arrays
    # ---------------------------------------------------------------------

    tx_ids = data["txId"].to_numpy()
    time_steps = pd.to_numeric(
        data["Time step"],
        errors="coerce"
    ).to_numpy()

# =============================================================================
# 2. BASIC DATA CHECK
# =============================================================================

print("\n[2] DATA CHECK")
print("-" * 90)

print("Rows:", len(data))

print(
    "Time range:",
    np.nanmin(time_steps),
    "to",
    np.nanmax(time_steps)
)

print("Missing txId:", pd.Series(tx_ids).isna().sum())
print("Missing time:", np.isnan(time_steps).sum())

if "class" in data.columns:
    print("\nPost-hoc class distribution:")
    print(data["class"].value_counts().sort_index())

# =============================================================================
# 3. IDENTIFY CANDIDATE FEATURES
# =============================================================================

print("\n[3] LABEL-INDEPENDENT FEATURE CANDIDATES")
print("-" * 90)

# IMPORTANT:
# class, txId and Time step are NOT feature candidates.
# Labels are completely excluded from feature selection.

exclude_columns = {
    "txId",
    "class",
    "Time step"
}

candidate_features = [
    c for c in data.columns
    if c not in exclude_columns
    and pd.api.types.is_numeric_dtype(data[c])
]

print("Candidate numeric features:", len(candidate_features))

# =============================================================================
# 4. NUMERIC MATRIX
# =============================================================================

X_raw = data[candidate_features].copy()

# Replace infinities
X_raw = X_raw.replace(
    [np.inf, -np.inf],
    np.nan
)

print("Raw feature matrix:", X_raw.shape)

# =============================================================================
# 5. MISSING VALUE HANDLING
# =============================================================================

print("\n[4] MEDIAN IMPUTATION")
print("-" * 90)

missing_before = int(X_raw.isna().sum().sum())

print("NaN before:", missing_before)

imputer = SimpleImputer(
    strategy="median"
)

X_imputed = imputer.fit_transform(X_raw)

print("NaN after:", np.isnan(X_imputed).sum())

# =============================================================================
# 6. LOW VARIANCE FILTER
# =============================================================================

print("\n[5] LOW-VARIANCE FILTER")
print("-" * 90)

variance_selector = VarianceThreshold(
    threshold=VARIANCE_THRESHOLD
)

X_variance = variance_selector.fit_transform(
    X_imputed
)

variance_features = [
    candidate_features[i]
    for i, keep in enumerate(
        variance_selector.get_support()
    )
    if keep
]

print("Before:", len(candidate_features))
print("After :", len(variance_features))
print("Removed:", len(candidate_features) - len(variance_features))

# =============================================================================
# 7. CORRELATION REDUNDANCY FILTER
# =============================================================================

print("\n[6] CORRELATION REDUNDANCY FILTER")
print("-" * 90)

X_var_df = pd.DataFrame(
    X_variance,
    columns=variance_features
)

print(
    "Calculating correlation matrix for",
    len(variance_features),
    "features..."
)

corr_matrix = X_var_df.corr(
    method="pearson"
).abs()

# Upper triangle
upper = corr_matrix.where(
    np.triu(
        np.ones(corr_matrix.shape),
        k=1
    ).astype(bool)
)

to_remove = [
    column
    for column in upper.columns
    if any(
        upper[column] >= CORRELATION_THRESHOLD
    )
]

selected_features = [
    f for f in variance_features
    if f not in to_remove
]

print("Before correlation filter:", len(variance_features))
print("Highly correlated removed :", len(to_remove))
print("Final selected features   :", len(selected_features))

print("\nSelected features:")

for i, f in enumerate(selected_features, 1):
    print(f"{i:2d}. {f}")

# =============================================================================
# 8. FINAL MATRIX
# =============================================================================

X_final = X_var_df[selected_features].to_numpy(
    dtype=np.float32
)

print("\nFinal X shape:", X_final.shape)

# =============================================================================
# 9. STANDARDIZATION
# =============================================================================

print("\n[7] STANDARDIZATION")
print("-" * 90)

scaler = StandardScaler()

X_scaled = scaler.fit_transform(
    X_final
).astype(np.float32)

print("Scaled matrix:", X_scaled.shape)

print(
    "NaN:",
    np.isnan(X_scaled).sum(),
    "| Inf:",
    np.isinf(X_scaled).sum()
)

# =============================================================================
# 10. TEMPORAL SPLIT
# =============================================================================

print("\n[8] TEMPORAL SPLIT")
print("-" * 90)

train_idx = np.where(
    time_steps <= 34
)[0]

val_idx = np.where(
    (time_steps >= 35) &
    (time_steps <= 41)
)[0]

test_idx = np.where(
    time_steps >= 42
)[0]

print(
    f"Train      : {len(train_idx):,} "
    "(time 1–34)"
)

print(
    f"Validation : {len(val_idx):,} "
    "(time 35–41)"
)

print(
    f"Test       : {len(test_idx):,} "
    "(time 42–49)"
)

# =============================================================================
# 11. TEMPORAL DATA
# =============================================================================

X_train = X_scaled[train_idx]
X_val = X_scaled[val_idx]
X_test = X_scaled[test_idx]

print("\nTrain matrix:", X_train.shape)
print("Val matrix  :", X_val.shape)
print("Test matrix :", X_test.shape)

# =============================================================================
# 12. ISOLATION FOREST
# =============================================================================

print("\n[9] TRAINING ISOLATION FOREST")
print("-" * 90)

if_model = IsolationForest(
    n_estimators=IF_N_ESTIMATORS,
    contamination=IF_CONTAMINATION,
    max_samples=IF_MAX_SAMPLES,
    random_state=RANDOM_STATE,
    n_jobs=-1
)

if_model.fit(X_train)

print("✓ Isolation Forest trained")
print("Trees:", IF_N_ESTIMATORS)

# -------------------------------------------------------------------------
# IMPORTANT:
# sklearn decision_function:
# higher = more normal
# lower  = more anomalous
#
# We reverse it so:
# higher anomaly score = more anomalous
# -------------------------------------------------------------------------

if_train = -if_model.decision_function(X_train)
if_val = -if_model.decision_function(X_val)
if_test = -if_model.decision_function(X_test)

print("IF train scores:", len(if_train))
print("IF val scores  :", len(if_val))
print("IF test scores :", len(if_test))

# =============================================================================
# 13. LOF
# =============================================================================

print("\n[10] TRAINING LOF")
print("-" * 90)

# LOF is computationally expensive.
# Use a representative training subset.

rng = np.random.RandomState(
    RANDOM_STATE
)

if len(X_train) > LOF_TRAIN_SAMPLE:

    lof_sample_idx = rng.choice(
        len(X_train),
        size=LOF_TRAIN_SAMPLE,
        replace=False
    )

    X_lof_train = X_train[
        lof_sample_idx
    ]

else:

    X_lof_train = X_train

print(
    "LOF training samples:",
    len(X_lof_train)
)

lof_model = LocalOutlierFactor(
    n_neighbors=LOF_NEIGHBORS,
    novelty=True,
    contamination="auto",
    n_jobs=-1
)

lof_model.fit(
    X_lof_train
)

print("✓ LOF trained")
print("Neighbors:", LOF_NEIGHBORS)

# Higher = more anomalous
lof_train = -lof_model.decision_function(X_train)
lof_val = -lof_model.decision_function(X_val)
lof_test = -lof_model.decision_function(X_test)

print("LOF train scores:", len(lof_train))
print("LOF val scores  :", len(lof_val))
print("LOF test scores :", len(lof_test))

# =============================================================================
# 14. SCORE NORMALIZATION
# =============================================================================

print("\n[11] NORMALIZING ANOMALY SCORES")
print("-" * 90)

# Use TRAIN statistics only.
# This prevents validation/test information leakage.

def robust_zscore(
    train_score,
    other_scores
):
    median = np.median(
        train_score
    )

    q1 = np.percentile(
        train_score,
        25
    )

    q3 = np.percentile(
        train_score,
        75
    )

    iqr = q3 - q1

    if iqr == 0:
        iqr = 1.0

    train_norm = (
        train_score - median
    ) / iqr

    other_norm = [
        (
            s - median
        ) / iqr
        for s in other_scores
    ]

    return (
        train_norm,
        *other_norm
    )

(
    if_train_n,
    if_val_n,
    if_test_n
) = robust_zscore(
    if_train,
    [if_val, if_test]
)

(
    lof_train_n,
    lof_val_n,
    lof_test_n
) = robust_zscore(
    lof_train,
    [lof_val, lof_test]
)

# =============================================================================
# 15. ENSEMBLE
# =============================================================================

print("\n[12] ENSEMBLE ANOMALY SCORE")
print("-" * 90)

ensemble_train = (
    IF_WEIGHT * if_train_n +
    LOF_WEIGHT * lof_train_n
)

ensemble_val = (
    IF_WEIGHT * if_val_n +
    LOF_WEIGHT * lof_val_n
)

ensemble_test = (
    IF_WEIGHT * if_test_n +
    LOF_WEIGHT * lof_test_n
)

print(
    "Weights:",
    f"Isolation Forest={IF_WEIGHT}",
    f"| LOF={LOF_WEIGHT}"
)

# =============================================================================
# 16. BUILD COMPLETE RESULTS DATAFRAME
# =============================================================================

print("\n[13] BUILDING RESULTS")
print("-" * 90)

results = pd.DataFrame({
    "txId": tx_ids,
    "Time step": time_steps,
    "class": data["class"].to_numpy(),
    "IF_score": np.nan,
    "LOF_score": np.nan,
    "ensemble_score": np.nan
})

results.loc[
    train_idx,
    "IF_score"
] = if_train

results.loc[
    val_idx,
    "IF_score"
] = if_val

results.loc[
    test_idx,
    "IF_score"
] = if_test

results.loc[
    train_idx,
    "LOF_score"
] = lof_train

results.loc[
    val_idx,
    "LOF_score"
] = lof_val

results.loc[
    test_idx,
    "LOF_score"
] = lof_test

results.loc[
    train_idx,
    "ensemble_score"
] = ensemble_train

results.loc[
    val_idx,
    "ensemble_score"
] = ensemble_val

results.loc[
    test_idx,
    "ensemble_score"
] = ensemble_test

print(
    "Results shape:",
    results.shape
)

# =============================================================================
# 17. SCORE DISTRIBUTION
# =============================================================================

print("\n[14] MODEL SCORE DISTRIBUTION")
print("=" * 90)

distribution = pd.DataFrame({
    "Split": [
        "Train",
        "Validation",
        "Test"
    ],
    "Samples": [
        len(ensemble_train),
        len(ensemble_val),
        len(ensemble_test)
    ],
    "IF_Mean": [
        np.mean(if_train),
        np.mean(if_val),
        np.mean(if_test)
    ],
    "IF_Std": [
        np.std(if_train),
        np.std(if_val),
        np.std(if_test)
    ],
    "LOF_Mean": [
        np.mean(lof_train),
        np.mean(lof_val),
        np.mean(lof_test)
    ],
    "LOF_Std": [
        np.std(lof_train),
        np.std(lof_val),
        np.std(lof_test)
    ],
    "Ensemble_Mean": [
        np.mean(ensemble_train),
        np.mean(ensemble_val),
        np.mean(ensemble_test)
    ],
    "Ensemble_Std": [
        np.std(ensemble_train),
        np.std(ensemble_val),
        np.std(ensemble_test)
    ]
})

print(
    distribution.to_string(
        index=False
    )
)

# =============================================================================
# 18. TOP ANOMALIES
# =============================================================================

print("\n[15] TOP ANOMALOUS TRANSACTIONS")
print("=" * 90)

test_results = results.loc[
    test_idx
].copy()

test_results = test_results.sort_values(
    "ensemble_score",
    ascending=False
)

print(
    test_results[
        [
            "txId",
            "Time step",
            "class",
            "IF_score",
            "LOF_score",
            "ensemble_score"
        ]
    ].head(30).to_string(
        index=False
    )
)

# =============================================================================
# 19. POST-HOC CLASS ANALYSIS
# =============================================================================
# IMPORTANT:
# Class labels were NOT used for model training or feature selection.
# They are used here ONLY to understand what the anomaly detector finds.

print("\n[16] POST-HOC CLASS ANALYSIS")
print("=" * 90)

overall_class3 = (
    results["class"].eq(3).mean()
)

print(
    f"Overall Class-3 percentage: "
    f"{overall_class3 * 100:.2f}%"
)

for k in [
    50,
    100,
    500,
    1000,
    5000
]:

    top = test_results.head(k)

    class3_count = (
        top["class"] == 3
    ).sum()

    class3_pct = (
        class3_count / k
    )

    enrichment = (
        class3_pct /
        overall_class3
        if overall_class3 > 0
        else np.nan
    )

    print(
        f"Top {k:5d}: "
        f"Class-3 = {class3_pct*100:6.2f}% | "
        f"Enrichment = {enrichment:.3f}x"
    )

# =============================================================================
# 20. CLASS-WISE SCORE ANALYSIS
# =============================================================================

print("\n[17] CLASS-WISE ENSEMBLE SCORE ANALYSIS")
print("=" * 90)

class_analysis = (
    test_results
    .groupby("class")["ensemble_score"]
    .agg(
        Samples="count",
        Mean="mean",
        Std="std",
        Min="min",
        Median="median",
        Max="max"
    )
)

print(
    class_analysis.to_string()
)

# =============================================================================
# 21. CLASS-WISE QUANTILES
# =============================================================================

print("\n[18] CLASS-WISE QUANTILES")
print("=" * 90)

class_quantiles = (
    test_results
    .groupby("class")["ensemble_score"]
    .quantile(
        [
            0.01,
            0.05,
            0.10,
            0.25,
            0.50,
            0.75,
            0.90,
            0.95,
            0.99
        ]
    )
    .unstack()
)

class_quantiles.columns = [
    "Q01",
    "Q05",
    "Q10",
    "Q25",
    "Q50",
    "Q75",
    "Q90",
    "Q95",
    "Q99"
]

print(
    class_quantiles.to_string()
)

# =============================================================================
# 22. TEMPORAL ANALYSIS
# =============================================================================

print("\n[19] TEMPORAL ANOMALY ANALYSIS")
print("=" * 90)

temporal = (
    test_results
    .groupby("Time step")
    .agg(
        Transactions=("txId", "count"),
        Mean_Anomaly=("ensemble_score", "mean"),
        Median_Anomaly=("ensemble_score", "median"),
        Max_Anomaly=("ensemble_score", "max"),
        Class3_Count=(
            "class",
            lambda x: (x == 3).sum()
        )
    )
)

temporal["Class3_%"] = (
    temporal["Class3_Count"] /
    temporal["Transactions"] *
    100
)

# Sort by anomaly intensity
temporal_sorted = temporal.sort_values(
    "Mean_Anomaly",
    ascending=False
)

print(
    temporal_sorted.head(15).to_string()
)

# =============================================================================
# 23. MODEL AGREEMENT
# =============================================================================

print("\n[20] MODEL AGREEMENT")
print("=" * 90)

# Top 1000 from each model

top_k = min(
    1000,
    len(test_results)
)

if_top = set(
    test_results
    .sort_values(
        "IF_score",
        ascending=False
    )
    .head(top_k)["txId"]
)

lof_top = set(
    test_results
    .sort_values(
        "LOF_score",
        ascending=False
    )
    .head(top_k)["txId"]
)

ensemble_top = set(
    test_results
    .sort_values(
        "ensemble_score",
        ascending=False
    )
    .head(top_k)["txId"]
)

if_lof_overlap = len(
    if_top & lof_top
)

if_ensemble_overlap = len(
    if_top & ensemble_top
)

lof_ensemble_overlap = len(
    lof_top & ensemble_top
)

all_three = len(
    if_top &
    lof_top &
    ensemble_top
)

print(
    f"Top {top_k} IF ∩ LOF:",
    if_lof_overlap
)

print(
    f"Top {top_k} IF ∩ Ensemble:",
    if_ensemble_overlap
)

print(
    f"Top {top_k} LOF ∩ Ensemble:",
    lof_ensemble_overlap
)

print(
    f"Top {top_k} common to all three:",
    all_three
)

# =============================================================================
# 24. FEATURE SUMMARY
# =============================================================================

print("\n[21] FINAL FEATURE SUMMARY")
print("=" * 90)

print(
    "Original numeric candidates:",
    len(candidate_features)
)

print(
    "After variance filtering:",
    len(variance_features)
)

print(
    "Removed by correlation:",
    len(to_remove)
)

print(
    "Final unsupervised features:",
    len(selected_features)
)

print("\nFinal features:")

for i, f in enumerate(
    selected_features,
    1
):
    print(
        f"{i:3d}. {f}"
    )

# =============================================================================
# 25. SAVE RESULTS
# =============================================================================

print("\n[22] SAVING RESULTS")
print("=" * 90)

results_file = (
    "ellipticpp_unsupervised_"
    "if_lof_ensemble_results.csv"
)

features_file = (
    "ellipticpp_unsupervised_"
    "selected_features.txt"
)

temporal_file = (
    "ellipticpp_unsupervised_"
    "temporal_analysis.csv"
)

results.to_csv(
    results_file,
    index=False
)

with open(
    features_file,
    "w",
    encoding="utf-8"
) as f:

    f.write(
        "BIT-TRACE LABEL-INDEPENDENT "
        "UNSUPERVISED FEATURE SET\n"
    )

    f.write(
        "=" * 70 + "\n"
    )

    for i, feature in enumerate(
        selected_features,
        1
    ):

        f.write(
            f"{i}. {feature}\n"
        )

temporal.to_csv(
    temporal_file
)

print("✓ Results saved:")
print(" ", results_file)
print(" ", features_file)
print(" ", temporal_file)

# =============================================================================
# 26. FINAL STATUS
# =============================================================================

print("\n")
print("=" * 90)
print("PIPELINE COMPLETE")
print("=" * 90)

print("✓ Label-independent feature selection")
print("✓ Median imputation")
print("✓ Low-variance filtering")
print("✓ Correlation redundancy filtering")
print("✓ Temporal split preserved")
print("✓ Isolation Forest")
print("✓ LOF")
print("✓ Ensemble anomaly score")
print("✓ Post-hoc class analysis")
print("✓ Temporal anomaly analysis")
print("✓ Model agreement analysis")
print("✓ Results exported")

print("\nIMPORTANT:")
print(
    "Class labels were NOT used for feature selection "
    "or model training."
)

print(
    "They were used only for post-hoc evaluation."
)

print("\nNEXT BIT-TRACE PHASE:")
print(
    "Use txs_edgelist.csv to generate graph/transaction-network "
    "features and combine them with this anomaly engine."
)

print("=" * 90)
```

```python
# =============================================================================
# BIT-TRACE — ELLIPTIC++ GRAPH + BEHAVIORAL UNSUPERVISED ANOMALY ENGINE
# SINGLE OPTIMIZED PIPELINE
# =============================================================================

import os
import gc
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd

from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.ensemble import IsolationForest
from sklearn.neighbors import LocalOutlierFactor

print("=" * 90)
print("BIT-TRACE — GRAPH + BEHAVIORAL UNSUPERVISED ANOMALY PIPELINE")
print("=" * 90)


# =============================================================================
# 1. FILE DISCOVERY
# =============================================================================

print("\n[1] LOCATING ELLIPTIC++ FILES")
print("-" * 90)

SEARCH_ROOT = r"C:\Users\YASHIKA\Desktop\bit-trace"

FEATURE_FILE = None
CLASS_FILE = None
EDGE_FILE = None

for root, dirs, files in os.walk(SEARCH_ROOT):

    for f in files:

        fl = f.lower()

        full = os.path.join(root, f)

        if fl == "txs_features.csv" and FEATURE_FILE is None:
            FEATURE_FILE = full

        elif fl == "txs_classes.csv" and CLASS_FILE is None:
            CLASS_FILE = full

        elif fl == "txs_edgelist.csv" and EDGE_FILE is None:
            EDGE_FILE = full


print("FEATURE :", FEATURE_FILE)
print("CLASS   :", CLASS_FILE)
print("EDGES   :", EDGE_FILE)

if EDGE_FILE is None:
    raise FileNotFoundError(
        "txs_edgelist.csv not found."
    )

if FEATURE_FILE is None:
    raise FileNotFoundError(
        "txs_features.csv not found."
    )

if CLASS_FILE is None:
    raise FileNotFoundError(
        "txs_classes.csv not found."
    )


# =============================================================================
# 2. SELECTED BEHAVIORAL FEATURES
# =============================================================================

BEHAVIOR_FEATURES = [
    "Local_feature_22",
    "Local_feature_25",
    "Local_feature_47",
    "Local_feature_52",
    "Local_feature_53",
    "Local_feature_54",
    "Local_feature_59",
    "Local_feature_77",
    "Local_feature_89",
    "Local_feature_90",
    "Aggregate_feature_14",
    "Aggregate_feature_21",
    "Aggregate_feature_32"
]

print("\n[2] BEHAVIORAL FEATURES")
print("-" * 90)

for i, f in enumerate(BEHAVIOR_FEATURES, 1):
    print(f"{i:2d}. {f}")


# =============================================================================
# 3. LOAD ONLY REQUIRED TRANSACTION DATA
# =============================================================================

print("\n[3] LOADING TRANSACTION DATA")
print("-" * 90)

header = pd.read_csv(FEATURE_FILE, nrows=0)

cols = header.columns.tolist()

id_col = "txId"
time_col = "Time step"

required = [id_col, time_col] + BEHAVIOR_FEATURES

missing = [c for c in required if c not in cols]

if missing:
    raise ValueError(
        f"Missing transaction columns: {missing}"
    )

data = pd.read_csv(
    FEATURE_FILE,
    usecols=required,
    low_memory=False
)

data[id_col] = pd.to_numeric(
    data[id_col],
    errors="coerce"
).astype("Int64")

data[time_col] = pd.to_numeric(
    data[time_col],
    errors="coerce"
)

print("Transaction data:", data.shape)

print("Missing txId:", data[id_col].isna().sum())
print("Missing time:", data[time_col].isna().sum())


# =============================================================================
# 4. NUMERIC CLEANING
# =============================================================================

print("\n[4] CLEANING BEHAVIORAL FEATURES")
print("-" * 90)

X_behavior = data[BEHAVIOR_FEATURES].copy()

X_behavior = X_behavior.replace(
    [np.inf, -np.inf],
    np.nan
)

imputer = SimpleImputer(strategy="median")

X_behavior = imputer.fit_transform(X_behavior)

X_behavior = np.asarray(
    X_behavior,
    dtype=np.float32
)

print("Behavior matrix:", X_behavior.shape)
print("NaN:", np.isnan(X_behavior).sum())
print("Inf:", np.isinf(X_behavior).sum())


# =============================================================================
# 5. LOAD EDGE LIST
# =============================================================================

print("\n[5] LOADING TRANSACTION EDGE LIST")
print("-" * 90)

edge_header = pd.read_csv(
    EDGE_FILE,
    nrows=5
)

print("Edge columns detected:")
print(edge_header.columns.tolist())

# Re-read header only
eh = pd.read_csv(
    EDGE_FILE,
    nrows=0
)

edge_cols = eh.columns.tolist()

print("Number of edge columns:", len(edge_cols))


# =============================================================================
# 6. AUTOMATIC EDGE COLUMN DETECTION
# =============================================================================

def find_source_target(columns):

    lower_map = {
        c.lower().strip(): c
        for c in columns
    }

    source_candidates = [
        "txid1",
        "txid_1",
        "source",
        "src",
        "from",
        "source_txid",
        "tx1",
        "id1"
    ]

    target_candidates = [
        "txid2",
        "txid_2",
        "target",
        "dst",
        "to",
        "target_txid",
        "tx2",
        "id2"
    ]

    source = None
    target = None

    for c in source_candidates:
        if c in lower_map:
            source = lower_map[c]
            break

    for c in target_candidates:
        if c in lower_map:
            target = lower_map[c]
            break

    return source, target


SOURCE_COL, TARGET_COL = find_source_target(edge_cols)

print("\nDetected source:", SOURCE_COL)
print("Detected target:", TARGET_COL)


# =============================================================================
# 7. FALLBACK EDGE DETECTION
# =============================================================================

if SOURCE_COL is None or TARGET_COL is None:

    numeric_candidates = []

    sample_edges = pd.read_csv(
        EDGE_FILE,
        nrows=10000
    )

    for c in sample_edges.columns:

        converted = pd.to_numeric(
            sample_edges[c],
            errors="coerce"
        )

        valid_ratio = converted.notna().mean()

        if valid_ratio > 0.95:
            numeric_candidates.append(c)

    print(
        "Numeric edge candidates:",
        numeric_candidates
    )

    if len(numeric_candidates) >= 2:

        SOURCE_COL = numeric_candidates[0]
        TARGET_COL = numeric_candidates[1]

    else:

        raise ValueError(
            "Could not automatically detect source/target edge columns. "
            f"Columns found: {edge_cols}"
        )


print(
    f"Using edge columns: {SOURCE_COL} -> {TARGET_COL}"
)


# =============================================================================
# 8. LOAD EDGE LIST
# =============================================================================

print("\n[6] READING EDGE LIST")
print("-" * 90)

edges = pd.read_csv(
    EDGE_FILE,
    usecols=[SOURCE_COL, TARGET_COL],
    low_memory=False
)

edges.columns = ["source", "target"]

edges["source"] = pd.to_numeric(
    edges["source"],
    errors="coerce"
)

edges["target"] = pd.to_numeric(
    edges["target"],
    errors="coerce"
)

edges = edges.dropna(
    subset=["source", "target"]
)

edges["source"] = edges["source"].astype(np.int64)
edges["target"] = edges["target"].astype(np.int64)

print("Raw edges:", len(edges))


# =============================================================================
# 9. REMOVE SELF LOOPS
# =============================================================================

edges = edges[
    edges["source"] != edges["target"]
]

print("After self-loop removal:", len(edges))


# =============================================================================
# 10. REMOVE DUPLICATE EDGES
# =============================================================================

edges = edges.drop_duplicates()

print("Unique edges:", len(edges))


# =============================================================================
# 11. GRAPH FEATURE ENGINEERING
# =============================================================================

print("\n[7] GENERATING GRAPH FEATURES")
print("-" * 90)

tx_ids_array = data[id_col].to_numpy(
    dtype=np.int64
)

n = len(data)

tx_to_idx = pd.Series(
    np.arange(n, dtype=np.int32),
    index=tx_ids_array
)

source_idx = tx_to_idx.reindex(
    edges["source"]
).to_numpy()

target_idx = tx_to_idx.reindex(
    edges["target"]
).to_numpy()

valid = (
    ~pd.isna(source_idx)
    &
    ~pd.isna(target_idx)
)

source_idx = source_idx[valid].astype(np.int32)
target_idx = target_idx[valid].astype(np.int32)

print("Valid graph edges:", len(source_idx))


# =============================================================================
# 12. DEGREE FEATURES
# =============================================================================

print("\n[8] DEGREE FEATURES")
print("-" * 90)

in_degree = np.bincount(
    target_idx,
    minlength=n
).astype(np.float32)

out_degree = np.bincount(
    source_idx,
    minlength=n
).astype(np.float32)

total_degree = (
    in_degree +
    out_degree
)

print(
    "Mean total degree:",
    total_degree.mean()
)

print(
    "Maximum total degree:",
    total_degree.max()
)


# =============================================================================
# 13. RECIPROCAL / BIDIRECTIONAL CONNECTION
# =============================================================================

print("\n[9] RECIPROCAL EDGE FEATURES")
print("-" * 90)

# Encode directed edge pairs into a structured array.
# This avoids Python set of millions of tuples.

edge_pairs = np.empty(
    len(source_idx),
    dtype=[
        ("s", np.int32),
        ("t", np.int32)
    ]
)

edge_pairs["s"] = source_idx
edge_pairs["t"] = target_idx

reverse_pairs = np.empty(
    len(source_idx),
    dtype=[
        ("s", np.int32),
        ("t", np.int32)
    ]
)

reverse_pairs["s"] = target_idx
reverse_pairs["t"] = source_idx

reverse_exists = np.isin(
    reverse_pairs,
    edge_pairs
)

reciprocal_degree = np.zeros(
    n,
    dtype=np.float32
)

np.add.at(
    reciprocal_degree,
    source_idx[reverse_exists],
    1
)

np.add.at(
    reciprocal_degree,
    target_idx[reverse_exists],
    1
)

print(
    "Nodes with reciprocal connections:",
    np.count_nonzero(reciprocal_degree)
)


# =============================================================================
# 14. APPROXIMATE NEIGHBORHOOD FEATURES
# =============================================================================

print("\n[10] NEIGHBORHOOD FEATURES")
print("-" * 90)

# Neighbor count already represented by degree.
# Add edge participation and degree imbalance.

degree_imbalance = np.abs(
    in_degree -
    out_degree
)

in_out_ratio = (
    (in_degree + 1.0) /
    (out_degree + 1.0)
)

log_total_degree = np.log1p(
    total_degree
)

log_in_degree = np.log1p(
    in_degree
)

log_out_degree = np.log1p(
    out_degree
)

log_reciprocal = np.log1p(
    reciprocal_degree
)


# =============================================================================
# 15. GRAPH FEATURE MATRIX
# =============================================================================

graph_features = np.column_stack([
    in_degree,
    out_degree,
    total_degree,
    degree_imbalance,
    in_out_ratio,
    log_total_degree,
    log_in_degree,
    log_out_degree,
    reciprocal_degree,
    log_reciprocal
]).astype(np.float32)

GRAPH_FEATURE_NAMES = [
    "graph_in_degree",
    "graph_out_degree",
    "graph_total_degree",
    "graph_degree_imbalance",
    "graph_in_out_ratio",
    "graph_log_total_degree",
    "graph_log_in_degree",
    "graph_log_out_degree",
    "graph_reciprocal_degree",
    "graph_log_reciprocal_degree"
]

print(
    "Graph feature matrix:",
    graph_features.shape
)


# =============================================================================
# 16. COMBINE BEHAVIOR + GRAPH
# =============================================================================

print("\n[11] COMBINING BEHAVIORAL + GRAPH FEATURES")
print("-" * 90)

X_combined = np.column_stack([
    X_behavior,
    graph_features
]).astype(np.float32)

ALL_FEATURE_NAMES = (
    BEHAVIOR_FEATURES +
    GRAPH_FEATURE_NAMES
)

print("Behavior features:", len(BEHAVIOR_FEATURES))
print("Graph features   :", len(GRAPH_FEATURE_NAMES))
print("Total features   :", len(ALL_FEATURE_NAMES))
print("Final X shape     :", X_combined.shape)


# =============================================================================
# 17. CLEAN FINAL MATRIX
# =============================================================================

X_combined = np.nan_to_num(
    X_combined,
    nan=0.0,
    posinf=0.0,
    neginf=0.0
).astype(np.float32)


# =============================================================================
# 18. STANDARDIZATION
# =============================================================================

print("\n[12] STANDARDIZATION")
print("-" * 90)

scaler = StandardScaler()

X_scaled = scaler.fit_transform(
    X_combined
).astype(np.float32)

print("Scaled:", X_scaled.shape)

del X_combined
gc.collect()


# =============================================================================
# 19. TEMPORAL SPLIT
# =============================================================================

print("\n[13] TEMPORAL SPLIT")
print("-" * 90)

time_steps = data[time_col].to_numpy(
    dtype=np.int16
)

train_idx = np.where(
    time_steps <= 34
)[0]

val_idx = np.where(
    (time_steps >= 35) &
    (time_steps <= 41)
)[0]

test_idx = np.where(
    time_steps >= 42
)[0]

print(
    "Train:",
    len(train_idx),
    "(1–34)"
)

print(
    "Validation:",
    len(val_idx),
    "(35–41)"
)

print(
    "Test:",
    len(test_idx),
    "(42–49)"
)


X_train = X_scaled[train_idx]
X_val = X_scaled[val_idx]
X_test = X_scaled[test_idx]


# =============================================================================
# 20. ISOLATION FOREST
# =============================================================================

print("\n[14] TRAINING ISOLATION FOREST")
print("-" * 90)

IF_MODEL = IsolationForest(
    n_estimators=300,
    contamination="auto",
    max_samples="auto",
    max_features=1.0,
    random_state=42,
    n_jobs=-1
)

IF_MODEL.fit(
    X_train
)

if_train = IF_MODEL.score_samples(X_train)
if_val = IF_MODEL.score_samples(X_val)
if_test = IF_MODEL.score_samples(X_test)

print("Isolation Forest complete.")


# =============================================================================
# 21. LOF
# =============================================================================

print("\n[15] TRAINING LOF")
print("-" * 90)

# Use bounded reference sample to keep runtime reasonable.
rng = np.random.default_rng(42)

LOF_SAMPLE_SIZE = min(
    30000,
    len(X_train)
)

lof_reference_idx = rng.choice(
    len(X_train),
    size=LOF_SAMPLE_SIZE,
    replace=False
)

X_lof_train = X_train[
    lof_reference_idx
]

LOF_MODEL = LocalOutlierFactor(
    n_neighbors=35,
    novelty=True,
    contamination="auto",
    n_jobs=-1
)

LOF_MODEL.fit(
    X_lof_train
)

lof_train = LOF_MODEL.score_samples(X_train)
lof_val = LOF_MODEL.score_samples(X_val)
lof_test = LOF_MODEL.score_samples(X_test)

print("LOF complete.")


# =============================================================================
# 22. SCORE NORMALIZATION
# =============================================================================

print("\n[16] NORMALIZING MODEL SCORES")
print("-" * 90)

def robust_normalize(train_score, score):

    median = np.median(
        train_score
    )

    q25, q75 = np.percentile(
        train_score,
        [25, 75]
    )

    iqr = q75 - q25

    if iqr == 0:
        iqr = 1.0

    return (
        score - median
    ) / iqr


if_train_n = robust_normalize(
    if_train,
    if_train
)

if_val_n = robust_normalize(
    if_train,
    if_val
)

if_test_n = robust_normalize(
    if_train,
    if_test
)

lof_train_n = robust_normalize(
    lof_train,
    lof_train
)

lof_val_n = robust_normalize(
    lof_train,
    lof_val
)

lof_test_n = robust_normalize(
    lof_train,
    lof_test
)


# =============================================================================
# 23. ANOMALY DIRECTION
# =============================================================================

# sklearn IsolationForest score_samples:
# lower = more abnormal.
#
# sklearn LOF score_samples:
# lower = more abnormal.
#
# Convert them so:
# HIGHER = MORE ANOMALOUS

if_train_anom = -if_train_n
if_val_anom = -if_val_n
if_test_anom = -if_test_n

lof_train_anom = -lof_train_n
lof_val_anom = -lof_val_n
lof_test_anom = -lof_test_n


# =============================================================================
# 24. ENSEMBLE
# =============================================================================

print("\n[17] BUILDING ENSEMBLE")
print("-" * 90)

W_IF = 0.60
W_LOF = 0.40

ensemble_train = (
    W_IF * if_train_anom +
    W_LOF * lof_train_anom
)

ensemble_val = (
    W_IF * if_val_anom +
    W_LOF * lof_val_anom
)

ensemble_test = (
    W_IF * if_test_anom +
    W_LOF * lof_test_anom
)

print(
    f"Weights: IF={W_IF} | LOF={W_LOF}"
)


# =============================================================================
# 25. CREATE FULL SCORE VECTOR
# =============================================================================

full_if = np.empty(
    len(data),
    dtype=np.float32
)

full_lof = np.empty(
    len(data),
    dtype=np.float32
)

full_ensemble = np.empty(
    len(data),
    dtype=np.float32
)

full_if[train_idx] = if_train_anom
full_if[val_idx] = if_val_anom
full_if[test_idx] = if_test_anom

full_lof[train_idx] = lof_train_anom
full_lof[val_idx] = lof_val_anom
full_lof[test_idx] = lof_test_anom

full_ensemble[train_idx] = ensemble_train
full_ensemble[val_idx] = ensemble_val
full_ensemble[test_idx] = ensemble_test


# =============================================================================
# 26. RESULTS DATAFRAME
# =============================================================================

print("\n[18] BUILDING RESULTS")
print("-" * 90)

results = pd.DataFrame({
    "txId": data[id_col].astype(np.int64),
    "Time step": data[time_col].astype(np.int16),
    "IF_anomaly": full_if,
    "LOF_anomaly": full_lof,
    "ensemble_anomaly": full_ensemble
})

results["split"] = "train"

results.loc[
    val_idx,
    "split"
] = "validation"

results.loc[
    test_idx,
    "split"
] = "test"


# =============================================================================
# 27. TOP ANOMALIES
# =============================================================================

print("\n[19] TOP ANOMALOUS TRANSACTIONS")
print("-" * 90)

top_n = min(
    100,
    len(results)
)

top_anomalies = results.nlargest(
    top_n,
    "ensemble_anomaly"
)

print(
    top_anomalies[
        [
            "txId",
            "Time step",
            "split",
            "IF_anomaly",
            "LOF_anomaly",
            "ensemble_anomaly"
        ]
    ].head(30).to_string(index=False)
)


# =============================================================================
# 28. TEMPORAL ANALYSIS
# =============================================================================

print("\n[20] TEMPORAL ANOMALY ANALYSIS")
print("-" * 90)

temporal = (
    results
    .groupby("Time step")
    .agg(
        Transactions=("txId", "size"),
        Mean_Anomaly=("ensemble_anomaly", "mean"),
        Median_Anomaly=("ensemble_anomaly", "median"),
        Max_Anomaly=("ensemble_anomaly", "max")
    )
    .sort_values(
        "Mean_Anomaly",
        ascending=False
    )
)

print(
    temporal.head(15).to_string()
)


# =============================================================================
# 29. GRAPH FEATURE SUMMARY
# =============================================================================

print("\n[21] GRAPH FEATURE SUMMARY")
print("-" * 90)

graph_summary = pd.DataFrame({
    "Feature": GRAPH_FEATURE_NAMES,
    "Mean": graph_features.mean(axis=0),
    "Std": graph_features.std(axis=0),
    "Min": graph_features.min(axis=0),
    "Median": np.median(graph_features, axis=0),
    "Max": graph_features.max(axis=0)
})

print(
    graph_summary.to_string(index=False)
)


# =============================================================================
# 30. SAVE GRAPH FEATURES
# =============================================================================

GRAPH_OUTPUT = (
    "ellipticpp_graph_features.csv"
)

graph_output = pd.DataFrame(
    graph_features,
    columns=GRAPH_FEATURE_NAMES
)

graph_output.insert(
    0,
    "txId",
    data[id_col].astype(np.int64).to_numpy()
)

graph_output.insert(
    1,
    "Time step",
    data[time_col].astype(np.int16).to_numpy()
)

graph_output.to_csv(
    GRAPH_OUTPUT,
    index=False
)


# =============================================================================
# 31. SAVE FINAL ANOMALY RESULTS
# =============================================================================

RESULT_OUTPUT = (
    "ellipticpp_graph_behavioral_anomaly_results.csv"
)

results.to_csv(
    RESULT_OUTPUT,
    index=False
)


# =============================================================================
# 32. SAVE FEATURE LIST
# =============================================================================

FEATURE_OUTPUT = (
    "ellipticpp_graph_behavioral_features.txt"
)

with open(
    FEATURE_OUTPUT,
    "w",
    encoding="utf-8"
) as f:

    f.write(
        "BIT-TRACE FINAL UNSUPERVISED FEATURES\n"
    )
    f.write("=" * 60 + "\n\n")

    f.write(
        "BEHAVIORAL FEATURES\n"
    )

    for x in BEHAVIOR_FEATURES:
        f.write(
            x + "\n"
        )

    f.write(
        "\nGRAPH FEATURES\n"
    )

    for x in GRAPH_FEATURE_NAMES:
        f.write(
            x + "\n"
        )


# =============================================================================
# 33. MEMORY CLEANUP
# =============================================================================

del X_train
del X_val
del X_test
del X_scaled
del X_lof_train

gc.collect()


# =============================================================================
# FINAL STATUS
# =============================================================================

print("\n")
print("=" * 90)
print("BIT-TRACE GRAPH + BEHAVIORAL PIPELINE COMPLETE")
print("=" * 90)

print(
    f"Behavioral features : {len(BEHAVIOR_FEATURES)}"
)

print(
    f"Graph features      : {len(GRAPH_FEATURE_NAMES)}"
)

print(
    f"Total features      : {len(ALL_FEATURE_NAMES)}"
)

print(
    f"Transactions        : {len(data):,}"
)

print(
    f"Graph edges         : {len(edges):,}"
)

print(
    f"Train               : {len(train_idx):,}"
)

print(
    f"Validation          : {len(val_idx):,}"
)

print(
    f"Test                : {len(test_idx):,}"
)

print("\nSaved files:")

print(
    "✓",
    GRAPH_OUTPUT
)

print(
    "✓",
    RESULT_OUTPUT
)

print(
    "✓",
    FEATURE_OUTPUT
)

print("\n")
print("NEXT PHASE:")
print(
    "Use the graph + behavioral anomaly score as the core"
)
print(
    "investigation signal, then add explainability / evidence"
)
print(
    "generation for the BIT-TRACE investigator dashboard."
)

print("=" * 90)
```

```python
# =============================================================================
# BIT-TRACE — EXPLAINABILITY + EVIDENCE GENERATION
# Uses already-generated Graph + Behavioral anomaly results
# Does NOT reload the 662 MB raw feature file
# =============================================================================

import os
import json
import numpy as np
import pandas as pd

from sklearn.preprocessing import RobustScaler

print("=" * 90)
print("BIT-TRACE — EXPLAINABILITY + EVIDENCE GENERATION")
print("=" * 90)

# =============================================================================
# 1. LOCATE SAVED FILES
# =============================================================================

BASE_DIR = r"C:\Users\YASHIKA\Desktop\bit-trace"

RESULT_FILE = os.path.join(
    BASE_DIR,
    "ellipticpp_graph_behavioral_anomaly_results.csv"
)

GRAPH_FILE = os.path.join(
    BASE_DIR,
    "ellipticpp_graph_features.csv"
)

# Also check src / current working directory
SEARCH_ROOTS = [
    BASE_DIR,
    os.path.join(BASE_DIR, "src"),
    os.path.join(BASE_DIR, "data"),
]

def find_file(filename):
    candidates = []

    for root in SEARCH_ROOTS:
        if not os.path.exists(root):
            continue

        for current_root, dirs, files in os.walk(root):
            if filename in files:
                candidates.append(
                    os.path.join(current_root, filename)
                )

    if candidates:
        # Prefer shortest path
        candidates.sort(key=len)
        return candidates[0]

    return None


if not os.path.exists(RESULT_FILE):
    RESULT_FILE = find_file(
        "ellipticpp_graph_behavioral_anomaly_results.csv"
    )

if not os.path.exists(GRAPH_FILE):
    GRAPH_FILE = find_file(
        "ellipticpp_graph_features.csv"
    )

print("\n[1] FILE CHECK")
print("-" * 90)

print("RESULT FILE:", RESULT_FILE)
print("GRAPH FILE :", GRAPH_FILE)

if RESULT_FILE is None or not os.path.exists(RESULT_FILE):
    raise FileNotFoundError(
        "ellipticpp_graph_behavioral_anomaly_results.csv not found."
    )

if GRAPH_FILE is None or not os.path.exists(GRAPH_FILE):
    raise FileNotFoundError(
        "ellipticpp_graph_features.csv not found."
    )

print("✓ Required files found")


# =============================================================================
# 2. LOAD ANOMALY RESULTS
# =============================================================================

print("\n[2] LOADING ANOMALY RESULTS")
print("-" * 90)

results = pd.read_csv(RESULT_FILE)

print("Shape:", results.shape)
print("Columns:")
print(results.columns.tolist())

# Normalize column names
results.columns = [
    str(c).strip()
    for c in results.columns
]

# Required columns
required_basic = [
    "txId",
    "Time step",
    "split",
]

missing = [
    c for c in required_basic
    if c not in results.columns
]

if missing:
    raise ValueError(
        f"Missing required columns in results file: {missing}"
    )

print("✓ Results loaded")


# =============================================================================
# 3. LOAD GRAPH FEATURES
# =============================================================================

print("\n[3] LOADING GRAPH FEATURES")
print("-" * 90)

graph = pd.read_csv(GRAPH_FILE)

print("Shape:", graph.shape)
print("Columns:")
print(graph.columns.tolist())

if "txId" not in graph.columns:
    raise ValueError(
        "txId column not found in graph feature file."
    )

print("✓ Graph features loaded")


# =============================================================================
# 4. MERGE GRAPH EVIDENCE
# =============================================================================

print("\n[4] MERGING GRAPH EVIDENCE")
print("-" * 90)

# Avoid duplicate columns
graph_cols = [
    c for c in graph.columns
    if c == "txId" or c.startswith("graph_")
]

graph_small = graph[graph_cols].copy()

# Ensure one row per txId
graph_small = graph_small.drop_duplicates(
    subset=["txId"]
)

before = len(results)

evidence = results.merge(
    graph_small,
    on="txId",
    how="left"
)

print("Rows before merge:", before)
print("Rows after merge :", len(evidence))

graph_feature_cols = [
    c for c in graph_small.columns
    if c.startswith("graph_")
]

for c in graph_feature_cols:
    print(
        f"{c}: missing = "
        f"{evidence[c].isna().sum()}"
    )

print("✓ Graph evidence attached")


# =============================================================================
# 5. IDENTIFY MODEL SCORES
# =============================================================================

print("\n[5] IDENTIFYING MODEL SCORES")
print("-" * 90)

print(
    "Available score columns:",
    [
        c for c in evidence.columns
        if "score" in c.lower()
        or "anomaly" in c.lower()
    ]
)

if "IF_anomaly" not in evidence.columns:
    raise ValueError(
        "IF_anomaly column not found."
    )

if "LOF_anomaly" not in evidence.columns:
    raise ValueError(
        "LOF_anomaly column not found."
    )

print("✓ IF and LOF scores found")


# =============================================================================
# 6. ROBUST SCORE NORMALIZATION
# =============================================================================
# IMPORTANT:
# LOF and Isolation Forest are on very different raw scales.
# We therefore normalize them before combining.
#
# We use percentile/rank-based normalization because it is robust
# to extreme LOF values.
# =============================================================================

print("\n[6] NORMALIZING IF + LOF SCORES")
print("-" * 90)

def percentile_score(series):
    """
    Converts a score into approximately [0,1]
    using percentile rank.
    Higher value = more anomalous.
    """
    s = pd.to_numeric(
        series,
        errors="coerce"
    )

    return s.rank(
        pct=True,
        method="average"
    )


# Determine anomaly direction.
#
# In the current saved output:
# IF_anomaly has larger values for stronger anomalies.
# LOF_anomaly has larger values for stronger anomalies.
#
# Therefore higher = more anomalous.

evidence["IF_normalized"] = percentile_score(
    evidence["IF_anomaly"]
)

evidence["LOF_normalized"] = percentile_score(
    evidence["LOF_anomaly"]
)

# Robust ensemble
evidence["ensemble_score"] = (
    0.60 * evidence["IF_normalized"]
    +
    0.40 * evidence["LOF_normalized"]
)

print(
    "IF normalized range:",
    round(evidence["IF_normalized"].min(), 4),
    "to",
    round(evidence["IF_normalized"].max(), 4)
)

print(
    "LOF normalized range:",
    round(evidence["LOF_normalized"].min(), 4),
    "to",
    round(evidence["LOF_normalized"].max(), 4)
)

print(
    "Ensemble range:",
    round(evidence["ensemble_score"].min(), 4),
    "to",
    round(evidence["ensemble_score"].max(), 4)
)

print("✓ Scores normalized")


# =============================================================================
# 7. ANOMALY PERCENTILE
# =============================================================================

print("\n[7] CALCULATING ANOMALY PERCENTILE")
print("-" * 90)

evidence["anomaly_percentile"] = (
    evidence["ensemble_score"]
    .rank(pct=True)
    * 100
)

evidence["anomaly_percentile"] = (
    evidence["anomaly_percentile"]
    .round(2)
)

# Higher percentile = more anomalous

print(
    evidence["anomaly_percentile"].describe()
)

print("✓ Percentile calculated")


# =============================================================================
# 8. INVESTIGATION PRIORITY
# =============================================================================

print("\n[8] ASSIGNING INVESTIGATION PRIORITY")
print("-" * 90)

def priority(p):
    if p >= 99.5:
        return "CRITICAL"
    elif p >= 99:
        return "HIGH"
    elif p >= 95:
        return "MEDIUM"
    else:
        return "LOW"

evidence["investigation_priority"] = (
    evidence["anomaly_percentile"]
    .apply(priority)
)

print(
    evidence["investigation_priority"]
    .value_counts()
)


# =============================================================================
# 9. GRAPH BEHAVIOR EVIDENCE
# =============================================================================

print("\n[9] GENERATING GRAPH EVIDENCE")
print("-" * 90)

graph_evidence_cols = [
    "graph_in_degree",
    "graph_out_degree",
    "graph_total_degree",
    "graph_degree_imbalance",
    "graph_in_out_ratio",
]

available_graph_evidence = [
    c for c in graph_evidence_cols
    if c in evidence.columns
]

print(
    "Graph evidence columns:",
    available_graph_evidence
)

# Calculate percentile for each graph feature
# so values are comparable.

for col in available_graph_evidence:

    numeric = pd.to_numeric(
        evidence[col],
        errors="coerce"
    )

    evidence[col + "_percentile"] = (
        numeric.rank(
            pct=True,
            method="average"
        ) * 100
    )

    evidence[col + "_percentile"] = (
        evidence[col + "_percentile"]
        .round(2)
    )

print("✓ Graph percentiles generated")


# =============================================================================
# 10. BEHAVIORAL EVIDENCE
# =============================================================================
# We use the already selected behavioral features.
#
# Since raw behavioral features are standardized in the pipeline,
# absolute magnitude can be used as a deviation signal.
# =============================================================================

print("\n[10] GENERATING BEHAVIORAL EVIDENCE")
print("-" * 90)

behavioral_features = [
    "Local_feature_22",
    "Local_feature_25",
    "Local_feature_47",
    "Local_feature_52",
    "Local_feature_53",
    "Local_feature_54",
    "Local_feature_59",
    "Local_feature_77",
    "Local_feature_89",
    "Local_feature_90",
    "Aggregate_feature_14",
    "Aggregate_feature_21",
    "Aggregate_feature_32",
]

available_behavioral = [
    c for c in behavioral_features
    if c in evidence.columns
]

print(
    "Behavioral features available directly:",
    available_behavioral
)

# If behavioral values aren't present in the saved result file,
# don't fabricate them.
#
# Instead use the model outputs already available.

if len(available_behavioral) == 0:

    print(
        "NOTE: Raw behavioral feature columns are not present "
        "in the saved anomaly-result file."
    )

    print(
        "Behavioral evidence will therefore use the "
        "existing model-level anomaly signal."
    )

else:

    for col in available_behavioral:

        numeric = pd.to_numeric(
            evidence[col],
            errors="coerce"
        )

        evidence[
            col + "_abs_deviation"
        ] = numeric.abs()

    print("✓ Behavioral deviations generated")


# =============================================================================
# 11. GRAPH ANOMALY SCORE
# =============================================================================

print("\n[11] GRAPH ANOMALY SCORE")
print("-" * 90)

graph_score_components = []

for col in [
    "graph_total_degree_percentile",
    "graph_degree_imbalance_percentile",
    "graph_in_out_ratio_percentile",
]:

    if col in evidence.columns:
        graph_score_components.append(
            evidence[col] / 100.0
        )

if graph_score_components:

    evidence["graph_anomaly_score"] = (
        pd.concat(
            graph_score_components,
            axis=1
        ).mean(axis=1)
    )

else:

    evidence["graph_anomaly_score"] = np.nan

print(
    "Graph anomaly score available:",
    evidence["graph_anomaly_score"]
    .notna()
    .sum()
)

print("✓ Graph score generated")


# =============================================================================
# 12. FINAL INVESTIGATION SCORE
# =============================================================================
#
# This is NOT a trained classifier.
# It is an investigator prioritization signal.
#
# 70% model anomaly
# 30% graph abnormality
#
# If graph score unavailable, use ensemble score.
# =============================================================================

print("\n[12] BUILDING INVESTIGATION SCORE")
print("-" * 90)

if evidence["graph_anomaly_score"].notna().any():

    evidence["investigation_score"] = (
        0.70 * evidence["ensemble_score"]
        +
        0.30 * evidence["graph_anomaly_score"].fillna(
            evidence["ensemble_score"]
        )
    )

else:

    evidence["investigation_score"] = (
        evidence["ensemble_score"]
    )

evidence["investigation_score"] = (
    evidence["investigation_score"]
    .clip(0, 1)
)

print(
    evidence["investigation_score"].describe()
)

print("✓ Investigation score generated")


# =============================================================================
# 13. EVIDENCE REASON GENERATOR
# =============================================================================

print("\n[13] GENERATING HUMAN-READABLE EVIDENCE")
print("-" * 90)

def generate_reason(row):

    reasons = []

    # Model anomaly
    p = row.get(
        "anomaly_percentile",
        np.nan
    )

    if pd.notna(p):

        if p >= 99.5:
            reasons.append(
                "Extremely high ensemble anomaly score"
            )

        elif p >= 99:
            reasons.append(
                "Very high ensemble anomaly score"
            )

        elif p >= 95:
            reasons.append(
                "Elevated ensemble anomaly score"
            )

    # Graph structure
    total_degree = row.get(
        "graph_total_degree",
        np.nan
    )

    if pd.notna(total_degree):

        if total_degree >= 50:
            reasons.append(
                "Very high transaction connectivity"
            )

        elif total_degree >= 20:
            reasons.append(
                "Elevated transaction connectivity"
            )

    imbalance = row.get(
        "graph_degree_imbalance",
        np.nan
    )

    if pd.notna(imbalance):

        if imbalance >= 20:
            reasons.append(
                "Strong inbound/outbound degree imbalance"
            )

    ratio = row.get(
        "graph_in_out_ratio",
        np.nan
    )

    if pd.notna(ratio):

        if ratio >= 10 or ratio <= 0.1:
            reasons.append(
                "Highly asymmetric inbound/outbound activity"
            )

    if not reasons:
        reasons.append(
            "Behavioral/graph anomaly signal requires further investigation"
        )

    return "; ".join(reasons)


evidence["evidence_summary"] = (
    evidence.apply(
        generate_reason,
        axis=1
    )
)

print("✓ Evidence descriptions generated")


# =============================================================================
# 14. TOP INVESTIGATION CASES
# =============================================================================

print("\n[14] TOP INVESTIGATION CASES")
print("=" * 90)

top_cases = (
    evidence
    .sort_values(
        "investigation_score",
        ascending=False
    )
    .head(50)
)

display_cols = [
    "txId",
    "Time step",
    "split",
    "ensemble_score",
    "graph_anomaly_score",
    "investigation_score",
    "anomaly_percentile",
    "investigation_priority",
]

display_cols = [
    c for c in display_cols
    if c in top_cases.columns
]

print(
    top_cases[display_cols].to_string(
        index=False
    )
)


# =============================================================================
# 15. TOP CASE DETAILS
# =============================================================================

print("\n[15] TOP 10 CASE EVIDENCE")
print("=" * 90)

for _, row in top_cases.head(10).iterrows():

    print("\n" + "-" * 80)

    print(
        "Transaction ID:",
        row["txId"]
    )

    print(
        "Time step:",
        row["Time step"]
    )

    print(
        "Split:",
        row["split"]
    )

    print(
        "Investigation priority:",
        row["investigation_priority"]
    )

    print(
        "Investigation score:",
        round(
            row["investigation_score"],
            4
        )
    )

    print(
        "Anomaly percentile:",
        row["anomaly_percentile"],
        "%"
    )

    print(
        "Evidence:",
        row["evidence_summary"]
    )

    if "graph_total_degree" in row:
        print(
            "Total graph degree:",
            row["graph_total_degree"]
        )

    if "graph_in_degree" in row:
        print(
            "In degree:",
            row["graph_in_degree"]
        )

    if "graph_out_degree" in row:
        print(
            "Out degree:",
            row["graph_out_degree"]
        )


# =============================================================================
# 16. SAVE INVESTIGATOR CSV
# =============================================================================

print("\n[16] SAVING INVESTIGATOR DATA")
print("-" * 90)

OUTPUT_DIR = os.path.join(
    BASE_DIR,
    "data",
    "processed"
)

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)

CSV_OUTPUT = os.path.join(
    OUTPUT_DIR,
    "bittrace_investigator_evidence.csv"
)

evidence.to_csv(
    CSV_OUTPUT,
    index=False
)

print(
    "✓ Saved:",
    CSV_OUTPUT
)


# =============================================================================
# 17. SAVE TOP CASES
# =============================================================================

TOP_OUTPUT = os.path.join(
    OUTPUT_DIR,
    "bittrace_top_investigation_cases.csv"
)

top_cases.to_csv(
    TOP_OUTPUT,
    index=False
)

print(
    "✓ Saved:",
    TOP_OUTPUT
)


# =============================================================================
# 18. JSON EVIDENCE FOR DASHBOARD/API
# =============================================================================

print("\n[18] BUILDING DASHBOARD JSON")
print("-" * 90)

json_cases = []

for _, row in top_cases.head(100).iterrows():

    case = {
        "txId": int(row["txId"])
        if pd.notna(row["txId"])
        else None,

        "time_step": int(row["Time step"])
        if pd.notna(row["Time step"])
        else None,

        "split": str(row["split"]),

        "ensemble_score": round(
            float(row["ensemble_score"]),
            6
        ),

        "investigation_score": round(
            float(row["investigation_score"]),
            6
        ),

        "anomaly_percentile": round(
            float(row["anomaly_percentile"]),
            2
        ),

        "priority": str(
            row["investigation_priority"]
        ),

        "evidence": str(
            row["evidence_summary"]
        ),
    }

    # Graph evidence
    for col in [
        "graph_in_degree",
        "graph_out_degree",
        "graph_total_degree",
        "graph_degree_imbalance",
        "graph_in_out_ratio",
    ]:

        if col in row and pd.notna(row[col]):

            case[col] = float(
                row[col]
            )

    json_cases.append(case)


JSON_OUTPUT = os.path.join(
    OUTPUT_DIR,
    "bittrace_investigation_cases.json"
)

with open(
    JSON_OUTPUT,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        json_cases,
        f,
        indent=2
    )

print(
    "✓ Saved:",
    JSON_OUTPUT
)


# =============================================================================
# 19. SUMMARY
# =============================================================================

print("\n")
print("=" * 90)
print("BIT-TRACE — EXPLAINABILITY + EVIDENCE COMPLETE")
print("=" * 90)

print(
    "Transactions analysed:",
    len(evidence)
)

print(
    "Graph features:",
    len(graph_feature_cols)
)

print(
    "Top investigation cases:",
    len(top_cases)
)

print(
    "\nPriority distribution:"
)

print(
    evidence[
        "investigation_priority"
    ].value_counts()
)

print("\nGenerated files:")
print(
    "✓ bittrace_investigator_evidence.csv"
)
print(
    "✓ bittrace_top_investigation_cases.csv"
)
print(
    "✓ bittrace_investigation_cases.json"
)

print("\nNEXT PHASE:")
print(
    "Use these evidence records to build the "
    "BIT-TRACE Investigator Dashboard."
)

print(
    "\nIMPORTANT:"
)
print(
    "The anomaly score is an investigation-prioritization "
    "signal, not a proof that a transaction is illicit."
)

print("=" * 90)
```

```python
# =============================================================================
# BIT-TRACE — TRANSACTION-ONLY FINAL UNSUPERVISED ANOMALY PIPELINE
# Elliptic++ Transactions Dataset
#
# IMPORTANT:
# - Uses ONLY txs_features.csv + txs_classes.csv
# - Does NOT use P2P/network observations
# - Does NOT use txs_edgelist.csv
# - Labels are used ONLY for post-hoc evaluation
# =============================================================================

import os
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd

from sklearn.ensemble import IsolationForest
from sklearn.neighbors import LocalOutlierFactor
from sklearn.preprocessing import RobustScaler
from sklearn.metrics import roc_auc_score, average_precision_score


# =============================================================================
# 0. CONFIGURATION
# =============================================================================

FEATURE_PATH = r"C:\Users\YASHIKA\Desktop\bit-trace\EllipticPlusPlus\Transactions Dataset\txs_features.csv"
CLASS_PATH   = r"C:\Users\YASHIKA\Desktop\bit-trace\EllipticPlusPlus\Transactions Dataset\txs_classes.csv"

SELECTED_FEATURES = [
    "Local_feature_53",
    "Local_feature_54",
    "Local_feature_52",
    "Local_feature_59",
    "Aggregate_feature_21",
    "Local_feature_25",
    "Local_feature_77",
    "Local_feature_89",
    "Local_feature_90",
    "Aggregate_feature_32",
    "Aggregate_feature_14",
    "Local_feature_22",
    "Local_feature_47"
]

TRAIN_END = 34
VAL_START = 35
VAL_END = 41
TEST_START = 42

N_ESTIMATORS = 300
CONTAMINATION = "auto"
RANDOM_STATE = 42

# Ensemble weights
IF_WEIGHT = 0.60
LOF_WEIGHT = 0.40

OUTPUT_DIR = os.path.join(
    os.path.dirname(os.path.dirname(FEATURE_PATH)),
    "BITTRACE_TRANSACTION_RESULTS"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)


# =============================================================================
# HELPER FUNCTIONS
# =============================================================================

def section(title, width=90):
    print("\n" + "=" * width)
    print(title)
    print("=" * width)


def minmax_normalize(x):
    """
    Converts any score array to [0,1].
    Higher = more anomalous.
    """
    x = np.asarray(x, dtype=float)

    lo = np.nanmin(x)
    hi = np.nanmax(x)

    if hi - lo < 1e-12:
        return np.zeros_like(x)

    return (x - lo) / (hi - lo)


def percentile_score(x):
    """
    Percentile/rank based score.
    Higher = more anomalous.
    """
    s = pd.Series(np.asarray(x))
    return s.rank(method="average", pct=True).to_numpy()


# =============================================================================
# 1. FILE CHECK
# =============================================================================

section("BIT-TRACE — TRANSACTION-ONLY UNSUPERVISED PIPELINE")

print("\n[1] FILE CHECK")
print("-" * 90)

if not os.path.exists(FEATURE_PATH):
    raise FileNotFoundError(f"Feature file not found:\n{FEATURE_PATH}")

if not os.path.exists(CLASS_PATH):
    raise FileNotFoundError(f"Class file not found:\n{CLASS_PATH}")

print("FEATURE:", FEATURE_PATH)
print("CLASS  :", CLASS_PATH)
print("✓ Required transaction files found")
print("✗ P2P / edgelist is NOT used")


# =============================================================================
# 2. LOAD CLASS DATA
# =============================================================================

print("\n[2] LOADING CLASS DATA")
print("-" * 90)

classes = pd.read_csv(CLASS_PATH)

classes.columns = [
    str(c).strip().replace(" ", "_")
    for c in classes.columns
]

# Handle possible naming variations
if "txId" not in classes.columns:
    candidates = [c for c in classes.columns if c.lower() == "txid"]
    if candidates:
        classes.rename(columns={candidates[0]: "txId"}, inplace=True)

if "class" not in classes.columns:
    candidates = [c for c in classes.columns if c.lower() == "class"]
    if candidates:
        classes.rename(columns={candidates[0]: "class"}, inplace=True)

classes["txId"] = pd.to_numeric(classes["txId"], errors="coerce")
classes["class"] = pd.to_numeric(classes["class"], errors="coerce")

classes = classes.dropna(subset=["txId", "class"]).copy()

classes["txId"] = classes["txId"].astype(np.int64)
classes["class"] = classes["class"].astype(int)

print("Shape:", classes.shape)
print("\nClass distribution:")
print(classes["class"].value_counts().sort_index())


# =============================================================================
# 3. READ FEATURE HEADER
# =============================================================================

print("\n[3] READING FEATURE HEADER")
print("-" * 90)

header = pd.read_csv(FEATURE_PATH, nrows=0)

feature_columns = list(header.columns)

print("Total columns:", len(feature_columns))

# Detect ID
id_candidates = ["txId", "txid", "TxId", "transaction_id"]

id_col = None
for c in id_candidates:
    if c in feature_columns:
        id_col = c
        break

if id_col is None:
    raise ValueError("Could not detect txId column.")

# Detect time
time_candidates = [
    "Time step",
    "Time_step",
    "time_step",
    "TimeStep",
    "time"
]

time_col = None
for c in time_candidates:
    if c in feature_columns:
        time_col = c
        break

if time_col is None:
    raise ValueError("Could not detect Time step column.")

print("ID column  :", id_col)
print("Time column:", time_col)


# =============================================================================
# 4. VERIFY FEATURES
# =============================================================================

print("\n[4] VERIFYING SELECTED FEATURES")
print("-" * 90)

missing_features = [
    f for f in SELECTED_FEATURES
    if f not in feature_columns
]

if missing_features:
    raise ValueError(
        "Missing selected features:\n" +
        "\n".join(missing_features)
    )

print("Selected features:", len(SELECTED_FEATURES))

for i, f in enumerate(SELECTED_FEATURES, 1):
    print(f"{i:2d}. {f}")

print("✓ All selected features found")


# =============================================================================
# 5. LOAD ONLY REQUIRED TRANSACTION COLUMNS
# =============================================================================

print("\n[5] LOADING TRANSACTION FEATURES")
print("-" * 90)

usecols = [id_col, time_col] + SELECTED_FEATURES

data = pd.read_csv(
    FEATURE_PATH,
    usecols=usecols
)

print("Loaded shape:", data.shape)


# =============================================================================
# 6. NORMALIZE COLUMN NAMES
# =============================================================================

if id_col != "txId":
    data.rename(columns={id_col: "txId"}, inplace=True)

if time_col != "Time step":
    data.rename(columns={time_col: "Time step"}, inplace=True)

data["txId"] = pd.to_numeric(data["txId"], errors="coerce")
data["Time step"] = pd.to_numeric(data["Time step"], errors="coerce")

for f in SELECTED_FEATURES:
    data[f] = pd.to_numeric(data[f], errors="coerce")


# =============================================================================
# 7. DATA QUALITY
# =============================================================================

print("\n[6] DATA QUALITY")
print("-" * 90)

print("Rows:", len(data))
print("Missing txId:", data["txId"].isna().sum())
print("Missing Time step:", data["Time step"].isna().sum())
print("Duplicate txId:", data["txId"].duplicated().sum())

feature_nan = data[SELECTED_FEATURES].isna().sum().sum()
feature_inf = np.isinf(
    data[SELECTED_FEATURES].to_numpy(dtype=float)
).sum()

print("Feature NaN:", feature_nan)
print("Feature Inf:", feature_inf)


# =============================================================================
# 8. MERGE LABELS
# =============================================================================

print("\n[7] MERGING LABELS")
print("-" * 90)

data = data.merge(
    classes[["txId", "class"]],
    on="txId",
    how="inner",
    validate="one_to_one"
)

print("Merged shape:", data.shape)
print("Missing class:", data["class"].isna().sum())

print("\nClass distribution:")
print(data["class"].value_counts().sort_index())


# =============================================================================
# 9. CLEAN FEATURES
# =============================================================================

print("\n[8] CLEANING FEATURES")
print("-" * 90)

data[SELECTED_FEATURES] = (
    data[SELECTED_FEATURES]
    .replace([np.inf, -np.inf], np.nan)
)

# Median imputation
for f in SELECTED_FEATURES:
    med = data[f].median()

    if pd.isna(med):
        med = 0.0

    data[f] = data[f].fillna(med)

print(
    "Remaining NaN:",
    data[SELECTED_FEATURES].isna().sum().sum()
)

print(
    "Remaining Inf:",
    np.isinf(
        data[SELECTED_FEATURES].to_numpy(dtype=float)
    ).sum()
)


# =============================================================================
# 10. SORT TEMPORALLY
# =============================================================================

data = data.sort_values(
    ["Time step", "txId"]
).reset_index(drop=True)

X_all = data[SELECTED_FEATURES].to_numpy(dtype=np.float32)
y_all = data["class"].to_numpy(dtype=np.int8)
time_steps = data["Time step"].to_numpy(dtype=np.int16)
tx_ids = data["txId"].to_numpy(dtype=np.int64)

print("\n[9] FINAL DATA")
print("-" * 90)

print("X:", X_all.shape)
print("y:", y_all.shape)
print("Time:", time_steps.shape)

print(
    "Time range:",
    time_steps.min(),
    "to",
    time_steps.max()
)


# =============================================================================
# 11. TEMPORAL SPLIT
# =============================================================================

train_idx = np.where(time_steps <= TRAIN_END)[0]

val_idx = np.where(
    (time_steps >= VAL_START) &
    (time_steps <= VAL_END)
)[0]

test_idx = np.where(
    time_steps >= TEST_START
)[0]

print("\n[10] TEMPORAL SPLIT")
print("-" * 90)

print(
    f"Train      : {len(train_idx):,} "
    f"(time 1–{TRAIN_END})"
)

print(
    f"Validation : {len(val_idx):,} "
    f"(time {VAL_START}–{VAL_END})"
)

print(
    f"Test       : {len(test_idx):,} "
    f"(time {TEST_START}–{time_steps.max()})"
)


# =============================================================================
# 12. ROBUST SCALING
# =============================================================================

print("\n[11] ROBUST SCALING")
print("-" * 90)

# Fit scaler ONLY on training data
scaler = RobustScaler()

X_train = scaler.fit_transform(
    X_all[train_idx]
).astype(np.float32)

X_val = scaler.transform(
    X_all[val_idx]
).astype(np.float32)

X_test = scaler.transform(
    X_all[test_idx]
).astype(np.float32)

print("Train:", X_train.shape)
print("Validation:", X_val.shape)
print("Test:", X_test.shape)

print("✓ Scaler fitted only on training data")


# =============================================================================
# 13. ISOLATION FOREST
# =============================================================================

print("\n[12] TRAINING ISOLATION FOREST")
print("-" * 90)

iforest = IsolationForest(
    n_estimators=N_ESTIMATORS,
    contamination=CONTAMINATION,
    random_state=RANDOM_STATE,
    n_jobs=-1,
    max_samples="auto"
)

# NO LABELS USED
iforest.fit(X_train)

print("✓ Isolation Forest trained")
print("Trees:", N_ESTIMATORS)


# =============================================================================
# 14. ISOLATION FOREST SCORES
# =============================================================================

print("\n[13] GENERATING IF SCORES")
print("-" * 90)

# sklearn decision_function:
# higher = more normal
# therefore multiply by -1
if_train_raw = -iforest.decision_function(X_train)
if_val_raw = -iforest.decision_function(X_val)
if_test_raw = -iforest.decision_function(X_test)

print("Train:", len(if_train_raw))
print("Validation:", len(if_val_raw))
print("Test:", len(if_test_raw))


# =============================================================================
# 15. LOF
# =============================================================================

print("\n[14] TRAINING LOF")
print("-" * 90)

# IMPORTANT:
# LOF with novelty=True can score unseen validation/test data.
#
# n_neighbors=35 gives local behavioral context without being excessively
# expensive on this dataset.

lof = LocalOutlierFactor(
    n_neighbors=35,
    contamination="auto",
    novelty=True,
    n_jobs=-1
)

lof.fit(X_train)

print("✓ LOF trained")


# =============================================================================
# 16. LOF SCORES
# =============================================================================

print("\n[15] GENERATING LOF SCORES")
print("-" * 90)

# sklearn LOF decision_function:
# higher = more normal
# multiply by -1
lof_train_raw = -lof.decision_function(X_train)
lof_val_raw = -lof.decision_function(X_val)
lof_test_raw = -lof.decision_function(X_test)

print("Train:", len(lof_train_raw))
print("Validation:", len(lof_val_raw))
print("Test:", len(lof_test_raw))


# =============================================================================
# 17. SCORE NORMALIZATION
# =============================================================================

print("\n[16] NORMALIZING SCORES")
print("-" * 90)

# Normalize each model separately.
#
# IMPORTANT:
# We use TRAIN distribution to establish the scale.
# Then apply same scale to validation/test.

def train_based_minmax(train_scores, scores):
    lo = np.percentile(train_scores, 1)
    hi = np.percentile(train_scores, 99)

    if hi - lo < 1e-12:
        return np.zeros_like(scores)

    result = (scores - lo) / (hi - lo)

    return np.clip(result, 0, 1)


if_train = train_based_minmax(
    if_train_raw,
    if_train_raw
)

if_val = train_based_minmax(
    if_train_raw,
    if_val_raw
)

if_test = train_based_minmax(
    if_train_raw,
    if_test_raw
)

lof_train = train_based_minmax(
    lof_train_raw,
    lof_train_raw
)

lof_val = train_based_minmax(
    lof_train_raw,
    lof_val_raw
)

lof_test = train_based_minmax(
    lof_train_raw,
    lof_test_raw
)

print("✓ Scores normalized to approximately [0,1]")


# =============================================================================
# 18. ENSEMBLE
# =============================================================================

print("\n[17] BUILDING ENSEMBLE")
print("-" * 90)

print(
    f"Isolation Forest weight: {IF_WEIGHT}"
)

print(
    f"LOF weight            : {LOF_WEIGHT}"
)

ensemble_train = (
    IF_WEIGHT * if_train +
    LOF_WEIGHT * lof_train
)

ensemble_val = (
    IF_WEIGHT * if_val +
    LOF_WEIGHT * lof_val
)

ensemble_test = (
    IF_WEIGHT * if_test +
    LOF_WEIGHT * lof_test
)

print("✓ Ensemble generated")


# =============================================================================
# 19. CREATE FULL RESULT DATAFRAME
# =============================================================================

print("\n[18] BUILDING RESULT TABLE")
print("-" * 90)

results = data[
    ["txId", "Time step", "class"]
].copy()

results["split"] = "train"

results.loc[val_idx, "split"] = "validation"
results.loc[test_idx, "split"] = "test"

results["IF_anomaly"] = np.nan
results["LOF_anomaly"] = np.nan
results["ensemble_anomaly"] = np.nan

results.loc[train_idx, "IF_anomaly"] = if_train
results.loc[val_idx, "IF_anomaly"] = if_val
results.loc[test_idx, "IF_anomaly"] = if_test

results.loc[train_idx, "LOF_anomaly"] = lof_train
results.loc[val_idx, "LOF_anomaly"] = lof_val
results.loc[test_idx, "LOF_anomaly"] = lof_test

results.loc[train_idx, "ensemble_anomaly"] = ensemble_train
results.loc[val_idx, "ensemble_anomaly"] = ensemble_val
results.loc[test_idx, "ensemble_anomaly"] = ensemble_test


# =============================================================================
# 20. SCORE DISTRIBUTION
# =============================================================================

print("\n[19] SCORE DISTRIBUTION")
print("-" * 90)

score_summary = pd.DataFrame({
    "Split": ["Train", "Validation", "Test"],
    "Samples": [
        len(ensemble_train),
        len(ensemble_val),
        len(ensemble_test)
    ],
    "Mean": [
        ensemble_train.mean(),
        ensemble_val.mean(),
        ensemble_test.mean()
    ],
    "Std": [
        ensemble_train.std(),
        ensemble_val.std(),
        ensemble_test.std()
    ],
    "Min": [
        ensemble_train.min(),
        ensemble_val.min(),
        ensemble_test.min()
    ],
    "Median": [
        np.median(ensemble_train),
        np.median(ensemble_val),
        np.median(ensemble_test)
    ],
    "Max": [
        ensemble_train.max(),
        ensemble_val.max(),
        ensemble_test.max()
    ]
})

print(
    score_summary.to_string(index=False)
)


# =============================================================================
# 21. POST-HOC CLASS-3 EVALUATION
# =============================================================================

print("\n[20] POST-HOC CLASS-3 EVALUATION")
print("-" * 90)

test_labels = y_all[test_idx]

# Higher anomaly score = more anomalous.
# Class 3 is evaluated as a reference class only.

class3_binary = (
    test_labels == 3
).astype(int)

try:
    test_roc_auc = roc_auc_score(
        class3_binary,
        ensemble_test
    )

    test_pr_auc = average_precision_score(
        class3_binary,
        ensemble_test
    )

    print(
        f"Test ROC-AUC : {test_roc_auc:.4f}"
    )

    print(
        f"Test PR-AUC  : {test_pr_auc:.4f}"
    )

except Exception as e:
    print("Evaluation error:", e)


# =============================================================================
# 22. CLASS-WISE SCORE ANALYSIS
# =============================================================================

print("\n[21] CLASS-WISE TEST SCORE ANALYSIS")
print("-" * 90)

test_eval = results.iloc[test_idx].copy()

class_summary = (
    test_eval
    .groupby("class")["ensemble_anomaly"]
    .agg(
        Samples="count",
        Mean="mean",
        Std="std",
        Min="min",
        Median="median",
        Max="max"
    )
)

print(class_summary)


# =============================================================================
# 23. TOP ANOMALIES — TEST SET ONLY
# =============================================================================

print("\n[22] TOP 50 MOST ANOMALOUS TEST TRANSACTIONS")
print("-" * 90)

top_test = (
    test_eval
    .sort_values(
        "ensemble_anomaly",
        ascending=False
    )
    .head(50)
)

print(
    top_test[
        [
            "txId",
            "Time step",
            "class",
            "IF_anomaly",
            "LOF_anomaly",
            "ensemble_anomaly"
        ]
    ].to_string(index=False)
)


# =============================================================================
# 24. TOP-K CLASS-3 ENRICHMENT
# =============================================================================

print("\n[23] TOP-K CLASS-3 ENRICHMENT")
print("-" * 90)

overall_class3_pct = (
    (test_labels == 3).mean() * 100
)

print(
    f"Test-set Class-3 baseline: "
    f"{overall_class3_pct:.2f}%"
)

for k in [50, 100, 500, 1000, 5000]:

    k = min(k, len(test_eval))

    top_k = (
        test_eval
        .sort_values(
            "ensemble_anomaly",
            ascending=False
        )
        .head(k)
    )

    class3_count = (
        top_k["class"] == 3
    ).sum()

    class3_pct = (
        class3_count / k * 100
    )

    enrichment = (
        class3_pct /
        overall_class3_pct
        if overall_class3_pct > 0
        else np.nan
    )

    print(
        f"Top {k:5d}: "
        f"Class-3 = {class3_count:5d} "
        f"({class3_pct:6.2f}%) | "
        f"Enrichment = {enrichment:.3f}x"
    )


# =============================================================================
# 25. TEMPORAL ANALYSIS
# =============================================================================

print("\n[24] TEMPORAL ANOMALY ANALYSIS — TEST")
print("-" * 90)

temporal = (
    test_eval
    .groupby("Time step")
    .agg(
        Transactions=("txId", "count"),
        Mean_Anomaly=("ensemble_anomaly", "mean"),
        Median_Anomaly=("ensemble_anomaly", "median"),
        Max_Anomaly=("ensemble_anomaly", "max"),
        Class3_Count=("class", lambda x: (x == 3).sum())
    )
)

temporal["Class3_%"] = (
    temporal["Class3_Count"] /
    temporal["Transactions"] * 100
)

temporal = temporal.sort_values(
    "Mean_Anomaly",
    ascending=False
)

print(
    temporal.head(15).to_string()
)


# =============================================================================
# 26. FEATURE STATISTICS
# =============================================================================

print("\n[25] FEATURE STATISTICS")
print("-" * 90)

feature_stats = pd.DataFrame({
    "Feature": SELECTED_FEATURES,
    "Train_Mean": X_all[train_idx].mean(axis=0),
    "Train_Std": X_all[train_idx].std(axis=0),
    "Train_Min": X_all[train_idx].min(axis=0),
    "Train_Median": np.median(
        X_all[train_idx],
        axis=0
    ),
    "Train_Max": X_all[train_idx].max(axis=0)
})

print(
    feature_stats.to_string(index=False)
)


# =============================================================================
# 27. TRANSACTION-LEVEL EVIDENCE
# =============================================================================

print("\n[26] GENERATING TRANSACTION-LEVEL EVIDENCE")
print("-" * 90)

# -------------------------------------------------------------------------
# Evidence idea:
#
# An anomalous transaction can be explained by identifying which selected
# behavioral features are most unusual relative to TRAIN behavior.
#
# We calculate robust deviation:
#
#   |x - train_median| / train_MAD
#
# Larger value = feature is farther from normal training behavior.
#
# This is NOT a supervised explanation.
# It does NOT use class labels.
# -------------------------------------------------------------------------

train_raw = X_all[train_idx].astype(np.float64)

train_median = np.median(
    train_raw,
    axis=0
)

q25 = np.percentile(
    train_raw,
    25,
    axis=0
)

q75 = np.percentile(
    train_raw,
    75,
    axis=0
)

iqr = q75 - q25

iqr[iqr < 1e-9] = 1.0


def feature_evidence_for_row(row):

    deviation = (
        np.abs(row - train_median) /
        iqr
    )

    order = np.argsort(
        deviation
    )[::-1]

    return [
        (
            SELECTED_FEATURES[i],
            float(deviation[i])
        )
        for i in order[:5]
    ]


# Only generate detailed evidence for TOP 500 test anomalies
# to keep runtime/output manageable.

top500 = (
    test_eval
    .sort_values(
        "ensemble_anomaly",
        ascending=False
    )
    .head(500)
)

evidence_records = []

for idx in top500.index:

    row_position = idx

    feature_values = X_all[
        row_position
    ].astype(np.float64)

    deviations = (
        np.abs(
            feature_values -
            train_median
        ) / iqr
    )

    order = np.argsort(
        deviations
    )[::-1]

    top_features = [
        SELECTED_FEATURES[i]
        for i in order[:5]
    ]

    top_deviations = [
        float(deviations[i])
        for i in order[:5]
    ]

    evidence_records.append({
        "txId":
            int(results.loc[idx, "txId"]),

        "Time step":
            int(results.loc[idx, "Time step"]),

        "class":
            int(results.loc[idx, "class"]),

        "ensemble_anomaly":
            float(results.loc[idx, "ensemble_anomaly"]),

        "evidence_feature_1":
            top_features[0],

        "evidence_strength_1":
            top_deviations[0],

        "evidence_feature_2":
            top_features[1],

        "evidence_strength_2":
            top_deviations[1],

        "evidence_feature_3":
            top_features[2],

        "evidence_strength_3":
            top_deviations[2],

        "evidence_feature_4":
            top_features[3],

        "evidence_strength_4":
            top_deviations[3],

        "evidence_feature_5":
            top_features[4],

        "evidence_strength_5":
            top_deviations[4]
    })


evidence_df = pd.DataFrame(
    evidence_records
)

print(
    "Evidence generated for:",
    len(evidence_df),
    "top test transactions"
)


# =============================================================================
# 28. COMBINE TOP ANOMALY + EVIDENCE
# =============================================================================

print("\n[27] BUILDING INVESTIGATION TABLE")
print("-" * 90)

investigation_df = top500.merge(
    evidence_df[
        [
            "txId",
            "evidence_feature_1",
            "evidence_strength_1",
            "evidence_feature_2",
            "evidence_strength_2",
            "evidence_feature_3",
            "evidence_strength_3",
            "evidence_feature_4",
            "evidence_strength_4",
            "evidence_feature_5",
            "evidence_strength_5"
        ]
    ],
    on="txId",
    how="left"
)

print(
    investigation_df.head(10).to_string(
        index=False
    )
)


# =============================================================================
# 29. SAVE MAIN RESULTS
# =============================================================================

print("\n[28] SAVING RESULTS")
print("-" * 90)

results_path = os.path.join(
    OUTPUT_DIR,
    "ellipticpp_transaction_anomaly_results.csv"
)

top_path = os.path.join(
    OUTPUT_DIR,
    "ellipticpp_top_test_anomalies.csv"
)

evidence_path = os.path.join(
    OUTPUT_DIR,
    "ellipticpp_transaction_evidence.csv"
)

temporal_path = os.path.join(
    OUTPUT_DIR,
    "ellipticpp_transaction_temporal_analysis.csv"
)

feature_stats_path = os.path.join(
    OUTPUT_DIR,
    "ellipticpp_transaction_feature_statistics.csv"
)

summary_path = os.path.join(
    OUTPUT_DIR,
    "ellipticpp_transaction_pipeline_summary.txt"
)

results.to_csv(
    results_path,
    index=False
)

top_test.to_csv(
    top_path,
    index=False
)

evidence_df.to_csv(
    evidence_path,
    index=False
)

temporal.to_csv(
    temporal_path
)

feature_stats.to_csv(
    feature_stats_path,
    index=False
)


# =============================================================================
# 30. SAVE SUMMARY
# =============================================================================

with open(
    summary_path,
    "w",
    encoding="utf-8"
) as f:

    f.write(
        "BIT-TRACE — TRANSACTION-ONLY UNSUPERVISED PIPELINE\n"
    )

    f.write(
        "=" * 70 + "\n\n"
    )

    f.write(
        "Dataset: Elliptic++ Transactions Dataset\n"
    )

    f.write(
        "P2P/network data used: NO\n"
    )

    f.write(
        "Edgelist used: NO\n\n"
    )

    f.write(
        f"Transactions: {len(data)}\n"
    )

    f.write(
        f"Selected features: {len(SELECTED_FEATURES)}\n"
    )

    f.write(
        f"Train samples: {len(train_idx)}\n"
    )

    f.write(
        f"Validation samples: {len(val_idx)}\n"
    )

    f.write(
        f"Test samples: {len(test_idx)}\n\n"
    )

    f.write(
        "Selected Features:\n"
    )

    for feature in SELECTED_FEATURES:
        f.write(
            f" - {feature}\n"
        )

    f.write("\n")

    f.write(
        f"Isolation Forest trees: {N_ESTIMATORS}\n"
    )

    f.write(
        f"IF weight: {IF_WEIGHT}\n"
    )

    f.write(
        f"LOF weight: {LOF_WEIGHT}\n\n"
    )

    try:
        f.write(
            f"Test ROC-AUC: {test_roc_auc:.6f}\n"
        )

        f.write(
            f"Test PR-AUC: {test_pr_auc:.6f}\n"
        )

    except:
        pass

    f.write("\n")
    f.write(
        "Labels were NOT used during model training.\n"
    )

    f.write(
        "Labels were used only for post-hoc evaluation.\n"
    )


# =============================================================================
# 31. FINAL STATUS
# =============================================================================

section("TRANSACTION-ONLY PIPELINE COMPLETE")

print("""
✓ Elliptic++ transaction features loaded
✓ 13 selected behavioral features used
✓ No P2P/network observations used
✓ No transaction edgelist used
✓ Missing/Inf values handled
✓ Temporal split preserved
✓ RobustScaler fitted only on training data
✓ Isolation Forest trained unsupervised
✓ LOF trained unsupervised
✓ IF + LOF ensemble generated
✓ Test anomaly ranking generated
✓ Class-3 evaluation performed post-hoc
✓ ROC-AUC / PR-AUC calculated
✓ Temporal anomaly analysis completed
✓ Feature-level anomaly evidence generated
✓ Investigation table generated
✓ Results saved

IMPORTANT:
Class labels were NOT used for model training.

OUTPUT DIRECTORY:
""")

print(OUTPUT_DIR)

print("\nFILES:")
print("1.", results_path)
print("2.", top_path)
print("3.", evidence_path)
print("4.", temporal_path)
print("5.", feature_stats_path)
print("6.", summary_path)

print("\nNext stage can use the saved investigation table")
print("for the BIT-TRACE investigator dashboard.")
```

```python
# =============================================================================
# BIT-TRACE — ELLIPTIC++ TRANSACTION-ONLY FINAL UNSUPERVISED PIPELINE
# =============================================================================
# Scope:
#   ONLY Elliptic++ Transactions Dataset
#   NO Bitcoin P2P/network dataset
#
# Pipeline:
#   Transaction features
#        ↓
#   Selected behavioral features (13)
#        +
#   Transaction graph features (10)
#        ↓
#   Robust preprocessing
#        ↓
#   Temporal split
#        ↓
#   Isolation Forest
#        +
#   LOF
#        ↓
#   Robust score normalization
#        ↓
#   Ensemble anomaly score
#        ↓
#   Temporal analysis
#        ↓
#   Post-hoc class evaluation
#        ↓
#   Feature-level explainability
#        ↓
#   Investigator evidence table
# =============================================================================

import os
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd

from sklearn.ensemble import IsolationForest
from sklearn.neighbors import LocalOutlierFactor
from sklearn.preprocessing import RobustScaler
from sklearn.metrics import roc_auc_score, average_precision_score

print("=" * 95)
print("BIT-TRACE — ELLIPTIC++ TRANSACTION-ONLY FINAL PIPELINE")
print("=" * 95)


# =============================================================================
# 1. LOCATE FILES
# =============================================================================

print("\n[1] LOCATING ELLIPTIC++ FILES")
print("-" * 95)

SEARCH_ROOTS = [
    r"C:\Users\YASHIKA\Desktop\bit-trace",
    os.getcwd()
]

def find_file(filename):
    found = []

    for root in SEARCH_ROOTS:
        if not os.path.exists(root):
            continue

        for dirpath, dirnames, filenames in os.walk(root):
            if filename in filenames:
                found.append(os.path.join(dirpath, filename))

    if not found:
        return None

    # Prefer EllipticPlusPlus / raw dataset paths
    found.sort(
        key=lambda p: (
            "EllipticPlusPlus" not in p,
            "elliptic-plus-plus" not in p.lower(),
            len(p)
        )
    )

    return found[0]


FEATURE_PATH = find_file("txs_features.csv")
CLASS_PATH   = find_file("txs_classes.csv")
EDGE_PATH    = find_file("txs_edgelist.csv")

print("FEATURE:", FEATURE_PATH)
print("CLASS  :", CLASS_PATH)
print("EDGES  :", EDGE_PATH)

if FEATURE_PATH is None:
    raise FileNotFoundError("txs_features.csv not found")

if CLASS_PATH is None:
    raise FileNotFoundError("txs_classes.csv not found")

if EDGE_PATH is None:
    raise FileNotFoundError("txs_edgelist.csv not found")


# =============================================================================
# 2. SELECTED BEHAVIORAL FEATURES
# =============================================================================

BEHAVIOR_FEATURES = [
    "Local_feature_53",
    "Local_feature_54",
    "Local_feature_52",
    "Local_feature_59",
    "Aggregate_feature_21",
    "Local_feature_25",
    "Local_feature_77",
    "Local_feature_89",
    "Local_feature_90",
    "Aggregate_feature_32",
    "Aggregate_feature_14",
    "Local_feature_22",
    "Local_feature_47"
]

print("\n[2] SELECTED BEHAVIORAL FEATURES")
print("-" * 95)

for i, f in enumerate(BEHAVIOR_FEATURES, 1):
    print(f"{i:2d}. {f}")


# =============================================================================
# 3. LOAD CLASS DATA
# =============================================================================

print("\n[3] LOADING CLASS DATA")
print("-" * 95)

classes = pd.read_csv(CLASS_PATH)

classes.columns = [
    str(c).strip().replace("\ufeff", "")
    for c in classes.columns
]

# Normalize column names
rename_map = {}

for c in classes.columns:
    cl = c.lower().replace(" ", "").replace("_", "")

    if cl == "txid":
        rename_map[c] = "txId"

    elif cl == "class":
        rename_map[c] = "class"

classes = classes.rename(columns=rename_map)

required_class_cols = {"txId", "class"}

if not required_class_cols.issubset(classes.columns):
    raise ValueError(
        f"Class file must contain {required_class_cols}, "
        f"found {classes.columns.tolist()}"
    )

classes["txId"] = pd.to_numeric(
    classes["txId"],
    errors="coerce"
)

classes["class"] = pd.to_numeric(
    classes["class"],
    errors="coerce"
)

classes = classes.dropna(
    subset=["txId", "class"]
)

classes["txId"] = classes["txId"].astype(np.int64)
classes["class"] = classes["class"].astype(int)

print("Shape:", classes.shape)
print("\nClass distribution:")
print(classes["class"].value_counts().sort_index())


# =============================================================================
# 4. READ FEATURE HEADER
# =============================================================================

print("\n[4] READING FEATURE HEADER")
print("-" * 95)

header = pd.read_csv(
    FEATURE_PATH,
    nrows=0
)

feature_columns = header.columns.tolist()

print("Total columns:", len(feature_columns))

# Detect ID
id_candidates = [
    c for c in feature_columns
    if str(c).strip().lower() in [
        "txid",
        "tx_id",
        "transactionid",
        "transaction_id"
    ]
]

if not id_candidates:
    raise ValueError("Could not detect txId column")

TX_COL = id_candidates[0]

# Detect time
time_candidates = [
    c for c in feature_columns
    if "time" in str(c).lower()
]

if not time_candidates:
    raise ValueError("Could not detect Time step column")

TIME_COL = time_candidates[0]

print("ID column  :", TX_COL)
print("Time column:", TIME_COL)


# =============================================================================
# 5. VERIFY FEATURES
# =============================================================================

missing_features = [
    f for f in BEHAVIOR_FEATURES
    if f not in feature_columns
]

if missing_features:
    raise ValueError(
        "Missing behavioral features:\n"
        + "\n".join(missing_features)
    )

print("✓ All 13 behavioral features found")


# =============================================================================
# 6. LOAD ONLY REQUIRED TRANSACTION COLUMNS
# =============================================================================

print("\n[5] LOADING TRANSACTION FEATURES")
print("-" * 95)

LOAD_COLUMNS = [
    TX_COL,
    TIME_COL
] + BEHAVIOR_FEATURES

tx = pd.read_csv(
    FEATURE_PATH,
    usecols=LOAD_COLUMNS
)

tx = tx.rename(
    columns={
        TX_COL: "txId",
        TIME_COL: "Time step"
    }
)

tx["txId"] = pd.to_numeric(
    tx["txId"],
    errors="coerce"
)

tx["Time step"] = pd.to_numeric(
    tx["Time step"],
    errors="coerce"
)

for f in BEHAVIOR_FEATURES:
    tx[f] = pd.to_numeric(
        tx[f],
        errors="coerce"
    )

tx = tx.dropna(
    subset=["txId", "Time step"]
)

tx["txId"] = tx["txId"].astype(np.int64)

print("Transaction shape:", tx.shape)
print("Missing txId:", tx["txId"].isna().sum())
print("Missing time:", tx["Time step"].isna().sum())
print("Duplicate txId:", tx["txId"].duplicated().sum())


# =============================================================================
# 7. MERGE LABELS
# =============================================================================

print("\n[6] MERGING TRANSACTION CLASSES")
print("-" * 95)

data = tx.merge(
    classes,
    on="txId",
    how="inner",
    validate="one_to_one"
)

print("Merged shape:", data.shape)
print("Missing class:", data["class"].isna().sum())

print("\nClass distribution:")
print(data["class"].value_counts().sort_index())


# =============================================================================
# 8. CLEAN BEHAVIORAL FEATURES
# =============================================================================

print("\n[7] CLEANING BEHAVIORAL FEATURES")
print("-" * 95)

X_behavior = data[BEHAVIOR_FEATURES].copy()

X_behavior = X_behavior.replace(
    [np.inf, -np.inf],
    np.nan
)

print("NaN before:", X_behavior.isna().sum().sum())

# Median imputation
X_behavior = X_behavior.fillna(
    X_behavior.median()
)

print("NaN after :", X_behavior.isna().sum().sum())


# =============================================================================
# 9. LOAD TRANSACTION EDGES
# =============================================================================

print("\n[8] LOADING TRANSACTION GRAPH")
print("-" * 95)

edges = pd.read_csv(
    EDGE_PATH
)

edges.columns = [
    str(c).strip().replace("\ufeff", "")
    for c in edges.columns
]

print("Edge columns:", edges.columns.tolist())

# Detect source/target columns
edge_lower = {
    c.lower().replace(" ", "").replace("_", ""): c
    for c in edges.columns
}

source_candidates = [
    "txid1",
    "source",
    "from",
    "src"
]

target_candidates = [
    "txid2",
    "target",
    "to",
    "dst"
]

SOURCE_COL = None
TARGET_COL = None

for c in source_candidates:
    if c in edge_lower:
        SOURCE_COL = edge_lower[c]
        break

for c in target_candidates:
    if c in edge_lower:
        TARGET_COL = edge_lower[c]
        break

if SOURCE_COL is None or TARGET_COL is None:

    if len(edges.columns) >= 2:
        SOURCE_COL = edges.columns[0]
        TARGET_COL = edges.columns[1]

    else:
        raise ValueError(
            "Could not detect edge source/target columns"
        )

print("Source:", SOURCE_COL)
print("Target:", TARGET_COL)

edges = edges[
    [SOURCE_COL, TARGET_COL]
].copy()

edges.columns = [
    "source",
    "target"
]

edges["source"] = pd.to_numeric(
    edges["source"],
    errors="coerce"
)

edges["target"] = pd.to_numeric(
    edges["target"],
    errors="coerce"
)

edges = edges.dropna()

edges["source"] = edges["source"].astype(np.int64)
edges["target"] = edges["target"].astype(np.int64)

# Remove self loops
edges = edges[
    edges["source"] != edges["target"]
]

# Remove duplicate directed edges
edges = edges.drop_duplicates()

print("Valid edges:", len(edges))


# =============================================================================
# 10. BUILD TRANSACTION GRAPH FEATURES
# =============================================================================

print("\n[9] BUILDING TRANSACTION GRAPH FEATURES")
print("-" * 95)

tx_ids_set = set(
    data["txId"].values
)

# Only edges involving known transactions
edges_valid = edges[
    edges["source"].isin(tx_ids_set)
    &
    edges["target"].isin(tx_ids_set)
].copy()

print("Edges connected to transactions:", len(edges_valid))

# In degree
in_degree = (
    edges_valid
    .groupby("target")
    .size()
)

# Out degree
out_degree = (
    edges_valid
    .groupby("source")
    .size()
)

# Map to transactions
graph = pd.DataFrame({
    "txId": data["txId"]
})

graph["graph_in_degree"] = (
    graph["txId"]
    .map(in_degree)
    .fillna(0)
)

graph["graph_out_degree"] = (
    graph["txId"]
    .map(out_degree)
    .fillna(0)
)

graph["graph_total_degree"] = (
    graph["graph_in_degree"]
    +
    graph["graph_out_degree"]
)

graph["graph_degree_imbalance"] = (
    abs(
        graph["graph_in_degree"]
        -
        graph["graph_out_degree"]
    )
)

# Ratio
graph["graph_in_out_ratio"] = (
    (graph["graph_in_degree"] + 1.0)
    /
    (graph["graph_out_degree"] + 1.0)
)

# Log transforms
graph["graph_log_total_degree"] = np.log1p(
    graph["graph_total_degree"]
)

graph["graph_log_in_degree"] = np.log1p(
    graph["graph_in_degree"]
)

graph["graph_log_out_degree"] = np.log1p(
    graph["graph_out_degree"]
)


# Reciprocal edges
# Build set only once
edge_set = set(
    zip(
        edges_valid["source"].values,
        edges_valid["target"].values
    )
)

reciprocal_counts = {}

for s, t in zip(
    edges_valid["source"].values,
    edges_valid["target"].values
):

    if (t, s) in edge_set:

        reciprocal_counts[s] = (
            reciprocal_counts.get(s, 0) + 1
        )

        reciprocal_counts[t] = (
            reciprocal_counts.get(t, 0) + 1
        )

graph["graph_reciprocal_degree"] = (
    graph["txId"]
    .map(reciprocal_counts)
    .fillna(0)
)

graph["graph_log_reciprocal_degree"] = np.log1p(
    graph["graph_reciprocal_degree"]
)

GRAPH_FEATURES = [
    "graph_in_degree",
    "graph_out_degree",
    "graph_total_degree",
    "graph_degree_imbalance",
    "graph_in_out_ratio",
    "graph_log_total_degree",
    "graph_log_in_degree",
    "graph_log_out_degree",
    "graph_reciprocal_degree",
    "graph_log_reciprocal_degree"
]

print("Graph features:", len(GRAPH_FEATURES))

print(
    graph[GRAPH_FEATURES]
    .describe()
    .T[
        ["mean", "std", "min", "50%", "max"]
    ]
)


# =============================================================================
# 11. COMBINE BEHAVIOR + GRAPH
# =============================================================================

print("\n[10] COMBINING BEHAVIORAL + GRAPH FEATURES")
print("-" * 95)

# Ensure same row order
if not np.array_equal(
    graph["txId"].values,
    data["txId"].values
):
    raise ValueError(
        "Graph/data transaction order mismatch"
    )

X_combined = pd.concat(
    [
        X_behavior.reset_index(drop=True),
        graph[GRAPH_FEATURES].reset_index(drop=True)
    ],
    axis=1
)

ALL_FEATURES = (
    BEHAVIOR_FEATURES
    +
    GRAPH_FEATURES
)

print("Behavior features:", len(BEHAVIOR_FEATURES))
print("Graph features   :", len(GRAPH_FEATURES))
print("Total features   :", len(ALL_FEATURES))
print("Combined shape    :", X_combined.shape)


# =============================================================================
# 12. FINAL CLEANING
# =============================================================================

X_combined = X_combined.replace(
    [np.inf, -np.inf],
    np.nan
)

X_combined = X_combined.fillna(
    X_combined.median()
)

print(
    "Remaining NaN:",
    X_combined.isna().sum().sum()
)


# =============================================================================
# 13. TEMPORAL SPLIT
# =============================================================================

print("\n[11] TEMPORAL SPLIT")
print("-" * 95)

time_steps = data["Time step"].to_numpy()

train_idx = np.where(
    time_steps <= 34
)[0]

val_idx = np.where(
    (time_steps >= 35)
    &
    (time_steps <= 41)
)[0]

test_idx = np.where(
    time_steps >= 42
)[0]

print(
    f"Train      : {len(train_idx):,} "
    f"(time 1–34)"
)

print(
    f"Validation : {len(val_idx):,} "
    f"(time 35–41)"
)

print(
    f"Test       : {len(test_idx):,} "
    f"(time 42–49)"
)


# =============================================================================
# 14. ROBUST SCALING
# =============================================================================
# Fit scaler ONLY on training data to prevent temporal leakage.

print("\n[12] ROBUST SCALING")
print("-" * 95)

X_np = X_combined[
    ALL_FEATURES
].to_numpy(
    dtype=np.float32
)

scaler = RobustScaler()

X_train = scaler.fit_transform(
    X_np[train_idx]
)

X_val = scaler.transform(
    X_np[val_idx]
)

X_test = scaler.transform(
    X_np[test_idx]
)

print("Train:", X_train.shape)
print("Val  :", X_val.shape)
print("Test :", X_test.shape)


# =============================================================================
# 15. ISOLATION FOREST
# =============================================================================

print("\n[13] TRAINING ISOLATION FOREST")
print("-" * 95)

# Labels are NOT used here.

iso = IsolationForest(
    n_estimators=300,
    max_samples="auto",
    contamination="auto",
    random_state=42,
    n_jobs=-1
)

iso.fit(
    X_train
)

print("✓ Isolation Forest trained")


# =============================================================================
# 16. IF ANOMALY SCORE
# =============================================================================
# sklearn decision_function:
# higher = more normal
#
# Therefore anomaly = negative decision_function.

if_train_raw = -iso.decision_function(X_train)
if_val_raw   = -iso.decision_function(X_val)
if_test_raw  = -iso.decision_function(X_test)

print("✓ IF scores generated")


# =============================================================================
# 17. LOF
# =============================================================================
# LOF does not support normal train/predict in the same way as IF.
# We fit LOF on training data and evaluate validation/test using
# novelty=True.
#
# n_neighbors=35 gives local neighborhood context while remaining
# computationally manageable for this dataset.

print("\n[14] TRAINING LOF")
print("-" * 95)

lof = LocalOutlierFactor(
    n_neighbors=35,
    contamination="auto",
    novelty=True,
    n_jobs=-1
)

lof.fit(X_train)

print("✓ LOF trained")

# LOF decision_function:
# higher = more normal
# Therefore anomaly = negative decision_function.

lof_train_raw = -lof.decision_function(X_train)
lof_val_raw   = -lof.decision_function(X_val)
lof_test_raw  = -lof.decision_function(X_test)

print("✓ LOF scores generated")


# =============================================================================
# 18. ROBUST SCORE NORMALIZATION
# =============================================================================
# IMPORTANT:
# Do NOT combine raw IF and LOF scores.
#
# LOF and IF have completely different numerical scales.
#
# We normalize each score using training quantiles.
# Higher normalized value = more anomalous.

print("\n[15] NORMALIZING ANOMALY SCORES")
print("-" * 95)

def robust_normalize(
    train_score,
    val_score,
    test_score
):

    q_low, q_high = np.quantile(
        train_score,
        [0.01, 0.99]
    )

    denominator = q_high - q_low

    if denominator <= 1e-12:
        denominator = 1.0

    def transform(x):
        z = (
            x - q_low
        ) / denominator

        return np.clip(
            z,
            0,
            1
        )

    return (
        transform(train_score),
        transform(val_score),
        transform(test_score)
    )


if_train, if_val, if_test = robust_normalize(
    if_train_raw,
    if_val_raw,
    if_test_raw
)

lof_train, lof_val, lof_test = robust_normalize(
    lof_train_raw,
    lof_val_raw,
    lof_test_raw
)

print("✓ IF normalized")
print("✓ LOF normalized")


# =============================================================================
# 19. ENSEMBLE
# =============================================================================
# Equal weighting avoids one model dominating because of scale.

print("\n[16] BUILDING ENSEMBLE")
print("-" * 95)

IF_WEIGHT = 0.60
LOF_WEIGHT = 0.40

ensemble_train = (
    IF_WEIGHT * if_train
    +
    LOF_WEIGHT * lof_train
)

ensemble_val = (
    IF_WEIGHT * if_val
    +
    LOF_WEIGHT * lof_val
)

ensemble_test = (
    IF_WEIGHT * if_test
    +
    LOF_WEIGHT * lof_test
)

print(
    f"Weights: IF={IF_WEIGHT} | LOF={LOF_WEIGHT}"
)

print(
    "Ensemble range:",
    float(ensemble_test.min()),
    "to",
    float(ensemble_test.max())
)


# =============================================================================
# 20. BUILD COMPLETE RESULTS TABLE
# =============================================================================

print("\n[17] BUILDING RESULTS TABLE")
print("-" * 95)

results = data[
    ["txId", "Time step", "class"]
].copy()

results["split"] = "train"

results.loc[
    val_idx,
    "split"
] = "validation"

results.loc[
    test_idx,
    "split"
] = "test"

results["IF_anomaly"] = np.nan
results["LOF_anomaly"] = np.nan
results["ensemble_anomaly"] = np.nan

results.loc[
    train_idx,
    "IF_anomaly"
] = if_train

results.loc[
    val_idx,
    "IF_anomaly"
] = if_val

results.loc[
    test_idx,
    "IF_anomaly"
] = if_test

results.loc[
    train_idx,
    "LOF_anomaly"
] = lof_train

results.loc[
    val_idx,
    "LOF_anomaly"
] = lof_val

results.loc[
    test_idx,
    "LOF_anomaly"
] = lof_test

results.loc[
    train_idx,
    "ensemble_anomaly"
] = ensemble_train

results.loc[
    val_idx,
    "ensemble_anomaly"
] = ensemble_val

results.loc[
    test_idx,
    "ensemble_anomaly"
] = ensemble_test


# =============================================================================
# 21. TOP ANOMALIES
# =============================================================================

print("\n[18] TOP 30 ANOMALOUS TRANSACTIONS")
print("-" * 95)

top30 = (
    results[
        results["split"] != "train"
    ]
    .sort_values(
        "ensemble_anomaly",
        ascending=False
    )
    .head(30)
)

print(
    top30[
        [
            "txId",
            "Time step",
            "split",
            "class",
            "IF_anomaly",
            "LOF_anomaly",
            "ensemble_anomaly"
        ]
    ].to_string(index=False)
)


# =============================================================================
# 22. TEMPORAL ANALYSIS
# =============================================================================

print("\n[19] TEMPORAL ANOMALY ANALYSIS")
print("-" * 95)

temporal = (
    results[
        results["split"] != "train"
    ]
    .groupby("Time step")
    .agg(
        Transactions=("txId", "count"),
        Mean_Anomaly=("ensemble_anomaly", "mean"),
        Median_Anomaly=("ensemble_anomaly", "median"),
        Max_Anomaly=("ensemble_anomaly", "max"),
        Class3_Count=(
            "class",
            lambda x: (x == 3).sum()
        )
    )
)

temporal["Class3_%"] = (
    temporal["Class3_Count"]
    /
    temporal["Transactions"]
    *
    100
)

temporal = temporal.sort_values(
    "Mean_Anomaly",
    ascending=False
)

print(
    temporal.head(15).to_string()
)


# =============================================================================
# 23. POST-HOC CLASS-3 EVALUATION
# =============================================================================
# IMPORTANT:
# Class labels are used ONLY here for evaluation.
# They were never used for model training.

print("\n[20] POST-HOC CLASS-3 EVALUATION")
print("-" * 95)

test_results = results.loc[
    test_idx
].copy()

y_test_binary = (
    test_results["class"].values == 3
).astype(int)

score_test = (
    test_results["ensemble_anomaly"].values
)

try:

    roc = roc_auc_score(
        y_test_binary,
        score_test
    )

    pr = average_precision_score(
        y_test_binary,
        score_test
    )

    print(
        f"Test ROC-AUC : {roc:.4f}"
    )

    print(
        f"Test PR-AUC  : {pr:.4f}"
    )

except Exception as e:

    print(
        "Evaluation unavailable:",
        e
    )


# =============================================================================
# 24. TOP-K ENRICHMENT
# =============================================================================

print("\n[21] TEST-SET CLASS-3 ENRICHMENT")
print("-" * 95)

overall_class3 = (
    y_test_binary.mean()
)

print(
    f"Overall Test Class-3: "
    f"{overall_class3 * 100:.2f}%"
)

for k in [
    50,
    100,
    500,
    1000,
    5000
]:

    topk = test_results.sort_values(
        "ensemble_anomaly",
        ascending=False
    ).head(k)

    class3_pct = (
        (topk["class"] == 3)
        .mean()
    )

    enrichment = (
        class3_pct
        /
        overall_class3
        if overall_class3 > 0
        else np.nan
    )

    print(
        f"Top {k:5d}: "
        f"Class3 = {class3_pct * 100:6.2f}% | "
        f"Enrichment = {enrichment:.3f}x"
    )


# =============================================================================
# 25. FEATURE-LEVEL EXPLAINABILITY
# =============================================================================
# This is not SHAP.
#
# We calculate a practical transaction-level deviation:
#
# |scaled feature - training median|
#
# Larger value = more unusual relative to training behavior.
#
# This provides evidence for the investigator dashboard.

print("\n[22] GENERATING FEATURE-LEVEL EVIDENCE")
print("-" * 95)

train_median = np.median(
    X_train,
    axis=0
)

train_q75 = np.quantile(
    X_train,
    0.75,
    axis=0
)

train_q25 = np.quantile(
    X_train,
    0.25,
    axis=0
)

train_iqr = (
    train_q75
    -
    train_q25
)

train_iqr[
    train_iqr < 1e-6
] = 1.0

# Robust deviation:
# distance from train median in IQR units

def feature_deviation(X):

    return np.abs(
        X - train_median
    ) / train_iqr


test_deviation = feature_deviation(
    X_test
)

# top 5 evidence features
top_evidence_indices = np.argsort(
    test_deviation,
    axis=1
)[:, -5:][:, ::-1]

evidence_rows = []

test_txids = data.iloc[
    test_idx
]["txId"].values

test_times = data.iloc[
    test_idx
]["Time step"].values

test_classes = data.iloc[
    test_idx
]["class"].values

for i in range(
    len(test_idx)
):

    inds = top_evidence_indices[i]

    evidence = []

    for j in inds:

        evidence.append(
            (
                ALL_FEATURES[j],
                float(
                    test_deviation[i, j]
                )
            )
        )

    evidence_rows.append(
        evidence
    )


# =============================================================================
# 26. BUILD INVESTIGATOR EVIDENCE TABLE
# =============================================================================

evidence_df = pd.DataFrame({

    "txId": test_txids,

    "Time step": test_times,

    "class": test_classes,

    "ensemble_anomaly":
        ensemble_test,

    "IF_anomaly":
        if_test,

    "LOF_anomaly":
        lof_test,

    "evidence_1":
        [
            x[0][0]
            for x in evidence_rows
        ],

    "evidence_1_deviation":
        [
            x[0][1]
            for x in evidence_rows
        ],

    "evidence_2":
        [
            x[1][0]
            for x in evidence_rows
        ],

    "evidence_2_deviation":
        [
            x[1][1]
            for x in evidence_rows
        ],

    "evidence_3":
        [
            x[2][0]
            for x in evidence_rows
        ],

    "evidence_3_deviation":
        [
            x[2][1]
            for x in evidence_rows
        ],

    "evidence_4":
        [
            x[3][0]
            for x in evidence_rows
        ],

    "evidence_4_deviation":
        [
            x[3][1]
            for x in evidence_rows
        ],

    "evidence_5":
        [
            x[4][0]
            for x in evidence_rows
        ],

    "evidence_5_deviation":
        [
            x[4][1]
            for x in evidence_rows
        ]
})


# =============================================================================
# 27. RISK BANDS
# =============================================================================

# Thresholds are derived from TRAINING anomaly distribution,
# not from class labels.

q90 = np.quantile(
    ensemble_train,
    0.90
)

q95 = np.quantile(
    ensemble_train,
    0.95
)

q99 = np.quantile(
    ensemble_train,
    0.99
)

def risk_band(score):

    if score >= q99:
        return "EXTREME"

    elif score >= q95:
        return "HIGH"

    elif score >= q90:
        return "ELEVATED"

    else:
        return "NORMAL"


evidence_df["risk_band"] = (
    evidence_df[
        "ensemble_anomaly"
    ].apply(risk_band)
)


# =============================================================================
# 28. FINAL INVESTIGATION QUEUE
# =============================================================================

print("\n[23] FINAL INVESTIGATION QUEUE")
print("-" * 95)

investigation_queue = (
    evidence_df
    .sort_values(
        "ensemble_anomaly",
        ascending=False
    )
    .head(50)
)

print(
    investigation_queue[
        [
            "txId",
            "Time step",
            "ensemble_anomaly",
            "risk_band",
            "evidence_1",
            "evidence_2",
            "evidence_3"
        ]
    ].head(20).to_string(index=False)
)


# =============================================================================
# 29. SAVE OUTPUTS
# =============================================================================

print("\n[24] SAVING RESULTS")
print("-" * 95)

OUTPUT_DIR = os.path.join(
    os.path.dirname(FEATURE_PATH),
    "bit_trace_transaction_outputs"
)

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)

# Main anomaly results
results_path = os.path.join(
    OUTPUT_DIR,
    "ellipticpp_transaction_anomaly_results.csv"
)

results.to_csv(
    results_path,
    index=False
)

# Graph features
graph_path = os.path.join(
    OUTPUT_DIR,
    "ellipticpp_transaction_graph_features.csv"
)

graph.to_csv(
    graph_path,
    index=False
)

# Temporal analysis
temporal_path = os.path.join(
    OUTPUT_DIR,
    "ellipticpp_temporal_anomaly_analysis.csv"
)

temporal.to_csv(
    temporal_path
)

# Investigator evidence
evidence_path = os.path.join(
    OUTPUT_DIR,
    "ellipticpp_investigation_evidence.csv"
)

evidence_df.to_csv(
    evidence_path,
    index=False
)

# Top anomalies
top_path = os.path.join(
    OUTPUT_DIR,
    "ellipticpp_top_investigation_queue.csv"
)

investigation_queue.to_csv(
    top_path,
    index=False
)

# Feature list
feature_info = pd.DataFrame({
    "feature": ALL_FEATURES,
    "type": (
        ["behavioral"] * len(BEHAVIOR_FEATURES)
        +
        ["graph"] * len(GRAPH_FEATURES)
    )
})

feature_info_path = os.path.join(
    OUTPUT_DIR,
    "ellipticpp_final_feature_list.csv"
)

feature_info.to_csv(
    feature_info_path,
    index=False
)

print("✓", results_path)
print("✓", graph_path)
print("✓", temporal_path)
print("✓", evidence_path)
print("✓", top_path)
print("✓", feature_info_path)


# =============================================================================
# 30. FINAL SUMMARY
# =============================================================================

print("\n")
print("=" * 95)
print("BIT-TRACE — TRANSACTION-ONLY PIPELINE COMPLETE")
print("=" * 95)

print(
    f"""
Transactions              : {len(data):,}
Transaction graph edges   : {len(edges_valid):,}

Behavioral features       : {len(BEHAVIOR_FEATURES)}
Graph features            : {len(GRAPH_FEATURES)}
Total ML features         : {len(ALL_FEATURES)}

Train transactions        : {len(train_idx):,}
Validation transactions   : {len(val_idx):,}
Test transactions         : {len(test_idx):,}

Models:
  ✓ Isolation Forest
  ✓ Local Outlier Factor
  ✓ Normalized ensemble

Labels used for training  : NO
Labels used for evaluation: YES (post-hoc only)

Outputs:
  ✓ anomaly scores
  ✓ temporal anomaly analysis
  ✓ graph features
  ✓ feature-level evidence
  ✓ investigation queue

P2P / network dataset     : NOT USED
"""
)

print("=" * 95)
print("NEXT MODULE: INVESTIGATOR DASHBOARD / EVIDENCE VISUALIZATION")
print("=" * 95)
```

```python
# =============================================================================
# BIT-TRACE — TRANSACTION-ONLY FINAL EVALUATION + EXPLAINABILITY
# Graph + Behavioral Unsupervised Anomaly Detection
# P2P DATA NOT USED
# =============================================================================

import os
import numpy as np
import pandas as pd
from sklearn.metrics import roc_auc_score, average_precision_score

print("=" * 100)
print("BIT-TRACE — TRANSACTION-ONLY FINAL EVALUATION + EXPLAINABILITY")
print("=" * 100)

# =============================================================================
# 1. LOCATE SAVED RESULT FILE
# =============================================================================

possible_files = [
    r"C:\Users\YASHIKA\Desktop\bit-trace\ellipticpp_graph_behavioral_anomaly_results.csv",
    r"C:\Users\YASHIKA\Desktop\bit-trace\data\processed\ellipticpp_graph_behavioral_anomaly_results.csv",
    r"C:\Users\YASHIKA\Desktop\bit-trace\src\ellipticpp_graph_behavioral_anomaly_results.csv",
]

result_path = None

for p in possible_files:
    if os.path.exists(p):
        result_path = p
        break

# Recursive fallback search
if result_path is None:
    for root, dirs, files in os.walk(
        r"C:\Users\YASHIKA\Desktop\bit-trace"
    ):
        if "ellipticpp_graph_behavioral_anomaly_results.csv" in files:
            result_path = os.path.join(
                root,
                "ellipticpp_graph_behavioral_anomaly_results.csv"
            )
            break

if result_path is None:
    raise FileNotFoundError(
        "ellipticpp_graph_behavioral_anomaly_results.csv not found."
    )

print("\n[1] RESULT FILE")
print("-" * 80)
print(result_path)

df = pd.read_csv(result_path)

print("Shape:", df.shape)
print("Columns:")
print(df.columns.tolist())


# =============================================================================
# 2. NORMALIZE COLUMN NAMES
# =============================================================================

df.columns = [str(c).strip() for c in df.columns]

# Find columns robustly
def find_col(candidates):
    for c in candidates:
        if c in df.columns:
            return c
    return None

tx_col = find_col(["txId", "txid", "TxId", "transaction_id"])
time_col = find_col(["Time step", "time_step", "Time_step", "time"])
class_col = find_col(["class", "Class", "label"])
score_col = find_col([
    "ensemble_anomaly",
    "Ensemble_Anomaly",
    "anomaly_score",
    "Anomaly_Score"
])

if tx_col is None:
    raise ValueError("Transaction ID column not found.")

if time_col is None:
    raise ValueError("Time step column not found.")

if class_col is None:
    raise ValueError("Class column not found.")

if score_col is None:
    raise ValueError(
        "Ensemble anomaly score column not found.\n"
        f"Available columns: {df.columns.tolist()}"
    )

print("\nDetected columns:")
print("txId       :", tx_col)
print("Time step  :", time_col)
print("Class      :", class_col)
print("Score      :", score_col)


# =============================================================================
# 3. BASIC VALIDATION
# =============================================================================

df = df.copy()

df[tx_col] = pd.to_numeric(df[tx_col], errors="coerce")
df[time_col] = pd.to_numeric(df[time_col], errors="coerce")
df[class_col] = pd.to_numeric(df[class_col], errors="coerce")
df[score_col] = pd.to_numeric(df[score_col], errors="coerce")

df = df.dropna(
    subset=[tx_col, time_col, class_col, score_col]
).copy()

print("\n[2] DATA VALIDATION")
print("-" * 80)

print("Rows:", len(df))
print("Missing scores:", df[score_col].isna().sum())
print("Time range:",
      int(df[time_col].min()),
      "to",
      int(df[time_col].max()))

print("\nClass distribution:")
print(df[class_col].value_counts().sort_index())


# =============================================================================
# 4. CREATE TEMPORAL SPLIT
# =============================================================================

df["split"] = np.select(
    [
        df[time_col] <= 34,
        (df[time_col] >= 35) & (df[time_col] <= 41),
        df[time_col] >= 42
    ],
    [
        "train",
        "validation",
        "test"
    ],
    default="unknown"
)

print("\n[3] TEMPORAL SPLIT")
print("-" * 80)

print(
    df["split"].value_counts()
)

test_df = df[df["split"] == "test"].copy()

if len(test_df) == 0:
    raise ValueError("No test samples found.")


# =============================================================================
# 5. IMPORTANT:
#    SCORE DIRECTION
# =============================================================================
#
# For this pipeline, larger ensemble_anomaly values represent stronger
# anomaly according to the generated ranking.
#
# Therefore:
# top anomalies = descending score
# =============================================================================

print("\n[4] SCORE DIRECTION")
print("-" * 80)

print(
    "Highest score:",
    test_df[score_col].max()
)

print(
    "Lowest score :",
    test_df[score_col].min()
)

print(
    "Top anomaly = highest ensemble anomaly score"
)


# =============================================================================
# 6. TEST CLASS-WISE ANALYSIS
# =============================================================================

print("\n" + "=" * 100)
print("TEST SET — CLASS-WISE ANOMALY ANALYSIS")
print("=" * 100)

class_summary = (
    test_df
    .groupby(class_col)[score_col]
    .agg(
        Samples="count",
        Mean="mean",
        Std="std",
        Min="min",
        Median="median",
        Max="max"
    )
)

print(class_summary.to_string())


# =============================================================================
# 7. ROC-AUC + PR-AUC
# =============================================================================
#
# Class 3 = positive class.
#
# These are POST-HOC evaluations only.
# Labels are NOT used for unsupervised model training.
# =============================================================================

print("\n" + "=" * 100)
print("TEST SET — CLASS-3 POST-HOC EVALUATION")
print("=" * 100)

y_true = (test_df[class_col] == 3).astype(int)
scores = test_df[score_col].to_numpy()

if y_true.nunique() == 2:

    roc_auc = roc_auc_score(
        y_true,
        scores
    )

    pr_auc = average_precision_score(
        y_true,
        scores
    )

    print(f"Test ROC-AUC : {roc_auc:.4f}")
    print(f"Test PR-AUC  : {pr_auc:.4f}")

else:
    roc_auc = np.nan
    pr_auc = np.nan

    print("ROC-AUC / PR-AUC unavailable.")


# =============================================================================
# 8. TOP-K ANOMALY ANALYSIS
# =============================================================================

print("\n" + "=" * 100)
print("TOP-K ANOMALY ANALYSIS")
print("=" * 100)

overall_class3_pct = (
    (test_df[class_col] == 3).mean() * 100
)

print(
    f"Overall Test Class-3 percentage: "
    f"{overall_class3_pct:.2f}%"
)

topk_results = []

for k in [50, 100, 250, 500, 1000, 2500, 5000]:

    k = min(k, len(test_df))

    topk = (
        test_df
        .nlargest(k, score_col)
        .copy()
    )

    class3_count = (
        (topk[class_col] == 3).sum()
    )

    class3_pct = (
        class3_count / k * 100
    )

    enrichment = (
        class3_pct / overall_class3_pct
        if overall_class3_pct > 0
        else np.nan
    )

    topk_results.append({
        "Top_K": k,
        "Class3_Count": int(class3_count),
        "Class3_%": class3_pct,
        "Overall_Class3_%": overall_class3_pct,
        "Enrichment": enrichment
    })

topk_df = pd.DataFrame(topk_results)

print(topk_df.to_string(index=False))


# =============================================================================
# 9. TOP 100 MOST ANOMALOUS TRANSACTIONS
# =============================================================================

print("\n" + "=" * 100)
print("TOP 100 MOST ANOMALOUS TRANSACTIONS")
print("=" * 100)

top100 = (
    test_df
    .nlargest(100, score_col)
    [[
        tx_col,
        time_col,
        class_col,
        score_col
    ]]
    .copy()
)

print(
    top100.head(30).to_string(index=False)
)


# =============================================================================
# 10. ANOMALY THRESHOLDS
# =============================================================================
#
# Unsupervised thresholding:
# 99th percentile
# 99.5th percentile
# 99.9th percentile
# =============================================================================

print("\n" + "=" * 100)
print("UNSUPERVISED ANOMALY THRESHOLDS — TEST SET")
print("=" * 100)

threshold_rows = []

for pct in [95, 97, 99, 99.5, 99.9]:

    threshold = np.percentile(
        test_df[score_col],
        pct
    )

    flagged = (
        test_df[score_col] >= threshold
    )

    flagged_df = test_df[flagged]

    class3_count = (
        (flagged_df[class_col] == 3).sum()
    )

    class3_pct = (
        class3_count / len(flagged_df) * 100
        if len(flagged_df) > 0
        else 0
    )

    threshold_rows.append({
        "Percentile": pct,
        "Threshold": threshold,
        "Flagged": len(flagged_df),
        "Class3_Count": int(class3_count),
        "Class3_%": class3_pct
    })

threshold_df = pd.DataFrame(threshold_rows)

print(
    threshold_df.to_string(index=False)
)


# =============================================================================
# 11. TEMPORAL ANOMALY ANALYSIS
# =============================================================================

print("\n" + "=" * 100)
print("TEMPORAL ANOMALY ANALYSIS — TEST")
print("=" * 100)

temporal = (
    test_df
    .groupby(time_col)
    .agg(
        Transactions=(tx_col, "count"),
        Mean_Anomaly=(score_col, "mean"),
        Median_Anomaly=(score_col, "median"),
        Max_Anomaly=(score_col, "max")
    )
    .sort_values(
        "Mean_Anomaly",
        ascending=False
    )
)

print(
    temporal.head(15).to_string()
)


# =============================================================================
# 12. TEMPORAL CLASS-3 CONCENTRATION
# =============================================================================

temporal_class3 = (
    test_df
    .groupby(time_col)
    .agg(
        Transactions=(tx_col, "count"),
        Class3_Count=(class_col, lambda x: (x == 3).sum()),
        Mean_Anomaly=(score_col, "mean"),
        Max_Anomaly=(score_col, "max")
    )
)

temporal_class3["Class3_%"] = (
    temporal_class3["Class3_Count"]
    / temporal_class3["Transactions"]
    * 100
)

temporal_class3 = temporal_class3.sort_values(
    "Mean_Anomaly",
    ascending=False
)

print("\nTop anomalous time steps:")
print(
    temporal_class3.head(15).to_string()
)


# =============================================================================
# 13. BEHAVIORAL FEATURE EXPLAINABILITY
# =============================================================================
#
# If original behavioral feature columns are present in saved results,
# calculate contribution using absolute standardized magnitude.
#
# If they are not present, reload ONLY the 13 selected features from
# txs_features.csv.
# =============================================================================

print("\n" + "=" * 100)
print("FEATURE-LEVEL EXPLAINABILITY")
print("=" * 100)

behavioral_features = [
    "Local_feature_22",
    "Local_feature_25",
    "Local_feature_47",
    "Local_feature_52",
    "Local_feature_53",
    "Local_feature_54",
    "Local_feature_59",
    "Local_feature_77",
    "Local_feature_89",
    "Local_feature_90",
    "Aggregate_feature_14",
    "Aggregate_feature_21",
    "Aggregate_feature_32"
]

available_features = [
    f for f in behavioral_features
    if f in df.columns
]

print(
    "Behavioral features already available:",
    len(available_features)
)

# -------------------------------------------------------------------------
# If feature columns are not saved in anomaly results, load them directly.
# -------------------------------------------------------------------------

if len(available_features) < len(behavioral_features):

    feature_path = (
        r"C:\Users\YASHIKA\Desktop\bit-trace"
        r"\data\raw\elliptic-plus-plus"
        r"\txs_features.csv"
    )

    if not os.path.exists(feature_path):

        feature_path = (
            r"C:\Users\YASHIKA\Desktop\bit-trace"
            r"\EllipticPlusPlus\Transactions Dataset"
            r"\txs_features.csv"
        )

    if os.path.exists(feature_path):

        print("\nLoading only required behavioral features...")

        needed_cols = [
            "txId"
        ] + behavioral_features

        feature_data = pd.read_csv(
            feature_path,
            usecols=needed_cols
        )

        feature_data["txId"] = pd.to_numeric(
            feature_data["txId"],
            errors="coerce"
        )

        for f in behavioral_features:
            feature_data[f] = pd.to_numeric(
                feature_data[f],
                errors="coerce"
            )

        df = df.merge(
            feature_data,
            left_on=tx_col,
            right_on="txId",
            how="left",
            suffixes=("", "_feature")
        )

        # Avoid duplicate txId column confusion
        if "txId_feature" in df.columns:
            df.drop(
                columns=["txId_feature"],
                inplace=True
            )

        test_df = df[
            df["split"] == "test"
        ].copy()

        print(
            "Behavioral features loaded successfully."
        )

    else:
        print(
            "WARNING: txs_features.csv not found."
        )

        print(
            "Skipping feature-level contribution analysis."
        )


# =============================================================================
# 14. STANDARDIZED FEATURE CONTRIBUTION
# =============================================================================

available_features = [
    f for f in behavioral_features
    if f in test_df.columns
]

if len(available_features) > 0:

    feature_matrix = (
        test_df[available_features]
        .apply(pd.to_numeric, errors="coerce")
    )

    feature_matrix = feature_matrix.replace(
        [np.inf, -np.inf],
        np.nan
    )

    feature_matrix = feature_matrix.fillna(
        feature_matrix.median()
    )

    # Dataset features were already standardized in the previous pipeline.
    # We still calculate robust z-score here to make contribution comparable.
    means = feature_matrix.mean()
    stds = feature_matrix.std().replace(0, 1)

    z = (
        feature_matrix - means
    ) / stds

    abs_z = z.abs()

    feature_importance = (
        abs_z
        .mean()
        .sort_values(ascending=False)
        .reset_index()
    )

    feature_importance.columns = [
        "Feature",
        "Mean_Absolute_Deviation"
    ]

    print("\nGlobal behavioral feature contribution:")
    print(
        feature_importance.to_string(index=False)
    )

else:

    feature_importance = pd.DataFrame(
        columns=[
            "Feature",
            "Mean_Absolute_Deviation"
        ]
    )


# =============================================================================
# 15. TOP TRANSACTION EXPLANATIONS
# =============================================================================

print("\n" + "=" * 100)
print("TOP TRANSACTION EXPLANATIONS")
print("=" * 100)

if len(available_features) > 0:

    # Get top 100 anomalous test transactions
    explain_df = (
        test_df
        .nlargest(100, score_col)
        .copy()
    )

    explain_matrix = (
        explain_df[available_features]
        .apply(pd.to_numeric, errors="coerce")
    )

    explain_matrix = explain_matrix.replace(
        [np.inf, -np.inf],
        np.nan
    )

    explain_matrix = explain_matrix.fillna(
        feature_matrix[available_features].median()
    )

    global_means = (
        feature_matrix[available_features].mean()
    )

    global_stds = (
        feature_matrix[available_features]
        .std()
        .replace(0, 1)
    )

    explain_z = (
        explain_matrix - global_means
    ) / global_stds

    explain_abs = explain_z.abs()

    # Top 3 contributing features per transaction
    top_feature_names = []

    for idx in explain_abs.index:

        row = explain_abs.loc[idx]

        top_features = (
            row
            .sort_values(ascending=False)
            .head(3)
            .index
            .tolist()
        )

        top_feature_names.append(
            ", ".join(top_features)
        )

    explain_df["Top_Contributing_Features"] = (
        top_feature_names
    )

    evidence_df = explain_df[
        [
            tx_col,
            time_col,
            class_col,
            score_col,
            "Top_Contributing_Features"
        ]
    ].copy()

    print(
        evidence_df.head(30).to_string(index=False)
    )

else:

    evidence_df = pd.DataFrame()


# =============================================================================
# 16. FINAL INVESTIGATION PRIORITY TABLE
# =============================================================================

print("\n" + "=" * 100)
print("BIT-TRACE — FINAL INVESTIGATION PRIORITY")
print("=" * 100)

priority_df = (
    test_df
    .nlargest(500, score_col)
    .copy()
)

priority_df["Anomaly_Rank"] = (
    np.arange(1, len(priority_df) + 1)
)

priority_df["Class3_Flag"] = (
    priority_df[class_col] == 3
).astype(int)

priority_columns = [
    "Anomaly_Rank",
    tx_col,
    time_col,
    class_col,
    score_col,
    "Class3_Flag"
]

priority_columns = [
    c for c in priority_columns
    if c in priority_df.columns
]

priority_df = priority_df[
    priority_columns
]


# =============================================================================
# 17. SAVE ALL RESULTS
# =============================================================================

print("\n" + "=" * 100)
print("SAVING FINAL TRANSACTION-ONLY RESULTS")
print("=" * 100)

output_dir = (
    r"C:\Users\YASHIKA\Desktop\bit-trace"
)

files_saved = []

def save_csv(dataframe, filename):

    path = os.path.join(
        output_dir,
        filename
    )

    dataframe.to_csv(
        path,
        index=False
    )

    files_saved.append(path)

    print("✓", path)


save_csv(
    topk_df,
    "bittrace_transaction_topk_analysis.csv"
)

save_csv(
    threshold_df,
    "bittrace_transaction_threshold_analysis.csv"
)

save_csv(
    temporal_class3.reset_index(),
    "bittrace_transaction_temporal_analysis.csv"
)

save_csv(
    class_summary.reset_index(),
    "bittrace_transaction_class_summary.csv"
)

save_csv(
    priority_df,
    "bittrace_transaction_investigation_priority.csv"
)

if len(feature_importance) > 0:

    save_csv(
        feature_importance,
        "bittrace_transaction_feature_importance.csv"
    )

if len(evidence_df) > 0:

    save_csv(
        evidence_df,
        "bittrace_transaction_explainability.csv"
    )


# =============================================================================
# 18. FINAL SUMMARY
# =============================================================================

print("\n" + "=" * 100)
print("BIT-TRACE — TRANSACTION-ONLY PIPELINE STATUS")
print("=" * 100)

print(f"""
Dataset
-------
Transactions             : {len(df):,}
Time steps               : {int(df[time_col].min())}–{int(df[time_col].max())}

Temporal split
-------------
Train                    : {(df["split"] == "train").sum():,}
Validation               : {(df["split"] == "validation").sum():,}
Test                     : {(df["split"] == "test").sum():,}

Models
------
Isolation Forest         : ✓
LOF                      : ✓
Ensemble                 : ✓
Graph features           : ✓
Behavioral features      : ✓

Post-hoc evaluation
-------------------
Class-3 ROC-AUC          : {roc_auc:.4f}
Class-3 PR-AUC           : {pr_auc:.4f}

Output
------
Top-K analysis           : ✓
Threshold analysis       : ✓
Temporal analysis        : ✓
Feature explainability   : {'✓' if len(feature_importance) > 0 else 'Not available'}
Investigation table      : ✓
""")

print("Saved files:")
for f in files_saved:
    print("✓", f)

print("\n" + "=" * 100)
print("TRANSACTION-ONLY ANALYSIS COMPLETE")
print("=" * 100)

print("""
Interpretation reminder:
The model is UNSUPERVISED.
Class labels are used only after scoring for evaluation/analysis.

The anomaly score should therefore be interpreted as:
"how unusual is this transaction according to the learned
behavioral + transaction-graph structure"

It should NOT automatically be interpreted as:
"this transaction is illicit."

Next stage after this:
investigator-facing evidence generation / dashboard integration.
P2P data is NOT used in this pipeline.
""")
```

```python
# =============================================================================
# BIT-TRACE — FINAL TRANSACTION-ONLY EVALUATION
# Uses SAVED anomaly results + txs_classes.csv
# Does NOT reload 662 MB feature file
# =============================================================================

import os
import numpy as np
import pandas as pd
from sklearn.metrics import roc_auc_score, average_precision_score

print("=" * 100)
print("BIT-TRACE — FINAL TRANSACTION-ONLY EVALUATION")
print("=" * 100)

# -------------------------------------------------------------------------
# PATHS
# -------------------------------------------------------------------------

RESULT_PATH = r"C:\Users\YASHIKA\Desktop\bit-trace\src\ellipticpp_graph_behavioral_anomaly_results.csv"

CLASS_PATH = r"C:\Users\YASHIKA\Desktop\bit-trace\EllipticPlusPlus\Transactions Dataset\txs_classes.csv"

OUTPUT_DIR = r"C:\Users\YASHIKA\Desktop\bit-trace"

# -------------------------------------------------------------------------
# 1. LOAD SAVED ANOMALY RESULTS
# -------------------------------------------------------------------------

print("\n[1] LOADING ANOMALY RESULTS")
print("-" * 80)

if not os.path.exists(RESULT_PATH):
    raise FileNotFoundError(RESULT_PATH)

results = pd.read_csv(RESULT_PATH)

print("Shape:", results.shape)
print("Columns:", results.columns.tolist())

# -------------------------------------------------------------------------
# 2. LOAD ONLY CLASS DATA
# -------------------------------------------------------------------------

print("\n[2] LOADING CLASS LABELS")
print("-" * 80)

if not os.path.exists(CLASS_PATH):
    raise FileNotFoundError(CLASS_PATH)

classes = pd.read_csv(CLASS_PATH)

classes.columns = [
    str(c).strip()
    for c in classes.columns
]

print("Shape:", classes.shape)
print("Columns:", classes.columns.tolist())

# Normalize names

if "txId" not in classes.columns:
    # try first column as transaction ID
    classes = classes.rename(
        columns={classes.columns[0]: "txId"}
    )

if "class" not in classes.columns:
    # try second column as class
    classes = classes.rename(
        columns={classes.columns[1]: "class"}
    )

classes["txId"] = pd.to_numeric(
    classes["txId"],
    errors="coerce"
)

classes["class"] = pd.to_numeric(
    classes["class"],
    errors="coerce"
)

classes = classes.dropna(
    subset=["txId", "class"]
)

print("\nClass distribution:")
print(
    classes["class"]
    .value_counts()
    .sort_index()
)

# -------------------------------------------------------------------------
# 3. MERGE LABELS
# -------------------------------------------------------------------------

print("\n[3] MERGING LABELS")
print("-" * 80)

results["txId"] = pd.to_numeric(
    results["txId"],
    errors="coerce"
)

results["Time step"] = pd.to_numeric(
    results["Time step"],
    errors="coerce"
)

results["ensemble_anomaly"] = pd.to_numeric(
    results["ensemble_anomaly"],
    errors="coerce"
)

results = results.dropna(
    subset=[
        "txId",
        "Time step",
        "ensemble_anomaly"
    ]
)

df = results.merge(
    classes[["txId", "class"]],
    on="txId",
    how="left"
)

print("Merged shape:", df.shape)
print(
    "Missing labels:",
    df["class"].isna().sum()
)

if df["class"].isna().sum() > 0:
    print("\nWARNING: Some transactions have no class label.")

df = df.dropna(
    subset=["class"]
).copy()

df["class"] = df["class"].astype(int)

print("\nFinal class distribution:")
print(
    df["class"]
    .value_counts()
    .sort_index()
)

# -------------------------------------------------------------------------
# 4. TEST SET
# -------------------------------------------------------------------------

print("\n[4] TEST SET")
print("-" * 80)

test_df = df[
    df["split"].astype(str).str.lower() == "test"
].copy()

if len(test_df) == 0:

    print(
        "Saved split column did not identify test rows."
    )

    # Recover from time steps
    test_df = df[
        df["Time step"] >= 42
    ].copy()

print("Test samples:", len(test_df))
print(
    "Test time:",
    test_df["Time step"].min(),
    "to",
    test_df["Time step"].max()
)

# -------------------------------------------------------------------------
# 5. CLASS-WISE SCORE ANALYSIS
# -------------------------------------------------------------------------

print("\n" + "=" * 100)
print("CLASS-WISE ANOMALY SCORE ANALYSIS")
print("=" * 100)

class_summary = (
    test_df
    .groupby("class")["ensemble_anomaly"]
    .agg(
        Samples="count",
        Mean="mean",
        Std="std",
        Min="min",
        Median="median",
        Max="max"
    )
)

print(
    class_summary.to_string()
)

# -------------------------------------------------------------------------
# 6. CLASS-3 ROC-AUC / PR-AUC
# -------------------------------------------------------------------------

print("\n" + "=" * 100)
print("CLASS-3 POST-HOC EVALUATION")
print("=" * 100)

y_true = (
    test_df["class"] == 3
).astype(int)

scores = test_df[
    "ensemble_anomaly"
].to_numpy()

roc_auc = roc_auc_score(
    y_true,
    scores
)

pr_auc = average_precision_score(
    y_true,
    scores
)

print(
    f"Test ROC-AUC : {roc_auc:.4f}"
)

print(
    f"Test PR-AUC  : {pr_auc:.4f}"
)

# -------------------------------------------------------------------------
# 7. TOP-K ANALYSIS
# -------------------------------------------------------------------------

print("\n" + "=" * 100)
print("TOP-K ANOMALY ANALYSIS")
print("=" * 100)

overall_class3_pct = (
    (test_df["class"] == 3).mean()
    * 100
)

print(
    f"Overall Test Class-3: "
    f"{overall_class3_pct:.2f}%"
)

topk_rows = []

for k in [50, 100, 250, 500, 1000, 2500, 5000]:

    k = min(k, len(test_df))

    topk = (
        test_df
        .nlargest(
            k,
            "ensemble_anomaly"
        )
    )

    c3 = (
        topk["class"] == 3
    ).sum()

    c3_pct = (
        c3 / k * 100
    )

    enrichment = (
        c3_pct /
        overall_class3_pct
    )

    topk_rows.append({
        "Top_K": k,
        "Class3_Count": int(c3),
        "Class3_%": c3_pct,
        "Enrichment": enrichment
    })

topk_df = pd.DataFrame(
    topk_rows
)

print(
    topk_df.to_string(index=False)
)

# -------------------------------------------------------------------------
# 8. TOP 100 ANOMALOUS TRANSACTIONS
# -------------------------------------------------------------------------

print("\n" + "=" * 100)
print("TOP 100 MOST ANOMALOUS TEST TRANSACTIONS")
print("=" * 100)

top100 = (
    test_df
    .nlargest(
        100,
        "ensemble_anomaly"
    )
    [
        [
            "txId",
            "Time step",
            "class",
            "IF_anomaly",
            "LOF_anomaly",
            "ensemble_anomaly"
        ]
    ]
)

print(
    top100.head(30).to_string(
        index=False
    )
)

# -------------------------------------------------------------------------
# 9. CLASS DISTRIBUTION OF TOP ANOMALIES
# -------------------------------------------------------------------------

print("\n" + "=" * 100)
print("CLASS DISTRIBUTION AMONG TOP ANOMALIES")
print("=" * 100)

for k in [50, 100, 500, 1000, 5000]:

    k = min(
        k,
        len(test_df)
    )

    topk = (
        test_df
        .nlargest(
            k,
            "ensemble_anomaly"
        )
    )

    counts = (
        topk["class"]
        .value_counts()
        .sort_index()
    )

    print(f"\nTop {k}:")
    print(counts)

# -------------------------------------------------------------------------
# 10. TEMPORAL ANALYSIS
# -------------------------------------------------------------------------

print("\n" + "=" * 100)
print("TEMPORAL ANOMALY ANALYSIS")
print("=" * 100)

temporal = (
    test_df
    .groupby("Time step")
    .agg(
        Transactions=("txId", "count"),
        Mean_Anomaly=(
            "ensemble_anomaly",
            "mean"
        ),
        Median_Anomaly=(
            "ensemble_anomaly",
            "median"
        ),
        Max_Anomaly=(
            "ensemble_anomaly",
            "max"
        ),
        Class3_Count=(
            "class",
            lambda x: (x == 3).sum()
        )
    )
)

temporal["Class3_%"] = (
    temporal["Class3_Count"]
    /
    temporal["Transactions"]
    * 100
)

temporal = temporal.sort_values(
    "Mean_Anomaly",
    ascending=False
)

print(
    temporal.head(15).to_string()
)

# -------------------------------------------------------------------------
# 11. ANOMALY THRESHOLDS
# -------------------------------------------------------------------------

print("\n" + "=" * 100)
print("UNSUPERVISED ANOMALY THRESHOLDS")
print("=" * 100)

threshold_rows = []

for percentile in [
    95,
    97,
    99,
    99.5,
    99.9
]:

    threshold = np.percentile(
        test_df["ensemble_anomaly"],
        percentile
    )

    flagged = test_df[
        test_df["ensemble_anomaly"]
        >= threshold
    ]

    class3_count = (
        flagged["class"] == 3
    ).sum()

    class3_pct = (
        class3_count /
        len(flagged) * 100
        if len(flagged) > 0
        else 0
    )

    threshold_rows.append({
        "Percentile": percentile,
        "Threshold": threshold,
        "Flagged": len(flagged),
        "Class3_Count": int(
            class3_count
        ),
        "Class3_%": class3_pct
    })

threshold_df = pd.DataFrame(
    threshold_rows
)

print(
    threshold_df.to_string(
        index=False
    )
)

# -------------------------------------------------------------------------
# 12. SAVE RESULTS
# -------------------------------------------------------------------------

print("\n" + "=" * 100)
print("SAVING FINAL RESULTS")
print("=" * 100)

outputs = {

    "bittrace_transaction_class_summary.csv":
        class_summary.reset_index(),

    "bittrace_transaction_topk_analysis.csv":
        topk_df,

    "bittrace_transaction_temporal_analysis.csv":
        temporal.reset_index(),

    "bittrace_transaction_threshold_analysis.csv":
        threshold_df,

    "bittrace_transaction_top100.csv":
        top100,

    "bittrace_transaction_final_evaluation.csv":
        test_df
}

for filename, data in outputs.items():

    path = os.path.join(
        OUTPUT_DIR,
        filename
    )

    data.to_csv(
        path,
        index=False
    )

    print("✓", path)

# -------------------------------------------------------------------------
# 13. FINAL STATUS
# -------------------------------------------------------------------------

print("\n" + "=" * 100)
print("BIT-TRACE — TRANSACTION-ONLY EVALUATION COMPLETE")
print("=" * 100)

print(f"""
Transactions evaluated : {len(test_df):,}
Test time steps         : 42–49

Class-3 ROC-AUC         : {roc_auc:.4f}
Class-3 PR-AUC          : {pr_auc:.4f}

Isolation Forest        : ✓
LOF                     : ✓
Graph features          : ✓
Behavioral features     : ✓
Ensemble score          : ✓
Temporal evaluation     : ✓
Top-K analysis          : ✓
Threshold analysis      : ✓

P2P / network data      : NOT USED
""")

print("=" * 100)
```

```python
# =============================================================================
# BIT-TRACE — TRANSACTION-ONLY EXPLAINABILITY + EVIDENCE ENGINE
# =============================================================================
# Purpose:
#   Convert transaction anomaly scores into investigator-readable evidence.
#
# Uses:
#   - Elliptic++ transaction features
#   - Transaction graph features
#   - Existing IF + LOF ensemble results
#
# Does NOT use:
#   - P2P/network observations
#   - Network crawl data
#   - Class labels for scoring/training
#
# Output:
#   1. Transaction evidence for top anomalous transactions
#   2. Feature-level explanations
#   3. Temporal context
#   4. Investigation priority
#   5. Dashboard-ready CSV
# =============================================================================

import os
import warnings
import numpy as np
import pandas as pd

warnings.filterwarnings("ignore")

print("=" * 100)
print("BIT-TRACE — TRANSACTION-ONLY EXPLAINABILITY + EVIDENCE ENGINE")
print("=" * 100)


# =============================================================================
# 1. PATHS
# =============================================================================

BASE = r"C:\Users\YASHIKA\Desktop\bit-trace"

FEATURE_PATH = (
    r"C:\Users\YASHIKA\Desktop\bit-trace"
    r"\data\raw\elliptic-plus-plus\txs_features.csv"
)

CLASS_PATH = (
    r"C:\Users\YASHIKA\Desktop\bit-trace"
    r"\EllipticPlusPlus\Transactions Dataset\txs_classes.csv"
)

EDGE_PATH = (
    r"C:\Users\YASHIKA\Desktop\bit-trace"
    r"\EllipticPlusPlus\Transactions Dataset\txs_edgelist.csv"
)

RESULT_PATH = os.path.join(
    BASE,
    "src",
    "ellipticpp_graph_behavioral_anomaly_results.csv"
)

FINAL_EVAL_PATH = os.path.join(
    BASE,
    "bittrace_transaction_final_evaluation.csv"
)

OUTPUT_DIR = os.path.join(BASE, "bittrace_evidence")

os.makedirs(OUTPUT_DIR, exist_ok=True)


# =============================================================================
# 2. SELECTED BEHAVIORAL FEATURES
# =============================================================================

BEHAVIOR_FEATURES = [
    "Local_feature_22",
    "Local_feature_25",
    "Local_feature_47",
    "Local_feature_52",
    "Local_feature_53",
    "Local_feature_54",
    "Local_feature_59",
    "Local_feature_77",
    "Local_feature_89",
    "Local_feature_90",
    "Aggregate_feature_14",
    "Aggregate_feature_21",
    "Aggregate_feature_32",
]

print("\n[1] SELECTED BEHAVIORAL FEATURES")
print("-" * 80)

for i, f in enumerate(BEHAVIOR_FEATURES, 1):
    print(f"{i:2d}. {f}")


# =============================================================================
# 3. CHECK FILES
# =============================================================================

print("\n[2] FILE CHECK")
print("-" * 80)

required_files = {
    "Feature": FEATURE_PATH,
    "Class": CLASS_PATH,
    "Edge": EDGE_PATH,
    "Anomaly Results": RESULT_PATH,
}

for name, path in required_files.items():
    exists = os.path.exists(path)
    print(f"{name:<18}: {exists}")
    if exists:
        print(f"Path               : {path}")

if not os.path.exists(RESULT_PATH):
    raise FileNotFoundError(
        f"Anomaly result file not found:\n{RESULT_PATH}"
    )


# =============================================================================
# 4. LOAD EXISTING ANOMALY RESULTS
# =============================================================================

print("\n[3] LOADING EXISTING ANOMALY RESULTS")
print("-" * 80)

results = pd.read_csv(RESULT_PATH)

print("Shape:", results.shape)
print("Columns:", list(results.columns))

required_result_columns = [
    "txId",
    "Time step",
    "IF_anomaly",
    "LOF_anomaly",
    "ensemble_anomaly",
    "split",
]

missing = [
    c for c in required_result_columns
    if c not in results.columns
]

if missing:
    raise ValueError(
        f"Missing columns in anomaly result file: {missing}"
    )

results["txId"] = pd.to_numeric(
    results["txId"], errors="coerce"
)

results["Time step"] = pd.to_numeric(
    results["Time step"], errors="coerce"
)

results = results.dropna(
    subset=["txId", "Time step", "ensemble_anomaly"]
).copy()

print("Clean result shape:", results.shape)


# =============================================================================
# 5. USE TEST SET ONLY
# =============================================================================

print("\n[4] SELECTING TEST SET")
print("-" * 80)

test_results = results[
    results["split"].astype(str).str.lower() == "test"
].copy()

if len(test_results) == 0:

    print("No explicit 'test' split found.")

    test_results = results[
        results["Time step"] >= 42
    ].copy()

print("Test transactions:", len(test_results))

print(
    "Test time range:",
    int(test_results["Time step"].min()),
    "to",
    int(test_results["Time step"].max())
)


# =============================================================================
# 6. LOAD TRANSACTION FEATURES
# =============================================================================

print("\n[5] LOADING TRANSACTION FEATURES")
print("-" * 80)

header = pd.read_csv(
    FEATURE_PATH,
    nrows=0
)

feature_columns = list(header.columns)

print("Total feature columns:", len(feature_columns))

required_columns = [
    "txId",
    "Time step",
] + BEHAVIOR_FEATURES

missing_features = [
    c for c in required_columns
    if c not in feature_columns
]

if missing_features:
    raise ValueError(
        f"Missing feature columns: {missing_features}"
    )

transaction_data = pd.read_csv(
    FEATURE_PATH,
    usecols=required_columns
)

print("Loaded shape:", transaction_data.shape)

transaction_data["txId"] = pd.to_numeric(
    transaction_data["txId"],
    errors="coerce"
)

transaction_data["Time step"] = pd.to_numeric(
    transaction_data["Time step"],
    errors="coerce"
)

for f in BEHAVIOR_FEATURES:
    transaction_data[f] = pd.to_numeric(
        transaction_data[f],
        errors="coerce"
    )


# =============================================================================
# 7. MERGE TEST RESULTS + FEATURES
# =============================================================================

print("\n[6] MERGING ANOMALY RESULTS + FEATURES")
print("-" * 80)

evidence = test_results.merge(
    transaction_data,
    on=["txId", "Time step"],
    how="left"
)

print("Merged shape:", evidence.shape)

missing_feature_rows = evidence[
    BEHAVIOR_FEATURES
].isna().any(axis=1).sum()

print(
    "Rows with missing behavioral features:",
    missing_feature_rows
)

if missing_feature_rows > 0:

    print("Applying median imputation...")

    for f in BEHAVIOR_FEATURES:

        median_value = transaction_data[f].median()

        evidence[f] = evidence[f].fillna(
            median_value
        )


# =============================================================================
# 8. LOAD GRAPH EDGE LIST
# =============================================================================

print("\n[7] LOADING TRANSACTION GRAPH")
print("-" * 80)

edges = pd.read_csv(EDGE_PATH)

print("Edge shape:", edges.shape)

edge_columns = list(edges.columns)

print("Edge columns:", edge_columns)

source_candidates = [
    "txId1",
    "source",
    "src",
    "from",
]

target_candidates = [
    "txId2",
    "target",
    "dst",
    "to",
]

source_col = next(
    (c for c in source_candidates if c in edge_columns),
    None
)

target_col = next(
    (c for c in target_candidates if c in edge_columns),
    None
)

if source_col is None or target_col is None:

    if len(edge_columns) >= 2:
        source_col = edge_columns[0]
        target_col = edge_columns[1]

    else:
        raise ValueError(
            "Could not identify graph edge columns."
        )

print("Source:", source_col)
print("Target:", target_col)


# =============================================================================
# 9. BUILD DEGREE FEATURES
# =============================================================================

print("\n[8] BUILDING GRAPH EVIDENCE")
print("-" * 80)

edges[source_col] = pd.to_numeric(
    edges[source_col],
    errors="coerce"
)

edges[target_col] = pd.to_numeric(
    edges[target_col],
    errors="coerce"
)

edges = edges.dropna(
    subset=[source_col, target_col]
).copy()

edges = edges[
    edges[source_col] != edges[target_col]
]

in_degree = (
    edges.groupby(target_col)
    .size()
)

out_degree = (
    edges.groupby(source_col)
    .size()
)

graph_evidence = pd.DataFrame({
    "txId": pd.Index(
        set(in_degree.index)
        .union(set(out_degree.index))
    )
})

graph_evidence["in_degree"] = (
    graph_evidence["txId"]
    .map(in_degree)
    .fillna(0)
)

graph_evidence["out_degree"] = (
    graph_evidence["txId"]
    .map(out_degree)
    .fillna(0)
)

graph_evidence["total_degree"] = (
    graph_evidence["in_degree"]
    + graph_evidence["out_degree"]
)

graph_evidence["degree_imbalance"] = (
    graph_evidence["in_degree"]
    - graph_evidence["out_degree"]
).abs()

graph_evidence["in_out_ratio"] = (
    (graph_evidence["in_degree"] + 1)
    /
    (graph_evidence["out_degree"] + 1)
)

print(
    "Graph nodes:",
    len(graph_evidence)
)


# =============================================================================
# 10. MERGE GRAPH FEATURES
# =============================================================================

evidence = evidence.merge(
    graph_evidence,
    on="txId",
    how="left"
)

GRAPH_FEATURES = [
    "in_degree",
    "out_degree",
    "total_degree",
    "degree_imbalance",
    "in_out_ratio",
]

for f in GRAPH_FEATURES:

    evidence[f] = evidence[f].fillna(0)


# =============================================================================
# 11. CREATE NORMAL BASELINE
# =============================================================================

print("\n[9] BUILDING NORMAL BEHAVIOR BASELINE")
print("-" * 80)

# Use training transactions from the original result set.
train_ids = set(
    results[
        results["split"].astype(str).str.lower() == "train"
    ]["txId"]
)

train_features = transaction_data[
    transaction_data["txId"].isin(train_ids)
].copy()

print("Training baseline transactions:", len(train_features))

if len(train_features) == 0:

    # Fallback
    train_features = transaction_data.copy()

    print(
        "Fallback baseline: complete transaction dataset"
    )


# =============================================================================
# 12. BEHAVIORAL Z-SCORES
# =============================================================================

behavior_stats = {}

for f in BEHAVIOR_FEATURES:

    mean_value = train_features[f].mean()

    std_value = train_features[f].std()

    if pd.isna(std_value) or std_value == 0:
        std_value = 1.0

    behavior_stats[f] = {
        "mean": mean_value,
        "std": std_value,
    }

    evidence[f"{f}_z"] = (
        evidence[f] - mean_value
    ) / std_value


# =============================================================================
# 13. GRAPH Z-SCORES
# =============================================================================

graph_stats = {}

for f in GRAPH_FEATURES:

    mean_value = train_features[f].mean() \
        if f in train_features.columns else evidence[f].mean()

    std_value = train_features[f].std() \
        if f in train_features.columns else evidence[f].std()

    if pd.isna(std_value) or std_value == 0:
        std_value = 1.0

    graph_stats[f] = {
        "mean": mean_value,
        "std": std_value,
    }

    evidence[f"{f}_z"] = (
        evidence[f] - mean_value
    ) / std_value


# =============================================================================
# 14. FEATURE DEVIATION
# =============================================================================

print("\n[10] CALCULATING FEATURE DEVIATIONS")
print("-" * 80)

ALL_EXPLAIN_FEATURES = (
    BEHAVIOR_FEATURES +
    GRAPH_FEATURES
)

Z_COLUMNS = [
    f"{f}_z"
    for f in ALL_EXPLAIN_FEATURES
]

evidence["max_feature_deviation"] = (
    evidence[Z_COLUMNS]
    .abs()
    .max(axis=1)
)

evidence["mean_feature_deviation"] = (
    evidence[Z_COLUMNS]
    .abs()
    .mean(axis=1)
)


# =============================================================================
# 15. TOP CONTRIBUTING FEATURES
# =============================================================================

def get_top_features(row, n=5):

    values = {}

    for feature in ALL_EXPLAIN_FEATURES:

        z_col = f"{feature}_z"

        value = row.get(z_col, np.nan)

        if pd.notna(value):
            values[feature] = abs(float(value))

    ranked = sorted(
        values.items(),
        key=lambda x: x[1],
        reverse=True
    )[:n]

    return ranked


top_feature_names = []
top_feature_scores = []

for _, row in evidence.iterrows():

    ranked = get_top_features(row, n=5)

    names = [
        x[0]
        for x in ranked
    ]

    scores = [
        round(x[1], 3)
        for x in ranked
    ]

    top_feature_names.append(
        " | ".join(names)
    )

    top_feature_scores.append(
        " | ".join(
            str(x)
            for x in scores
        )
    )

evidence["top_features"] = top_feature_names

evidence["top_feature_deviations"] = top_feature_scores


# =============================================================================
# 16. NORMALIZE ENSEMBLE SCORE
# =============================================================================

print("\n[11] NORMALIZING ANOMALY SCORE")
print("-" * 80)

score = evidence["ensemble_anomaly"]

score_min = score.min()
score_max = score.max()

if score_max == score_min:

    evidence["anomaly_percentile"] = 50.0

else:

    evidence["anomaly_percentile"] = (
        score.rank(pct=True) * 100
    )


# =============================================================================
# 17. INVESTIGATION LEVEL
# =============================================================================

def investigation_level(percentile):

    if percentile >= 99.5:
        return "CRITICAL"

    if percentile >= 99:
        return "HIGH"

    if percentile >= 95:
        return "MEDIUM"

    return "LOW"


evidence["investigation_level"] = (
    evidence["anomaly_percentile"]
    .apply(investigation_level)
)


# =============================================================================
# 18. TEMPORAL CONTEXT
# =============================================================================

print("\n[12] BUILDING TEMPORAL CONTEXT")
print("-" * 80)

temporal_stats = (
    evidence.groupby("Time step")
    .agg(
        temporal_mean_score=(
            "ensemble_anomaly",
            "mean"
        ),
        temporal_median_score=(
            "ensemble_anomaly",
            "median"
        ),
        temporal_transaction_count=(
            "txId",
            "count"
        ),
    )
    .reset_index()
)

evidence = evidence.merge(
    temporal_stats,
    on="Time step",
    how="left"
)

evidence["score_vs_time_mean"] = (
    evidence["ensemble_anomaly"]
    -
    evidence["temporal_mean_score"]
)


# =============================================================================
# 19. GRAPH EVIDENCE TEXT
# =============================================================================

def graph_reason(row):

    reasons = []

    total_degree = row["total_degree"]
    imbalance = row["degree_imbalance"]

    if total_degree >= 20:
        reasons.append(
            f"high transaction connectivity "
            f"(total degree={total_degree:.0f})"
        )

    if imbalance >= 10:
        reasons.append(
            f"strong in/out degree imbalance "
            f"(imbalance={imbalance:.0f})"
        )

    if row["in_out_ratio"] >= 5:
        reasons.append(
            "inbound connectivity substantially "
            "exceeds outbound connectivity"
        )

    if row["in_out_ratio"] <= 0.2:
        reasons.append(
            "outbound connectivity substantially "
            "exceeds inbound connectivity"
        )

    if len(reasons) == 0:
        reasons.append(
            "no extreme graph-degree pattern detected"
        )

    return "; ".join(reasons)


evidence["graph_evidence"] = (
    evidence.apply(
        graph_reason,
        axis=1
    )
)


# =============================================================================
# 20. BEHAVIORAL EVIDENCE TEXT
# =============================================================================

def behavioral_reason(row):

    ranked = get_top_features(
        row,
        n=3
    )

    reasons = []

    for feature, deviation in ranked:

        z = row[f"{feature}_z"]

        direction = (
            "above"
            if z > 0
            else "below"
        )

        reasons.append(
            f"{feature} {direction} training baseline "
            f"(abs-z={abs(z):.2f})"
        )

    if len(reasons) == 0:
        return "No strong behavioral deviation."

    return "; ".join(reasons)


evidence["behavioral_evidence"] = (
    evidence.apply(
        behavioral_reason,
        axis=1
    )
)


# =============================================================================
# 21. TEMPORAL EVIDENCE
# =============================================================================

def temporal_reason(row):

    diff = row["score_vs_time_mean"]

    if diff > 5:
        return (
            f"anomaly score substantially above "
            f"time-step {int(row['Time step'])} average"
        )

    if diff > 1:
        return (
            f"anomaly score above "
            f"time-step {int(row['Time step'])} average"
        )

    return (
        f"time-step {int(row['Time step'])} "
        f"has {int(row['temporal_transaction_count'])} "
        f"transactions in evaluation set"
    )


evidence["temporal_evidence"] = (
    evidence.apply(
        temporal_reason,
        axis=1
    )
)


# =============================================================================
# 22. FINAL EVIDENCE SUMMARY
# =============================================================================

def build_evidence(row):

    return (
        f"Anomaly percentile: "
        f"{row['anomaly_percentile']:.2f}%. "
        f"Behavioral evidence: "
        f"{row['behavioral_evidence']}. "
        f"Graph evidence: "
        f"{row['graph_evidence']}. "
        f"Temporal context: "
        f"{row['temporal_evidence']}."
    )


evidence["evidence_summary"] = (
    evidence.apply(
        build_evidence,
        axis=1
    )
)


# =============================================================================
# 23. INVESTIGATION PRIORITY SCORE
# =============================================================================

# This is NOT a probability of illicit activity.
# It is only an internal ranking signal for investigation UI.

evidence["investigation_priority"] = (
    0.60 * evidence["anomaly_percentile"]
    +
    0.25 * np.clip(
        evidence["max_feature_deviation"] * 10,
        0,
        100
    )
    +
    0.15 * np.clip(
        evidence["total_degree"],
        0,
        100
    )
)


# =============================================================================
# 24. SORT
# =============================================================================

evidence = evidence.sort_values(
    "investigation_priority",
    ascending=False
).reset_index(drop=True)


# =============================================================================
# 25. TOP 100 INVESTIGATION CASES
# =============================================================================

top100 = evidence.head(100).copy()

print("\n[13] TOP 20 INVESTIGATION CANDIDATES")
print("-" * 100)

display_columns = [
    "txId",
    "Time step",
    "ensemble_anomaly",
    "anomaly_percentile",
    "investigation_level",
    "top_features",
    "max_feature_deviation",
    "total_degree",
    "investigation_priority",
]

print(
    top100[
        display_columns
    ].head(20).to_string(
        index=False
    )
)


# =============================================================================
# 26. SUMMARY
# =============================================================================

print("\n[14] INVESTIGATION SUMMARY")
print("-" * 100)

print(
    "CRITICAL:",
    (evidence["investigation_level"] == "CRITICAL").sum()
)

print(
    "HIGH:",
    (evidence["investigation_level"] == "HIGH").sum()
)

print(
    "MEDIUM:",
    (evidence["investigation_level"] == "MEDIUM").sum()
)

print(
    "LOW:",
    (evidence["investigation_level"] == "LOW").sum()
)


# =============================================================================
# 27. DASHBOARD-READY COLUMNS
# =============================================================================

dashboard_columns = [
    "txId",
    "Time step",
    "split",

    "IF_anomaly",
    "LOF_anomaly",
    "ensemble_anomaly",

    "anomaly_percentile",
    "investigation_level",
    "investigation_priority",

    "top_features",
    "top_feature_deviations",

    "max_feature_deviation",
    "mean_feature_deviation",

    "in_degree",
    "out_degree",
    "total_degree",
    "degree_imbalance",
    "in_out_ratio",

    "temporal_mean_score",
    "temporal_median_score",
    "score_vs_time_mean",

    "behavioral_evidence",
    "graph_evidence",
    "temporal_evidence",
    "evidence_summary",
]

dashboard_columns = [
    c for c in dashboard_columns
    if c in evidence.columns
]

dashboard = evidence[
    dashboard_columns
].copy()


# =============================================================================
# 28. SAVE RESULTS
# =============================================================================

print("\n[15] SAVING EXPLAINABILITY RESULTS")
print("-" * 100)

full_path = os.path.join(
    OUTPUT_DIR,
    "bittrace_transaction_explainability_full.csv"
)

top100_path = os.path.join(
    OUTPUT_DIR,
    "bittrace_transaction_top100_evidence.csv"
)

dashboard_path = os.path.join(
    OUTPUT_DIR,
    "bittrace_transaction_dashboard_data.csv"
)

summary_path = os.path.join(
    OUTPUT_DIR,
    "bittrace_transaction_evidence_summary.csv"
)

dashboard.to_csv(
    full_path,
    index=False
)

top100.to_csv(
    top100_path,
    index=False
)

dashboard.head(1000).to_csv(
    dashboard_path,
    index=False
)

summary = pd.DataFrame({
    "metric": [
        "Total test transactions",
        "Critical cases",
        "High cases",
        "Medium cases",
        "Low cases",
        "Top evidence transactions",
    ],
    "value": [
        len(evidence),
        (evidence["investigation_level"] == "CRITICAL").sum(),
        (evidence["investigation_level"] == "HIGH").sum(),
        (evidence["investigation_level"] == "MEDIUM").sum(),
        (evidence["investigation_level"] == "LOW").sum(),
        min(100, len(evidence)),
    ],
})

summary.to_csv(
    summary_path,
    index=False
)


# =============================================================================
# 29. FINAL OUTPUT
# =============================================================================

print("\n" + "=" * 100)
print("BIT-TRACE — TRANSACTION EXPLAINABILITY COMPLETE")
print("=" * 100)

print(
    "\nTransactions analyzed:",
    len(evidence)
)

print(
    "Behavioral features:",
    len(BEHAVIOR_FEATURES)
)

print(
    "Graph evidence features:",
    len(GRAPH_FEATURES)
)

print(
    "\nSaved:"
)

print("✓", full_path)
print("✓", top100_path)
print("✓", dashboard_path)
print("✓", summary_path)

print("\n" + "=" * 100)
print("NEXT STEP")
print("=" * 100)

print(
    """
Use bittrace_transaction_top100_evidence.csv
for the investigator dashboard.

Each transaction now contains:

  Transaction ID
        ↓
  Ensemble anomaly score
        ↓
  Anomaly percentile
        ↓
  Investigation level
        ↓
  Top deviating behavioral features
        ↓
  Graph evidence
        ↓
  Temporal context
        ↓
  Human-readable evidence summary

No P2P/network data was used.
Class labels are NOT used to generate anomaly scores.
"""
)

print("=" * 100)
```

```python
# =============================================================================
# BIT-TRACE — LARGE-SCALE TRANSACTION-ONLY INVESTIGATION TEST
# =============================================================================
# Uses:
#   - Existing transaction anomaly results
#   - Elliptic++ transaction labels
#
# Does NOT use:
#   - P2P/network observations
#   - P2P IP/port data
#   - Any new ML training
#
# Purpose:
#   Test how BIT-TRACE prioritizes a large number of unseen transactions.
# =============================================================================

import os
import numpy as np
import pandas as pd

print("=" * 100)
print("BIT-TRACE — LARGE-SCALE TRANSACTION-ONLY INVESTIGATION TEST")
print("=" * 100)

# -----------------------------------------------------------------------------
# 1. PATHS
# -----------------------------------------------------------------------------

RESULT_PATH = r"C:\Users\YASHIKA\Desktop\bit-trace\src\ellipticpp_graph_behavioral_anomaly_results.csv"

CLASS_PATH = r"C:\Users\YASHIKA\Desktop\bit-trace\EllipticPlusPlus\Transactions Dataset\txs_classes.csv"

OUTPUT_DIR = r"C:\Users\YASHIKA\Desktop\bit-trace\bittrace_large_scale_test"

os.makedirs(OUTPUT_DIR, exist_ok=True)

print("\n[1] FILE CHECK")
print("-" * 80)

print("Anomaly results :", os.path.exists(RESULT_PATH))
print("Class labels    :", os.path.exists(CLASS_PATH))
print("Output folder   :", OUTPUT_DIR)

if not os.path.exists(RESULT_PATH):
    raise FileNotFoundError(
        f"Anomaly result file not found:\n{RESULT_PATH}"
    )

if not os.path.exists(CLASS_PATH):
    raise FileNotFoundError(
        f"Class file not found:\n{CLASS_PATH}"
    )

# -----------------------------------------------------------------------------
# 2. LOAD ANOMALY RESULTS
# -----------------------------------------------------------------------------

print("\n[2] LOADING TRANSACTION ANOMALY RESULTS")
print("-" * 80)

results = pd.read_csv(RESULT_PATH)

required_result_cols = [
    "txId",
    "Time step",
    "IF_anomaly",
    "LOF_anomaly",
    "ensemble_anomaly",
    "split"
]

missing = [
    c for c in required_result_cols
    if c not in results.columns
]

if missing:
    raise ValueError(
        f"Missing columns in anomaly result file: {missing}"
    )

print("Shape:", results.shape)
print("Columns:", results.columns.tolist())

# Remove accidental duplicate txIds if any
before = len(results)

results = (
    results
    .drop_duplicates(subset=["txId"])
    .copy()
)

print("Duplicate txIds removed:", before - len(results))

# -----------------------------------------------------------------------------
# 3. LOAD CLASS LABELS
# -----------------------------------------------------------------------------

print("\n[3] LOADING TRANSACTION LABELS")
print("-" * 80)

classes = pd.read_csv(CLASS_PATH)

# Normalize column names
classes.columns = [
    str(c).strip()
    for c in classes.columns
]

# Handle possible variations
tx_col = None
class_col = None

for c in classes.columns:
    if c.lower() == "txid":
        tx_col = c
    if c.lower() == "class":
        class_col = c

if tx_col is None or class_col is None:
    raise ValueError(
        f"Could not detect txId/class columns. Found: {classes.columns.tolist()}"
    )

classes = classes.rename(
    columns={
        tx_col: "txId",
        class_col: "class"
    }
)

classes = classes[["txId", "class"]].copy()

print("Class shape:", classes.shape)

print("\nClass distribution:")
print(classes["class"].value_counts().sort_index())

# -----------------------------------------------------------------------------
# 4. MERGE
# -----------------------------------------------------------------------------

print("\n[4] MERGING TRANSACTION RESULTS + LABELS")
print("-" * 80)

df = results.merge(
    classes,
    on="txId",
    how="left",
    validate="one_to_one"
)

missing_labels = df["class"].isna().sum()

print("Merged shape:", df.shape)
print("Missing labels:", missing_labels)

if missing_labels > 0:
    raise ValueError(
        f"{missing_labels} transactions have missing class labels."
    )

# -----------------------------------------------------------------------------
# 5. BASIC DATA VALIDATION
# -----------------------------------------------------------------------------

print("\n[5] DATA VALIDATION")
print("-" * 80)

print("Total transactions:", len(df))
print("Time range:",
      int(df["Time step"].min()),
      "to",
      int(df["Time step"].max()))

print("\nSplit distribution:")
print(df["split"].value_counts())

if df["ensemble_anomaly"].isna().any():
    raise ValueError("Missing ensemble anomaly scores found.")

print("\nAnomaly score statistics:")
print(
    df["ensemble_anomaly"]
    .describe()
    .to_string()
)

# -----------------------------------------------------------------------------
# 6. SELECT FINAL UNSEEN TEST SET
# -----------------------------------------------------------------------------

print("\n[6] SELECTING UNSEEN TEST SET")
print("-" * 80)

test_df = df[
    df["split"].astype(str).str.lower().eq("test")
].copy()

if len(test_df) == 0:
    raise ValueError(
        "No rows with split='test' found."
    )

test_df = test_df.sort_values(
    "ensemble_anomaly",
    ascending=False
).reset_index(drop=True)

print("Test transactions:", len(test_df))
print(
    "Test time:",
    int(test_df["Time step"].min()),
    "to",
    int(test_df["Time step"].max())
)

# -----------------------------------------------------------------------------
# 7. ANOMALY PERCENTILE
# -----------------------------------------------------------------------------

print("\n[7] CALCULATING ANOMALY PERCENTILES")
print("-" * 80)

# Highest score = highest anomaly percentile
test_df["anomaly_percentile"] = (
    test_df["ensemble_anomaly"]
    .rank(
        method="average",
        pct=True
    ) * 100
)

# -----------------------------------------------------------------------------
# 8. INVESTIGATION LEVEL
# -----------------------------------------------------------------------------

def investigation_level(percentile):
    if percentile >= 99.0:
        return "CRITICAL"
    elif percentile >= 95.0:
        return "HIGH"
    elif percentile >= 80.0:
        return "MEDIUM"
    else:
        return "LOW"

test_df["investigation_level"] = (
    test_df["anomaly_percentile"]
    .apply(investigation_level)
)

# -----------------------------------------------------------------------------
# 9. INVESTIGATION PRIORITY
# -----------------------------------------------------------------------------

# Priority is deliberately based only on anomaly percentile.
# No class labels are used for ranking.

test_df["investigation_priority"] = (
    test_df["anomaly_percentile"]
)

# -----------------------------------------------------------------------------
# 10. TOP-K ANALYSIS
# -----------------------------------------------------------------------------

print("\n[8] LARGE-SCALE TOP-K ANALYSIS")
print("-" * 80)

TOP_K_VALUES = [
    100,
    500,
    1000,
    5000,
    10000,
    20000
]

overall_class3_pct = (
    (test_df["class"] == 3).mean() * 100
)

topk_rows = []

for k in TOP_K_VALUES:

    k_actual = min(k, len(test_df))

    topk = test_df.head(k_actual)

    class3_count = (
        topk["class"] == 3
    ).sum()

    class3_pct = (
        class3_count / k_actual * 100
    )

    enrichment = (
        class3_pct / overall_class3_pct
        if overall_class3_pct > 0
        else np.nan
    )

    topk_rows.append({
        "Top_K": k_actual,
        "Transactions": k_actual,
        "Class1_Count": int((topk["class"] == 1).sum()),
        "Class2_Count": int((topk["class"] == 2).sum()),
        "Class3_Count": int(class3_count),
        "Class1_%": (topk["class"] == 1).mean() * 100,
        "Class2_%": (topk["class"] == 2).mean() * 100,
        "Class3_%": class3_pct,
        "Overall_Test_Class3_%": overall_class3_pct,
        "Class3_Enrichment": enrichment,
        "Mean_Anomaly": topk["ensemble_anomaly"].mean(),
        "Median_Anomaly": topk["ensemble_anomaly"].median(),
        "Max_Anomaly": topk["ensemble_anomaly"].max()
    })

topk_df = pd.DataFrame(topk_rows)

print(
    topk_df.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)

# -----------------------------------------------------------------------------
# 11. INVESTIGATION LEVEL DISTRIBUTION
# -----------------------------------------------------------------------------

print("\n[9] INVESTIGATION LEVEL DISTRIBUTION")
print("-" * 80)

level_counts = (
    test_df["investigation_level"]
    .value_counts()
    .reindex(
        ["CRITICAL", "HIGH", "MEDIUM", "LOW"],
        fill_value=0
    )
)

level_pct = (
    level_counts / len(test_df) * 100
)

level_df = pd.DataFrame({
    "Count": level_counts,
    "Percentage": level_pct
})

print(level_df.to_string())

# -----------------------------------------------------------------------------
# 12. TEMPORAL DISTRIBUTION OF HIGH-PRIORITY TRANSACTIONS
# -----------------------------------------------------------------------------

print("\n[10] TEMPORAL INVESTIGATION ANALYSIS")
print("-" * 80)

high_priority = test_df[
    test_df["anomaly_percentile"] >= 95
].copy()

temporal = (
    high_priority
    .groupby("Time step")
    .agg(
        High_Priority_Transactions=("txId", "count"),
        Mean_Anomaly=("ensemble_anomaly", "mean"),
        Median_Anomaly=("ensemble_anomaly", "median"),
        Max_Anomaly=("ensemble_anomaly", "max"),
        Class1_Count=("class", lambda x: (x == 1).sum()),
        Class2_Count=("class", lambda x: (x == 2).sum()),
        Class3_Count=("class", lambda x: (x == 3).sum())
    )
    .reset_index()
)

temporal["Class3_%"] = (
    temporal["Class3_Count"]
    / temporal["High_Priority_Transactions"]
    * 100
)

temporal = temporal.sort_values(
    "Mean_Anomaly",
    ascending=False
)

print(
    temporal.head(20).to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)

# -----------------------------------------------------------------------------
# 13. SCORE DISTRIBUTION BY INVESTIGATION LEVEL
# -----------------------------------------------------------------------------

print("\n[11] SCORE DISTRIBUTION BY INVESTIGATION LEVEL")
print("-" * 80)

level_scores = (
    test_df
    .groupby("investigation_level")["ensemble_anomaly"]
    .agg(
        Transactions="count",
        Mean="mean",
        Std="std",
        Min="min",
        Median="median",
        Max="max"
    )
    .reindex(
        ["CRITICAL", "HIGH", "MEDIUM", "LOW"]
    )
)

print(
    level_scores.to_string(
        float_format=lambda x: f"{x:.6f}"
    )
)

# -----------------------------------------------------------------------------
# 14. TOP 100 TRANSACTIONS
# -----------------------------------------------------------------------------

print("\n[12] TOP 100 TRANSACTIONS")
print("-" * 80)

top100 = test_df.head(100).copy()

display_columns = [
    "txId",
    "Time step",
    "ensemble_anomaly",
    "anomaly_percentile",
    "investigation_level",
    "investigation_priority",
    "class"
]

print(
    top100[display_columns]
    .head(30)
    .to_string(
        index=False,
        float_format=lambda x: f"{x:.6f}"
    )
)

# -----------------------------------------------------------------------------
# 15. SAVE TOP-K REPORTS
# -----------------------------------------------------------------------------

print("\n[13] SAVING LARGE-SCALE REPORTS")
print("-" * 80)

# Full test ranking
full_test_path = os.path.join(
    OUTPUT_DIR,
    "bittrace_transaction_test_full_ranking.csv"
)

test_df.to_csv(
    full_test_path,
    index=False
)

print("✓ Full test ranking:")
print(full_test_path)

# Top-K individual files
for k in TOP_K_VALUES:

    k_actual = min(k, len(test_df))

    path = os.path.join(
        OUTPUT_DIR,
        f"bittrace_transaction_top_{k_actual}.csv"
    )

    test_df.head(k_actual).to_csv(
        path,
        index=False
    )

    print(f"✓ Top {k_actual}: {path}")

# Summary
topk_path = os.path.join(
    OUTPUT_DIR,
    "bittrace_transaction_large_scale_topk_summary.csv"
)

topk_df.to_csv(
    topk_path,
    index=False
)

print("✓ Top-K summary:")
print(topk_path)

# Investigation levels
level_path = os.path.join(
    OUTPUT_DIR,
    "bittrace_transaction_investigation_levels.csv"
)

level_df.to_csv(
    level_path
)

print("✓ Investigation levels:")
print(level_path)

# Temporal analysis
temporal_path = os.path.join(
    OUTPUT_DIR,
    "bittrace_transaction_high_priority_temporal.csv"
)

temporal.to_csv(
    temporal_path,
    index=False
)

print("✓ Temporal analysis:")
print(temporal_path)

# -----------------------------------------------------------------------------
# 16. FINAL SUMMARY
# -----------------------------------------------------------------------------

print("\n" + "=" * 100)
print("BIT-TRACE — LARGE-SCALE TRANSACTION TEST COMPLETE")
print("=" * 100)

print(f"""
Total transactions in dataset : {len(df):,}
Unseen test transactions      : {len(test_df):,}
Test time range               : {int(test_df['Time step'].min())}–{int(test_df['Time step'].max())}

Models already used:
  ✓ Isolation Forest
  ✓ LOF
  ✓ Ensemble anomaly score

Transaction evidence:
  ✓ Behavioral anomaly score
  ✓ Transaction graph features
  ✓ Temporal context
  ✓ Anomaly percentile
  ✓ Investigation priority

Large-scale rankings generated:
  ✓ Top 100
  ✓ Top 500
  ✓ Top 1,000
  ✓ Top 5,000
  ✓ Top 10,000
  ✓ Top 20,000

P2P / network data:
  ✗ NOT USED

Labels:
  ✗ NOT used to generate anomaly scores
  ✓ Used only for post-hoc evaluation
""")

print("=" * 100)
print("NEXT STEP:")
print("Use the Top-K CSVs to demonstrate transaction-scale investigation.")
print("After this is validated, P2P correlation can be added as a separate layer.")
print("=" * 100)
```

```python
# =============================================================================
# BIT-TRACE — TRANSACTION-ONLY LARGE-SCALE FINAL TEST
# Elliptic++ | Isolation Forest + LOF | NO P2P DATA
# =============================================================================

import os
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd

from sklearn.ensemble import IsolationForest
from sklearn.neighbors import LocalOutlierFactor
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import roc_auc_score, average_precision_score

# -----------------------------------------------------------------------------
# CONFIGURATION
# -----------------------------------------------------------------------------

FEATURE_PATH = (
    r"C:\Users\YASHIKA\Desktop\bit-trace\data\raw"
    r"\elliptic-plus-plus\txs_features.csv"
)

CLASS_PATH = (
    r"C:\Users\YASHIKA\Desktop\bit-trace\EllipticPlusPlus"
    r"\Transactions Dataset\txs_classes.csv"
)

OUTPUT_DIR = (
    r"C:\Users\YASHIKA\Desktop\bit-trace"
    r"transaction_final_test"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

BEHAVIORAL_FEATURES = [
    "Local_feature_22",
    "Local_feature_25",
    "Local_feature_47",
    "Local_feature_52",
    "Local_feature_53",
    "Local_feature_54",
    "Local_feature_59",
    "Local_feature_77",
    "Local_feature_89",
    "Local_feature_90",
    "Aggregate_feature_14",
    "Aggregate_feature_21",
    "Aggregate_feature_32"
]

RANDOM_STATE = 42

# Temporal split
TRAIN_START = 1
TRAIN_END   = 34

VAL_START = 35
VAL_END   = 41

TEST_START = 42
TEST_END   = 49

print("=" * 100)
print("BIT-TRACE — TRANSACTION-ONLY LARGE-SCALE FINAL TEST")
print("=" * 100)

# =============================================================================
# 1. FILE CHECK
# =============================================================================

print("\n[1] FILE CHECK")
print("-" * 100)

if not os.path.exists(FEATURE_PATH):
    raise FileNotFoundError(f"Feature file not found:\n{FEATURE_PATH}")

if not os.path.exists(CLASS_PATH):
    raise FileNotFoundError(f"Class file not found:\n{CLASS_PATH}")

print("FEATURE :", FEATURE_PATH)
print("CLASS   :", CLASS_PATH)

# =============================================================================
# 2. LOAD CLASS LABELS
# =============================================================================

print("\n[2] LOADING CLASS LABELS")
print("-" * 100)

classes = pd.read_csv(CLASS_PATH)

classes.columns = [str(c).strip() for c in classes.columns]

if "txId" not in classes.columns:
    raise ValueError(f"txId not found in class file: {classes.columns.tolist()}")

if "class" not in classes.columns:
    raise ValueError(f"class not found in class file: {classes.columns.tolist()}")

classes["txId"] = pd.to_numeric(classes["txId"], errors="coerce")
classes["class"] = pd.to_numeric(classes["class"], errors="coerce")

classes = classes.dropna(subset=["txId", "class"]).copy()

classes["txId"] = classes["txId"].astype(np.int64)
classes["class"] = classes["class"].astype(int)

print("Shape:", classes.shape)
print("\nClass distribution:")
print(classes["class"].value_counts().sort_index())

# =============================================================================
# 3. READ FEATURE HEADER
# =============================================================================

print("\n[3] READING FEATURE HEADER")
print("-" * 100)

header = pd.read_csv(FEATURE_PATH, nrows=0)

header.columns = [str(c).strip() for c in header.columns]

print("Total columns:", len(header.columns))

required = ["txId", "Time step"] + BEHAVIORAL_FEATURES

missing = [c for c in required if c not in header.columns]

if missing:
    raise ValueError(
        "Missing required columns:\n" + "\n".join(missing)
    )

print("✓ All required transaction features found")

# =============================================================================
# 4. LOAD ONLY REQUIRED TRANSACTION FEATURES
# =============================================================================

print("\n[4] LOADING TRANSACTION FEATURES")
print("-" * 100)

data = pd.read_csv(
    FEATURE_PATH,
    usecols=required,
    low_memory=False
)

data.columns = [str(c).strip() for c in data.columns]

print("Loaded shape:", data.shape)

# =============================================================================
# 5. DATA CLEANING
# =============================================================================

print("\n[5] CLEANING TRANSACTION DATA")
print("-" * 100)

data["txId"] = pd.to_numeric(data["txId"], errors="coerce")
data["Time step"] = pd.to_numeric(data["Time step"], errors="coerce")

for col in BEHAVIORAL_FEATURES:
    data[col] = pd.to_numeric(data[col], errors="coerce")

data = data.dropna(subset=["txId", "Time step"]).copy()

data["txId"] = data["txId"].astype(np.int64)
data["Time step"] = data["Time step"].astype(int)

# Replace infinities
data[BEHAVIORAL_FEATURES] = (
    data[BEHAVIORAL_FEATURES]
    .replace([np.inf, -np.inf], np.nan)
)

# Median imputation
for col in BEHAVIORAL_FEATURES:
    median_value = data[col].median()

    if pd.isna(median_value):
        median_value = 0.0

    data[col] = data[col].fillna(median_value)

print("Rows:", len(data))
print("NaN:", data[BEHAVIORAL_FEATURES].isna().sum().sum())
print("Inf:", np.isinf(data[BEHAVIORAL_FEATURES].to_numpy()).sum())

# =============================================================================
# 6. MERGE LABELS
# =============================================================================

print("\n[6] MERGING CLASS LABELS")
print("-" * 100)

data = data.merge(
    classes[["txId", "class"]],
    on="txId",
    how="inner",
    validate="one_to_one"
)

print("Merged shape:", data.shape)
print("Missing labels:", data["class"].isna().sum())

print("\nFinal class distribution:")
print(data["class"].value_counts().sort_index())

# =============================================================================
# 7. SORT TEMPORALLY
# =============================================================================

print("\n[7] TEMPORAL ORDERING")
print("-" * 100)

data = data.sort_values(
    ["Time step", "txId"]
).reset_index(drop=True)

print("Time range:",
      data["Time step"].min(),
      "to",
      data["Time step"].max())

# =============================================================================
# 8. TEMPORAL SPLIT
# =============================================================================

print("\n[8] TEMPORAL SPLIT")
print("-" * 100)

train_mask = (
    (data["Time step"] >= TRAIN_START) &
    (data["Time step"] <= TRAIN_END)
)

val_mask = (
    (data["Time step"] >= VAL_START) &
    (data["Time step"] <= VAL_END)
)

test_mask = (
    (data["Time step"] >= TEST_START) &
    (data["Time step"] <= TEST_END)
)

train_idx = np.where(train_mask.to_numpy())[0]
val_idx   = np.where(val_mask.to_numpy())[0]
test_idx  = np.where(test_mask.to_numpy())[0]

print(
    f"Train      : {len(train_idx):,} "
    f"(time {TRAIN_START}–{TRAIN_END})"
)

print(
    f"Validation : {len(val_idx):,} "
    f"(time {VAL_START}–{VAL_END})"
)

print(
    f"Test       : {len(test_idx):,} "
    f"(time {TEST_START}–{TEST_END})"
)

# =============================================================================
# 9. CREATE FEATURE MATRICES
# =============================================================================

print("\n[9] BUILDING FEATURE MATRICES")
print("-" * 100)

X = data[BEHAVIORAL_FEATURES].to_numpy(dtype=np.float32)

X_train = X[train_idx]
X_val   = X[val_idx]
X_test  = X[test_idx]

print("X:", X.shape)
print("Features:", X.shape[1])

# =============================================================================
# 10. STANDARDIZATION
# =============================================================================

print("\n[10] STANDARDIZATION")
print("-" * 100)

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled   = scaler.transform(X_val)
X_test_scaled  = scaler.transform(X_test)

print("✓ Scaler fitted ONLY on training data")

# =============================================================================
# 11. ISOLATION FOREST
# =============================================================================

print("\n[11] TRAINING ISOLATION FOREST")
print("-" * 100)

iso = IsolationForest(
    n_estimators=300,
    contamination="auto",
    max_samples="auto",
    random_state=RANDOM_STATE,
    n_jobs=-1
)

iso.fit(X_train_scaled)

print("✓ Isolation Forest trained")
print("Trees:", 300)

# sklearn decision_function:
# higher = more normal
#
# For anomaly ranking:
# anomaly = -decision_function
#
# Therefore higher anomaly score = more anomalous.

if_train = -iso.decision_function(X_train_scaled)
if_val   = -iso.decision_function(X_val_scaled)
if_test  = -iso.decision_function(X_test_scaled)

# =============================================================================
# 12. LOF
# =============================================================================

print("\n[12] TRAINING LOF")
print("-" * 100)

# LOF is computationally expensive.
# It is trained only on the training set.
#
# n_neighbors=35 gives a reasonable local neighborhood.

lof = LocalOutlierFactor(
    n_neighbors=35,
    contamination="auto",
    novelty=True,
    n_jobs=-1
)

lof.fit(X_train_scaled)

print("✓ LOF trained")

lof_train_raw = -lof.decision_function(X_train_scaled)
lof_val_raw   = -lof.decision_function(X_val_scaled)
lof_test_raw  = -lof.decision_function(X_test_scaled)

# =============================================================================
# 13. ROBUST SCORE NORMALIZATION
# =============================================================================

print("\n[13] NORMALIZING MODEL SCORES")
print("-" * 100)

def robust_normalize(train_score, other_scores):
    """
    Normalize scores using training-set median and IQR.
    This avoids letting extreme outliers dominate the ensemble.
    """

    train_score = np.asarray(train_score, dtype=np.float64)

    q25, q75 = np.percentile(train_score, [25, 75])
    iqr = q75 - q25

    if iqr <= 1e-12:
        iqr = np.std(train_score)

    if iqr <= 1e-12:
        iqr = 1.0

    median = np.median(train_score)

    train_norm = (train_score - median) / iqr

    normalized = []

    for score in other_scores:
        score = np.asarray(score, dtype=np.float64)
        normalized.append((score - median) / iqr)

    return train_norm, normalized


if_train_norm, [if_val_norm, if_test_norm] = robust_normalize(
    if_train,
    [if_val, if_test]
)

lof_train_norm, [lof_val_norm, lof_test_norm] = robust_normalize(
    lof_train_raw,
    [lof_val_raw, lof_test_raw]
)

# =============================================================================
# 14. ENSEMBLE
# =============================================================================

print("\n[14] BUILDING ENSEMBLE")
print("-" * 100)

IF_WEIGHT = 0.60
LOF_WEIGHT = 0.40

ensemble_train = (
    IF_WEIGHT * if_train_norm +
    LOF_WEIGHT * lof_train_norm
)

ensemble_val = (
    IF_WEIGHT * if_val_norm +
    LOF_WEIGHT * lof_val_norm
)

ensemble_test = (
    IF_WEIGHT * if_test_norm +
    LOF_WEIGHT * lof_test_norm
)

print(f"Isolation Forest weight : {IF_WEIGHT}")
print(f"LOF weight              : {LOF_WEIGHT}")

# =============================================================================
# 15. BUILD COMPLETE RESULTS
# =============================================================================

print("\n[15] BUILDING FINAL RESULTS")
print("-" * 100)

results = data[
    ["txId", "Time step", "class"]
].copy()

results["split"] = "train"

results.loc[val_idx, "split"] = "validation"
results.loc[test_idx, "split"] = "test"

results["IF_anomaly"] = np.nan
results["LOF_anomaly"] = np.nan
results["ensemble_anomaly"] = np.nan

results.loc[train_idx, "IF_anomaly"] = if_train_norm
results.loc[val_idx, "IF_anomaly"]   = if_val_norm
results.loc[test_idx, "IF_anomaly"]  = if_test_norm

results.loc[train_idx, "LOF_anomaly"] = lof_train_norm
results.loc[val_idx, "LOF_anomaly"]   = lof_val_norm
results.loc[test_idx, "LOF_anomaly"]  = lof_test_norm

results.loc[train_idx, "ensemble_anomaly"] = ensemble_train
results.loc[val_idx, "ensemble_anomaly"]   = ensemble_val
results.loc[test_idx, "ensemble_anomaly"]  = ensemble_test

# =============================================================================
# 16. TEST SET
# =============================================================================

test_results = results[
    results["split"] == "test"
].copy()

test_results = test_results.sort_values(
    "ensemble_anomaly",
    ascending=False
).reset_index(drop=True)

print("\nTest transactions:", len(test_results))

# =============================================================================
# 17. CLASS-WISE ANALYSIS
# =============================================================================

print("\n" + "=" * 100)
print("CLASS-WISE TEST ANOMALY ANALYSIS")
print("=" * 100)

class_summary = (
    test_results
    .groupby("class")["ensemble_anomaly"]
    .agg(
        Samples="count",
        Mean="mean",
        Std="std",
        Min="min",
        Median="median",
        Max="max"
    )
)

print(class_summary)

# =============================================================================
# 18. POST-HOC CLASS-3 EVALUATION
# =============================================================================

print("\n" + "=" * 100)
print("POST-HOC CLASS-3 EVALUATION")
print("=" * 100)

y_test_class3 = (
    test_results["class"].to_numpy() == 3
).astype(int)

score_test = test_results[
    "ensemble_anomaly"
].to_numpy()

if len(np.unique(y_test_class3)) == 2:

    roc_auc = roc_auc_score(
        y_test_class3,
        score_test
    )

    pr_auc = average_precision_score(
        y_test_class3,
        score_test
    )

    print(f"Class-3 ROC-AUC : {roc_auc:.4f}")
    print(f"Class-3 PR-AUC  : {pr_auc:.4f}")

else:

    roc_auc = np.nan
    pr_auc = np.nan

    print("ROC-AUC/PR-AUC unavailable.")

# =============================================================================
# 19. TOP-K ANALYSIS
# =============================================================================

print("\n" + "=" * 100)
print("TOP-K ANOMALY ANALYSIS")
print("=" * 100)

overall_class3_pct = (
    y_test_class3.mean() * 100
)

top_k_values = [
    50,
    100,
    250,
    500,
    1000,
    2500,
    5000,
    10000
]

topk_rows = []

for k in top_k_values:

    k = min(k, len(test_results))

    top = test_results.iloc[:k]

    class3_count = (
        (top["class"] == 3).sum()
    )

    class3_pct = (
        class3_count / k * 100
    )

    enrichment = (
        class3_pct / overall_class3_pct
    )

    topk_rows.append({
        "Top_K": k,
        "Class3_Count": int(class3_count),
        "Class3_%": class3_pct,
        "Overall_Class3_%": overall_class3_pct,
        "Enrichment": enrichment
    })

topk_df = pd.DataFrame(topk_rows)

print(topk_df.to_string(index=False))

# =============================================================================
# 20. TOP 100
# =============================================================================

print("\n" + "=" * 100)
print("TOP 100 MOST ANOMALOUS TRANSACTIONS")
print("=" * 100)

top100 = test_results.head(100).copy()

print(
    top100[
        [
            "txId",
            "Time step",
            "class",
            "IF_anomaly",
            "LOF_anomaly",
            "ensemble_anomaly"
        ]
    ].to_string(index=False)
)

# =============================================================================
# 21. TEMPORAL ANALYSIS
# =============================================================================

print("\n" + "=" * 100)
print("TEMPORAL ANOMALY ANALYSIS")
print("=" * 100)

temporal = (
    test_results
    .groupby("Time step")
    .agg(
        Transactions=("txId", "count"),
        Mean_Anomaly=("ensemble_anomaly", "mean"),
        Median_Anomaly=("ensemble_anomaly", "median"),
        Max_Anomaly=("ensemble_anomaly", "max"),
        Class3_Count=("class", lambda x: (x == 3).sum())
    )
)

temporal["Class3_%"] = (
    temporal["Class3_Count"] /
    temporal["Transactions"] * 100
)

temporal = temporal.sort_values(
    "Mean_Anomaly",
    ascending=False
)

print(temporal)

# =============================================================================
# 22. UNSUPERVISED THRESHOLDS
# =============================================================================

print("\n" + "=" * 100)
print("UNSUPERVISED ANOMALY THRESHOLDS")
print("=" * 100)

threshold_rows = []

for percentile in [
    90,
    95,
    97,
    98,
    99,
    99.5,
    99.9
]:

    threshold = np.percentile(
        score_test,
        percentile
    )

    flagged = test_results[
        test_results["ensemble_anomaly"] >= threshold
    ]

    class3_count = (
        flagged["class"] == 3
    ).sum()

    class3_pct = (
        class3_count / len(flagged) * 100
        if len(flagged) > 0
        else 0
    )

    threshold_rows.append({
        "Percentile": percentile,
        "Threshold": threshold,
        "Flagged": len(flagged),
        "Class3_Count": int(class3_count),
        "Class3_%": class3_pct
    })

threshold_df = pd.DataFrame(threshold_rows)

print(threshold_df.to_string(index=False))

# =============================================================================
# 23. MODEL SCORE DISTRIBUTION
# =============================================================================

print("\n" + "=" * 100)
print("MODEL SCORE DISTRIBUTION")
print("=" * 100)

score_distribution = pd.DataFrame({
    "Model": [
        "Isolation Forest",
        "LOF",
        "Ensemble"
    ],
    "Mean": [
        np.mean(if_test_norm),
        np.mean(lof_test_norm),
        np.mean(ensemble_test)
    ],
    "Std": [
        np.std(if_test_norm),
        np.std(lof_test_norm),
        np.std(ensemble_test)
    ],
    "Min": [
        np.min(if_test_norm),
        np.min(lof_test_norm),
        np.min(ensemble_test)
    ],
    "Median": [
        np.median(if_test_norm),
        np.median(lof_test_norm),
        np.median(ensemble_test)
    ],
    "Max": [
        np.max(if_test_norm),
        np.max(lof_test_norm),
        np.max(ensemble_test)
    ]
})

print(score_distribution.to_string(index=False))

# =============================================================================
# 24. SAVE COMPLETE RESULT
# =============================================================================

print("\n" + "=" * 100)
print("SAVING FINAL TRANSACTION RESULTS")
print("=" * 100)

full_results_path = os.path.join(
    OUTPUT_DIR,
    "bittrace_transaction_large_scale_results.csv"
)

top100_path = os.path.join(
    OUTPUT_DIR,
    "bittrace_transaction_top100.csv"
)

topk_path = os.path.join(
    OUTPUT_DIR,
    "bittrace_transaction_topk.csv"
)

temporal_path = os.path.join(
    OUTPUT_DIR,
    "bittrace_transaction_temporal.csv"
)

threshold_path = os.path.join(
    OUTPUT_DIR,
    "bittrace_transaction_thresholds.csv"
)

class_summary_path = os.path.join(
    OUTPUT_DIR,
    "bittrace_transaction_class_summary.csv"
)

score_distribution_path = os.path.join(
    OUTPUT_DIR,
    "bittrace_transaction_score_distribution.csv"
)

results.to_csv(
    full_results_path,
    index=False
)

top100.to_csv(
    top100_path,
    index=False
)

topk_df.to_csv(
    topk_path,
    index=False
)

temporal.to_csv(
    temporal_path
)

threshold_df.to_csv(
    threshold_path,
    index=False
)

class_summary.to_csv(
    class_summary_path
)

score_distribution.to_csv(
    score_distribution_path,
    index=False
)

print("✓", full_results_path)
print("✓", top100_path)
print("✓", topk_path)
print("✓", temporal_path)
print("✓", threshold_path)
print("✓", class_summary_path)
print("✓", score_distribution_path)

# =============================================================================
# 25. FINAL SUMMARY
# =============================================================================

print("\n" + "=" * 100)
print("BIT-TRACE — TRANSACTION-ONLY LARGE-SCALE TEST COMPLETE")
print("=" * 100)

print(f"""
Transactions loaded       : {len(data):,}
Features used              : {len(BEHAVIORAL_FEATURES)}
Training transactions     : {len(train_idx):,}
Validation transactions   : {len(val_idx):,}
Test transactions         : {len(test_idx):,}

Training time             : {TRAIN_START}–{TRAIN_END}
Validation time           : {VAL_START}–{VAL_END}
Test time                 : {TEST_START}–{TEST_END}

Isolation Forest          : ✓
LOF                       : ✓
Ensemble                  : ✓
Temporal split            : ✓
Top-K analysis            : ✓
Threshold analysis        : ✓
Temporal analysis         : ✓
Post-hoc evaluation       : ✓

Class-3 ROC-AUC            : {roc_auc:.4f}
Class-3 PR-AUC             : {pr_auc:.4f}

P2P / network data        : NOT USED
Labels during training    : NOT USED
""")

print("=" * 100)
print("IMPORTANT:")
print("The class labels are used ONLY for post-hoc evaluation.")
print("They are NOT used to train Isolation Forest or LOF.")
print("=" * 100)
```

```python
# =============================================================================
# BIT-TRACE — TRANSACTION-ONLY LARGE-SCALE FINAL TEST
# 203,769 TRANSACTIONS
# =============================================================================

import os
import numpy as np
import pandas as pd

from pathlib import Path

print("=" * 100)
print("BIT-TRACE — TRANSACTION-ONLY LARGE-SCALE FINAL TEST")
print("=" * 100)

# -----------------------------------------------------------------------------
# 1. PATHS
# -----------------------------------------------------------------------------

BASE = Path(r"C:\Users\YASHIKA\Desktop\bit-trace")

RESULT_PATH = BASE / "src" / "ellipticpp_graph_behavioral_anomaly_results.csv"
CLASS_PATH = (
    BASE
    / "EllipticPlusPlus"
    / "Transactions Dataset"
    / "txs_classes.csv"
)

FEATURE_PATH = (
    BASE
    / "data"
    / "raw"
    / "elliptic-plus-plus"
    / "txs_features.csv"
)

OUTPUT_DIR = BASE / "bittrace_transaction_stress_test"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# -----------------------------------------------------------------------------
# 2. FILE CHECK
# -----------------------------------------------------------------------------

print("\n[1] FILE CHECK")
print("-" * 100)

for name, path in {
    "Anomaly results": RESULT_PATH,
    "Class labels": CLASS_PATH,
    "Feature data": FEATURE_PATH
}.items():

    print(f"{name:20}: {path.exists()}")

    if path.exists():
        print(f"Path                : {path}")

# -----------------------------------------------------------------------------
# 3. LOAD EXISTING ANOMALY RESULTS
# -----------------------------------------------------------------------------

print("\n[2] LOADING ANOMALY RESULTS")
print("-" * 100)

results = pd.read_csv(RESULT_PATH)

print("Shape:", results.shape)
print("Columns:", results.columns.tolist())

required_result_cols = [
    "txId",
    "Time step",
    "IF_anomaly",
    "LOF_anomaly",
    "ensemble_anomaly",
    "split"
]

missing = [
    c for c in required_result_cols
    if c not in results.columns
]

if missing:
    raise ValueError(f"Missing result columns: {missing}")

results = results.drop_duplicates("txId").copy()

print("Clean shape:", results.shape)

# -----------------------------------------------------------------------------
# 4. LOAD LABELS
# -----------------------------------------------------------------------------

print("\n[3] LOADING CLASS LABELS")
print("-" * 100)

classes = pd.read_csv(CLASS_PATH)

classes.columns = [
    str(c).strip()
    for c in classes.columns
]

print("Shape:", classes.shape)
print("Columns:", classes.columns.tolist())

classes = classes[["txId", "class"]].drop_duplicates("txId")

print("\nClass distribution:")
print(classes["class"].value_counts().sort_index())

# -----------------------------------------------------------------------------
# 5. MERGE
# -----------------------------------------------------------------------------

print("\n[4] MERGING RESULTS + LABELS")
print("-" * 100)

df = results.merge(
    classes,
    on="txId",
    how="left",
    validate="one_to_one"
)

print("Merged shape:", df.shape)
print("Missing labels:", df["class"].isna().sum())

if df["class"].isna().any():
    raise ValueError("Some transactions have missing class labels.")

# -----------------------------------------------------------------------------
# 6. VERIFY ALL 203K TRANSACTIONS
# -----------------------------------------------------------------------------

print("\n[5] FULL DATASET VERIFICATION")
print("-" * 100)

print("Total transactions:", len(df))

print(
    "Unique transactions:",
    df["txId"].nunique()
)

print(
    "Time range:",
    df["Time step"].min(),
    "to",
    df["Time step"].max()
)

print("\nSplit distribution:")
print(df["split"].value_counts())

print("\nClass distribution:")
print(df["class"].value_counts().sort_index())

# -----------------------------------------------------------------------------
# 7. CHECK SCORE QUALITY
# -----------------------------------------------------------------------------

print("\n[6] SCORE QUALITY CHECK")
print("-" * 100)

score_cols = [
    "IF_anomaly",
    "LOF_anomaly",
    "ensemble_anomaly"
]

for col in score_cols:

    print(f"\n{col}")

    print(
        "NaN:",
        df[col].isna().sum()
    )

    print(
        "Inf:",
        np.isinf(df[col].to_numpy()).sum()
    )

    print(
        "Min:",
        df[col].min()
    )

    print(
        "Median:",
        df[col].median()
    )

    print(
        "Max:",
        df[col].max()
    )

# -----------------------------------------------------------------------------
# 8. GLOBAL ANOMALY PERCENTILE
# -----------------------------------------------------------------------------

print("\n[7] CALCULATING GLOBAL ANOMALY PERCENTILE")
print("-" * 100)

df["anomaly_percentile"] = (
    df["ensemble_anomaly"]
    .rank(pct=True) * 100
)

# -----------------------------------------------------------------------------
# 9. INVESTIGATION LEVEL
# -----------------------------------------------------------------------------

def investigation_level(p):

    if p >= 99.5:
        return "CRITICAL"

    elif p >= 99:
        return "HIGH"

    elif p >= 95:
        return "MEDIUM"

    else:
        return "LOW"


df["investigation_level"] = (
    df["anomaly_percentile"]
    .apply(investigation_level)
)

print(
    df["investigation_level"]
    .value_counts()
)

# -----------------------------------------------------------------------------
# 10. FULL DATASET TOP ANOMALIES
# -----------------------------------------------------------------------------

print("\n[8] TOP ANOMALOUS TRANSACTIONS — FULL DATASET")
print("-" * 100)

top100 = (
    df
    .sort_values(
        "ensemble_anomaly",
        ascending=False
    )
    .head(100)
)

display_cols = [
    "txId",
    "Time step",
    "class",
    "split",
    "IF_anomaly",
    "LOF_anomaly",
    "ensemble_anomaly",
    "anomaly_percentile",
    "investigation_level"
]

print(
    top100[display_cols]
    .head(30)
    .to_string(index=False)
)

# -----------------------------------------------------------------------------
# 11. TOP-K DISTRIBUTION
# -----------------------------------------------------------------------------

print("\n[9] TOP-K ANALYSIS — FULL 203K DATASET")
print("-" * 100)

overall_class3 = (
    (df["class"] == 3).mean() * 100
)

topk_values = [
    50,
    100,
    250,
    500,
    1000,
    2500,
    5000,
    10000,
    25000
]

topk_rows = []

for k in topk_values:

    subset = (
        df
        .sort_values(
            "ensemble_anomaly",
            ascending=False
        )
        .head(k)
    )

    class3_count = (
        subset["class"] == 3
    ).sum()

    class3_percentage = (
        class3_count / k * 100
    )

    enrichment = (
        class3_percentage /
        overall_class3
    )

    topk_rows.append({

        "Top_K": k,

        "Samples": k,

        "Class1_Count":
            (subset["class"] == 1).sum(),

        "Class2_Count":
            (subset["class"] == 2).sum(),

        "Class3_Count":
            class3_count,

        "Class3_%":
            class3_percentage,

        "Overall_Class3_%":
            overall_class3,

        "Enrichment":
            enrichment
    })


topk_df = pd.DataFrame(topk_rows)

print(
    topk_df.to_string(index=False)
)

# -----------------------------------------------------------------------------
# 12. SCORE THRESHOLD ANALYSIS
# -----------------------------------------------------------------------------

print("\n[10] UNSUPERVISED THRESHOLD ANALYSIS")
print("-" * 100)

percentiles = [
    90,
    95,
    97,
    98,
    99,
    99.5,
    99.9
]

threshold_rows = []

for p in percentiles:

    threshold = np.percentile(
        df["ensemble_anomaly"],
        p
    )

    flagged = df[
        df["ensemble_anomaly"] >= threshold
    ]

    class3_count = (
        flagged["class"] == 3
    ).sum()

    threshold_rows.append({

        "Percentile": p,

        "Threshold":
            threshold,

        "Flagged":
            len(flagged),

        "Flagged_%":
            len(flagged) / len(df) * 100,

        "Class1_Count":
            (flagged["class"] == 1).sum(),

        "Class2_Count":
            (flagged["class"] == 2).sum(),

        "Class3_Count":
            class3_count,

        "Class3_%":
            class3_count / len(flagged) * 100
            if len(flagged) else 0
    })


threshold_df = pd.DataFrame(threshold_rows)

print(
    threshold_df.to_string(index=False)
)

# -----------------------------------------------------------------------------
# 13. TEMPORAL ANALYSIS
# -----------------------------------------------------------------------------

print("\n[11] TEMPORAL STRESS TEST")
print("-" * 100)

temporal = (
    df
    .groupby("Time step")
    .agg(

        Transactions=(
            "txId",
            "count"
        ),

        Mean_Anomaly=(
            "ensemble_anomaly",
            "mean"
        ),

        Median_Anomaly=(
            "ensemble_anomaly",
            "median"
        ),

        Max_Anomaly=(
            "ensemble_anomaly",
            "max"
        ),

        Std_Anomaly=(
            "ensemble_anomaly",
            "std"
        ),

        Class3_Count=(
            "class",
            lambda x: (x == 3).sum()
        )
    )
)

temporal["Class3_%"] = (
    temporal["Class3_Count"] /
    temporal["Transactions"] *
    100
)

temporal = (
    temporal
    .sort_values(
        "Mean_Anomaly",
        ascending=False
    )
)

print(
    temporal.head(20).to_string()
)

# -----------------------------------------------------------------------------
# 14. SPLIT-WISE ANALYSIS
# -----------------------------------------------------------------------------

print("\n[12] TRAIN / VALIDATION / TEST ANALYSIS")
print("-" * 100)

split_summary = (
    df
    .groupby("split")
    .agg(

        Transactions=(
            "txId",
            "count"
        ),

        Mean_Anomaly=(
            "ensemble_anomaly",
            "mean"
        ),

        Median_Anomaly=(
            "ensemble_anomaly",
            "median"
        ),

        Max_Anomaly=(
            "ensemble_anomaly",
            "max"
        ),

        P95=(
            "ensemble_anomaly",
            lambda x: np.percentile(x, 95)
        ),

        P99=(
            "ensemble_anomaly",
            lambda x: np.percentile(x, 99)
        )
    )
)

print(
    split_summary.to_string()
)

# -----------------------------------------------------------------------------
# 15. HIGH PRIORITY INVESTIGATION QUEUE
# -----------------------------------------------------------------------------

print("\n[13] INVESTIGATION QUEUE")
print("-" * 100)

investigation_queue = (
    df[
        df["investigation_level"]
        .isin(["CRITICAL", "HIGH"])
    ]
    .sort_values(
        "ensemble_anomaly",
        ascending=False
    )
    .copy()
)

print(
    "CRITICAL + HIGH:",
    len(investigation_queue)
)

print(
    "\nDistribution:"
)

print(
    investigation_queue[
        "investigation_level"
    ].value_counts()
)

# -----------------------------------------------------------------------------
# 16. TRANSACTION SUMMARY
# -----------------------------------------------------------------------------

print("\n[14] FINAL TRANSACTION SUMMARY")
print("-" * 100)

summary = {

    "total_transactions":
        len(df),

    "unique_transactions":
        df["txId"].nunique(),

    "time_steps":
        df["Time step"].nunique(),

    "min_time":
        df["Time step"].min(),

    "max_time":
        df["Time step"].max(),

    "critical":
        (
            df["investigation_level"]
            == "CRITICAL"
        ).sum(),

    "high":
        (
            df["investigation_level"]
            == "HIGH"
        ).sum(),

    "medium":
        (
            df["investigation_level"]
            == "MEDIUM"
        ).sum(),

    "low":
        (
            df["investigation_level"]
            == "LOW"
        ).sum()
}

for key, value in summary.items():

    print(
        f"{key:25}: {value}"
    )

# -----------------------------------------------------------------------------
# 17. SAVE FULL STRESS TEST RESULTS
# -----------------------------------------------------------------------------

print("\n[15] SAVING RESULTS")
print("-" * 100)

full_path = (
    OUTPUT_DIR /
    "bittrace_transaction_full_stress_test.csv"
)

topk_path = (
    OUTPUT_DIR /
    "bittrace_transaction_large_scale_topk.csv"
)

threshold_path = (
    OUTPUT_DIR /
    "bittrace_transaction_large_scale_thresholds.csv"
)

temporal_path = (
    OUTPUT_DIR /
    "bittrace_transaction_large_scale_temporal.csv"
)

split_path = (
    OUTPUT_DIR /
    "bittrace_transaction_large_scale_split_summary.csv"
)

queue_path = (
    OUTPUT_DIR /
    "bittrace_transaction_investigation_queue.csv"
)

summary_path = (
    OUTPUT_DIR /
    "bittrace_transaction_large_scale_summary.csv"
)

df.to_csv(
    full_path,
    index=False
)

topk_df.to_csv(
    topk_path,
    index=False
)

threshold_df.to_csv(
    threshold_path,
    index=False
)

temporal.to_csv(
    temporal_path
)

split_summary.to_csv(
    split_path
)

investigation_queue.to_csv(
    queue_path,
    index=False
)

pd.DataFrame(
    [summary]
).to_csv(
    summary_path,
    index=False
)

print("✓ Full results     :", full_path)
print("✓ Top-K analysis   :", topk_path)
print("✓ Thresholds       :", threshold_path)
print("✓ Temporal         :", temporal_path)
print("✓ Split summary    :", split_path)
print("✓ Investigation    :", queue_path)
print("✓ Summary          :", summary_path)

# -----------------------------------------------------------------------------
# 18. FINAL STATUS
# -----------------------------------------------------------------------------

print("\n")
print("=" * 100)
print("BIT-TRACE — LARGE-SCALE TRANSACTION TEST COMPLETE")
print("=" * 100)

print(f"""
Transactions tested       : {len(df):,}
Unique transactions       : {df['txId'].nunique():,}
Time steps                : {df['Time step'].min()}–{df['Time step'].max()}

Models already used:
✓ Isolation Forest
✓ LOF
✓ Ensemble anomaly score

Features:
✓ Behavioral features
✓ Graph transaction features
✓ Temporal information

Testing:
✓ Full 203K transaction dataset
✓ Top-K anomaly ranking
✓ Threshold testing
✓ Temporal stress testing
✓ Investigation queue generation

P2P / network data:
✗ NOT USED

Labels:
✗ NOT USED FOR MODEL TRAINING
✓ Used only for post-hoc evaluation

Next:
Use the generated investigation queue
and dashboard data for transaction-level investigation.
""")

print("=" * 100)
```

```python
# =============================================================================
# BIT-TRACE — TRANSACTION-ONLY LARGE-SCALE TEST / STRESS EVALUATION
# =============================================================================

import os
import time
import gc
import numpy as np
import pandas as pd

from sklearn.metrics import roc_auc_score, average_precision_score

print("=" * 100)
print("BIT-TRACE — TRANSACTION-ONLY LARGE-SCALE TEST")
print("=" * 100)

# -----------------------------------------------------------------------------
# 1. PATHS
# -----------------------------------------------------------------------------

BASE = r"C:\Users\YASHIKA\Desktop\bit-trace"

ANOMALY_PATH = os.path.join(
    BASE,
    "src",
    "ellipticpp_graph_behavioral_anomaly_results.csv"
)

CLASS_PATH = os.path.join(
    BASE,
    "EllipticPlusPlus",
    "Transactions Dataset",
    "txs_classes.csv"
)

OUTPUT_DIR = os.path.join(
    BASE,
    "bittrace_large_scale_test"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

print("\n[1] FILE CHECK")
print("-" * 80)

print("Anomaly results:", os.path.exists(ANOMALY_PATH))
print("Class labels   :", os.path.exists(CLASS_PATH))

if not os.path.exists(ANOMALY_PATH):
    raise FileNotFoundError(ANOMALY_PATH)

if not os.path.exists(CLASS_PATH):
    raise FileNotFoundError(CLASS_PATH)


# -----------------------------------------------------------------------------
# 2. LOAD EXISTING ANOMALY RESULTS
# -----------------------------------------------------------------------------

print("\n[2] LOADING EXISTING TRANSACTION RESULTS")
print("-" * 80)

start_time = time.perf_counter()

results = pd.read_csv(ANOMALY_PATH)

print("Shape:", results.shape)
print("Columns:", list(results.columns))

required_result_cols = [
    "txId",
    "Time step",
    "IF_anomaly",
    "LOF_anomaly",
    "ensemble_anomaly",
    "split"
]

missing = [
    c for c in required_result_cols
    if c not in results.columns
]

if missing:
    raise ValueError(f"Missing columns in anomaly results: {missing}")

print("✓ Required columns verified")


# -----------------------------------------------------------------------------
# 3. LOAD CLASS LABELS
# -----------------------------------------------------------------------------

print("\n[3] LOADING CLASS LABELS")
print("-" * 80)

classes = pd.read_csv(CLASS_PATH)

classes.columns = [
    str(c).strip()
    for c in classes.columns
]

print("Shape:", classes.shape)
print("Columns:", list(classes.columns))

if "txId" not in classes.columns or "class" not in classes.columns:
    raise ValueError(
        "Class file must contain txId and class columns."
    )

classes = classes[["txId", "class"]].copy()

classes["txId"] = pd.to_numeric(
    classes["txId"],
    errors="coerce"
)

classes["class"] = pd.to_numeric(
    classes["class"],
    errors="coerce"
)

classes = classes.dropna(
    subset=["txId", "class"]
)

classes["txId"] = classes["txId"].astype(np.int64)
classes["class"] = classes["class"].astype(np.int8)

print("Class distribution:")
print(classes["class"].value_counts().sort_index())


# -----------------------------------------------------------------------------
# 4. MERGE
# -----------------------------------------------------------------------------

print("\n[4] MERGING RESULTS + LABELS")
print("-" * 80)

df = results.merge(
    classes,
    on="txId",
    how="left",
    validate="one_to_one"
)

missing_labels = df["class"].isna().sum()

print("Merged shape:", df.shape)
print("Missing labels:", missing_labels)

if missing_labels > 0:
    raise ValueError(
        f"{missing_labels} transactions have no class label."
    )

df["class"] = df["class"].astype(np.int8)

print("✓ Labels successfully attached")


# -----------------------------------------------------------------------------
# 5. SELECT TEST SET ONLY
# -----------------------------------------------------------------------------

print("\n[5] SELECTING FULL TEST SET")
print("-" * 80)

test_df = df[
    df["split"].astype(str).str.lower().eq("test")
].copy()

if len(test_df) == 0:
    raise ValueError(
        "No rows found with split='test'."
    )

test_df = test_df.sort_values(
    "ensemble_anomaly",
    ascending=False
).reset_index(drop=True)

print("Test transactions:", len(test_df))
print(
    "Test time range:",
    test_df["Time step"].min(),
    "to",
    test_df["Time step"].max()
)

print("\nTest class distribution:")
print(test_df["class"].value_counts().sort_index())


# -----------------------------------------------------------------------------
# 6. VERIFY ALL TEST TRANSACTIONS
# -----------------------------------------------------------------------------

print("\n[6] DATA QUALITY CHECK")
print("-" * 80)

score_cols = [
    "IF_anomaly",
    "LOF_anomaly",
    "ensemble_anomaly"
]

for col in score_cols:
    nan_count = test_df[col].isna().sum()
    inf_count = np.isinf(
        test_df[col].to_numpy(dtype=float)
    ).sum()

    print(
        f"{col:20s} | NaN={nan_count:,} | Inf={inf_count:,}"
    )

if test_df[score_cols].isna().any().any():
    raise ValueError("Missing anomaly scores detected.")

if np.isinf(
    test_df[score_cols]
    .to_numpy(dtype=float)
).any():
    raise ValueError("Infinite anomaly scores detected.")

print("✓ All test scores valid")


# -----------------------------------------------------------------------------
# 7. BASIC LARGE-SCALE STATISTICS
# -----------------------------------------------------------------------------

print("\n[7] LARGE-SCALE SCORE STATISTICS")
print("-" * 80)

stats = test_df[score_cols].describe().T

print(stats.to_string())


# -----------------------------------------------------------------------------
# 8. CLASS-WISE ANALYSIS
# -----------------------------------------------------------------------------

print("\n[8] CLASS-WISE TEST ANALYSIS")
print("-" * 80)

class_summary = (
    test_df
    .groupby("class")["ensemble_anomaly"]
    .agg(
        Samples="count",
        Mean="mean",
        Std="std",
        Min="min",
        Median="median",
        Max="max"
    )
)

print(class_summary.to_string())


# -----------------------------------------------------------------------------
# 9. CLASS-3 POST-HOC METRICS
# -----------------------------------------------------------------------------

print("\n[9] CLASS-3 POST-HOC EVALUATION")
print("-" * 80)

y_true = (
    test_df["class"].to_numpy() == 3
).astype(np.int8)

scores = test_df[
    "ensemble_anomaly"
].to_numpy(dtype=np.float64)

try:
    roc_auc = roc_auc_score(
        y_true,
        scores
    )

    pr_auc = average_precision_score(
        y_true,
        scores
    )

    print(f"ROC-AUC : {roc_auc:.4f}")
    print(f"PR-AUC  : {pr_auc:.4f}")

except Exception as e:
    roc_auc = np.nan
    pr_auc = np.nan

    print("Metric calculation failed:", e)


# -----------------------------------------------------------------------------
# 10. TOP-K ANALYSIS
# -----------------------------------------------------------------------------

print("\n[10] TOP-K LARGE-SCALE ANALYSIS")
print("-" * 80)

overall_class3_pct = (
    y_true.mean() * 100
)

top_k_values = [
    50,
    100,
    250,
    500,
    1000,
    2500,
    5000,
    10000,
    15000,
    20000
]

top_k_values = [
    k for k in top_k_values
    if k <= len(test_df)
]

topk_rows = []

for k in top_k_values:

    topk = test_df.head(k)

    class3_count = (
        (topk["class"] == 3)
        .sum()
    )

    class3_pct = (
        class3_count / k * 100
    )

    enrichment = (
        class3_pct /
        overall_class3_pct
    )

    topk_rows.append({
        "Top_K": k,
        "Transactions": k,
        "Class3_Count": int(class3_count),
        "Class3_%": class3_pct,
        "Overall_Class3_%": overall_class3_pct,
        "Enrichment": enrichment
    })

topk_df = pd.DataFrame(topk_rows)

print(
    topk_df.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)


# -----------------------------------------------------------------------------
# 11. UNSUPERVISED THRESHOLD ANALYSIS
# -----------------------------------------------------------------------------

print("\n[11] THRESHOLD / ALERT-VOLUME ANALYSIS")
print("-" * 80)

percentiles = [
    90,
    92,
    95,
    97,
    98,
    99,
    99.5,
    99.9
]

threshold_rows = []

for p in percentiles:

    threshold = np.percentile(
        scores,
        p
    )

    flagged = test_df[
        test_df["ensemble_anomaly"] >= threshold
    ]

    flagged_count = len(flagged)

    class3_count = (
        flagged["class"] == 3
    ).sum()

    class3_pct = (
        class3_count / flagged_count * 100
        if flagged_count > 0
        else 0
    )

    alert_rate = (
        flagged_count /
        len(test_df) * 100
    )

    threshold_rows.append({
        "Percentile": p,
        "Threshold": threshold,
        "Flagged": flagged_count,
        "Alert_Rate_%": alert_rate,
        "Class3_Count": int(class3_count),
        "Class3_%": class3_pct
    })

threshold_df = pd.DataFrame(
    threshold_rows
)

print(
    threshold_df.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)


# -----------------------------------------------------------------------------
# 12. TEMPORAL SCALABILITY ANALYSIS
# -----------------------------------------------------------------------------

print("\n[12] TEMPORAL TEST ANALYSIS")
print("-" * 80)

temporal_df = (
    test_df
    .groupby("Time step")
    .agg(
        Transactions=("txId", "count"),
        Mean_Anomaly=("ensemble_anomaly", "mean"),
        Median_Anomaly=("ensemble_anomaly", "median"),
        Max_Anomaly=("ensemble_anomaly", "max"),
        Class3_Count=("class", lambda x: (x == 3).sum())
    )
)

temporal_df["Class3_%"] = (
    temporal_df["Class3_Count"] /
    temporal_df["Transactions"] *
    100
)

temporal_df = temporal_df.sort_values(
    "Mean_Anomaly",
    ascending=False
)

print(
    temporal_df.to_string(
        float_format=lambda x: f"{x:.4f}"
    )
)


# -----------------------------------------------------------------------------
# 13. TOP 1000 TRANSACTIONS
# -----------------------------------------------------------------------------

print("\n[13] SAVING TOP SUSPICIOUS TRANSACTIONS")
print("-" * 80)

top1000 = test_df.head(
    min(1000, len(test_df))
).copy()

top1000_path = os.path.join(
    OUTPUT_DIR,
    "transaction_test_top1000.csv"
)

top1000.to_csv(
    top1000_path,
    index=False
)

print("✓", top1000_path)


# -----------------------------------------------------------------------------
# 14. SAVE ALL TEST TRANSACTIONS
# -----------------------------------------------------------------------------

all_test_path = os.path.join(
    OUTPUT_DIR,
    "transaction_test_all_36824.csv"
)

test_df.to_csv(
    all_test_path,
    index=False
)

print("✓", all_test_path)


# -----------------------------------------------------------------------------
# 15. SAVE TOP-K RESULTS
# -----------------------------------------------------------------------------

topk_path = os.path.join(
    OUTPUT_DIR,
    "transaction_test_topk_analysis.csv"
)

topk_df.to_csv(
    topk_path,
    index=False
)

print("✓", topk_path)


# -----------------------------------------------------------------------------
# 16. SAVE THRESHOLD RESULTS
# -----------------------------------------------------------------------------

threshold_path = os.path.join(
    OUTPUT_DIR,
    "transaction_test_threshold_analysis.csv"
)

threshold_df.to_csv(
    threshold_path,
    index=False
)

print("✓", threshold_path)


# -----------------------------------------------------------------------------
# 17. SAVE TEMPORAL RESULTS
# -----------------------------------------------------------------------------

temporal_path = os.path.join(
    OUTPUT_DIR,
    "transaction_test_temporal_analysis.csv"
)

temporal_df.to_csv(
    temporal_path
)

print("✓", temporal_path)


# -----------------------------------------------------------------------------
# 18. SAVE SUMMARY
# -----------------------------------------------------------------------------

elapsed = time.perf_counter() - start_time

summary = pd.DataFrame([{
    "Total_Test_Transactions": len(test_df),
    "Test_Time_Start": test_df["Time step"].min(),
    "Test_Time_End": test_df["Time step"].max(),
    "Class1_Count": int((test_df["class"] == 1).sum()),
    "Class2_Count": int((test_df["class"] == 2).sum()),
    "Class3_Count": int((test_df["class"] == 3).sum()),
    "Class3_Percentage": overall_class3_pct,
    "ROC_AUC_Class3": roc_auc,
    "PR_AUC_Class3": pr_auc,
    "Processing_Time_Seconds": elapsed
}])

summary_path = os.path.join(
    OUTPUT_DIR,
    "transaction_test_summary.csv"
)

summary.to_csv(
    summary_path,
    index=False
)

print("✓", summary_path)


# -----------------------------------------------------------------------------
# 19. FINAL REPORT
# -----------------------------------------------------------------------------

print("\n")
print("=" * 100)
print("BIT-TRACE — LARGE-SCALE TRANSACTION TEST COMPLETE")
print("=" * 100)

print(f"""
Transactions evaluated : {len(test_df):,}
Test time steps         : {test_df["Time step"].min()}–{test_df["Time step"].max()}

Class distribution:
  Class 1 : {(test_df["class"] == 1).sum():,}
  Class 2 : {(test_df["class"] == 2).sum():,}
  Class 3 : {(test_df["class"] == 3).sum():,}

Class-3 percentage      : {overall_class3_pct:.2f}%

ROC-AUC                 : {roc_auc:.4f}
PR-AUC                  : {pr_auc:.4f}

Processing time         : {elapsed:.2f} seconds

P2P / network data      : NOT USED
Transaction data        : USED
Graph transaction edges : USED
Behavioral features    : USED
Isolation Forest        : USED
LOF                     : USED
Ensemble                : USED

Outputs saved to:
{OUTPUT_DIR}
""")

print("=" * 100)

# Memory cleanup
gc.collect()

print("✓ Large-scale transaction-only test finished.")
```
