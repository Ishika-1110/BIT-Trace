```python
import zipfile
from collections import Counter

with zipfile.ZipFile(ZIP_PATH, "r") as z:

    with z.open(sample_file) as f:

        field_counts = Counter()
        total_lines = 0

        for raw_line in f:

            line = raw_line.decode("utf-8", errors="replace").rstrip("\r\n")

            if not line:
                continue

            fields = line.split(",")

            field_counts[len(fields)] += 1
            total_lines += 1

print("Total rows:", total_lines)

print("\nField-count distribution:")

for fields, count in sorted(field_counts.items()):
    percentage = count / total_lines * 100

    print(
        f"{fields} fields : "
        f"{count:,} rows "
        f"({percentage:.2f}%)"
    )
```

```python
import zipfile
import json
import pandas as pd

pilot_output = os.path.join(
    PROCESSED_DIR,
    "p2p_observations_20200215.parquet"
)

rows = []

with zipfile.ZipFile(ZIP_PATH, "r") as z:

    for i, json_file in enumerate(pilot_files, start=1):

        timestamp = snapshot_datetime(json_file)

        with z.open(json_file) as f:
            snapshot = json.load(f)

        for record in snapshot:

            node_info, port, service, height = record
            node_id, ip = node_info

            rows.append({
                "timestamp": timestamp,
                "node_id": node_id,
                "ip": ip,
                "port": port,
                "service": service,
                "reported_height": height
            })

        if i % 50 == 0:
            print(
                f"Processed {i}/{len(pilot_files)} snapshots | "
                f"rows={len(rows):,}"
            )

pilot_df = pd.DataFrame(rows)

pilot_df.to_parquet(
    pilot_output,
    index=False
)

print("\nFinished.")
print("Rows:", len(pilot_df))
print("Output:", pilot_output)
print("Size MB:", os.path.getsize(pilot_output) / (1024**2))
```

```python
import zipfile
import json
import os
import pandas as pd
import pyarrow as pa
import pyarrow.parquet as pq

# ---------------------------------------------------------
# Output schema
# ---------------------------------------------------------

schema = pa.schema([
    ("timestamp", pa.timestamp("ns")),
    ("node_id", pa.string()),
    ("ip", pa.string()),
    ("port", pa.int64()),
    ("service", pa.int64()),
    ("reported_height", pa.int64()),
])

# ---------------------------------------------------------
# Group snapshots by date
# ---------------------------------------------------------

snapshots_by_date = {}

for name in crawl_jsons_sorted:

    parts = name.split("/")
    date = parts[2]

    snapshots_by_date.setdefault(date, []).append(name)

print("Dates to process:", len(snapshots_by_date))
print("Total snapshots:",
      sum(len(v) for v in snapshots_by_date.values()))

# ---------------------------------------------------------
# Process ZIP
# ---------------------------------------------------------

metadata_rows = []

with zipfile.ZipFile(ZIP_PATH, "r") as z:

    for date_index, (date, files) in enumerate(
        sorted(snapshots_by_date.items()),
        start=1
    ):

        output_file = os.path.join(
            OBS_DIR,
            f"p2p_observations_{date}.parquet"
        )

        print("\n" + "=" * 80)
        print(
            f"[{date_index}/{len(snapshots_by_date)}] "
            f"Processing {date}"
        )
        print("Snapshots:", len(files))
        print("=" * 80)

        writer = None
        total_rows = 0

        try:

            for i, json_file in enumerate(files, start=1):

                timestamp = snapshot_datetime(json_file)

                with z.open(json_file) as f:
                    snapshot = json.load(f)

                rows = []

                heights = []

                for record in snapshot:

                    node_info, port, service, height = record

                    node_id, ip = node_info

                    rows.append({
                        "timestamp": timestamp,
                        "node_id": node_id,
                        "ip": ip,
                        "port": port,
                        "service": service,
                        "reported_height": height
                    })

                    if isinstance(height, int) and height > 0:
                        heights.append(height)

                # -----------------------------------------
                # Calculate snapshot reference height
                # -----------------------------------------

                reference_height = (
                    int(pd.Series(heights).median())
                    if heights
                    else None
                )

                node_ids = [
                    row["node_id"]
                    for row in rows
                ]

                unique_nodes = len(set(node_ids))

                duplicate_nodes = (
                    len(node_ids) - unique_nodes
                )

                # -----------------------------------------
                # Record metadata
                # -----------------------------------------

                metadata_rows.append({
                    "timestamp": timestamp,
                    "date": date,
                    "snapshot_file": json_file,
                    "record_count": len(rows),
                    "unique_node_count": unique_nodes,
                    "duplicate_node_records": duplicate_nodes,
                    "reference_height": reference_height
                })

                # -----------------------------------------
                # Convert only this snapshot to Arrow
                # -----------------------------------------

                df = pd.DataFrame(rows)

                table = pa.Table.from_pandas(
                    df,
                    schema=schema,
                    preserve_index=False
                )

                if writer is None:

                    writer = pq.ParquetWriter(
                        output_file,
                        schema,
                        compression="snappy"
                    )

                writer.write_table(table)

                total_rows += len(df)

                # Explicitly release snapshot objects
                del rows
                del df
                del table
                del snapshot

                if i % 50 == 0 or i == len(files):

                    print(
                        f"  {i:3d}/{len(files)} snapshots | "
                        f"rows={total_rows:,}"
                    )

        finally:

            if writer is not None:
                writer.close()

        print(
            f"Finished {date}: "
            f"{total_rows:,} rows"
        )

        print(
            f"File size: "
            f"{os.path.getsize(output_file) / (1024**2):.2f} MB"
        )

# ---------------------------------------------------------
# Save snapshot metadata
# ---------------------------------------------------------

metadata_df = pd.DataFrame(metadata_rows)

metadata_file = os.path.join(
    META_DIR,
    "p2p_snapshot_metadata.parquet"
)

metadata_df.to_parquet(
    metadata_file,
    index=False
)

print("\n" + "=" * 80)
print("FULL EXTRACTION COMPLETE")
print("=" * 80)

print("Total snapshots:", len(metadata_df))

print(
    "Total observations:",
    metadata_df["record_count"].sum()
)

print(
    "Metadata file:",
    metadata_file
)
```

```python
feature_output = os.path.join(
    PROCESSED_DIR,
    "p2p_peer_features_v1.parquet"
)

query = f"""
WITH observations AS (

    SELECT
        o.timestamp,
        o.node_id,
        o.ip,
        o.port,
        o.service,
        o.reported_height,

        m.reference_height,

        o.reported_height - m.reference_height
            AS height_deviation,

        ABS(
            o.reported_height - m.reference_height
        )
            AS absolute_height_deviation

    FROM read_parquet('{observation_glob}') o

    LEFT JOIN read_parquet('{metadata_glob}') m
        ON o.timestamp = m.timestamp
)

SELECT

    node_id,

    -- ---------------------------------------------
    -- Activity
    -- ---------------------------------------------

    COUNT(*) AS observation_count,

    COUNT(DISTINCT DATE(timestamp))
        AS active_days,

    COUNT(
        DISTINCT DATE_TRUNC('hour', timestamp)
    )
        AS active_hours,

    COUNT(DISTINCT timestamp)
        AS active_snapshots,

    MIN(timestamp)
        AS first_observed,

    MAX(timestamp)
        AS last_observed,

    DATE_DIFF(
        'second',
        MIN(timestamp),
        MAX(timestamp)
    )
        AS observation_span_seconds,

    -- ---------------------------------------------
    -- Endpoint behaviour
    -- ---------------------------------------------

    COUNT(DISTINCT ip)
        AS unique_ips,

    COUNT(DISTINCT port)
        AS unique_ports,

    COUNT(DISTINCT service)
        AS unique_services,

    COUNT(
        DISTINCT CAST(ip AS VARCHAR)
        || ':'
        || CAST(port AS VARCHAR)
    )
        AS unique_endpoints,

    -- ---------------------------------------------
    -- Port behaviour
    -- ---------------------------------------------

    COUNT(
        CASE
            WHEN port != 8333 THEN 1
        END
    )
        AS non_default_port_observations,

    COUNT(DISTINCT
        CASE
            WHEN port != 8333 THEN port
        END
    )
        AS non_default_ports,

    -- ---------------------------------------------
    -- Service behaviour
    -- ---------------------------------------------

    COUNT(
        CASE
            WHEN service != 1037 THEN 1
        END
    )
        AS non_1037_service_observations,

    -- ---------------------------------------------
    -- Height behaviour
    -- ---------------------------------------------

    AVG(reported_height)
        AS mean_reported_height,

    MIN(reported_height)
        AS min_reported_height,

    MAX(reported_height)
        AS max_reported_height,

    AVG(height_deviation)
        AS mean_height_deviation,

    AVG(absolute_height_deviation)
        AS mean_absolute_height_deviation,

    MAX(absolute_height_deviation)
        AS max_absolute_height_deviation,

    COUNT(
        CASE
            WHEN absolute_height_deviation > 100 THEN 1
        END
    )
        AS height_deviation_over_100,

    COUNT(
        CASE
            WHEN absolute_height_deviation > 1000 THEN 1
        END
    )
        AS height_deviation_over_1000,

    COUNT(
        CASE
            WHEN absolute_height_deviation > 10000 THEN 1
        END
    )
        AS height_deviation_over_10000

FROM observations

GROUP BY node_id
"""

print("Building P2P peer feature table...")

con.execute(
    f"""
    COPY (
        {query}
    )
    TO '{feature_output}'
    (FORMAT PARQUET, COMPRESSION ZSTD)
    """
)

print("\n✅ Feature table created:")
print(feature_output)
```

```python
# ============================================================
# STEP 54C — CORRECTED LIGHTWEIGHT TEMPORAL AGGREGATION
# ============================================================

temporal_basic_output = os.path.join(
    PROCESSED_DIR,
    "p2p_peer_temporal_basic_v1.parquet"
)

con_temporal.execute("SET memory_limit='5GB'")
con_temporal.execute("SET threads=1")
con_temporal.execute("SET preserve_insertion_order=false")

print("Building lightweight temporal features...")
print("RAM limit: 5 GB")
print("Threads: 1")

query_basic = f"""
WITH hourly_counts AS (

    SELECT
        node_id,
        DATE_TRUNC('hour', timestamp) AS hour_bucket,
        COUNT(*) AS observations_in_hour

    FROM read_parquet('{observation_glob}')

    GROUP BY
        node_id,
        DATE_TRUNC('hour', timestamp)
),

hourly_stats AS (

    SELECT
        node_id,

        COUNT(*) AS active_hours,

        AVG(observations_in_hour)
            AS mean_observations_per_hour,

        MEDIAN(observations_in_hour)
            AS median_observations_per_hour,

        MAX(observations_in_hour)
            AS max_observations_per_hour,

        STDDEV_SAMP(observations_in_hour)
            AS hourly_activity_std,

        VAR_SAMP(observations_in_hour)
            AS hourly_activity_variance

    FROM hourly_counts

    GROUP BY node_id
),

daily_counts AS (

    SELECT
        node_id,
        DATE(timestamp) AS day_bucket,
        COUNT(*) AS observations_in_day

    FROM read_parquet('{observation_glob}')

    GROUP BY
        node_id,
        DATE(timestamp)
),

daily_stats AS (

    SELECT
        node_id,

        COUNT(*) AS active_days,

        AVG(observations_in_day)
            AS mean_observations_per_day,

        MEDIAN(observations_in_day)
            AS median_observations_per_day,

        MAX(observations_in_day)
            AS max_observations_per_day,

        STDDEV_SAMP(observations_in_day)
            AS daily_activity_std

    FROM daily_counts

    GROUP BY node_id
)

SELECT
    h.node_id,

    h.active_hours,
    h.mean_observations_per_hour,
    h.median_observations_per_hour,
    h.max_observations_per_hour,
    h.hourly_activity_std,
    h.hourly_activity_variance,

    d.active_days,
    d.mean_observations_per_day,
    d.median_observations_per_day,
    d.max_observations_per_day,
    d.daily_activity_std

FROM hourly_stats h

LEFT JOIN daily_stats d
    ON h.node_id = d.node_id
"""

con_temporal.execute(
    f"""
    COPY (
        {query_basic}
    )
    TO '{temporal_basic_output}'
    (FORMAT PARQUET, COMPRESSION ZSTD)
    """
)

print("\n✅ Basic temporal features created:")
print(temporal_basic_output)
```

```python
import pandas as pd
import numpy as np

PROCESSED_DIR = "/content/drive/MyDrive/Bitcoin P2P Dataset/processed"

# Load data
ml_matrix = pd.read_parquet(
    f"{PROCESSED_DIR}/p2p_ml_matrix_v2.parquet"
)

scores = pd.read_parquet(
    f"{PROCESSED_DIR}/p2p_isolation_forest_v1_scores.parquet"
)

df = ml_matrix.merge(
    scores[["node_id", "anomaly_score", "anomaly_rank"]],
    on="node_id",
    how="inner"
)

# ---------------------------------------------------------
# Population percentile for each feature
# ---------------------------------------------------------

feature_cols = [
    c for c in ml_matrix.columns
    if c != "node_id"
]

percentiles = df[feature_cols].rank(pct=True)

# ---------------------------------------------------------
# Feature -> BIT-TRACE reason-code mapping
# ---------------------------------------------------------

reason_map = {
    "observation_count": "HIGH_OBSERVATION_ACTIVITY",
    "active_days": "HIGH_TEMPORAL_PERSISTENCE",
    "active_hours": "HIGH_TEMPORAL_PERSISTENCE",
    "observation_span_seconds": "LONG_OBSERVATION_SPAN",

    "unique_ports": "HIGH_PORT_DIVERSITY",
    "unique_services": "HIGH_SERVICE_DIVERSITY",

    "non_default_port_observations":
        "HIGH_NON_DEFAULT_PORT_ACTIVITY",

    "non_default_ports":
        "HIGH_NON_DEFAULT_PORT_DIVERSITY",

    "non_1037_service_observations":
        "HIGH_NON_DEFAULT_SERVICE_ACTIVITY",

    "min_reported_height":
        "UNUSUAL_LOW_REPORTED_HEIGHT",

    "max_reported_height":
        "UNUSUAL_HIGH_REPORTED_HEIGHT",

    "mean_height_deviation":
        "UNUSUAL_MEAN_HEIGHT_DEVIATION",

    "mean_absolute_height_deviation":
        "HIGH_HEIGHT_DEVIATION",

    "max_absolute_height_deviation":
        "EXTREME_HEIGHT_DEVIATION",

    "height_deviation_over_100":
        "FREQUENT_HEIGHT_DEVIATION",

    "height_deviation_over_1000":
        "LARGE_HEIGHT_DEVIATION",

    "height_deviation_over_10000":
        "EXTREME_HEIGHT_DEVIATION",

    "mean_observations_per_hour":
        "HIGH_HOURLY_ACTIVITY",

    "max_observations_per_hour":
        "HIGH_HOURLY_BURST",

    "hourly_activity_std":
        "HIGH_HOURLY_ACTIVITY_VARIABILITY",

    "hourly_activity_variance":
        "HIGH_HOURLY_ACTIVITY_VARIABILITY",

    "mean_observations_per_day":
        "HIGH_DAILY_ACTIVITY",

    "max_observations_per_day":
        "HIGH_DAILY_BURST",

    "daily_activity_std":
        "HIGH_DAILY_ACTIVITY_VARIABILITY",

    "hourly_burst_ratio":
        "HIGH_HOURLY_BURST",

    "daily_burst_ratio":
        "HIGH_DAILY_BURST",

    "hourly_activity_cv":
        "HIGH_HOURLY_ACTIVITY_VARIABILITY",

    "daily_activity_cv":
        "HIGH_DAILY_ACTIVITY_VARIABILITY",
}

# ---------------------------------------------------------
# Generate reason codes
# ---------------------------------------------------------

reason_rows = []

for idx, row in df.iterrows():

    reasons = []

    for feature in feature_cols:

        percentile = percentiles.loc[idx, feature]

        # Extreme upper-tail behavior
        if percentile >= 0.99:
            if feature in reason_map:
                reasons.append(reason_map[feature])

        # Extreme lower-tail behavior
        elif percentile <= 0.01:

            # Only use lower-tail reason codes where
            # low values are actually meaningful.
            if feature == "min_reported_height":
                reasons.append("UNUSUAL_LOW_REPORTED_HEIGHT")

            elif feature == "max_reported_height":
                reasons.append("UNUSUAL_LOW_REPORTED_HEIGHT")

            elif feature == "mean_height_deviation":
                reasons.append("UNUSUAL_NEGATIVE_HEIGHT_DEVIATION")

            elif feature == "mean_absolute_height_deviation":
                reasons.append("LOW_HEIGHT_VARIABILITY")

    # Remove duplicate reason codes while preserving order
    reasons = list(dict.fromkeys(reasons))

    reason_rows.append({
        "node_id": row["node_id"],
        "anomaly_score": row["anomaly_score"],
        "anomaly_rank": row["anomaly_rank"],
        "reason_codes": reasons
    })

reason_df = pd.DataFrame(reason_rows)

# ---------------------------------------------------------
# Save explainability layer
# ---------------------------------------------------------

output_path = (
    f"{PROCESSED_DIR}/p2p_isolation_forest_v1_explanations.parquet"
)

reason_df.to_parquet(
    output_path,
    index=False
)

print("✅ Reason-code layer created")
print("Shape:", reason_df.shape)
print("Saved:", output_path)

print("\nTop 20 explainable anomalies:")
display(
    reason_df
    .sort_values("anomaly_score", ascending=False)
    .head(20)
)
```

```python
import zipfile
import os
import pandas as pd
import numpy as np
from collections import Counter

ZIP_PATH = "/content/drive/MyDrive/Bitcoin P2P Dataset/dataset.zip"
PROCESSED_DIR = "/content/drive/MyDrive/Bitcoin P2P Dataset/processed"

# ---------------------------------------------------------
# Safe CSV parser for variable-width discovery files
# ---------------------------------------------------------

def parse_variable_csv(raw_bytes):

    text = raw_bytes.decode(
        "utf-8",
        errors="replace"
    )

    rows = []

    for line in text.splitlines():

        line = line.strip()

        if not line:
            continue

        parts = line.split(",")

        rows.append(parts)

    return rows


# ---------------------------------------------------------
# Process discovery files
# ---------------------------------------------------------

getaddr_profiles = []
up_profiles = []

with zipfile.ZipFile(ZIP_PATH, "r") as z:

    names = z.namelist()

    getaddr_files = [
        n for n in names
        if "/nodes_per_getADDR_" in n
        and n.endswith(".csv")
    ]

    up_files = [
        n for n in names
        if "/up_nodes_per_seconds_" in n
        and n.endswith(".csv")
    ]

    print("getADDR files:", len(getaddr_files))
    print("up-node files:", len(up_files))

    # -----------------------------------------------------
    # A. nodes_per_getADDR
    # -----------------------------------------------------

    for i, path in enumerate(getaddr_files):

        raw = z.read(path)

        rows = parse_variable_csv(raw)

        row_widths = []

        numeric_values = []

        peer_count = 0

        for row in rows:

            if len(row) < 2:
                continue

            peer_count += 1

            # First field = row index
            # Second field = node endpoint
            numeric = []

            for value in row[2:]:

                try:
                    numeric.append(
                        float(value)
                    )
                except:
                    pass

            row_widths.append(
                len(numeric)
            )

            numeric_values.extend(
                numeric
            )

        if numeric_values:

            arr = np.array(
                numeric_values,
                dtype=float
            )

            valid = arr[arr >= 0]

            profile = {
                "file": path,
                "timestamp": os.path.basename(path)
                    .replace("nodes_per_getADDR_", "")
                    .replace(".csv", ""),
                "peer_rows": peer_count,
                "min_numeric_fields": min(row_widths),
                "max_numeric_fields": max(row_widths),
                "mean_numeric_fields": np.mean(row_widths),
                "total_numeric_values": len(arr),
                "valid_numeric_values": len(valid),
                "missing_numeric_values": int(
                    np.sum(arr < 0)
                ),
                "mean_valid_value": (
                    np.mean(valid)
                    if len(valid) > 0
                    else np.nan
                ),
                "median_valid_value": (
                    np.median(valid)
                    if len(valid) > 0
                    else np.nan
                ),
                "max_valid_value": (
                    np.max(valid)
                    if len(valid) > 0
                    else np.nan
                )
            }

        else:

            profile = {
                "file": path,
                "timestamp": os.path.basename(path)
                    .replace("nodes_per_getADDR_", "")
                    .replace(".csv", ""),
                "peer_rows": peer_count,
                "min_numeric_fields": 0,
                "max_numeric_fields": 0,
                "mean_numeric_fields": 0,
                "total_numeric_values": 0,
                "valid_numeric_values": 0,
                "missing_numeric_values": 0,
                "mean_valid_value": np.nan,
                "median_valid_value": np.nan,
                "max_valid_value": np.nan
            }

        getaddr_profiles.append(profile)

        if (i + 1) % 500 == 0:
            print(
                f"Processed getADDR: {i+1}/{len(getaddr_files)}"
            )

    # -----------------------------------------------------
    # B. up_nodes_per_seconds
    # -----------------------------------------------------

    for i, path in enumerate(up_files):

        raw = z.read(path)

        rows = parse_variable_csv(raw)

        seconds = []
        up_nodes = []

        for row in rows:

            if len(row) < 2:
                continue

            try:

                sec = float(row[0])
                nodes = float(row[1])

                seconds.append(sec)
                up_nodes.append(nodes)

            except:
                continue

        if up_nodes:

            arr = np.array(
                up_nodes,
                dtype=float
            )

            profile = {
                "file": path,
                "timestamp": os.path.basename(path)
                    .replace("up_nodes_per_seconds_", "")
                    .replace(".csv", ""),
                "time_points": len(seconds),
                "duration_seconds": (
                    max(seconds) - min(seconds)
                    if seconds
                    else 0
                ),
                "mean_up_nodes": np.mean(arr),
                "median_up_nodes": np.median(arr),
                "max_up_nodes": np.max(arr),
                "min_up_nodes": np.min(arr),
                "std_up_nodes": (
                    np.std(arr)
                    if len(arr) > 1
                    else 0
                )
            }

        else:

            profile = {
                "file": path,
                "timestamp": os.path.basename(path)
                    .replace("up_nodes_per_seconds_", "")
                    .replace(".csv", ""),
                "time_points": 0,
                "duration_seconds": 0,
                "mean_up_nodes": np.nan,
                "median_up_nodes": np.nan,
                "max_up_nodes": np.nan,
                "min_up_nodes": np.nan,
                "std_up_nodes": np.nan
            }

        up_profiles.append(profile)

        if (i + 1) % 500 == 0:
            print(
                f"Processed up-nodes: {i+1}/{len(up_files)}"
            )


# ---------------------------------------------------------
# Convert to DataFrames
# ---------------------------------------------------------

getaddr_df = pd.DataFrame(
    getaddr_profiles
)

up_df = pd.DataFrame(
    up_profiles
)

print("\n=== getADDR profile ===")
print(getaddr_df.shape)
display(getaddr_df.head())

print("\n=== up_nodes profile ===")
print(up_df.shape)
display(up_df.head())

# ---------------------------------------------------------
# Schema statistics
# ---------------------------------------------------------

print("\n=== getADDR numeric-field distribution ===")

print(
    getaddr_df[
        [
            "min_numeric_fields",
            "max_numeric_fields",
            "mean_numeric_fields"
        ]
    ].describe()
)

print("\n=== getADDR peer-row distribution ===")

print(
    getaddr_df["peer_rows"].describe()
)

print("\n=== up-node statistics ===")

print(
    up_df[
        [
            "time_points",
            "duration_seconds",
            "mean_up_nodes",
            "max_up_nodes",
            "std_up_nodes"
        ]
    ].describe()
)

# ---------------------------------------------------------
# Save
# ---------------------------------------------------------

getaddr_path = (
    f"{PROCESSED_DIR}/p2p3_getaddr_profile_v1.parquet"
)

up_path = (
    f"{PROCESSED_DIR}/p2p3_up_nodes_profile_v1.parquet"
)

getaddr_df.to_parquet(
    getaddr_path,
    index=False
)

up_df.to_parquet(
    up_path,
    index=False
)

print("\n✅ Step 65E complete")

print(
    "getADDR profile saved:",
    getaddr_path
)

print(
    "up-node profile saved:",
    up_path
)
```

```python
import zipfile
import os
import numpy as np
import pandas as pd
from collections import defaultdict

ZIP_PATH = "/content/drive/MyDrive/Bitcoin P2P Dataset/dataset.zip"
PROCESSED_DIR = "/content/drive/MyDrive/Bitcoin P2P Dataset/processed"

OUTPUT_PATH = (
    f"{PROCESSED_DIR}/p2p3_peer_getaddr_features_v1.parquet"
)

# ---------------------------------------------------------
# Node-level running aggregates
# ---------------------------------------------------------

stats = defaultdict(lambda: {
    "observation_count": 0,
    "active_snapshots": set(),

    "sum_mean_value": 0.0,
    "count_mean_value": 0,

    "valid_value_count": 0,
    "missing_value_count": 0,

    "max_value": np.nan,

    "sum_std_value": 0.0,
    "count_std_value": 0,

    # For an approximate median later
    "sample_values": []
})


def parse_node_id(endpoint):
    """
    Observed endpoint format:
    node_id-IP-port-service
    """
    parts = endpoint.split("-")

    if len(parts) < 4:
        return None

    return parts[0]


# ---------------------------------------------------------
# Process ZIP incrementally
# ---------------------------------------------------------

with zipfile.ZipFile(ZIP_PATH, "r") as z:

    getaddr_files = [
        n for n in z.namelist()
        if "/nodes_per_getADDR_" in n
        and n.endswith(".csv")
    ]

    getaddr_files.sort()

    print("getADDR files:", len(getaddr_files))
    print("Starting RAM-safe aggregation...\n")

    for i, path in enumerate(getaddr_files):

        filename = os.path.basename(path)

        timestamp = (
            filename
            .replace("nodes_per_getADDR_", "")
            .replace(".csv", "")
        )

        # Read ONE file only
        with z.open(path) as f:

            for raw_line in f:

                try:
                    line = raw_line.decode(
                        "utf-8",
                        errors="replace"
                    ).strip()

                except Exception:
                    continue

                if not line:
                    continue

                parts = line.split(",")

                if len(parts) < 3:
                    continue

                endpoint = parts[1]

                node_id = parse_node_id(endpoint)

                if node_id is None:
                    continue

                # -------------------------------------------------
                # Parse numeric fields
                # -------------------------------------------------

                valid_values = []
                missing_count = 0

                for value in parts[2:]:

                    try:
                        v = float(value)

                    except (ValueError, TypeError):
                        continue

                    if v < 0:
                        missing_count += 1
                    else:
                        valid_values.append(v)

                if not valid_values and missing_count == 0:
                    continue

                s = stats[node_id]

                s["observation_count"] += 1
                s["active_snapshots"].add(timestamp)

                # -------------------------------------------------
                # Valid values
                # -------------------------------------------------

                if valid_values:

                    arr = np.asarray(
                        valid_values,
                        dtype=np.float64
                    )

                    mean_value = float(np.mean(arr))
                    std_value = (
                        float(np.std(arr))
                        if len(arr) > 1
                        else 0.0
                    )

                    s["sum_mean_value"] += mean_value
                    s["count_mean_value"] += 1

                    s["sum_std_value"] += std_value
                    s["count_std_value"] += 1

                    s["valid_value_count"] += len(arr)

                    current_max = float(np.max(arr))

                    if (
                        np.isnan(s["max_value"])
                        or current_max > s["max_value"]
                    ):
                        s["max_value"] = current_max

                    # Keep only a bounded sample for median
                    if len(s["sample_values"]) < 100:
                        remaining = 100 - len(s["sample_values"])
                        s["sample_values"].extend(
                            arr[:remaining].tolist()
                        )

                s["missing_value_count"] += missing_count

        # ---------------------------------------------------------
        # Progress
        # ---------------------------------------------------------

        if (i + 1) % 250 == 0:

            print(
                f"Processed "
                f"{i + 1}/{len(getaddr_files)} files | "
                f"Unique nodes: {len(stats):,}"
            )


# ---------------------------------------------------------
# Build final peer-level DataFrame
# ---------------------------------------------------------

rows = []

for node_id, s in stats.items():

    total_measurements = (
        s["valid_value_count"]
        +
        s["missing_value_count"]
    )

    if s["sample_values"]:

        median_value = float(
            np.median(
                np.asarray(
                    s["sample_values"],
                    dtype=np.float64
                )
            )
        )

    else:
        median_value = np.nan

    rows.append({

        "node_id": node_id,

        "getaddr_observation_count":
            s["observation_count"],

        "getaddr_active_snapshots":
            len(s["active_snapshots"]),

        "getaddr_mean_value":
            (
                s["sum_mean_value"]
                /
                s["count_mean_value"]
                if s["count_mean_value"] > 0
                else np.nan
            ),

        "getaddr_median_value":
            median_value,

        "getaddr_max_value":
            s["max_value"],

        "getaddr_std_value":
            (
                s["sum_std_value"]
                /
                s["count_std_value"]
                if s["count_std_value"] > 0
                else 0.0
            ),

        "getaddr_valid_value_count":
            s["valid_value_count"],

        "getaddr_missing_value_count":
            s["missing_value_count"],

        "getaddr_missing_ratio":
            (
                s["missing_value_count"]
                /
                total_measurements
                if total_measurements > 0
                else 0.0
            )
    })


peer_getaddr = pd.DataFrame(rows)

# ---------------------------------------------------------
# Save
# ---------------------------------------------------------

peer_getaddr.to_parquet(
    OUTPUT_PATH,
    index=False
)

print("\n========================================")
print("✅ Step 66A complete")
print("========================================")

print(
    "Peer feature shape:",
    peer_getaddr.shape
)

print(
    "Missing values:",
    int(peer_getaddr.isna().sum().sum())
)

print(
    "Saved:",
    OUTPUT_PATH
)

display(peer_getaddr.head(20))
```

```python
import pandas as pd
import numpy as np

GETADDR_PATH = (
    "/content/drive/MyDrive/Bitcoin P2P Dataset/processed/"
    "p2p3_peer_getaddr_features_v1.parquet"
)

df = pd.read_parquet(GETADDR_PATH)

numeric_cols = df.select_dtypes(
    include=[np.number]
).columns.tolist()

# ---------------------------------------------------------
# 1. Constant / near-constant features
# ---------------------------------------------------------

quality = []

for col in numeric_cols:

    s = df[col]

    quality.append({
        "feature": col,
        "missing": int(s.isna().sum()),
        "unique": int(s.nunique()),
        "mean": float(s.mean()),
        "std": float(s.std()),
        "min": float(s.min()),
        "max": float(s.max()),
        "skew": float(s.skew())
    })

quality_df = pd.DataFrame(quality)

print("=== FEATURE QUALITY ===")
display(
    quality_df.sort_values(
        "skew",
        key=lambda x: x.abs(),
        ascending=False
    )
)

# ---------------------------------------------------------
# 2. Highly correlated features
# ---------------------------------------------------------

corr = df[numeric_cols].corr()

pairs = []

for i in range(len(numeric_cols)):

    for j in range(i + 1, len(numeric_cols)):

        value = corr.iloc[i, j]

        if abs(value) >= 0.90:

            pairs.append({
                "feature_1": numeric_cols[i],
                "feature_2": numeric_cols[j],
                "correlation": value
            })

corr_df = pd.DataFrame(pairs)

print("\n=== HIGH CORRELATIONS (|r| >= 0.90) ===")

if len(corr_df) > 0:
    display(
        corr_df.sort_values(
            "correlation",
            key=lambda x: x.abs(),
            ascending=False
        )
    )
else:
    print("No correlations >= 0.90")

# ---------------------------------------------------------
# 3. Extreme missing-ratio groups
# ---------------------------------------------------------

print("\n=== MISSING-RATIO DISTRIBUTION ===")

display(
    df["getaddr_missing_ratio"].describe(
        percentiles=[
            .01, .05, .10,
            .25, .50, .75,
            .90, .95, .99
        ]
    )
)

print("\n=== PEERS WITH 100% MISSING GETADDR VALUES ===")

print(
    (
        df["getaddr_missing_ratio"] == 1.0
    ).sum()
)

# ---------------------------------------------------------
# 4. Zero-valid-value peers
# ---------------------------------------------------------

print("\n=== PEERS WITH ZERO VALID VALUES ===")

print(
    (
        df["getaddr_valid_value_count"] == 0
    ).sum()
)
```

```python
import pandas as pd
import numpy as np

PROCESSED_DIR = (
    "/content/drive/MyDrive/Bitcoin P2P Dataset/processed"
)

GETADDR_PATH = (
    f"{PROCESSED_DIR}/p2p3_getaddr_profile_v1.parquet"
)

UPNODES_PATH = (
    f"{PROCESSED_DIR}/p2p3_up_nodes_profile_v1.parquet"
)

OUTPUT_PATH = (
    f"{PROCESSED_DIR}/p2p3_snapshot_features_v1.parquet"
)

# ---------------------------------------------------------
# Load
# ---------------------------------------------------------

getaddr = pd.read_parquet(GETADDR_PATH)
upnodes = pd.read_parquet(UPNODES_PATH)

# ---------------------------------------------------------
# Merge snapshot-level data
# ---------------------------------------------------------

df = getaddr.merge(
    upnodes,
    on=["file", "timestamp"],
    how="inner",
    validate="one_to_one"
)

print("Merged snapshot shape:", df.shape)

# ---------------------------------------------------------
# Derived getADDR features
# ---------------------------------------------------------

df["getaddr_valid_ratio"] = np.where(
    df["total_numeric_values"] > 0,
    df["valid_numeric_values"]
    / df["total_numeric_values"],
    0.0
)

df["getaddr_missing_ratio"] = np.where(
    df["total_numeric_values"] > 0,
    df["missing_numeric_values"]
    / df["total_numeric_values"],
    0.0
)

df["getaddr_values_per_peer"] = np.where(
    df["peer_rows"] > 0,
    df["total_numeric_values"]
    / df["peer_rows"],
    0.0
)

# ---------------------------------------------------------
# Derived up-node features
# ---------------------------------------------------------

df["up_nodes_range"] = (
    df["max_up_nodes"]
    -
    df["median_up_nodes"]
)

df["up_nodes_variability_ratio"] = np.where(
    df["mean_up_nodes"] > 0,
    df["std_up_nodes"]
    / df["mean_up_nodes"],
    0.0
)

# min_up_nodes contains extreme negative values.
# Preserve the raw column for auditing but create a
# non-negative bounded version for ML use.

df["min_up_nodes_clipped"] = (
    df["min_up_nodes"]
    .clip(lower=0)
)

# ---------------------------------------------------------
# Validation
# ---------------------------------------------------------

print("\nFeature shape:", df.shape)

print("\nMissing values:")
print(
    df.isna().sum()
    .sort_values(ascending=False)
    .head(20)
)

print("\nDuplicate snapshots:")

print(
    df.duplicated(
        subset=["file", "timestamp"]
    ).sum()
)

print("\nDerived feature summary:")

derived_cols = [
    "getaddr_valid_ratio",
    "getaddr_missing_ratio",
    "getaddr_values_per_peer",
    "up_nodes_range",
    "up_nodes_variability_ratio",
    "min_up_nodes_clipped"
]

display(
    df[derived_cols].describe().T
)

# ---------------------------------------------------------
# Save
# ---------------------------------------------------------

df.to_parquet(
    OUTPUT_PATH,
    index=False
)

print("\n✅ Step 69 complete")
print("Saved:", OUTPUT_PATH)
```

```python
import pandas as pd
import numpy as np

PROCESSED_DIR = "/content/drive/MyDrive/Bitcoin P2P Dataset/processed"

GETADDR_PATH = f"{PROCESSED_DIR}/p2p3_getaddr_profile_v1.parquet"
UPNODES_PATH = f"{PROCESSED_DIR}/p2p3_up_nodes_profile_v1.parquet"

# Load snapshot profiles
getaddr = pd.read_parquet(GETADDR_PATH)
upnodes = pd.read_parquet(UPNODES_PATH)

# Parse timestamps
getaddr["timestamp"] = pd.to_datetime(getaddr["timestamp"])
upnodes["timestamp"] = pd.to_datetime(upnodes["timestamp"])

# Add date/hour
for df in [getaddr, upnodes]:
    df["date"] = df["timestamp"].dt.date
    df["hour"] = df["timestamp"].dt.hour

# ---------------------------------------------------------
# GETADDR DAILY FEATURES
# ---------------------------------------------------------
getaddr_daily = (
    getaddr
    .groupby("date")
    .agg(
        getaddr_snapshots=("timestamp", "count"),
        getaddr_peer_rows_mean=("peer_rows", "mean"),
        getaddr_peer_rows_std=("peer_rows", "std"),
        getaddr_peer_rows_min=("peer_rows", "min"),
        getaddr_peer_rows_max=("peer_rows", "max"),
        getaddr_valid_values_mean=("valid_numeric_values", "mean"),
        getaddr_missing_values_mean=("missing_numeric_values", "mean"),
        getaddr_missing_ratio_mean=(
            "missing_numeric_values",
            lambda x: np.nan
        ),
        getaddr_mean_value_mean=("mean_valid_value", "mean"),
        getaddr_max_value_max=("max_valid_value", "max"),
    )
    .reset_index()
)

# Calculate missing ratio correctly from totals
daily_missing = (
    getaddr
    .groupby("date")
    .agg(
        total_numeric_values=("total_numeric_values", "sum"),
        total_missing_values=("missing_numeric_values", "sum"),
    )
    .reset_index()
)

getaddr_daily = getaddr_daily.drop(
    columns=["getaddr_missing_ratio_mean"]
)

getaddr_daily = getaddr_daily.merge(
    daily_missing,
    on="date",
    how="left"
)

getaddr_daily["getaddr_missing_ratio"] = np.where(
    getaddr_daily["total_numeric_values"] > 0,
    getaddr_daily["total_missing_values"] /
    getaddr_daily["total_numeric_values"],
    0
)

# ---------------------------------------------------------
# UP-NODES DAILY FEATURES
# ---------------------------------------------------------
upnodes_daily = (
    upnodes
    .groupby("date")
    .agg(
        upnodes_snapshots=("timestamp", "count"),
        upnodes_mean_mean=("mean_up_nodes", "mean"),
        upnodes_mean_std=("mean_up_nodes", "std"),
        upnodes_mean_min=("mean_up_nodes", "min"),
        upnodes_mean_max=("mean_up_nodes", "max"),
        upnodes_max_mean=("max_up_nodes", "mean"),
        upnodes_max_max=("max_up_nodes", "max"),
        upnodes_std_mean=("std_up_nodes", "mean"),
        upnodes_std_max=("std_up_nodes", "max"),
    )
    .reset_index()
)

# ---------------------------------------------------------
# SAVE INDEPENDENT DAILY FEATURES
# ---------------------------------------------------------
getaddr_daily_path = (
    f"{PROCESSED_DIR}/p2p3_getaddr_daily_features_v1.parquet"
)

upnodes_daily_path = (
    f"{PROCESSED_DIR}/p2p3_upnodes_daily_features_v1.parquet"
)

getaddr_daily.to_parquet(
    getaddr_daily_path,
    index=False
)

upnodes_daily.to_parquet(
    upnodes_daily_path,
    index=False
)

print("=== GETADDR DAILY ===")
print("Shape:", getaddr_daily.shape)
print(getaddr_daily.head().to_string(index=False))

print("\n=== UP-NODES DAILY ===")
print("Shape:", upnodes_daily.shape)
print(upnodes_daily.head().to_string(index=False))

print("\nSaved:")
print(getaddr_daily_path)
print(upnodes_daily_path)
```

```python
import pandas as pd
import numpy as np

PROCESSED_DIR = "/content/drive/MyDrive/Bitcoin P2P Dataset/processed"

GETADDR_PATH = f"{PROCESSED_DIR}/p2p3_getaddr_daily_features_v1.parquet"
UPNODES_PATH = f"{PROCESSED_DIR}/p2p3_upnodes_daily_features_v1.parquet"

getaddr = pd.read_parquet(GETADDR_PATH)
upnodes = pd.read_parquet(UPNODES_PATH)

# ---------------------------------------------------------
# GETADDR TEMPORAL FEATURES
# ---------------------------------------------------------
getaddr = getaddr.sort_values("date").reset_index(drop=True)

getaddr["getaddr_snapshot_rate"] = (
    getaddr["getaddr_snapshots"] /
    getaddr["getaddr_snapshots"].median()
)

getaddr["getaddr_peer_row_change"] = (
    getaddr["getaddr_peer_rows_mean"].diff()
)

getaddr["getaddr_peer_row_change_pct"] = (
    getaddr["getaddr_peer_rows_mean"].pct_change()
)

getaddr["getaddr_valid_value_change"] = (
    getaddr["getaddr_valid_values_mean"].diff()
)

getaddr["getaddr_missing_ratio_change"] = (
    getaddr["getaddr_missing_ratio"].diff()
)

# Robust z-score using median/MAD
def robust_z(series):
    median = series.median()
    mad = np.median(np.abs(series - median))

    if mad == 0:
        return pd.Series(np.zeros(len(series)), index=series.index)

    return 0.6745 * (series - median) / mad

getaddr["getaddr_peer_row_robust_z"] = robust_z(
    getaddr["getaddr_peer_rows_mean"]
)

getaddr["getaddr_missing_ratio_robust_z"] = robust_z(
    getaddr["getaddr_missing_ratio"]
)

# ---------------------------------------------------------
# UP-NODES TEMPORAL FEATURES
# ---------------------------------------------------------
upnodes = upnodes.sort_values("date").reset_index(drop=True)

upnodes["upnodes_mean_change"] = (
    upnodes["upnodes_mean_mean"].diff()
)

upnodes["upnodes_mean_change_pct"] = (
    upnodes["upnodes_mean_mean"].pct_change()
)

upnodes["upnodes_max_change"] = (
    upnodes["upnodes_max_mean"].diff()
)

upnodes["upnodes_std_change"] = (
    upnodes["upnodes_std_mean"].diff()
)

upnodes["upnodes_mean_robust_z"] = robust_z(
    upnodes["upnodes_mean_mean"]
)

upnodes["upnodes_max_robust_z"] = robust_z(
    upnodes["upnodes_max_mean"]
)

upnodes["upnodes_variability_ratio"] = np.where(
    upnodes["upnodes_mean_mean"] > 0,
    upnodes["upnodes_mean_std"] /
    upnodes["upnodes_mean_mean"],
    0
)

# ---------------------------------------------------------
# SAVE
# ---------------------------------------------------------
GETADDR_OUT = (
    f"{PROCESSED_DIR}/p2p3_getaddr_temporal_features_v1.parquet"
)

UPNODES_OUT = (
    f"{PROCESSED_DIR}/p2p3_upnodes_temporal_features_v1.parquet"
)

getaddr.to_parquet(GETADDR_OUT, index=False)
upnodes.to_parquet(UPNODES_OUT, index=False)

print("=== GETADDR TEMPORAL ===")
print("Shape:", getaddr.shape)
print(
    getaddr[
        [
            "date",
            "getaddr_peer_rows_mean",
            "getaddr_peer_row_change",
            "getaddr_peer_row_change_pct",
            "getaddr_peer_row_robust_z",
            "getaddr_missing_ratio",
            "getaddr_missing_ratio_robust_z"
        ]
    ].head(10).to_string(index=False)
)

print("\n=== UP-NODES TEMPORAL ===")
print("Shape:", upnodes.shape)
print(
    upnodes[
        [
            "date",
            "upnodes_mean_mean",
            "upnodes_mean_change",
            "upnodes_mean_change_pct",
            "upnodes_max_change",
            "upnodes_mean_robust_z",
            "upnodes_variability_ratio"
        ]
    ].head(10).to_string(index=False)
)

print("\nSaved:")
print(GETADDR_OUT)
print(UPNODES_OUT)
```

```python
import pandas as pd
import numpy as np

PROCESSED_DIR = "/content/drive/MyDrive/Bitcoin P2P Dataset/processed"

GETADDR_PATH = (
    f"{PROCESSED_DIR}/p2p3_getaddr_temporal_features_v1.parquet"
)

UPNODES_PATH = (
    f"{PROCESSED_DIR}/p2p3_upnodes_temporal_features_v1.parquet"
)

getaddr = pd.read_parquet(GETADDR_PATH)
upnodes = pd.read_parquet(UPNODES_PATH)

# Numeric-only matrices
getaddr_num = getaddr.select_dtypes(include=[np.number])
upnodes_num = upnodes.select_dtypes(include=[np.number])

# ---------------------------------------------------------
# BASIC QUALITY
# ---------------------------------------------------------
print("=== GETADDR QUALITY ===")
print("Shape:", getaddr.shape)
print("\nMissing values:")
print(getaddr.isna().sum()[getaddr.isna().sum() > 0])

print("\nConstant features:")
getaddr_constants = [
    c for c in getaddr_num.columns
    if getaddr_num[c].nunique(dropna=True) <= 1
]
print(getaddr_constants)

print("\nSkewed features:")
getaddr_skew = (
    getaddr_num
    .replace([np.inf, -np.inf], np.nan)
    .skew()
    .sort_values(key=lambda x: x.abs(), ascending=False)
)
print(getaddr_skew.head(15).to_string())

print("\n=== UPNODES QUALITY ===")
print("Shape:", upnodes.shape)
print("\nMissing values:")
print(upnodes.isna().sum()[upnodes.isna().sum() > 0])

print("\nConstant features:")
upnodes_constants = [
    c for c in upnodes_num.columns
    if upnodes_num[c].nunique(dropna=True) <= 1
]
print(upnodes_constants)

print("\nSkewed features:")
upnodes_skew = (
    upnodes_num
    .replace([np.inf, -np.inf], np.nan)
    .skew()
    .sort_values(key=lambda x: x.abs(), ascending=False)
)
print(upnodes_skew.head(15).to_string())

# ---------------------------------------------------------
# HIGH CORRELATIONS
# ---------------------------------------------------------
def high_correlations(df, threshold=0.90):
    numeric = (
        df.select_dtypes(include=[np.number])
        .replace([np.inf, -np.inf], np.nan)
    )

    corr = numeric.corr()

    pairs = []

    cols = corr.columns

    for i in range(len(cols)):
        for j in range(i + 1, len(cols)):
            value = corr.iloc[i, j]

            if pd.notna(value) and abs(value) >= threshold:
                pairs.append(
                    (
                        cols[i],
                        cols[j],
                        round(float(value), 4)
                    )
                )

    return sorted(
        pairs,
        key=lambda x: abs(x[2]),
        reverse=True
    )

print("\n=== GETADDR HIGH CORRELATIONS >= 0.90 ===")
getaddr_corr = high_correlations(getaddr)
for row in getaddr_corr[:20]:
    print(row)

print("\n=== UPNODES HIGH CORRELATIONS >= 0.90 ===")
upnodes_corr = high_correlations(upnodes)
for row in upnodes_corr[:20]:
    print(row)
```

```python
import pandas as pd
import numpy as np

PROCESSED_DIR = "/content/drive/MyDrive/Bitcoin P2P Dataset/processed"

GETADDR_PATH = (
    f"{PROCESSED_DIR}/p2p3_getaddr_features_clean_v1.parquet"
)

UPNODES_PATH = (
    f"{PROCESSED_DIR}/p2p3_upnodes_features_clean_v1.parquet"
)

getaddr = pd.read_parquet(GETADDR_PATH)
upnodes = pd.read_parquet(UPNODES_PATH)

# ---------------------------------------------------------
# Robust z-score
# ---------------------------------------------------------
def robust_z(series):
    median = series.median()
    mad = np.median(np.abs(series - median))

    if mad == 0:
        return pd.Series(
            np.zeros(len(series)),
            index=series.index
        )

    return 0.6745 * (series - median) / mad


# ---------------------------------------------------------
# Features suitable for temporal screening
# ---------------------------------------------------------
getaddr_features = [
    "getaddr_peer_rows_mean",
    "getaddr_peer_rows_std",
    "getaddr_valid_values_mean",
    "getaddr_missing_ratio",
    "getaddr_peer_row_change_pct",
    "getaddr_valid_value_change",
]

upnodes_features = [
    "upnodes_mean_mean",
    "upnodes_mean_std",
    "upnodes_max_mean",
    "upnodes_std_mean",
    "upnodes_mean_change_pct",
    "upnodes_max_change",
    "upnodes_std_change",
]

# ---------------------------------------------------------
# Calculate robust z-scores
# ---------------------------------------------------------
for col in getaddr_features:
    getaddr[f"rz_{col}"] = robust_z(getaddr[col])

for col in upnodes_features:
    upnodes[f"rz_{col}"] = robust_z(upnodes[col])

getaddr_rz_cols = [
    f"rz_{c}" for c in getaddr_features
]

upnodes_rz_cols = [
    f"rz_{c}" for c in upnodes_features
]

# ---------------------------------------------------------
# Count extreme deviations
# Threshold = |robust z| >= 3
# ---------------------------------------------------------
getaddr["extreme_feature_count"] = (
    getaddr[getaddr_rz_cols].abs() >= 3
).sum(axis=1)

upnodes["extreme_feature_count"] = (
    upnodes[upnodes_rz_cols].abs() >= 3
).sum(axis=1)

# Maximum absolute deviation
getaddr["max_abs_robust_z"] = (
    getaddr[getaddr_rz_cols].abs().max(axis=1)
)

upnodes["max_abs_robust_z"] = (
    upnodes[upnodes_rz_cols].abs().max(axis=1)
)

# ---------------------------------------------------------
# Sort unusual historical days
# ---------------------------------------------------------
getaddr_top = getaddr.sort_values(
    ["extreme_feature_count", "max_abs_robust_z"],
    ascending=False
)

upnodes_top = upnodes.sort_values(
    ["extreme_feature_count", "max_abs_robust_z"],
    ascending=False
)

print("=== GETADDR TEMPORAL SCREENING ===")
print(
    getaddr_top[
        [
            "date",
            "extreme_feature_count",
            "max_abs_robust_z",
            "getaddr_peer_rows_mean",
            "getaddr_peer_rows_std",
            "getaddr_missing_ratio",
            "getaddr_peer_row_change_pct"
        ]
    ].head(10).to_string(index=False)
)

print("\n=== UP-NODES TEMPORAL SCREENING ===")
print(
    upnodes_top[
        [
            "date",
            "extreme_feature_count",
            "max_abs_robust_z",
            "upnodes_mean_mean",
            "upnodes_mean_std",
            "upnodes_max_mean",
            "upnodes_std_mean",
            "upnodes_mean_change_pct"
        ]
    ].head(10).to_string(index=False)
)

print("\n=== EXTREME DAY COUNTS ===")
print(
    "GetADDR days with >=1 extreme feature:",
    int((getaddr["extreme_feature_count"] >= 1).sum())
)

print(
    "GetADDR days with >=2 extreme features:",
    int((getaddr["extreme_feature_count"] >= 2).sum())
)

print(
    "Up-nodes days with >=1 extreme feature:",
    int((upnodes["extreme_feature_count"] >= 1).sum())
)

print(
    "Up-nodes days with >=2 extreme features:",
    int((upnodes["extreme_feature_count"] >= 2).sum())
)

# ---------------------------------------------------------
# Save screening results
# ---------------------------------------------------------
GETADDR_OUT = (
    f"{PROCESSED_DIR}/p2p3_getaddr_temporal_screening_v1.parquet"
)

UPNODES_OUT = (
    f"{PROCESSED_DIR}/p2p3_upnodes_temporal_screening_v1.parquet"
)

getaddr.to_parquet(GETADDR_OUT, index=False)
upnodes.to_parquet(UPNODES_OUT, index=False)

print("\nSaved:")
print(GETADDR_OUT)
print(UPNODES_OUT)
```

```python
import pandas as pd
import numpy as np

from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler

PROCESSED_DIR = "/content/drive/MyDrive/Bitcoin P2P Dataset/processed"

FEATURE_PATH = (
    f"{PROCESSED_DIR}/p2p_ml_matrix_v2.parquet"
)

df = pd.read_parquet(FEATURE_PATH)

print("Dataset shape:", df.shape)
print("Columns:", df.columns.tolist())

# =========================================================
# IDENTIFY NODE ID + FEATURES
# =========================================================

NODE_COL = "node_id"

feature_cols = [
    c for c in df.columns
    if c != NODE_COL
]

X = df[feature_cols].copy()

X = (
    X
    .replace([np.inf, -np.inf], np.nan)
    .fillna(0)
)

# =========================================================
# LOG TRANSFORM
# Same configuration used for P2P-2
# =========================================================

LOG_FEATURES = [
    "observation_count",
    "observation_span_seconds",
    "unique_ports",
    "unique_services",
    "non_default_port_observations",
    "non_default_ports",
    "non_1037_service_observations",
    "height_deviation_over_100",
    "height_deviation_over_1000",
    "height_deviation_over_10000",
    "max_absolute_height_deviation",
    "max_observations_per_hour",
    "max_observations_per_day",
    "hourly_burst_ratio",
    "daily_burst_ratio",
]

for col in LOG_FEATURES:
    if col in X.columns:
        X[col] = np.log1p(
            X[col].clip(lower=0)
        )

# =========================================================
# CREATE NODE-LEVEL TEMPORAL SPLIT
# =========================================================

# We need observation dates for each node.
# Load the normalized observations incrementally.

OBS_DIR = f"{PROCESSED_DIR}/observations"

import glob

observation_files = sorted(
    glob.glob(f"{OBS_DIR}/p2p_observations_*.parquet")
)

print("\nObservation files:", len(observation_files))

# ---------------------------------------------------------
# Find first and last observation date per node
# ---------------------------------------------------------

node_dates = []

for path in observation_files:

    part = pd.read_parquet(
        path,
        columns=["node_id", "timestamp"]
    )

    part["timestamp"] = pd.to_datetime(
        part["timestamp"]
    )

    grouped = (
        part
        .groupby("node_id")["timestamp"]
        .agg(["min", "max"])
        .reset_index()
    )

    node_dates.append(grouped)

node_dates = pd.concat(
    node_dates,
    ignore_index=True
)

# Combine repeated node records
node_dates = (
    node_dates
    .groupby("node_id")
    .agg(
        first_observed=("min", "min"),
        last_observed=("max", "max")
    )
    .reset_index()
)

node_dates["node_id"] = (
    node_dates["node_id"].astype(str)
)

df[NODE_COL] = df[NODE_COL].astype(str)

df = df.merge(
    node_dates,
    on=NODE_COL,
    how="left"
)

print("\nMerged shape:", df.shape)

print(
    "Missing first_observed:",
    df["first_observed"].isna().sum()
)

# =========================================================
# TEMPORAL HOLDOUT
# =========================================================

TRAIN_END = pd.Timestamp("2020-02-25")

train_mask = (
    df["first_observed"] <= TRAIN_END
)

test_mask = (
    df["first_observed"] > TRAIN_END
)

print("\n=== TEMPORAL SPLIT ===")
print("Train nodes:", train_mask.sum())
print("Test nodes :", test_mask.sum())

# =========================================================
# BUILD MATRICES
# =========================================================

X_train = X.loc[train_mask].copy()
X_test = X.loc[test_mask].copy()

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)

# =========================================================
# SCALE USING TRAIN ONLY
# =========================================================

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(
    X_train
)

X_test_scaled = scaler.transform(
    X_test
)

# =========================================================
# TRAIN ISOLATION FOREST
# =========================================================

model = IsolationForest(
    n_estimators=300,
    contamination="auto",
    random_state=42,
    n_jobs=1
)

model.fit(X_train_scaled)

# =========================================================
# SCORE TRAIN + TEST
# =========================================================

train_scores = -model.decision_function(
    X_train_scaled
)

test_scores = -model.decision_function(
    X_test_scaled
)

print("\n=== SCORE DISTRIBUTION ===")

print(
    "Train mean:",
    round(train_scores.mean(), 6)
)

print(
    "Train std:",
    round(train_scores.std(), 6)
)

print(
    "Test mean:",
    round(test_scores.mean(), 6)
)

print(
    "Test std:",
    round(test_scores.std(), 6)
)

print(
    "Train 95th percentile:",
    round(np.percentile(train_scores, 95), 6)
)

print(
    "Test 95th percentile:",
    round(np.percentile(test_scores, 95), 6)
)

# =========================================================
# TOP-K HOLDOUT
# =========================================================

test_results = df.loc[
    test_mask,
    ["node_id", "first_observed", "last_observed"]
].copy()

test_results["anomaly_score"] = test_scores

print("\n=== TOP 20 HOLDOUT CANDIDATES ===")

print(
    test_results
    .nlargest(20, "anomaly_score")
    .to_string(index=False)
)

# =========================================================
# SAVE HOLDOUT RESULTS
# =========================================================

OUT_PATH = (
    f"{PROCESSED_DIR}/p2p_temporal_holdout_v1.parquet"
)

test_results.to_parquet(
    OUT_PATH,
    index=False
)

print("\nSaved:")
print(OUT_PATH)
```

```python
# ============================================================
# STEP 82 — FINAL TRAIN-ONLY P2P-2 ISOLATION FOREST
# ============================================================

import os
import json
import joblib
import numpy as np
import pandas as pd

from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import IsolationForest

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

BASE = "/content/drive/MyDrive/Bitcoin P2P Dataset"
PROCESSED = os.path.join(BASE, "processed")

ML_PATH = os.path.join(PROCESSED, "p2p_ml_matrix_v2.parquet")
OUT_MODEL = os.path.join(PROCESSED, "p2p_final_isolation_forest_v1.joblib")
OUT_SCALER = os.path.join(PROCESSED, "p2p_final_scaler_v1.joblib")
OUT_CONFIG = os.path.join(PROCESSED, "p2p_final_feature_config_v1.json")
OUT_SCORES = os.path.join(PROCESSED, "p2p_final_train_holdout_scores_v1.parquet")
OUT_METADATA = os.path.join(PROCESSED, "p2p_final_model_metadata_v1.json")

# ------------------------------------------------------------
# LOAD FEATURE MATRIX
# ------------------------------------------------------------

df = pd.read_parquet(ML_PATH)

print("Loaded:", df.shape)

NODE_COL = "node_id"

feature_cols = [c for c in df.columns if c != NODE_COL]

print("Feature count:", len(feature_cols))
print("Features:")
for i, c in enumerate(feature_cols, 1):
    print(f"{i:02d}. {c}")

# ------------------------------------------------------------
# TEMPORAL SPLIT
# Same split used in Step 80
# ------------------------------------------------------------

# Load first/last observation timestamps from normalized P2P data
obs_dir = os.path.join(PROCESSED, "observations")

files = sorted([
    os.path.join(obs_dir, f)
    for f in os.listdir(obs_dir)
    if f.endswith(".parquet")
])

print("\nObservation files:", len(files))

first_seen_parts = []

for path in files:
    tmp = pd.read_parquet(
        path,
        columns=["timestamp", "node_id"]
    )

    tmp["timestamp"] = pd.to_datetime(
        tmp["timestamp"],
        errors="coerce"
    )

    g = (
        tmp.groupby("node_id", sort=False)["timestamp"]
        .agg(["min", "max"])
        .reset_index()
    )

    g.columns = ["node_id", "first_observed", "last_observed"]

    first_seen_parts.append(g)

    del tmp, g

node_times = pd.concat(
    first_seen_parts,
    ignore_index=True
)

# Consolidate because a node can occur in multiple files
node_times = (
    node_times
    .groupby("node_id", as_index=False)
    .agg(
        first_observed=("first_observed", "min"),
        last_observed=("last_observed", "max")
    )
)

print("Node time table:", node_times.shape)

# ------------------------------------------------------------
# MERGE
# ------------------------------------------------------------

data = df.merge(
    node_times,
    on="node_id",
    how="left",
    validate="one_to_one"
)

assert data["first_observed"].notna().all()

# Same cutoff used in Step 80
TRAIN_END = pd.Timestamp("2020-02-25")

train_mask = data["first_observed"] <= TRAIN_END
test_mask = ~train_mask

train = data.loc[train_mask].copy()
holdout = data.loc[test_mask].copy()

print("\n================================================")
print("TEMPORAL SPLIT")
print("================================================")
print("Train:", len(train))
print("Holdout:", len(holdout))
print("Total:", len(data))

assert len(train) == 14871
assert len(holdout) == 6251

# ------------------------------------------------------------
# FEATURES
# ------------------------------------------------------------

X_train = train[feature_cols].copy()
X_holdout = holdout[feature_cols].copy()

# ------------------------------------------------------------
# LOG TRANSFORM
# Same transformation used in Step 80
# ------------------------------------------------------------

log_features = [
    "observation_count",
    "observation_span_seconds",
    "unique_ports",
    "unique_services",
    "non_default_port_observations",
    "non_default_ports",
    "non_1037_service_observations",
    "height_deviation_over_100",
    "height_deviation_over_1000",
    "height_deviation_over_10000",
    "max_absolute_height_deviation",
    "max_observations_per_hour",
    "max_observations_per_day",
    "hourly_burst_ratio",
    "daily_burst_ratio"
]

print("\nLog-transform features:", len(log_features))

for col in log_features:
    X_train[col] = np.log1p(
        np.clip(X_train[col].astype(float), 0, None)
    )

    X_holdout[col] = np.log1p(
        np.clip(X_holdout[col].astype(float), 0, None)
    )

# ------------------------------------------------------------
# FILL UNDEFINED TEMPORAL VARIABILITY VALUES
# ------------------------------------------------------------

for col in feature_cols:
    if X_train[col].isna().any():
        median_value = X_train[col].median()

        X_train[col] = X_train[col].fillna(median_value)
        X_holdout[col] = X_holdout[col].fillna(median_value)

# ------------------------------------------------------------
# STANDARD SCALER
# FIT ONLY ON TRAIN
# ------------------------------------------------------------

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_holdout_scaled = scaler.transform(X_holdout)

print("\nScaled train:", X_train_scaled.shape)
print("Scaled holdout:", X_holdout_scaled.shape)

# ------------------------------------------------------------
# FINAL ISOLATION FOREST
# ------------------------------------------------------------

model = IsolationForest(
    n_estimators=300,
    contamination="auto",
    random_state=42,
    n_jobs=1
)

print("\nTraining Isolation Forest...")

model.fit(X_train_scaled)

print("Training complete.")

# ------------------------------------------------------------
# ANOMALY SCORES
#
# sklearn decision_function:
#   larger = more normal
#
# We invert it:
#   larger = more anomalous
# ------------------------------------------------------------

train_scores = -model.decision_function(
    X_train_scaled
)

holdout_scores = -model.decision_function(
    X_holdout_scaled
)

# ------------------------------------------------------------
# PREDICTIONS
# ------------------------------------------------------------

train_predictions = model.predict(X_train_scaled)
holdout_predictions = model.predict(X_holdout_scaled)

# ------------------------------------------------------------
# THRESHOLDS FROM TRAIN ONLY
# ------------------------------------------------------------

threshold_95 = np.percentile(
    train_scores,
    95
)

threshold_99 = np.percentile(
    train_scores,
    99
)

print("\n================================================")
print("TRAIN REFERENCE")
print("================================================")

print("95th percentile:", round(threshold_95, 6))
print("99th percentile:", round(threshold_99, 6))

# ------------------------------------------------------------
# HOLDOUT THRESHOLD RESULTS
# ------------------------------------------------------------

holdout_above_95 = holdout_scores >= threshold_95
holdout_above_99 = holdout_scores >= threshold_99

print("\n================================================")
print("HOLDOUT RESULTS")
print("================================================")

print(
    "Above train 95th:",
    int(holdout_above_95.sum()),
    "/",
    len(holdout),
    f"({holdout_above_95.mean()*100:.2f}%)"
)

print(
    "Above train 99th:",
    int(holdout_above_99.sum()),
    "/",
    len(holdout),
    f"({holdout_above_99.mean()*100:.2f}%)"
)

# ------------------------------------------------------------
# SCORE DISTRIBUTION
# ------------------------------------------------------------

print("\n================================================")
print("SCORE DISTRIBUTION")
print("================================================")

for q in [50, 75, 90, 95, 99, 99.5, 99.9]:
    print(
        f"{q:>5.1f}th percentile:",
        f"{np.percentile(holdout_scores, q):.6f}"
    )

# ------------------------------------------------------------
# SAVE TRAIN + HOLDOUT SCORES
# ------------------------------------------------------------

train_result = pd.DataFrame({
    "node_id": train[NODE_COL].values,
    "first_observed": train["first_observed"].values,
    "last_observed": train["last_observed"].values,
    "dataset_split": "train",
    "anomaly_score": train_scores,
    "model_prediction": train_predictions,
    "above_train_95_threshold": train_scores >= threshold_95,
    "above_train_99_threshold": train_scores >= threshold_99
})

holdout_result = pd.DataFrame({
    "node_id": holdout[NODE_COL].values,
    "first_observed": holdout["first_observed"].values,
    "last_observed": holdout["last_observed"].values,
    "dataset_split": "holdout",
    "anomaly_score": holdout_scores,
    "model_prediction": holdout_predictions,
    "above_train_95_threshold": holdout_above_95,
    "above_train_99_threshold": holdout_above_99
})

scores_output = pd.concat(
    [train_result, holdout_result],
    ignore_index=True
)

scores_output.to_parquet(
    OUT_SCORES,
    index=False
)

# ------------------------------------------------------------
# SAVE MODEL
# ------------------------------------------------------------

joblib.dump(
    model,
    OUT_MODEL
)

joblib.dump(
    scaler,
    OUT_SCALER
)

# ------------------------------------------------------------
# SAVE FEATURE CONFIGURATION
# ------------------------------------------------------------

feature_config = {
    "model_name": "BIT-TRACE P2P-2 Isolation Forest",
    "version": "v1",
    "node_identifier": NODE_COL,
    "feature_count": len(feature_cols),
    "features": feature_cols,
    "log1p_features": log_features,
    "scaler": "StandardScaler",
    "model": "IsolationForest",
    "n_estimators": 300,
    "contamination": "auto",
    "random_state": 42,
    "n_jobs": 1,
    "train_cutoff": "2020-02-25",
    "train_nodes": int(len(train)),
    "holdout_nodes": int(len(holdout)),
    "score_definition": "-IsolationForest.decision_function(X)",
    "higher_score_means": "more_anomalous",
    "thresholds": {
        "train_95_percentile": float(threshold_95),
        "train_99_percentile": float(threshold_99)
    }
}

with open(
    OUT_CONFIG,
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        feature_config,
        f,
        indent=2
    )

# ------------------------------------------------------------
# MODEL METADATA
# ------------------------------------------------------------

metadata = {
    "model": "Isolation Forest",
    "branch": "P2P",
    "feature_set": "P2P-2",
    "dataset": "Bitcoin P2P historical crawler observations",
    "dataset_period": "2020-02-07 to 2020-07-09",
    "training_split": {
        "rule": "first_observed <= 2020-02-25",
        "nodes": int(len(train))
    },
    "holdout_split": {
        "rule": "first_observed > 2020-02-25",
        "nodes": int(len(holdout))
    },
    "parameters": {
        "n_estimators": 300,
        "contamination": "auto",
        "random_state": 42
    },
    "preprocessing": {
        "log1p_feature_count": len(log_features),
        "scaler": "StandardScaler",
        "fit_only_on_training_data": True
    },
    "results": {
        "train_95_threshold": float(threshold_95),
        "train_99_threshold": float(threshold_99),
        "holdout_above_95_count": int(holdout_above_95.sum()),
        "holdout_above_95_rate": float(holdout_above_95.mean()),
        "holdout_above_99_count": int(holdout_above_99.sum()),
        "holdout_above_99_rate": float(holdout_above_99.mean())
    },
    "evaluation_note": (
        "The real P2P dataset has no verified anomaly labels. "
        "Therefore conventional classification accuracy is not reported. "
        "Evaluation uses temporal holdout scoring, threshold exceedance, "
        "score distributions, ranking stability, and period analysis."
    )
}

with open(
    OUT_METADATA,
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        metadata,
        f,
        indent=2
    )

# ------------------------------------------------------------
# VERIFY ARTIFACTS
# ------------------------------------------------------------

print("\n================================================")
print("SAVED ARTIFACTS")
print("================================================")

for path in [
    OUT_MODEL,
    OUT_SCALER,
    OUT_CONFIG,
    OUT_SCORES,
    OUT_METADATA
]:
    print(
        os.path.basename(path),
        "->",
        f"{os.path.getsize(path) / 1024**2:.2f} MB"
    )

print("\n================================================")
print("STEP 82 COMPLETE")
print("================================================")
```

```python
# ============================================================
# STEP 83 — FINAL MODEL REPRODUCIBILITY TEST
# ============================================================

import os
import json
import joblib
import numpy as np
import pandas as pd

BASE = "/content/drive/MyDrive/Bitcoin P2P Dataset"
PROCESSED = os.path.join(BASE, "processed")

MODEL_PATH = os.path.join(
    PROCESSED,
    "p2p_final_isolation_forest_v1.joblib"
)

SCALER_PATH = os.path.join(
    PROCESSED,
    "p2p_final_scaler_v1.joblib"
)

CONFIG_PATH = os.path.join(
    PROCESSED,
    "p2p_final_feature_config_v1.json"
)

SCORES_PATH = os.path.join(
    PROCESSED,
    "p2p_final_train_holdout_scores_v1.parquet"
)

REPRO_PATH = os.path.join(
    PROCESSED,
    "p2p_final_reproducibility_v1.json"
)

# ------------------------------------------------------------
# LOAD ARTIFACTS
# ------------------------------------------------------------

print("Loading saved model...")

model = joblib.load(MODEL_PATH)
scaler = joblib.load(SCALER_PATH)

with open(CONFIG_PATH, "r", encoding="utf-8") as f:
    config = json.load(f)

saved_scores = pd.read_parquet(SCORES_PATH)

print("Model loaded.")
print("Scaler loaded.")
print("Saved score table:", saved_scores.shape)

# ------------------------------------------------------------
# LOAD FEATURE MATRIX
# ------------------------------------------------------------

ML_PATH = os.path.join(
    PROCESSED,
    "p2p_ml_matrix_v2.parquet"
)

df = pd.read_parquet(ML_PATH)

NODE_COL = config["node_identifier"]
feature_cols = config["features"]

print("\nFeature count:", len(feature_cols))

assert len(feature_cols) == 28

# ------------------------------------------------------------
# RECREATE TEMPORAL SPLIT
# ------------------------------------------------------------

obs_dir = os.path.join(
    PROCESSED,
    "observations"
)

obs_files = sorted([
    os.path.join(obs_dir, f)
    for f in os.listdir(obs_dir)
    if f.endswith(".parquet")
])

first_seen_parts = []

for path in obs_files:

    tmp = pd.read_parquet(
        path,
        columns=["timestamp", "node_id"]
    )

    tmp["timestamp"] = pd.to_datetime(
        tmp["timestamp"],
        errors="coerce"
    )

    g = (
        tmp.groupby("node_id")["timestamp"]
        .min()
        .reset_index()
    )

    g.columns = [
        "node_id",
        "first_observed"
    ]

    first_seen_parts.append(g)

    del tmp, g

first_seen = pd.concat(
    first_seen_parts,
    ignore_index=True
)

first_seen = (
    first_seen
    .groupby("node_id", as_index=False)
    .first()
)

data = df.merge(
    first_seen,
    on="node_id",
    how="left",
    validate="one_to_one"
)

TRAIN_END = pd.Timestamp("2020-02-25")

holdout = data[
    data["first_observed"] > TRAIN_END
].copy()

print("Holdout nodes:", len(holdout))

assert len(holdout) == 6251

# ------------------------------------------------------------
# PREPROCESS EXACTLY AS TRAINING
# ------------------------------------------------------------

X_holdout = holdout[feature_cols].copy()

log_features = config["log1p_features"]

for col in log_features:
    X_holdout[col] = np.log1p(
        np.clip(
            X_holdout[col].astype(float),
            0,
            None
        )
    )

# Fill missing using scaler-training compatible values.
# These features had no problematic missing values in the
# final ML matrix, but keep the safeguard.

for col in feature_cols:

    if X_holdout[col].isna().any():

        # Use scaler center as an approximate standardized-space
        # fallback converted back to original space is unnecessary
        # here because final matrix should be complete.
        X_holdout[col] = X_holdout[col].fillna(
            X_holdout[col].median()
        )

# ------------------------------------------------------------
# TRANSFORM USING SAVED SCALER
# ------------------------------------------------------------

X_holdout_scaled = scaler.transform(
    X_holdout
)

# ------------------------------------------------------------
# REGENERATE SCORES
# ------------------------------------------------------------

reproduced_scores = -model.decision_function(
    X_holdout_scaled
)

# ------------------------------------------------------------
# GET ORIGINAL HOLDOUT SCORES
# ------------------------------------------------------------

original = saved_scores[
    saved_scores["dataset_split"] == "holdout"
].copy()

original = (
    original
    .sort_values("node_id")
    .reset_index(drop=True)
)

reproduced = pd.DataFrame({
    "node_id": holdout["node_id"].values,
    "reproduced_score": reproduced_scores
})

reproduced = (
    reproduced
    .sort_values("node_id")
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# COMPARE
# ------------------------------------------------------------

comparison = original[
    ["node_id", "anomaly_score"]
].copy()

comparison = comparison.merge(
    reproduced,
    on="node_id",
    how="inner",
    validate="one_to_one"
)

comparison["absolute_difference"] = (
    comparison["anomaly_score"]
    - comparison["reproduced_score"]
).abs()

max_difference = comparison[
    "absolute_difference"
].max()

mean_difference = comparison[
    "absolute_difference"
].mean()

exact_matches = (
    comparison["absolute_difference"] == 0
).sum()

total = len(comparison)

print("\n================================================")
print("REPRODUCIBILITY RESULT")
print("================================================")

print("Compared scores:", total)
print("Exact matches:", exact_matches)
print("Maximum absolute difference:", max_difference)
print("Mean absolute difference:", mean_difference)

# Floating-point tolerance
PASS_TOLERANCE = 1e-10

passed = max_difference <= PASS_TOLERANCE

print("\nTolerance:", PASS_TOLERANCE)
print("STATUS:", "PASS" if passed else "FAIL")

# ------------------------------------------------------------
# SAVE RESULT
# ------------------------------------------------------------

repro_result = {
    "model": "BIT-TRACE P2P-2 Isolation Forest",
    "holdout_nodes": int(total),
    "exact_matches": int(exact_matches),
    "max_absolute_difference": float(max_difference),
    "mean_absolute_difference": float(mean_difference),
    "tolerance": PASS_TOLERANCE,
    "status": "PASS" if passed else "FAIL"
}

with open(
    REPRO_PATH,
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        repro_result,
        f,
        indent=2
    )

print("\nSaved:")
print(REPRO_PATH)

print("\n================================================")
print("STEP 83 COMPLETE")
print("================================================")
```

```python
# ============================================================
# STEP 85 — FINAL P2P EXPLAINABLE ANOMALY CANDIDATES
# ============================================================

import os
import json
import numpy as np
import pandas as pd

BASE = "/content/drive/MyDrive/Bitcoin P2P Dataset"
PROCESSED = os.path.join(BASE, "processed")

ML_PATH = os.path.join(
    PROCESSED,
    "p2p_ml_matrix_v2.parquet"
)

SCORES_PATH = os.path.join(
    PROCESSED,
    "p2p_final_train_holdout_scores_v1.parquet"
)

OUT_PATH = os.path.join(
    PROCESSED,
    "p2p_final_explainable_candidates_v1.parquet"
)

# ------------------------------------------------------------
# LOAD
# ------------------------------------------------------------

features = pd.read_parquet(ML_PATH)
scores = pd.read_parquet(SCORES_PATH)

holdout_scores = scores[
    scores["dataset_split"] == "holdout"
].copy()

print("Feature matrix:", features.shape)
print("Holdout scores:", holdout_scores.shape)

# ------------------------------------------------------------
# MERGE
# ------------------------------------------------------------

holdout = holdout_scores.merge(
    features,
    on="node_id",
    how="inner",
    validate="one_to_one"
)

print("Merged holdout:", holdout.shape)

# ------------------------------------------------------------
# TRAIN REFERENCE
# ------------------------------------------------------------

train_scores = scores[
    scores["dataset_split"] == "train"
]["anomaly_score"].values

threshold_95 = np.percentile(
    train_scores,
    95
)

threshold_99 = np.percentile(
    train_scores,
    99
)

# ------------------------------------------------------------
# FEATURE → REASON CODE
# ------------------------------------------------------------

reason_map = {

    "observation_count":
        "HIGH_OBSERVATION_ACTIVITY",

    "active_days":
        "HIGH_TEMPORAL_PERSISTENCE",

    "active_hours":
        "HIGH_TEMPORAL_ACTIVITY",

    "observation_span_seconds":
        "LONG_OBSERVATION_SPAN",

    "unique_ports":
        "HIGH_PORT_DIVERSITY",

    "unique_services":
        "HIGH_SERVICE_DIVERSITY",

    "non_default_port_observations":
        "HIGH_NON_DEFAULT_PORT_ACTIVITY",

    "non_default_ports":
        "HIGH_NON_DEFAULT_PORT_DIVERSITY",

    "non_1037_service_observations":
        "HIGH_NON_DEFAULT_SERVICE_ACTIVITY",

    "min_reported_height":
        "UNUSUAL_LOW_REPORTED_HEIGHT",

    "max_reported_height":
        "UNUSUAL_HIGH_REPORTED_HEIGHT",

    "mean_height_deviation":
        "UNUSUAL_MEAN_HEIGHT_DEVIATION",

    "mean_absolute_height_deviation":
        "HIGH_MEAN_HEIGHT_DEVIATION",

    "max_absolute_height_deviation":
        "EXTREME_HEIGHT_DEVIATION",

    "height_deviation_over_100":
        "FREQUENT_HEIGHT_DEVIATION",

    "height_deviation_over_1000":
        "LARGE_HEIGHT_DEVIATION",

    "height_deviation_over_10000":
        "EXTREME_HEIGHT_DEVIATION",

    "mean_observations_per_hour":
        "HIGH_HOURLY_ACTIVITY",

    "max_observations_per_hour":
        "HIGH_HOURLY_PEAK_ACTIVITY",

    "hourly_activity_std":
        "HIGH_HOURLY_VARIABILITY",

    "hourly_activity_variance":
        "HIGH_HOURLY_VARIANCE",

    "mean_observations_per_day":
        "HIGH_DAILY_ACTIVITY",

    "max_observations_per_day":
        "HIGH_DAILY_PEAK_ACTIVITY",

    "daily_activity_std":
        "HIGH_DAILY_VARIABILITY",

    "hourly_burst_ratio":
        "HIGH_HOURLY_BURST",

    "daily_burst_ratio":
        "HIGH_DAILY_BURST",

    "hourly_activity_cv":
        "HIGH_HOURLY_ACTIVITY_VARIABILITY",

    "daily_activity_cv":
        "HIGH_DAILY_ACTIVITY_VARIABILITY"
}

# ------------------------------------------------------------
# REFERENCE VALUES
#
# Use training population to determine unusually high/low
# feature values.
# ------------------------------------------------------------

train_nodes = scores[
    scores["dataset_split"] == "train"
]["node_id"]

train_features = features[
    features["node_id"].isin(train_nodes)
].copy()

# ------------------------------------------------------------
# ROBUST THRESHOLDS
# Median + 3*MAD
# ------------------------------------------------------------

robust_thresholds = {}

for feature in reason_map:

    if feature not in train_features.columns:
        continue

    values = pd.to_numeric(
        train_features[feature],
        errors="coerce"
    ).dropna()

    if len(values) == 0:
        continue

    median = values.median()

    mad = np.median(
        np.abs(values - median)
    )

    # Prevent zero MAD
    if mad == 0:
        q99 = values.quantile(0.99)

        robust_thresholds[feature] = {
            "median": float(median),
            "mad": float(mad),
            "upper": float(q99),
            "lower": float(values.quantile(0.01))
        }

    else:

        robust_thresholds[feature] = {
            "median": float(median),
            "mad": float(mad),
            "upper": float(
                median + 3 * 1.4826 * mad
            ),
            "lower": float(
                median - 3 * 1.4826 * mad
            )
        }

# ------------------------------------------------------------
# GENERATE REASONS
# ------------------------------------------------------------

def generate_reasons(row):

    reasons = []

    for feature, code in reason_map.items():

        if feature not in robust_thresholds:
            continue

        value = row[feature]

        if pd.isna(value):
            continue

        ref = robust_thresholds[feature]

        # For height lower bound
        if feature in [
            "min_reported_height"
        ]:

            if value < ref["lower"]:
                reasons.append(code)

        # For max-type features
        else:

            if value > ref["upper"]:
                reasons.append(code)

    # Remove duplicates while preserving order
    reasons = list(dict.fromkeys(reasons))

    return reasons


holdout["reason_codes"] = holdout.apply(
    generate_reasons,
    axis=1
)

# ------------------------------------------------------------
# CANDIDATE CLASS
# ------------------------------------------------------------

holdout["candidate_level"] = np.select(
    [
        holdout["anomaly_score"] >= threshold_99,
        holdout["anomaly_score"] >= threshold_95
    ],
    [
        "HIGH",
        "ELEVATED"
    ],
    default="BASELINE"
)

# ------------------------------------------------------------
# REASON COUNT
# ------------------------------------------------------------

holdout["reason_count"] = (
    holdout["reason_codes"]
    .apply(len)
)

# ------------------------------------------------------------
# TOP CANDIDATES
# ------------------------------------------------------------

holdout = holdout.sort_values(
    "anomaly_score",
    ascending=False
).reset_index(drop=True)

holdout["rank"] = (
    np.arange(len(holdout)) + 1
)

# Convert list to readable string for Parquet/GitHub
holdout["reason_codes_text"] = (
    holdout["reason_codes"]
    .apply(
        lambda x: "|".join(x)
        if len(x) > 0
        else "NO_EXTREME_FEATURE_REASON"
    )
)

# ------------------------------------------------------------
# SELECT OUTPUT
# ------------------------------------------------------------

output_cols = [
    "rank",
    "node_id",
    "first_observed",
    "last_observed",
    "anomaly_score",
    "candidate_level",
    "reason_count",
    "reason_codes_text"
]

final_candidates = holdout[
    output_cols
].copy()

# ------------------------------------------------------------
# SAVE
# ------------------------------------------------------------

final_candidates.to_parquet(
    OUT_PATH,
    index=False
)

# ------------------------------------------------------------
# DISPLAY
# ------------------------------------------------------------

print("\n================================================")
print("TOP 20 FINAL HOLDOUT CANDIDATES")
print("================================================")

print(
    final_candidates.head(20).to_string(
        index=False
    )
)

print("\n================================================")
print("CANDIDATE LEVELS")
print("================================================")

print(
    final_candidates[
        "candidate_level"
    ].value_counts()
)

print("\n================================================")
print("REASON CODE FREQUENCY")
print("================================================")

reason_frequency = {}

for codes in final_candidates["reason_codes_text"]:

    for code in codes.split("|"):

        if code == "NO_EXTREME_FEATURE_REASON":
            continue

        reason_frequency[code] = (
            reason_frequency.get(code, 0) + 1
        )

reason_frequency = dict(
    sorted(
        reason_frequency.items(),
        key=lambda x: x[1],
        reverse=True
    )
)

for code, count in list(
    reason_frequency.items()
)[:20]:

    print(
        f"{code}: {count}"
    )

print("\nSaved:")
print(OUT_PATH)

print("\n================================================")
print("STEP 85 COMPLETE")
print("================================================")
```

```python
# ============================================================
# STEP 88 — CREATE DEPLOYABLE P2P MODEL BUNDLE
# ============================================================

import os
import json
import joblib
import shutil

BASE = "/content/drive/MyDrive/Bitcoin P2P Dataset"
PROCESSED = os.path.join(BASE, "processed")

PACKAGE = os.path.join(
    BASE,
    "BIT-TRACE-P2P-GitHub"
)

MODEL_DIR = os.path.join(
    PACKAGE,
    "model"
)

INFERENCE_DIR = os.path.join(
    PACKAGE,
    "inference"
)

os.makedirs(
    MODEL_DIR,
    exist_ok=True
)

os.makedirs(
    INFERENCE_DIR,
    exist_ok=True
)

# ------------------------------------------------------------
# EXISTING ARTIFACTS
# ------------------------------------------------------------

MODEL_PATH = os.path.join(
    PACKAGE,
    "model",
    "isolation_forest.joblib"
)

SCALER_PATH = os.path.join(
    PACKAGE,
    "model",
    "scaler.joblib"
)

CONFIG_PATH = os.path.join(
    PACKAGE,
    "model",
    "feature_config.json"
)

METADATA_PATH = os.path.join(
    PACKAGE,
    "model",
    "model_metadata.json"
)

# ------------------------------------------------------------
# LOAD EXISTING ARTIFACTS
# ------------------------------------------------------------

print("Loading saved model artifacts...")

model = joblib.load(
    MODEL_PATH
)

scaler = joblib.load(
    SCALER_PATH
)

with open(
    CONFIG_PATH,
    "r",
    encoding="utf-8"
) as f:
    config = json.load(f)

with open(
    METADATA_PATH,
    "r",
    encoding="utf-8"
) as f:
    metadata = json.load(f)

print("Model loaded.")
print("Scaler loaded.")
print("Features:", len(config["features"]))

# ------------------------------------------------------------
# CREATE SINGLE DEPLOYMENT BUNDLE
# ------------------------------------------------------------

bundle = {
    "model": model,
    "scaler": scaler,
    "feature_columns": config["features"],
    "log1p_features": config["log1p_features"],
    "train_95_threshold": config["thresholds"][
        "train_95_percentile"
    ],
    "train_99_threshold": config["thresholds"][
        "train_99_percentile"
    ],
    "score_definition": config[
        "score_definition"
    ],
    "version": "v1"
}

BUNDLE_PATH = os.path.join(
    MODEL_DIR,
    "p2p_anomaly_detector.joblib"
)

joblib.dump(
    bundle,
    BUNDLE_PATH
)

print("\nDeployment bundle created:")
print(BUNDLE_PATH)

print(
    "Size:",
    f"{os.path.getsize(BUNDLE_PATH)/1024**2:.2f} MB"
)

# ------------------------------------------------------------
# CREATE INFERENCE MODULE
# ------------------------------------------------------------

INFERENCE_CODE = r'''
import joblib
import numpy as np
import pandas as pd


class P2PAnomalyDetector:
    """
    BIT-TRACE P2P anomaly detector.

    Loads the trained Isolation Forest together with:
    - StandardScaler
    - feature ordering
    - log1p preprocessing configuration
    - train-derived anomaly thresholds
    """

    def __init__(self, model_path):
        bundle = joblib.load(model_path)

        self.model = bundle["model"]
        self.scaler = bundle["scaler"]

        self.feature_columns = bundle[
            "feature_columns"
        ]

        self.log1p_features = bundle[
            "log1p_features"
        ]

        self.threshold_95 = bundle[
            "train_95_threshold"
        ]

        self.threshold_99 = bundle[
            "train_99_threshold"
        ]

    def _preprocess(self, data):
        """
        Apply the same preprocessing used during training.
        """

        if isinstance(data, dict):
            data = pd.DataFrame([data])

        elif not isinstance(data, pd.DataFrame):
            data = pd.DataFrame(data)

        missing = [
            feature
            for feature in self.feature_columns
            if feature not in data.columns
        ]

        if missing:
            raise ValueError(
                f"Missing required features: {missing}"
            )

        X = data[
            self.feature_columns
        ].copy()

        # Same log1p preprocessing used during training
        for feature in self.log1p_features:

            X[feature] = np.log1p(
                np.clip(
                    X[feature].astype(float),
                    0,
                    None
                )
            )

        # Safety handling for missing values
        for feature in self.feature_columns:

            if X[feature].isna().any():

                X[feature] = X[
                    feature
                ].fillna(
                    X[feature].median()
                )

        X_scaled = self.scaler.transform(X)

        return X_scaled

    def predict(self, data):
        """
        Return anomaly score and candidate level.
        """

        X_scaled = self._preprocess(data)

        scores = -self.model.decision_function(
            X_scaled
        )

        results = []

        for score in scores:

            if score >= self.threshold_99:
                level = "HIGH"

            elif score >= self.threshold_95:
                level = "ELEVATED"

            else:
                level = "BASELINE"

            results.append({
                "anomaly_score": float(score),
                "candidate_level": level
            })

        return results
'''

INFERENCE_PATH = os.path.join(
    INFERENCE_DIR,
    "p2p_detector.py"
)

with open(
    INFERENCE_PATH,
    "w",
    encoding="utf-8"
) as f:
    f.write(INFERENCE_CODE)

print("\nInference module created:")
print(INFERENCE_PATH)

# ------------------------------------------------------------
# CREATE SIMPLE TEST SCRIPT
# ------------------------------------------------------------

TEST_CODE = r'''
from p2p_detector import P2PAnomalyDetector


MODEL_PATH = "../model/p2p_anomaly_detector.joblib"

detector = P2PAnomalyDetector(
    MODEL_PATH
)

# Example:
# Replace these values with a real 28-feature observation.

sample = {
    "observation_count": 100,
    "active_days": 5,
    "active_hours": 20,
    "observation_span_seconds": 100000,

    "unique_ports": 1,
    "unique_services": 1,

    "non_default_port_observations": 0,
    "non_default_ports": 0,
    "non_1037_service_observations": 0,

    "min_reported_height": 620000,
    "max_reported_height": 620100,

    "mean_height_deviation": 10,
    "mean_absolute_height_deviation": 10,
    "max_absolute_height_deviation": 20,

    "height_deviation_over_100": 0,
    "height_deviation_over_1000": 0,
    "height_deviation_over_10000": 0,

    "mean_observations_per_hour": 5,
    "max_observations_per_hour": 10,

    "hourly_activity_std": 2,
    "hourly_activity_variance": 4,

    "mean_observations_per_day": 20,
    "max_observations_per_day": 30,

    "daily_activity_std": 5,

    "hourly_burst_ratio": 1,
    "daily_burst_ratio": 1,

    "hourly_activity_cv": 0.2,
    "daily_activity_cv": 0.2
}

result = detector.predict(sample)

print(result)
'''

TEST_PATH = os.path.join(
    INFERENCE_DIR,
    "test_inference.py"
)

with open(
    TEST_PATH,
    "w",
    encoding="utf-8"
) as f:
    f.write(TEST_CODE)

print("\nTest script created:")
print(TEST_PATH)

# ------------------------------------------------------------
# FINAL MODEL CARD
# ------------------------------------------------------------

MODEL_CARD = {
    "name": "BIT-TRACE P2P Anomaly Detector",
    "version": "v1",
    "algorithm": "Isolation Forest",
    "features": 28,
    "estimators": 300,

    "training_nodes": 14871,
    "holdout_nodes": 6251,

    "train_95_threshold": bundle[
        "train_95_threshold"
    ],

    "train_99_threshold": bundle[
        "train_99_threshold"
    ],

    "holdout_above_95": {
        "count": 269,
        "rate": 0.0430
    },

    "holdout_above_99": {
        "count": 46,
        "rate": 0.0074
    },

    "ground_truth_available": False,

    "accuracy": None,

    "evaluation_note":
        "Ground-truth anomaly labels are unavailable. "
        "Temporal holdout and threshold-based evaluation "
        "are used instead.",

    "reproducibility": {
        "tested_nodes": 6251,
        "exact_matches": 6251,
        "maximum_difference": 0.0,
        "status": "PASS"
    }
}

MODEL_CARD_PATH = os.path.join(
    MODEL_DIR,
    "model_card.json"
)

with open(
    MODEL_CARD_PATH,
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        MODEL_CARD,
        f,
        indent=2
    )

print("\nModel card created.")

# ------------------------------------------------------------
# FINAL STRUCTURE
# ------------------------------------------------------------

print("\n================================================")
print("DEPLOYABLE MODEL PACKAGE")
print("================================================")

for root, dirs, files in os.walk(PACKAGE):

    level = root.replace(
        PACKAGE,
        ""
    ).count(os.sep)

    indent = "  " * level

    print(
        f"{indent}{os.path.basename(root)}/"
    )

    for file in files:

        path = os.path.join(
            root,
            file
        )

        size = (
            os.path.getsize(path)
            / 1024**2
        )

        print(
            f"{indent}  {file} "
            f"({size:.2f} MB)"
        )

print("\n================================================")
print("STEP 88 COMPLETE")
print("================================================")
```

```python
# ============================================================
# STEP 89 — FINAL FIXED PACKAGE + VERIFIED ZIP
# ============================================================

import os
import shutil
import json
from google.colab import files

DRIVE_PACKAGE = (
    "/content/drive/MyDrive/Bitcoin P2P Dataset/"
    "BIT-TRACE-P2P-GitHub"
)

LOCAL_PACKAGE = "/content/BIT-TRACE-P2P-GitHub"
ZIP_PATH = "/content/BIT-TRACE-P2P-GitHub-FINAL.zip"

# ------------------------------------------------------------
# 1. CHECK DRIVE PACKAGE
# ------------------------------------------------------------

print("Checking Google Drive package...")

if not os.path.exists(DRIVE_PACKAGE):
    raise FileNotFoundError(
        "BIT-TRACE-P2P-GitHub package not found on Drive."
    )

# ------------------------------------------------------------
# 2. SHOW EVERYTHING CURRENTLY ON DRIVE
# ------------------------------------------------------------

print("\nCurrent Drive package contents:\n")

for root, dirs, files_in_dir in os.walk(DRIVE_PACKAGE):

    relative = os.path.relpath(
        root,
        DRIVE_PACKAGE
    )

    print(
        f"\n[{relative}]"
    )

    for f in files_in_dir:
        path = os.path.join(root, f)

        print(
            f"  {f} "
            f"({os.path.getsize(path)/1024**2:.2f} MB)"
        )

# ------------------------------------------------------------
# 3. REMOVE OLD LOCAL COPY
# ------------------------------------------------------------

if os.path.exists(LOCAL_PACKAGE):
    shutil.rmtree(LOCAL_PACKAGE)

if os.path.exists(ZIP_PATH):
    os.remove(ZIP_PATH)

# ------------------------------------------------------------
# 4. COPY DRIVE PACKAGE TO LOCAL COLAB STORAGE
# ------------------------------------------------------------

print("\nCopying package from Drive to local storage...")

shutil.copytree(
    DRIVE_PACKAGE,
    LOCAL_PACKAGE
)

print("Local copy created.")

# ------------------------------------------------------------
# 5. CHECK REQUIRED FILES
# ------------------------------------------------------------

required_files = [

    "model/p2p_anomaly_detector.joblib",
    "model/isolation_forest.joblib",
    "model/scaler.joblib",
    "model/feature_config.json",
    "model/model_metadata.json",
    "model/model_card.json",

    "inference/p2p_detector.py",
    "inference/test_inference.py",

    "results/evaluation_report.json",
    "results/reproducibility.json",
    "results/topk_evaluation.parquet",
    "results/explainable_candidates.parquet",

    "requirements.txt",
    "README.md"
]

print("\n================================================")
print("REQUIRED FILE CHECK")
print("================================================")

missing = []

for relative_path in required_files:

    full_path = os.path.join(
        LOCAL_PACKAGE,
        relative_path
    )

    if os.path.exists(full_path):

        size = (
            os.path.getsize(full_path)
            / 1024**2
        )

        print(
            f"OK      {relative_path:<55}"
            f"{size:.2f} MB"
        )

    else:

        print(
            f"MISSING {relative_path}"
        )

        missing.append(
            relative_path
        )

# ------------------------------------------------------------
# STOP IF PACKAGE IS INCOMPLETE
# ------------------------------------------------------------

if missing:

    print("\n================================================")
    print("PACKAGE INCOMPLETE")
    print("================================================")

    print(
        "\nMissing files:"
    )

    for f in missing:
        print(
            " -",
            f
        )

    raise RuntimeError(
        "Package is incomplete. ZIP was NOT created."
    )

# ------------------------------------------------------------
# 6. ZIP LOCAL COPY
# ------------------------------------------------------------

print("\n================================================")
print("CREATING VERIFIED ZIP")
print("================================================")

shutil.make_archive(
    "/content/BIT-TRACE-P2P-GitHub-FINAL",
    "zip",
    root_dir="/content",
    base_dir="BIT-TRACE-P2P-GitHub"
)

# ------------------------------------------------------------
# 7. VERIFY ZIP SIZE
# ------------------------------------------------------------

zip_size = (
    os.path.getsize(ZIP_PATH)
    / 1024**2
)

print(
    f"\nZIP size: {zip_size:.2f} MB"
)

# ------------------------------------------------------------
# 8. VERIFY ZIP CONTENTS
# ------------------------------------------------------------

import zipfile

print("\n================================================")
print("ZIP CONTENT VERIFICATION")
print("================================================")

with zipfile.ZipFile(
    ZIP_PATH,
    "r"
) as z:

    zip_files = z.namelist()

    print(
        "Files inside ZIP:",
        len(zip_files)
    )

    for required in required_files:

        zip_path = (
            "BIT-TRACE-P2P-GitHub/"
            + required
        )

        if zip_path in zip_files:
            print(
                "OK     ",
                required
            )

        else:
            print(
                "MISSING",
                required
            )

# ------------------------------------------------------------
# 9. DOWNLOAD
# ------------------------------------------------------------

print("\n================================================")
print("DOWNLOAD STARTING")
print("================================================")

files.download(
    ZIP_PATH
)

print("\n================================================")
print("STEP 89 COMPLETE")
print("================================================")
```
